# Project 14: AI Railway Track Monitoring & Defect Detection
### Vibration Spectral Telemetry & Real-Time Anomaly Scoring

Monitoring railway track health to trigger preventative maintenance before structural failure.

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest

# Simulate 1,000 km track vibration sensor points
np.random.seed(42)
n_points = 1000
track_km = np.linspace(0, 100, n_points)
vibration_g = np.random.normal(0.45, 0.08, n_points)
temperature_c = np.random.normal(28, 4, n_points)

# Inject track defects at specific locations
defect_indices = [150, 420, 890]
for idx in defect_indices:
    vibration_g[idx] += 1.8 # Severe vibration spike
    temperature_c[idx] += 15 # Friction overheating

df_track = pd.DataFrame({"KM": track_km, "Vibration_G": vibration_g, "Temp_C": temperature_c})
iso = IsolationForest(contamination=0.01, random_state=42)
df_track["Anomaly_Score"] = iso.fit_predict(df_track[["Vibration_G", "Temp_C"]])

flagged = df_track[df_track["Anomaly_Score"] == -1]
print("Railway Anomaly Detection Summary:")
print(f"Total track points scanned: {n_points}")
print(f"Defects flagged: {len(flagged)}")
print("\nFlagged Critical Track Locations:")
for _, row in flagged.iterrows():
    print(f"-> Alert at KM {row['KM']:.2f}: Vibration={row['Vibration_G']:.2f}g, Temp={row['Temp_C']:.1f}C")
```
