### Automated Material Sorting & Smart Municipal Routing

Classifying discarded items and scheduling dynamic collection routes based on bin fill telemetry.

```python
waste_categories = ["Organic", "Plastic", "Paper", "Metal", "Glass", "E-Waste"]
bin_telemetry = [
    {"bin_id": "B-101", "location": "Market Square", "fill_level": 0.92, "category": "Organic"},
    {"bin_id": "B-102", "location": "Tech Park", "fill_level": 0.45, "category": "Paper"},
    {"bin_id": "B-103", "location": "University Campus", "fill_level": 0.88, "category": "Plastic"},
    {"bin_id": "B-104", "location": "Railway Station", "fill_level": 0.95, "category": "Mixed"},
    {"bin_id": "B-105", "location": "Residential Zone", "fill_level": 0.30, "category": "Organic"}
]

# Identify bins exceeding 80% threshold for immediate dispatch
urgent_pickup = [b for b in bin_telemetry if b["fill_level"] >= 0.80]
urgent_pickup.sort(key=lambda x: x["fill_level"], reverse=True)

print("Optimized Collection Dispatch Queue (Threshold >= 80%):")
for idx, b in enumerate(urgent_pickup, start=1):
    print(f"Stop #{idx}: Bin {b['bin_id']} ({b['location']}) - Fill: {b['fill_level']:.0%} [{b['category']}]")
```
