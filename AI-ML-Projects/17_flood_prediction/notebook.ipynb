# Project 17: Urban Flood Prediction & Risk Mapping
### Hydrological Risk Assessment & Early Disaster Warning

Evaluating rainfall and watershed topography to issue tiered flood warning alerts.

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import GradientBoostingClassifier

# Generate synthetic watershed risk data (Rainfall_mm, Elevation_m, Drainage_Score)
np.random.seed(42)
n_zones = 500
rainfall = np.random.uniform(10, 250, n_zones)
elevation = np.random.uniform(2, 80, n_zones)
drainage = np.random.uniform(1, 10, n_zones)

# High risk formula
risk_prob = (rainfall * 0.004) - (elevation * 0.02) - (drainage * 0.05) + 0.3
y_flood = (risk_prob > 0.5).astype(int)

model = GradientBoostingClassifier(n_estimators=50, random_state=42)
model.fit(np.column_stack([rainfall, elevation, drainage]), y_flood)

# Test 3 municipal zones
zones_test = np.array([
    [210, 5, 2.0],  # Heavy rain, low elevation, poor drainage -> High Risk
    [45, 65, 8.5],  # Low rain, high elevation -> Safe
    [130, 25, 5.0]  # Moderate
])
probs = model.predict_proba(zones_test)[:, 1]
for idx, p in enumerate(probs):
    alert = "RED: SEVERE INUNDATION" if p > 0.75 else "YELLOW: ADVISORY" if p > 0.40 else "GREEN: SAFE"
    print(f"Zone {idx + 1}: Flood Probability = {p:.1%} [{alert}]")
```
