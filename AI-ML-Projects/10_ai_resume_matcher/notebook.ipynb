# Project 10: AI Resume & Job Matching Platform
### Skill Extraction, Semantic Proximity & Gap Analysis

Matching candidate profiles to job requisitions with transparent gap identification.

```python
def analyze_resume_fit(resume_skills: set, job_skills: set, experience_years: float, min_exp: float):
    matched_skills = resume_skills & job_skills
    missing_skills = job_skills - resume_skills
    skill_score = len(matched_skills) / len(job_skills)
    exp_score = min(1.0, experience_years / min_exp)
    composite_score = (0.7 * skill_score) + (0.3 * exp_score)
    
    return {
        "match_percentage": f"{composite_score:.1%}",
        "matched_skills": sorted(list(matched_skills)),
        "skill_gaps": sorted(list(missing_skills)),
        "recommendation": "STRONG FIT" if composite_score >= 0.75 else "CONSIDER" if composite_score >= 0.50 else "NOT RECOMMENDED"
    }

job_reqs = {"Python", "PyTorch", "Docker", "Kubernetes", "ChromaDB", "FastAPI"}
candidate_skills = {"Python", "PyTorch", "Docker", "FastAPI", "SQL"}

report = analyze_resume_fit(candidate_skills, job_reqs, experience_years=4.5, min_exp=3.0)
print("Candidate Talent Fit Assessment:")
for k, v in report.items():
    print(f"- {k}: {v}")
```
