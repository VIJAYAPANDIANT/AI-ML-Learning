# Project 08: Smart Traffic Vehicle Detection & Counting
### Real-Time Object Detection, Tracking & Traffic Congestion Analytics

Implementing bounding box detection, centroid tracking, and traffic density estimation.

```python
import numpy as np

class TrafficFlowAnalyzer:
    def __init__(self, count_line_y: int = 300):
        self.count_line_y = count_line_y
        self.total_counted = 0
        
    def process_frame(self, detections: list):
        """detections: list of (x, y, w, h, class_name)"""
        frame_counts = {"car": 0, "truck": 0, "bus": 0}
        for x, y, w, h, cname in detections:
            cy = y + (h // 2)
            # Check if centroid crosses detection line
            if abs(cy - self.count_line_y) < 15:
                self.total_counted += 1
            frame_counts[cname] = frame_counts.get(cname, 0) + 1
            
        density = "HEAVY" if len(detections) > 10 else "MODERATE" if len(detections) > 4 else "LIGHT"
        return {"frame_vehicles": len(detections), "density": density, "cumulative_count": self.total_counted}

analyzer = TrafficFlowAnalyzer(count_line_y=300)
mock_detections = [
    (120, 290, 80, 50, "car"),
    (240, 295, 120, 70, "truck"),
    (450, 180, 75, 45, "car"),
    (520, 285, 90, 60, "bus"),
    (600, 310, 80, 50, "car")
]
stats = analyzer.process_frame(mock_detections)
print("Traffic Analytics Frame Telemetry:")
for k, v in stats.items():
    print(f"- {k}: {v}")
```
