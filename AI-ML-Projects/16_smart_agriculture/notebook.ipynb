# Project 16: Smart Agriculture Recommendation System
### Precision Crop Recommendation & Agronomic Yield Modeling

Empowering farmers with data-driven crop selection and yield estimation.

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# Sample training data for crop recommendation
crops = ["Rice", "Maize", "Chickpea", "Coffee", "Cotton"]
# N, P, K, Temp, Humidity, pH, Rainfall
training_profiles = np.array([
    [80, 40, 40, 26, 82, 6.5, 220], # Rice (high water, warm)
    [70, 45, 20, 24, 65, 6.2, 100], # Maize
    [30, 60, 80, 18, 16, 7.0, 75],  # Chickpea
    [100, 30, 30, 25, 55, 6.8, 160],# Coffee
    [120, 50, 20, 24, 75, 7.2, 80]  # Cotton
])

clf = RandomForestClassifier(n_estimators=50, random_state=42)
clf.fit(training_profiles, crops)

# Test query: High rainfall, warm temperature field
field_sensor = np.array([[82, 38, 42, 25.5, 80, 6.4, 210]])
recommended_crop = clf.predict(field_sensor)[0]
print(f"Optimal Crop Recommendation for field conditions: '{recommended_crop}'")
```
