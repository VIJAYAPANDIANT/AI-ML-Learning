### Text-to-Pandas Operations & Automated Narrative Analytics

Translating non-technical executive queries into verified DataFrame transformations and statistical summaries.

```python
import pandas as pd
import numpy as np

# Sample retail transaction dataset
df_biz = pd.DataFrame({
    "Region": ["North", "North", "South", "East", "West", "East", "South"],
    "Category": ["Tech", "Furniture", "Tech", "Office", "Tech", "Furniture", "Tech"],
    "Revenue": [45000, 18000, 62000, 12000, 51000, 22000, 48000],
    "Profit": [14000, 3200, 21000, 2500, 17500, 4100, 16200]
})

def natural_language_query_engine(df: pd.DataFrame, intent: str):
    if "top region by revenue" in intent.lower():
        res = df.groupby("Region")["Revenue"].sum().sort_values(ascending=False)
        top_r = res.index[0]
        return f"The top performing region is '{top_r}' with ${res.iloc[0]:,.2f} total revenue."
    elif "profit margin by category" in intent.lower():
        grouped = df.groupby("Category")[["Revenue", "Profit"]].sum()
        grouped["Margin"] = (grouped["Profit"] / grouped["Revenue"]).map("{:.1%}".format)
        return grouped
    return "Query intent not recognized."

print("Natural Language BI Queries:")
print("Q1: Which is the top region by revenue?")
print("A1:", natural_language_query_engine(df_biz, "top region by revenue"))
print("\nQ2: What is the profit margin by category?")
print(natural_language_query_engine(df_biz, "profit margin by category"))
```
