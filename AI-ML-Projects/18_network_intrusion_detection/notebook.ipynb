# Project 18: Network Intrusion Detection System (NIDS)
### Telemetry Packet Classification & Cyber Threat Mitigation

Classifying malicious network traffic signatures to safeguard enterprise infrastructure.

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

# Synthetic packet features: duration, src_bytes, dst_bytes, count, srv_count
np.random.seed(42)
n_packets = 1500
normal_packets = np.random.normal([10, 300, 500, 5, 5], [2, 50, 80, 2, 2], (1200, 5)).clip(0)
dos_packets = np.random.normal([0.1, 10, 0, 150, 150], [0.05, 5, 0, 20, 20], (200, 5)).clip(0)
probe_packets = np.random.normal([1, 0, 0, 45, 1], [0.5, 0, 0, 10, 1], (100, 5)).clip(0)

X = np.vstack([normal_packets, dos_packets, probe_packets])
y = ["Normal"] * 1200 + ["DoS_Attack"] * 200 + ["Port_Scan_Probe"] * 100

clf = RandomForestClassifier(n_estimators=50, random_state=42)
clf.fit(X, y)

test_traffic = np.array([
    [8, 290, 480, 4, 4],    # Normal connection
    [0.05, 5, 0, 160, 160], # DoS Syn-Flood
    [1.2, 0, 0, 50, 1]      # Port Probe
])
preds = clf.predict(test_traffic)
for idx, p in enumerate(preds):
    print(f"Packet Stream #{idx + 1}: Classification Result = '{p}'")
```
