# Project 03: Customer Segmentation & RFM Clustering
### Unsupervised Cohort Discovery with K-Means & PCA

Discovering behavioral customer segments to guide tailored retention campaigns.

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Generate synthetic RFM dataset for 1,200 retail customers
np.random.seed(42)
n_cust = 1200
recency = np.random.exponential(45, n_cust).clip(1, 365) # Days since last purchase
frequency = np.random.poisson(6, n_cust).clip(1, 40)      # Orders per year
monetary = frequency * np.random.normal(85, 20, n_cust).clip(20, 500) # Total spend ($)

df_rfm = pd.DataFrame({"Recency": recency, "Frequency": frequency, "Monetary": monetary})
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_rfm)

# Benchmark K = 2 to 5
for k in [2, 3, 4, 5]:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    print(f"K = {k} Clusters | Silhouette Score: {score:.4f}")

# Fit optimal K=4
final_km = KMeans(n_clusters=4, random_state=42, n_init=10)
df_rfm["Cluster"] = final_km.fit_predict(X_scaled)
print("\nCluster Centers Summary:")
print(df_rfm.groupby("Cluster").mean().round(1))
```
