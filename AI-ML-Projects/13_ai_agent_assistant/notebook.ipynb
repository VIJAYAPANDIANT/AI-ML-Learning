# Project 13: Autonomous AI Task Assistant
### Multi-Step ReAct Planning, Tool Calling & Human-in-the-Loop Governance

Building an autonomous assistant capable of executing complex workflows safely.

```python
class AutonomousTaskAgent:
    def __init__(self):
        self.tools = {
            "fetch_user_balance": lambda uid: f"User {uid} balance: $4,500.00",
            "transfer_funds": lambda from_id, to_id, amt: f"Transferred ${amt} from {from_id} to {to_id}"
        }
        self.logs = []
        
    def execute_plan(self, task: str):
        self.logs.append(f"[PLANNER] Decomposing task: {task}")
        # Step 1: Read-only tool
        self.logs.append("[STEP 1] Executing tool: fetch_user_balance(uid=101)")
        bal = self.tools["fetch_user_balance"](101)
        self.logs.append(f"[OBSERVATION 1] {bal}")
        
        # Step 2: Destructive action -> Human approval
        self.logs.append("[STEP 2] Preparing tool: transfer_funds(from=101, to=202, amt=200)")
        self.logs.append("[HITL GATE] Approval requested from human supervisor: APPROVED")
        tx = self.tools["transfer_funds"](101, 202, 200)
        self.logs.append(f"[OBSERVATION 2] {tx}")
        self.logs.append("[FINAL] Task executed successfully.")
        return "\n".join(self.logs)

agent = AutonomousTaskAgent()
trajectory = agent.execute_plan("Transfer $200 from account 101 to account 202.")
print(trajectory)
```
