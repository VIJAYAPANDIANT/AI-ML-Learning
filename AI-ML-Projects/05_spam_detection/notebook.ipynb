# Project 05: Spam & Phishing Detection System
### Text Feature Extraction & Probabilistic Classification

Protecting enterprise communication channels with automated spam filtering and URL risk scoring.

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report

# Sample corpus of emails
texts = [
    "URGENT: Your account has been suspended! Click http://secure-login.xyz to verify credentials.",
    "Congratulations, you won a free $1000 Walmart gift card! Claim immediately.",
    "Meeting agenda for tomorrow's quarterly product roadmap review attached.",
    "Please review the updated pull request on GitHub when you have a moment.",
    "Exclusive deal on luxury watches! 90% discount for the next 24 hours only.",
    "Can you send over the updated sales projections spreadsheet for Q3?"
]
labels = [1, 1, 0, 0, 1, 0] # 1: Spam, 0: Ham

pipe = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2))),
    ("nb", MultinomialNB(alpha=0.1))
])

pipe.fit(texts, labels)
test_emails = [
    "Hey team, don't forget our 2pm standup meeting.",
    "Claim your lottery winnings right now at http://win-big.tk"
]
preds = pipe.predict(test_emails)
probs = pipe.predict_proba(test_emails)[:, 1]

for email, pred, p in zip(test_emails, preds, probs):
    tag = "SPAM" if pred == 1 else "HAM"
    print(f"[{tag}] (Spam Prob: {p:.2%}): \"{email}\"")
```
