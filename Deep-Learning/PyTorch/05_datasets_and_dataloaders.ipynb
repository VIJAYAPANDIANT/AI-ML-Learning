# 05. PyTorch Datasets & DataLoaders

PyTorch decouples dataset storage from model training loops via two primitives:
1. **`torch.utils.data.Dataset`**: Custom dataset class implementing `__len__()` and `__getitem__()`.
2. **`torch.utils.data.DataLoader`**: Wraps a Dataset with multi-process batching, shuffling, and memory pinning for GPUs.

```python
import torch
from torch.utils.data import Dataset, DataLoader

class SyntheticTabularDataset(Dataset):
    def __init__(self, n_samples=500, n_features=6):
        self.X = torch.randn(n_samples, n_features)
        self.y = (self.X[:, 0] + self.X[:, 1] > 0).float().unsqueeze(1)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

dataset = SyntheticTabularDataset(n_samples=500, n_features=6)
loader = DataLoader(dataset, batch_size=32, shuffle=True)

first_batch_x, first_batch_y = next(iter(loader))
print(f"Batch X Shape: {first_batch_x.shape}, Batch y Shape: {first_batch_y.shape}")
```
