# 10. Model Evaluation in PyTorch

Evaluating a trained PyTorch model requires:
1. Setting `model.eval()` to freeze Dropout and BatchNorm layers
2. Wrapping evaluation loops with `with torch.no_grad():` to disable gradient computation graphs and save memory
3. Computing accuracy and confusion metrics over holdout batches

```python
import torch

model.eval()
X_test = torch.randn(100, 4)
y_test = (X_test[:, 0] * 2 + X_test[:, 1] > 0).long()

correct = 0
total = len(y_test)

with torch.no_grad():
    logits = model(X_test)
    predictions = torch.argmax(logits, dim=1)
    correct = (predictions == y_test).sum().item()

acc = (correct / total) * 100
print(f"Holdout Test Accuracy: {acc:.2f}% ({correct}/{total} correct)")
```
