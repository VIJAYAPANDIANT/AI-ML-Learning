# 04. Automatic Differentiation with PyTorch Autograd

**Autograd** is PyTorch's automatic differentiation engine. It records all operations executed on tensors with `requires_grad=True` into a Directed Acyclic Graph (DAG) whose leaves are input tensors and root is the output.

### Core Mechanics:
- `tensor.requires_grad = True`: Tracks computation history
- `loss.backward()`: Computes vector-Jacobian products backward through the graph
- `tensor.grad`: Stores accumulated partial derivatives $\frac{\partial \text{loss}}{\partial \text{tensor}}$
- `torch.no_grad()`: Disables gradient tracking during inference to save memory

```python
import torch

# Define leaf tensors
x = torch.tensor(3.0, requires_grad=True)
w = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)

# Forward pass: y = w * x + b
y = w * x + b
loss = y ** 2

# Backward pass
loss.backward()

print("dy/dw = x = 3.0, d(loss)/dw = 2*y*x = 2*(7)*3 = 42.0")
print("Computed w.grad: ", w.grad.item())
print("Computed x.grad: ", x.grad.item())
print("Computed b.grad: ", b.grad.item())
```
