# 08. PyTorch Optimizers & Learning Rate Schedulers

The `torch.optim` package provides state-of-the-art optimization algorithms that update model weights using their accumulated gradients.

### Common Optimizers:
- `optim.SGD(model.parameters(), lr=0.01, momentum=0.9, weight_decay=1e-4)`
- `optim.Adam(model.parameters(), lr=0.001, betas=(0.9, 0.999))`
- `optim.AdamW(model.parameters(), lr=0.001)`: Corrected weight decay decoupling

### Learning Rate Scheduling:
- `optim.lr_scheduler.CosineAnnealingLR`
- `optim.lr_scheduler.ReduceLROnPlateau`

```python
import torch
import torch.nn as nn
import torch.optim as optim

model = nn.Linear(5, 1)
optimizer = optim.AdamW(model.parameters(), lr=0.01, weight_decay=1e-3)
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=5, gamma=0.5)

print("Initial LR:", optimizer.param_groups[0]['lr'])
for epoch in range(10):
    optimizer.step()
    scheduler.step()
print("LR after 10 epochs:", optimizer.param_groups[0]['lr'])
```
