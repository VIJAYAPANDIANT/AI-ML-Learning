# 11. Convolutional Neural Networks (CNNs) in PyTorch

CNNs in PyTorch use **`nn.Conv2d`** and **`nn.MaxPool2d`** layers.

### Tensor Shape Conventions in PyTorch:
PyTorch expects image tensors in **NCHW** format:
$$\text{Batch Size } (N) \times \text{Channels } (C) \times \text{Height } (H) \times \text{Width } (W)$$
*(Note: TensorFlow defaults to NHWC)*

```python
import torch
import torch.nn as nn

class ConvNet(nn.Module):
    def __init__(self, in_channels=1, num_classes=10):
        super(ConvNet, self).__init__()
        self.features = nn.Sequential(
            nn.Conv2d(in_channels, 32, kernel_size=3, padding=1), # (N, 32, 28, 28)
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2, 2), # (N, 32, 14, 14)
            
            nn.Conv2d(32, 64, kernel_size=3, padding=1),  # (N, 64, 14, 14)
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2, 2)  # (N, 64, 7, 7)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64 * 7 * 7, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        return self.classifier(x)

cnn = ConvNet()
x_img = torch.randn(4, 1, 28, 28)
out = cnn(x_img)
print("CNN Output Logits shape:", out.shape)
```
