# 07. PyTorch Loss Functions

PyTorch provides loss criteria in `torch.nn`.

### Critical PyTorch Nuance: CrossEntropyLoss
- **`nn.CrossEntropyLoss`**: Combines `nn.LogSoftmax()` and `nn.NLLLoss()` in a single mathematically stable class. **Do NOT apply Softmax to final logits before CrossEntropyLoss!**
- **`nn.BCEWithLogitsLoss`**: Combines Sigmoid activation and Binary Cross Entropy into one class, avoiding numerical overflow.

```python
import torch
import torch.nn as nn

# Multi-class criterion
criterion_multi = nn.CrossEntropyLoss()
# Raw unnormalized logits (Batch=3, Classes=4)
raw_logits = torch.randn(3, 4)
target_classes = torch.tensor([0, 2, 1])  # 1D class indices
loss_multi = criterion_multi(raw_logits, target_classes)
print("CrossEntropyLoss value:", loss_multi.item())

# Binary criterion with logits
criterion_bce = nn.BCEWithLogitsLoss()
raw_binary_logit = torch.tensor([[1.2], [-0.5], [2.1]])
binary_target = torch.tensor([[1.0], [0.0], [1.0]])
loss_bce = criterion_bce(raw_binary_logit, binary_target)
print("BCEWithLogitsLoss value:", loss_bce.item())
```
