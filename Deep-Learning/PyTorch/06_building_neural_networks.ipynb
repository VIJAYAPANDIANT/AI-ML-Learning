# 06. Building Neural Networks with torch.nn.Module

In PyTorch, custom neural networks subclass **`torch.nn.Module`**.

### The Anatomy of an nn.Module:
- `__init__(self)`: Instantiates sub-layers (e.g., `nn.Linear`, `nn.Dropout`, `nn.ReLU`).
- `forward(self, x)`: Defines computation flow executed when calling `model(x)`.
- Automatically tracks trainable parameters accessible via `model.parameters()`.

```python
import torch
import torch.nn as nn

class MultiLayerPerceptron(nn.Module):
    def __init__(self, in_features, hidden_dim, out_classes):
        super(MultiLayerPerceptron, self).__init__()
        self.network = nn.Sequential(
            nn.Linear(in_features, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.ReLU(),
            nn.Linear(hidden_dim // 2, out_classes)
        )

    def forward(self, x):
        return self.network(x)

mlp = MultiLayerPerceptron(in_features=20, hidden_dim=64, out_classes=2)
print(mlp)
total_params = sum(p.numel() for p in mlp.parameters() if p.requires_grad)
print(f"Trainable Parameters: {total_params:,}")
```
