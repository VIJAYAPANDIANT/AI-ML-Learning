# 02. PyTorch Tensors

A **Tensor** is a specialized data structure that is very similar to arrays and matrices, used to encode inputs, outputs, and model parameters.

### Creation Methods:
- From Python lists: `torch.tensor([1, 2, 3])`
- From NumPy ndarrays: `torch.from_numpy(arr)` (zero-copy memory share)
- Initialization routines: `torch.zeros()`, `torch.ones()`, `torch.randn()`, `torch.arange()`

```python
import torch
import numpy as np

# Creating tensors
t1 = torch.tensor([[1.0, 2.0], [3.0, 4.0]], dtype=torch.float32)
t_rand = torch.randn(2, 3)
t_ones = torch.ones(3, 3)

print("Shape:   ", t1.shape)
print("Dtype:   ", t1.dtype)
print("Device:  ", t1.device)

# Zero-copy memory bridge with NumPy
np_arr = np.array([10, 20, 30])
t_from_np = torch.from_numpy(np_arr)
np_arr[0] = 999
print("Memory Shared Mutation:", t_from_np)
```
