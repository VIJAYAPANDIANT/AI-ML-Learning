# 03. PyTorch Tensor Operations

PyTorch supports over 200 mathematical tensor operations including indexing, slicing, reshaping, matrix multiplications, and in-place mutations.

### Key Operations:
- Matrix Multiplication: `torch.matmul(A, B)` or `A @ B`
- In-place Operations: Postfixed with an underscore (e.g., `t.add_()`, `t.zero_()`). Saves memory, but mutates autograd history!
- Reshaping: `t.view()` (contiguous memory required) vs `t.reshape()`

```python
import torch

A = torch.randn(3, 4)
B = torch.randn(4, 2)
C = A @ B  # Matrix multiplication (3, 2)
print("Matrix Product shape:", C.shape)

# Reshaping and flattening
t_flat = A.view(-1)
print("Flattened tensor shape:", t_flat.shape)

# Broadcasting
row_bias = torch.tensor([1.0, 2.0, 3.0, 4.0])
A_biased = A + row_bias
print("Broadcasted Addition shape:", A_biased.shape)
```
