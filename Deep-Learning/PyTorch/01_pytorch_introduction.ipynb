# 01. Introduction to PyTorch

**PyTorch** is an open-source machine learning framework developed primarily by Meta AI Research. It has become the dominant framework across AI academic research, LLM development, and modern production deployments.

### Why PyTorch Dominates AI Research:
- **Pythonic & Intuitive**: Operates like NumPy with seamless autograd integration and standard Python debugging.
- **Dynamic Computation Graphs (Define-by-Run)**: Dynamic execution enables flexible input dimensions and dynamic control flow loops.
- **Native CUDA Acceleration**: Seamless multi-GPU computing with torch.cuda and distributed training (FSDP/DDP).

### Learning Objectives:
- Verify PyTorch environment and CUDA GPU acceleration
- Understand CPU vs GPU device transfers via `torch.device`

```python
import torch

print("PyTorch Version:", torch.__version__)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Active Device: {device}")
if torch.cuda.is_available():
    print("GPU Name:", torch.cuda.get_device_name(0))
    print("CUDA Capability:", torch.cuda.get_device_capability(0))
```
