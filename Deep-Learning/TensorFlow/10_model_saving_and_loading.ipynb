# 10. Model Persistence: Saving & Loading

Serializing trained models to disk is necessary for production serving and reproducible deployments.

### Keras Persistence Formats:
- **Keras V3 format (`.keras`)**: The recommended modern native format storing architecture, weights, and optimizer state in a single zip archive.
- **SavedModel format**: TensorFlow standard folder format for C++, TF Serving, and mobile.
- **Weights-only (`.weights.h5`)**: Saves only learnable parameters without graph architecture.

```python
import tensorflow as tf
import tempfile
import os

# Save model in native .keras format
tmp_path = os.path.join(tempfile.gettempdir(), 'my_keras_model.keras')
model.save(tmp_path)
print(f"Model saved successfully to: {tmp_path}")

# Reload model
reloaded_model = tf.keras.models.load_model(tmp_path)
print("Model reloaded successfully! Trainable params:", reloaded_model.count_params())
```
