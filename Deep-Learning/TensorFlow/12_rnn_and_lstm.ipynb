# 12. Recurrent Neural Networks (RNN & LSTM) in TensorFlow

Recurrent neural architectures process sequential data (time-series, natural language) by maintaining an internal hidden recurrence state: $\mathbf{h}_t = \tanh(\mathbf{W} \mathbf{x}_t + \mathbf{U} \mathbf{h}_{t-1} + \mathbf{b})$.

### LSTM Gating Architecture:
Long Short-Term Memory networks solve vanishing gradients across long sequences via three adaptive gates:
1. **Forget Gate ($f_t$)**: Decides what information to discard from cell state
2. **Input Gate ($i_t$)**: Decides what new information to store in cell state
3. **Output Gate ($o_t$)**: Regulates which parts of the cell state constitute the hidden output

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Sequence of 50 timesteps with 8 features each
seq_model = models.Sequential([
    layers.Input(shape=(50, 8)),
    layers.LSTM(64, return_sequences=True, name='lstm_1'),
    layers.Dropout(0.2),
    layers.LSTM(32, return_sequences=False, name='lstm_2'),
    layers.Dense(1, activation='linear', name='forecast_output')
])

seq_model.summary()
```
