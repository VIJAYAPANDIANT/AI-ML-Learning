# 11. Convolutional Neural Networks (CNNs) in TensorFlow

Convolutional Neural Networks preserve spatial pixel hierarchies using learnable **convolution kernels** and **pooling layers**.

### Key Layers:
- `layers.Conv2D(filters, kernel_size, activation)`: Spatial 2D feature detector
- `layers.MaxPooling2D(pool_size)`: Downsamples spatial resolution, creating translation invariance
- `layers.Flatten()`: Converts 2D feature maps to 1D vector for dense classification heads

```python
import tensorflow as tf
from tensorflow.keras import layers, models

cnn_model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, kernel_size=(3, 3), activation='relu', name='conv1'),
    layers.MaxPooling2D(pool_size=(2, 2), name='pool1'),
    layers.Conv2D(64, kernel_size=(3, 3), activation='relu', name='conv2'),
    layers.MaxPooling2D(pool_size=(2, 2), name='pool2'),
    layers.Flatten(name='flatten'),
    layers.Dropout(0.3, name='dropout'),
    layers.Dense(64, activation='relu', name='dense'),
    layers.Dense(10, activation='softmax', name='output')
])

cnn_model.summary()
```
