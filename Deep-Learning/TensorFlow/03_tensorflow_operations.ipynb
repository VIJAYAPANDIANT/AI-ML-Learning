# 03. TensorFlow Operations & Broadcasting

TensorFlow provides high-throughput vectorized mathematical routines optimized for GPU execution and automatic broadcasting.

### Common Tensor Operations:
- Arithmetic: `tf.add`, `tf.subtract`, `tf.multiply`, `tf.divide`
- Matrix Multiplication: `tf.matmul` or the `@` operator
- Reductions: `tf.reduce_mean`, `tf.reduce_sum`, `tf.argmax`
- Reshaping: `tf.reshape`, `tf.squeeze`, `tf.expand_dims`

```python
import tensorflow as tf

A = tf.constant([[1.0, 2.0], [3.0, 4.0]])
B = tf.constant([[5.0, 6.0], [7.0, 8.0]])

# Matrix multiplication
C = tf.matmul(A, B)
print("Matrix Product A @ B:\n", C.numpy())

# Broadcasting vector addition
v = tf.constant([10.0, 20.0])
print("Broadcasted Addition:\n", (A + v).numpy())

# Reductions across axis
print("Row-wise mean:", tf.reduce_mean(A, axis=1).numpy())
print("Overall sum:   ", tf.reduce_sum(A).numpy())
```
