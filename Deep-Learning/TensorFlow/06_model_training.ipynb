# 06. Model Compilation & Training

Before training a Keras model, you must **compile** it by declaring:
1. **Optimizer**: Algorithm to update weights (`'adam'`, `'rmsprop'`, `'sgd'`)
2. **Loss Function**: Objective to minimize (`'binary_crossentropy'`, `'categorical_crossentropy'`, `'mse'`)
3. **Metrics**: Performance metrics tracked during training (`['accuracy']`, `[tf.keras.metrics.AUC()]`)

Training is executed via `model.fit(X_train, y_train, epochs=..., batch_size=..., validation_data=...)`.

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

# Generate synthetic binary dataset
np.random.seed(42)
X_train = np.random.randn(800, 10)
y_train = (X_train[:, 0] + X_train[:, 1] > 0).astype(int)
X_val = np.random.randn(200, 10)
y_val = (X_val[:, 0] + X_val[:, 1] > 0).astype(int)

model = models.Sequential([
    layers.Input(shape=(10,)),
    layers.Dense(32, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

history = model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_val, y_val), verbose=1)
```

## 🎯 Exercises & Practice
1. Plot the training loss vs validation loss from `history.history` using Matplotlib.

```python
# Exercise Solution
import matplotlib.pyplot as plt
plt.figure(figsize=(6, 3), dpi=100)
plt.plot(history.history['loss'], label='Train Loss', lw=2)
plt.plot(history.history['val_loss'], label='Val Loss', lw=2)
plt.title('Training & Validation Loss Convergence')
plt.xlabel('Epoch')
plt.ylabel('BCE Loss')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
