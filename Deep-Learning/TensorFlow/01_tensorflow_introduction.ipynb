# 01. Introduction to TensorFlow & Keras

**TensorFlow** is an open-source end-to-end platform for machine learning and deep learning developed by Google Brain. **Keras** is the official high-level, human-centric API tightly integrated into TensorFlow 2.x as `tf.keras`.

### Core Advantages of TensorFlow 2.x:
- **Eager Execution by Default**: Immediate evaluation of tensor operations without constructing static computation graphs manually.
- **Production Ecosystem**: Complete ecosystem spanning TF Serving (production deployment), TFLite (edge/mobile), TF.js (web browsers), and TensorBoard (visualization).
- **High-Performance Acceleration**: Automated hardware distribution across multi-GPU and TPU clusters via `tf.distribute.Strategy`.

### Learning Objectives:
- Verify TensorFlow installation and GPU availability
- Understand the TensorFlow execution model

```python
import tensorflow as tf

print("TensorFlow Version:", tf.__version__)
print("Eager Execution Enabled:", tf.executing_eagerly())

gpus = tf.config.list_physical_devices('GPU')
print(f"GPUs Detected: {len(gpus)}")
for gpu in gpus:
    print("  Device:", gpu)
```
