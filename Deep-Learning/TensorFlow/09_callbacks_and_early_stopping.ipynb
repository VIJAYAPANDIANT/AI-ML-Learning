# 09. Keras Callbacks & Early Stopping

**Callbacks** are objects that can perform actions at various stages of training (at epoch end, batch begin, etc.).

### Essential Callbacks:
1. **EarlyStopping**: Halts training when validation metric stops improving, restoring optimal weights.
2. **ModelCheckpoint**: Automatically saves model weights when validation loss reaches a new minimum.
3. **ReduceLROnPlateau**: Dynamically decays the learning rate when training plateaus.

```python
import tensorflow as tf
from tensorflow.keras import layers, models, callbacks

model_cb = models.Sequential([
    layers.Input(shape=(10,)),
    layers.Dense(64, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])
model_cb.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

es = callbacks.EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=2,
    verbose=1
)

print("Configured Callbacks: EarlyStopping & ReduceLROnPlateau")
```
