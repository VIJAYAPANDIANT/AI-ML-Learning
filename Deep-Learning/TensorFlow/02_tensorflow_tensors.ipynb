# 02. TensorFlow Tensors

A **Tensor** is a multi-dimensional array of elements with a uniform data type (`dtype`). Tensors are immutable in shape and value once instantiated.

### Tensor Types:
- **Constants (`tf.constant`)**: Immutable tensor values.
- **Variables (`tf.Variable`)**: Mutable tensors used to store and update trainable weights during backpropagation.
- **Ranks**: Scalar (Rank 0), Vector (Rank 1), Matrix (Rank 2), 3D/4D Tensor (Rank 3/4).

```python
import tensorflow as tf

# Scalar, Vector, Matrix
scalar = tf.constant(7, dtype=tf.int32)
vector = tf.constant([1.5, 2.8, 3.1], dtype=tf.float32)
matrix = tf.constant([[1, 2], [3, 4]], dtype=tf.float32)

print("Scalar Rank:", scalar.ndim, "Shape:", scalar.shape)
print("Vector Rank:", vector.ndim, "Shape:", vector.shape)
print("Matrix Rank:", matrix.ndim, "Shape:", matrix.shape)

# Trainable Variable
w = tf.Variable([[0.5], [0.8]], dtype=tf.float32, name='weight')
print("Variable Before:", w.numpy())
w.assign_add([[0.1], [-0.2]])
print("Variable After assign_add:", w.numpy())
```
