# 05. Building Neural Networks: Sequential vs Functional API

While `Sequential` is ideal for plain feedforward pipelines, non-linear topologies (multi-input, multi-output, residual skip connections) require the **Keras Functional API**.

### Functional API Mechanics:
Layers are callable objects that take tensors as inputs and return tensors as outputs: `x = layers.Dense(...)(inputs)`.

### Learning Objectives:
- Construct a Multi-Layer Perceptron using the Functional API
- Understand residual skip connections ($x + \text{layer}(x)$)

```python
import tensorflow as tf
from tensorflow.keras import layers, Model

# Functional API Model
inputs = layers.Input(shape=(30,), name='feature_input')
x = layers.Dense(64, activation='relu')(inputs)
x = layers.BatchNormalization()(x)

# Residual block
residual = x
x = layers.Dense(64, activation='relu')(x)
x = layers.add([x, residual])  # Skip connection

outputs = layers.Dense(3, activation='softmax', name='class_probs')(x)

functional_model = Model(inputs=inputs, outputs=outputs, name='Residual_MLP')
functional_model.summary()
```
