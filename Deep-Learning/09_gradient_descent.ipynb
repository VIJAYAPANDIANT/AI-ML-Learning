# 09. Gradient Descent Optimization

**Gradient Descent** is the core iterative optimization algorithm used to minimize the cost function $J(\mathbf{w})$ by updating weights in the opposite direction of the gradient $\nabla J(\mathbf{w})$.

### Weight Update Formula:
$$\mathbf{w}^{(t+1)} = \mathbf{w}^{(t)} - \eta \nabla J(\mathbf{w}^{(t)})$$
where $\eta$ is the **learning rate**.

### The Three Variants of Gradient Descent:
1. **Batch Gradient Descent**: Computes gradients across the entire training dataset. Exact gradients, but slow on large datasets.
2. **Stochastic Gradient Descent (SGD)**: Updates weights using one sample at a time. Fast, noisy trajectories, escapes shallow local minima.
3. **Mini-Batch Gradient Descent**: Computes gradients over small subsets (e.g., 32, 64, 128 samples). Combines vectorization efficiency with stochastic regularization.

```python
import numpy as np
import matplotlib.pyplot as plt

# Objective function: J(w) = w^2
w = 4.0
lr = 0.2
trajectory = [w]

for epoch in range(15):
    grad = 2 * w  # Analytical derivative dJ/dw = 2w
    w = w - lr * grad
    trajectory.append(w)

plt.figure(figsize=(7, 3.5), dpi=100)
plt.plot(trajectory, marker='o', color='#2980b9', lw=2)
plt.title("Gradient Descent Convergence to Minimum (w = 0)", fontweight='bold')
plt.xlabel("Iteration")
plt.ylabel("Weight Value (w)")
plt.grid(True, alpha=0.3)
plt.show()
print(f"Final optimized weight: {w:.6f}")
```
