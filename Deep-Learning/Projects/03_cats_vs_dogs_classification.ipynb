# Project 03: Binary Real-World Image Classification: Cats vs. Dogs

### Objective:
Build a binary image classification pipeline to distinguish feline vs canine images, demonstrating transfer learning, data augmentation, and threshold calibration.

### Pipeline Stages:
1. Preprocessing and resizing heterogeneous input photographs to $128 \times 128$
2. Data Augmentation to prevent overfitting on photographic backgrounds
3. Pretrained feature extractor backbone
4. Sigmoid classification head with binary cross entropy

```python
import torch
import torch.nn as nn

class CatDogClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, 3, stride=2, padding=1), nn.BatchNorm2d(32), nn.ReLU(),
            nn.Conv2d(32, 64, 3, stride=2, padding=1), nn.BatchNorm2d(64), nn.ReLU(),
            nn.Conv2d(64, 128, 3, stride=2, padding=1), nn.BatchNorm2d(128), nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1))
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.4),
            nn.Linear(128, 1)
        )
    def forward(self, x):
        return self.classifier(self.features(x))

cat_dog_net = CatDogClassifier()
photo_batch = torch.randn(4, 3, 128, 128)
logits = cat_dog_net(photo_batch)
probs = torch.sigmoid(logits)
print("Predicted Probabilities (Dog=1, Cat=0):\n", probs.detach().numpy())
```
