# Project 02: Multi-Class Color Image Classification (CIFAR-10)

### Objective:
Construct a Deep ResNet-inspired Convolutional Network for 3-channel RGB color image classification across 10 object classes (airplanes, automobiles, birds, cats, deer, dogs, frogs, horses, ships, trucks).

### Architectural Highlights:
- 3-Channel RGB Input $(3 \times 32 \times 32)$
- Spatial Data Augmentation (Random Horizontal Flips, Random Crops)
- Batch Normalization after each convolution
- Global Average Pooling (`AdaptiveAvgPool2d`) prior to dense classification

```python
import torch
import torch.nn as nn

class ColorImageNet(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.backbone = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1), nn.BatchNorm2d(32), nn.ReLU(),
            nn.Conv2d(32, 64, 3, padding=1), nn.BatchNorm2d(64), nn.ReLU(),
            nn.MaxPool2d(2, 2),
            nn.Conv2d(64, 128, 3, padding=1), nn.BatchNorm2d(128), nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1))
        )
        self.head = nn.Linear(128, num_classes)

    def forward(self, x):
        feat = self.backbone(x)
        return self.head(feat.squeeze())

model_cifar = ColorImageNet()
rgb_sample = torch.randn(8, 3, 32, 32)
out = model_cifar(rgb_sample)
print("Color Image Classifier Output shape:", out.shape)
```
