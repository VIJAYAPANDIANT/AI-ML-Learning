# 06. Activation Functions

Activation functions introduce **non-linearity** into neural networks. Without non-linear activations, an arbitrarily deep neural network collapses mathematically into a single linear regression matrix product: $\mathbf{W}_2 (\mathbf{W}_1 \mathbf{x}) = (\mathbf{W}_2 \mathbf{W}_1) \mathbf{x} = \mathbf{W}^* \mathbf{x}$.

### Primary Activation Functions:
1. **Sigmoid**: $\sigma(z) = \frac{1}{1 + e^{-z}} \in (0, 1)$. Vanishing gradient problem for $|z| > 4$.
2. **Tanh**: $\tanh(z) = \frac{e^z - e^{-z}}{e^z + e^{-z}} \in (-1, 1)$. Zero-centered.
3. **ReLU (Rectified Linear Unit)**: $\text{ReLU}(z) = \max(0, z)$. Fast gradient $\frac{\partial}{\partial z} = 1$ for $z > 0$; solves vanishing gradients.
4. **Leaky ReLU**: $\max(\alpha z, z)$ with $\alpha \approx 0.01$. Prevents dying ReLU neurons.
5. **Softmax**: $\frac{e^{z_i}}{\sum_j e^{z_j}}$. Converts raw logits into normalized categorical probability distributions.

```python
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-5, 5, 200)

sigmoid = 1 / (1 + np.exp(-z))
tanh = np.tanh(z)
relu = np.maximum(0, z)
leaky_relu = np.where(z > 0, z, 0.1 * z)

fig, axes = plt.subplots(1, 4, figsize=(14, 3.5), dpi=100)
plots = [(sigmoid, 'Sigmoid (0, 1)', '#2980b9'), 
         (tanh, 'Tanh (-1, 1)', '#e67e22'), 
         (relu, 'ReLU [0, inf)', '#27ae60'), 
         (leaky_relu, 'Leaky ReLU', '#8e44ad')]

for ax, (data, title, col) in zip(axes, plots):
    ax.plot(z, data, color=col, lw=2.5)
    ax.set_title(title, fontweight='bold')
    ax.grid(True, alpha=0.3)
    ax.axhline(0, color='black', lw=0.8, ls='--')
    ax.axvline(0, color='black', lw=0.8, ls='--')
plt.tight_layout()
plt.show()
```

## 🎯 Exercises & Practice
1. Implement a numerically stable Softmax function in NumPy that subtracts $\max(z)$ from logits before exponentiation to prevent overflow.

```python
# Exercise Solution
def softmax_stable(z):
    exp_z = np.exp(z - np.max(z, axis=-1, keepdims=True))
    return exp_z / np.sum(exp_z, axis=-1, keepdims=True)

logits = np.array([1000.0, 1002.0, 999.0])
print("Softmax Probabilities:", softmax_stable(logits))
```
