# 11. Deep Learning Optimizers

Vanilla Stochastic Gradient Descent struggles in ravines, saddle points, and pathological curvature where gradients osciliate wildly. Advanced optimizers incorporate **momentum** and **adaptive per-parameter learning rates**.

### Evolutionary Hierarchy of Optimizers:
1. **SGD with Momentum**: Accelerates in directions of consistent gradient and dampens oscillations:
   $$\mathbf{v}_t = \beta \mathbf{v}_{t-1} + (1 - \beta) \mathbf{g}_t, \quad \mathbf{w}_{t+1} = \mathbf{w}_t - \eta \mathbf{v}_t$$
2. **RMSprop (Hinton, 2012)**: Adapts learning rate inversely proportional to the moving average of squared gradients:
   $$\mathbf{s}_t = \beta \mathbf{s}_{t-1} + (1 - \beta) \mathbf{g}_t^2, \quad \mathbf{w}_{t+1} = \mathbf{w}_t - \frac{\eta}{\sqrt{\mathbf{s}_t + \epsilon}} \odot \mathbf{g}_t$$
3. **Adam (Kingma & Ba, 2014)**: Combines 1st-moment momentum and 2nd-moment adaptive scales with bias correction.

```python
import numpy as np

class AdamOptimizer:
    def __init__(self, lr=0.001, beta1=0.9, beta2=0.999, eps=1e-8):
        self.lr = lr
        self.beta1 = beta1
        self.beta2 = beta2
        self.eps = eps
        self.m = None
        self.v = None
        self.t = 0

    def update(self, w, grad):
        if self.m is None:
            self.m = np.zeros_like(w)
            self.v = np.zeros_like(w)
        self.t += 1
        self.m = self.beta1 * self.m + (1 - self.beta1) * grad
        self.v = self.beta2 * self.v + (1 - self.beta2) * (grad ** 2)
        # Bias correction
        m_hat = self.m / (1 - self.beta1 ** self.t)
        v_hat = self.v / (1 - self.beta2 ** self.t)
        return w - self.lr * m_hat / (np.sqrt(v_hat) + self.eps)

opt = AdamOptimizer(lr=0.05)
w = np.array([5.0, -3.0])
for _ in range(20):
    grad = 2 * w  # For J(w) = w1^2 + w2^2
    w = opt.update(w, grad)
print("Adam Optimized weights:", w.round(4))
```
