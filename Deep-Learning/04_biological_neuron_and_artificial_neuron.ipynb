# 04. Biological Neuron vs. Artificial Neuron

The fundamental inspiration for artificial neural networks stems from the biological nervous system, although artificial models are mathematical abstractions rather than literal biological simulations.

### Structural Analogies:

| Biological Structure | Artificial Analog | Mathematical Role |
| :--- | :--- | :--- |
| **Dendrites** | Input Vector ($\mathbf{x}$) | Receives incoming feature signals |
| **Synapses** | Weights ($\mathbf{w}$) | Modulates signal strength / conductivity |
| **Soma (Cell Body)** | Linear Combiner ($\sum w_i x_i + b$) | Summates incoming weighted electrical potentials |
| **Action Potential Threshold** | Activation Function ($\sigma(z)$) | Determines firing rate and non-linear output |
| **Axon** | Output Signal ($y$) | Propagates signal to subsequent connected neurons |

### Learning Objectives:
- Understand the biological roots of McCulloch-Pitts and Rosenblatt neurons
- Implement a mathematical single-neuron forward pass from scratch

```python
import numpy as np

def artificial_neuron(inputs, weights, bias, activation_fn):
    # Soma aggregation
    z = np.dot(weights, inputs) + bias
    # Axon action potential firing
    output = activation_fn(z)
    return output

# Biological inputs (action potentials)
x = np.array([1.2, 0.5, -0.8])
# Synaptic strengths
w = np.array([0.4, 0.8, -0.3])
b = 0.1

# Step activation function (threshold firing)
step_fn = lambda z: 1 if z >= 0 else 0
sigmoid_fn = lambda z: 1 / (1 + np.exp(-z))

print("Linear potential (z):       ", np.dot(w, x) + b)
print("Neuron Fired (Step Fn):    ", artificial_neuron(x, w, b, step_fn))
print("Firing Rate Prob (Sigmoid):", round(artificial_neuron(x, w, b, sigmoid_fn), 4))
```
