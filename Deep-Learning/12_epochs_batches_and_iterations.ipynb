# 12. Epochs, Batches, and Iterations

Training deep neural networks requires partitioning training data into manageable subsets to fit GPU memory and achieve stochastic gradient regularization.

### Key Definitions:
- **Epoch**: Exactly **one complete forward and backward pass** through all training samples in the dataset.
- **Batch Size ($B$)**: The number of training samples processed in one forward/backward pass before updating weights (e.g., 32, 64, 128).
- **Iteration (Step)**: A single weight update step.

$$\text{Iterations per Epoch} = \left\lceil \frac{\text{Total Dataset Size } N}{\text{Batch Size } B} \right\rceil$$

### Learning Objectives:
- Construct an iterable mini-batch data generator in pure Python/NumPy
- Understand the memory and convergence trade-offs of batch size selection

```python
import numpy as np

def get_mini_batches(X, y, batch_size=32, shuffle=True):
    n_samples = len(X)
    indices = np.arange(n_samples)
    if shuffle:
        np.random.shuffle(indices)
    
    for start_idx in range(0, n_samples, batch_size):
        batch_idx = indices[start_idx:start_idx + batch_size]
        yield X[batch_idx], y[batch_idx]

# Simulate dataset with 1,000 samples
X_dummy = np.random.randn(1000, 10)
y_dummy = np.random.randint(0, 2, 1000)

batch_size = 64
steps = len(list(get_mini_batches(X_dummy, y_dummy, batch_size)))
print(f"Dataset Size: {len(X_dummy)}")
print(f"Batch Size:   {batch_size}")
print(f"Steps / Iterations per Epoch: {steps}")
print(f"Total Steps across 10 Epochs: {steps * 10}")
```

## 🎯 Exercises & Practice
1. If a dataset has 50,000 samples and batch size is 128, how many updates are performed over 25 epochs? Calculate programmatically.

```python
# Exercise Solution
import math
N = 50000
B = 128
epochs = 25
steps_per_epoch = math.ceil(N / B)
total_updates = steps_per_epoch * epochs
print(f"Steps per epoch: {steps_per_epoch}, Total updates: {total_updates:,}")
```
