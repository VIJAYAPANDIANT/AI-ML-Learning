# 07. Forward Propagation

**Forward Propagation** is the sequential computation through the computational graph from input features to output predictions.

### Step-by-Step Flow:
1. Input layer receives feature batch $\mathbf{X} \in \mathbb{R}^{B \times d}$.
2. Hidden layer 1 computes pre-activation: $\mathbf{Z}^{[1]} = \mathbf{X} \mathbf{W}^{[1]} + \mathbf{b}^{[1]}$.
3. Hidden layer 1 applies activation: $\mathbf{A}^{[1]} = \text{ReLU}(\mathbf{Z}^{[1]})$.
4. Output layer computes logits: $\mathbf{Z}^{[2]} = \mathbf{A}^{[1]} \mathbf{W}^{[2]} + \mathbf{b}^{[2]}$.
5. Output activation produces predictions: $\hat{\mathbf{Y}} = \sigma(\mathbf{Z}^{[2]})$.

```python
import numpy as np

def relu(Z):
    return np.maximum(0, Z)

def sigmoid(Z):
    return 1 / (1 + np.exp(-Z))

# Batch of 3 samples with 4 features each
X = np.array([
    [0.5, 1.2, 0.3, 0.8],
    [0.9, 0.1, 1.5, 0.2],
    [0.2, 0.8, 0.4, 1.1]
])

np.random.seed(42)
W1 = np.random.randn(4, 3) * 0.5
b1 = np.zeros((1, 3))
W2 = np.random.randn(3, 1) * 0.5
b2 = np.zeros((1, 1))

# Forward Propagation
Z1 = np.dot(X, W1) + b1
A1 = relu(Z1)
Z2 = np.dot(A1, W2) + b2
Y_hat = sigmoid(Z2)

print("Hidden Activations A1 shape:", A1.shape)
print("Predicted Probabilities Y_hat:\n", Y_hat)
```
