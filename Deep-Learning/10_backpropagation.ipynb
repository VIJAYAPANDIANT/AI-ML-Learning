# 10. Backpropagation & The Chain Rule

**Backpropagation** (Rumelhart, Hinton, & Williams, 1986) computes the analytical gradient of the scalar loss with respect to every weight in a deep neural network by systematically applying the **multivariate calculus chain rule** backward through the computational graph.

### Derivation for 2-Layer Perceptron:
Given $\mathbf{a}^{[1]} = \text{ReLU}(\mathbf{z}^{[1]})$, $\hat{y} = \sigma(\mathbf{z}^{[2]})$, and Cross-Entropy loss $\mathcal{L}$:
$$\delta^{[2]} = \frac{\partial \mathcal{L}}{\partial \mathbf{z}^{[2]}} = (\hat{y} - y)$$
$$\frac{\partial \mathcal{L}}{\partial \mathbf{W}^{[2]}} = (\mathbf{a}^{[1]})^T \delta^{[2]}$$
$$\delta^{[1]} = \frac{\partial \mathcal{L}}{\partial \mathbf{z}^{[1]}} = (\delta^{[2]} (\mathbf{W}^{[2]})^T) \odot \text{ReLU}'(\mathbf{z}^{[1]})$$
$$\frac{\partial \mathcal{L}}{\partial \mathbf{W}^{[1]}} = \mathbf{X}^T \delta^{[1]}$$

```python
import numpy as np

# 2-Layer Neural Network Backpropagation from Scratch
np.random.seed(42)
X = np.array([[0.5, 0.8], [0.1, 0.3]])
y = np.array([[1.0], [0.0]])

W1 = np.random.randn(2, 3) * 0.5
W2 = np.random.randn(3, 1) * 0.5

# Forward
Z1 = np.dot(X, W1)
A1 = np.maximum(0, Z1)
Z2 = np.dot(A1, W2)
y_hat = 1 / (1 + np.exp(-Z2))

# Backward
dZ2 = y_hat - y
dW2 = np.dot(A1.T, dZ2) / len(X)
dZ1 = np.dot(dZ2, W2.T) * (Z1 > 0)
dW1 = np.dot(X.T, dZ1) / len(X)

print("Gradient dW2:\n", dW2)
print("Gradient dW1:\n", dW1)
```
