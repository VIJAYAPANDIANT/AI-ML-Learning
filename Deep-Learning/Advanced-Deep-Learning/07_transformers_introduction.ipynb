# 07. Introduction to Transformers

The **Transformer** architecture replaces recurrent loops entirely with **Multi-Head Self-Attention** and **Positional Encodings**, enabling 100% parallel training across sequence tokens.

### The Transformer Block Anatomy:
1. Multi-Head Self-Attention (captures relationships across multiple subspace perspectives)
2. Add & Layer Normalization (Residual Highway)
3. Position-Wise Feed-Forward Network (FFN: Linear $\to$ GELU $\to$ Linear)
4. Add & Layer Normalization

```python
import torch
import torch.nn as nn

# Using PyTorch native TransformerEncoderLayer
encoder_layer = nn.TransformerEncoderLayer(d_model=64, nhead=4, dim_feedforward=128, batch_first=True)
transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=2)

# Sequence of 10 tokens with embedding dim 64
seq_tokens = torch.randn(2, 10, 64)
out = transformer_encoder(seq_tokens)
print("Transformer Output Shape:", out.shape)
print("Trainable Transformer Params:", sum(p.numel() for p in transformer_encoder.parameters()):,)
```

1. Why do Transformers require Positional Encodings, whereas RNNs do not?

```python
# Exercise Solution
# Answer: Attention is permutation-equivariant (bag-of-words without order). 
# RNNs process sequentially step-by-step, naturally encoding order.
# Transformers require explicit positional embeddings added to token vectors to encode word order!
```
