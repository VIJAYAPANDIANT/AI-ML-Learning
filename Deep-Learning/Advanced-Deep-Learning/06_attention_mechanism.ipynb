# 06. The Attention Mechanism

Introduced by Bahdanau et al. (2014) and revolutionized in "Attention Is All You Need" (Vaswani et al., 2017), the **Attention Mechanism** allows a model to dynamically focus on relevant parts of the input sequence regardless of distance, solving the fixed-length vector bottleneck of RNNs.

### Scaled Dot-Product Attention Formula:
$$\text{Attention}(\mathbf{Q}, \mathbf{K}, \mathbf{V}) = \text{softmax}\left( \frac{\mathbf{Q} \mathbf{K}^T}{\sqrt{d_k}} \right) \mathbf{V}$$
where:
- $\mathbf{Q}$ (Query): What am I looking for?
- $\mathbf{K}$ (Key): What information do you contain?
- $\mathbf{V}$ (Value): What is the actual content payload?

```python
import torch
import math

def scaled_dot_product_attention(Q, K, V, mask=None):
    d_k = Q.size(-1)
    # Compute similarity scores
    scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(d_k)
    if mask is not None:
        scores = scores.masked_fill(mask == 0, -1e9)
    # Attention weights
    weights = torch.softmax(scores, dim=-1)
    # Weighted sum of values
    output = torch.matmul(weights, V)
    return output, weights

# Batch=1, SeqLen=4, Dim=8
Q = torch.randn(1, 4, 8)
K = torch.randn(1, 4, 8)
V = torch.randn(1, 4, 8)

out, attn_weights = scaled_dot_product_attention(Q, K, V)
print("Attention Output Shape: ", out.shape)
print("Attention Weights (Row Sum=1):\n", attn_weights[0].detach().numpy().round(3))
```
