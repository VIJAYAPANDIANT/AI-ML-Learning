# 09. Deep Learning Regularization Techniques

Deep neural networks possess millions to billions of parameters, making them prone to memorizing training noise. **Regularization** ensures robust generalization to unseen distributions.

### Taxonomy of Regularization:
1. **Dropout (Srivastava et al., 2014)**: Randomly zeros activations during forward pass, breaking co-adaptation.
2. **Weight Decay ($L_2$)**: Penalizes large weight magnitudes (Ridge penalty).
3. **Batch Normalization (Ioffe & Szegedy, 2015)**: Normalizes layer inputs across batch, smoothing the loss landscape.
4. **Data Augmentation**: Artificially expands training diversity (rotations, flips, crops).
5. **Label Smoothing**: Replaces hard 0/1 targets with soft targets to prevent overconfidence.

```python
import torch
import torch.nn as nn

class RegularizedNet(nn.Module):
    def __init__(self, in_features=50, out_classes=2):
        super(RegularizedNet, self).__init__()
        self.net = nn.Sequential(
            nn.Linear(in_features, 64),
            nn.BatchNorm1d(64),          # 1. Batch Normalization
            nn.ReLU(),
            nn.Dropout(p=0.3),            # 2. Dropout
            nn.Linear(64, 32),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.Dropout(p=0.2),
            nn.Linear(32, out_classes)
        )

    def forward(self, x):
        return self.net(x)

model = RegularizedNet()
print(model)
```
