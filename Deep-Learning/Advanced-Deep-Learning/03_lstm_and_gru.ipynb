# 03. Gated Architectures: LSTM vs. GRU

To overcome vanishing gradients across long time horizons, gated architectures introduce internal highway memory cells.

### Comparison:
- **LSTM (Hochreiter & Schmidhuber, 1997)**: 3 Gates (Forget, Input, Output) + Separate Cell State ($c_t$) and Hidden State ($h_t$).
- **GRU (Cho et al., 2014)**: 2 Gates (Reset, Update) + Merged Hidden State. ~25% fewer parameters, faster training, comparable accuracy on most sequence tasks.

```python
import torch
import torch.nn as nn

lstm = nn.LSTM(input_size=32, hidden_size=64, batch_first=True)
gru = nn.GRU(input_size=32, hidden_size=64, batch_first=True)

lstm_params = sum(p.numel() for p in lstm.parameters())
gru_params = sum(p.numel() for p in gru.parameters())

print(f"LSTM Parameter Count: {lstm_params:,} (4 internal linear transformations)")
print(f"GRU Parameter Count:  {gru_params:,} (3 internal linear transformations)")
print(f"GRU Parameter Reduction: {(1 - gru_params/lstm_params)*100:.1f}% fewer weights")
```
