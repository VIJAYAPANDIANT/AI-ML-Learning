# 01. Convolutional Neural Networks (CNNs): Advanced Architectural Paradigms

CNNs revolutionize computer vision through spatial parameter sharing and equivariant representations.

### Mathematical Formulation of 2D Convolution:
$$S(i, j) = (I * K)(i, j) = \sum_{m} \sum_{n} I(i-m, j-n) K(m, n)$$

### Structural Evolution:
- **LeNet-5 (1998)**: Pioneer handwritten digit recognizer
- **AlexNet (2012)**: Deep 8-layer network with ReLU and Dropout, winning ImageNet
- **VGG (2014)**: Demonstrates that small $3 \times 3$ receptive fields stacked deeply outperform large $7 \times 7$ kernels
- **ResNet (2015)**: Residual connections solving the degradation problem up to 152+ layers

```python
import torch
import torch.nn as nn

# 3x3 Convolution parameter count comparison vs 7x7
conv_3x3_stacked = sum(p.numel() for p in nn.Sequential(
    nn.Conv2d(64, 64, kernel_size=3, padding=1),
    nn.Conv2d(64, 64, kernel_size=3, padding=1),
    nn.Conv2d(64, 64, kernel_size=3, padding=1)
).parameters())

conv_7x7 = sum(p.numel() for p in nn.Conv2d(64, 64, kernel_size=7, padding=3).parameters())

print(f"Three stacked 3x3 Conv Layers Params: {conv_3x3_stacked:,}")
print(f"Single 7x7 Conv Layer Params:        {conv_7x7:,}")
print(f"Effective Receptive Field: Both cover 7x7, but 3x3 has 3x non-linearities and fewer weights!")
```
