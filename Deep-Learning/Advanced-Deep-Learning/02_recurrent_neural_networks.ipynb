# 02. Recurrent Neural Networks (RNNs): Sequence Modeling & Vanishing Gradients

Standard feedforward networks assume independent inputs. **Recurrent Neural Networks (RNNs)** process sequential inputs $\mathbf{x}_1, \dots, \mathbf{x}_T$ by sharing transition parameters $\mathbf{W}_{hh}$ across all time steps.

### The Mathematical Bottleneck: Vanishing Gradients
When computing gradients across $T$ timesteps via Backpropagation Through Time (BPTT):
$$\frac{\partial \mathcal{L}_T}{\partial \mathbf{h}_1} = \prod_{t=2}^T \frac{\partial \mathbf{h}_t}{\partial \mathbf{h}_{t-1}} = \prod_{t=2}^T \mathbf{W}_{hh}^T \text{diag}(\sigma'(\mathbf{z}_t))$$
If largest eigenvalue of $\mathbf{W}_{hh} < 1$, gradient vanishes exponentially ($0.9^{50} \approx 0.005$); if $> 1$, gradient explodes ($1.1^{50} \approx 117.4$).

```python
import numpy as np
import matplotlib.pyplot as plt

timesteps = 50
grad_norm_vanish = [0.85 ** t for t in range(timesteps)]
grad_norm_explode = [min(1.15 ** t, 500) for t in range(timesteps)]

plt.figure(figsize=(8, 3.5), dpi=100)
plt.plot(grad_norm_vanish, label='Vanishing Gradient (Norm < 1)', color='#e74c3c', lw=2)
plt.title("The Fundamental Flaw of Vanilla RNN: Exponential Gradient Decay", fontweight='bold')
plt.xlabel("Backpropagation Time Step (T)")
plt.ylabel("Gradient Magnitude")
plt.yscale('log')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
