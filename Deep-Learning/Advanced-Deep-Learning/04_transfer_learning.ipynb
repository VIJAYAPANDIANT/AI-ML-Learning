# 04. Transfer Learning & Fine-Tuning

**Transfer Learning** repurposes a neural network pre-trained on a massive source dataset (e.g., ImageNet with 1.4M images across 1,000 categories) to a target downstream task with limited labeled data.

### Transfer Learning Strategies:
1. **Feature Extraction**: Freeze all pre-trained convolutional backbone weights (`param.requires_grad = False`) and train only a new linear classification head.
2. **Fine-Tuning**: Unfreeze the top layers of the backbone and train with a very small learning rate (e.g., $10^{-5}$) to subtly adjust pre-trained filters without catastrophic forgetting.

```python
import torch
import torch.nn as nn

# Simulated pre-trained backbone
class DummyResNetBackbone(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv = nn.Conv2d(3, 512, 3)
    def forward(self, x): return self.conv(x)

backbone = DummyResNetBackbone()

# 1. Feature Extraction: Freeze backbone
for param in backbone.parameters():
    param.requires_grad = False

# 2. Add custom downstream task head
custom_classifier = nn.Sequential(
    backbone,
    nn.AdaptiveAvgPool2d((1, 1)),
    nn.Flatten(),
    nn.Linear(512, 5) # 5 downstream target classes
)

trainable = sum(p.numel() for p in custom_classifier.parameters() if p.requires_grad)
frozen = sum(p.numel() for p in custom_classifier.parameters() if not p.requires_grad)
print(f"Frozen Backbone Params: {frozen:,}")
print(f"Trainable Head Params:  {trainable:,}")
```
