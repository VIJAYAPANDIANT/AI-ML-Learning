# 05. Autoencoders & Dimensionality Compression

An **Autoencoder** is an unsupervised neural network trained to reconstruct its input through an informational **bottleneck latent space** $\mathbf{z} \in \mathbb{R}^k$ ($k \ll d$).

### Key Components:
- **Encoder**: Compresses input features: $\mathbf{z} = f_{\theta}(\mathbf{x})$
- **Decoder**: Reconstructs original data: $\hat{\mathbf{x}} = g_{\phi}(\mathbf{z})$
- **Reconstruction Loss**: $\mathcal{L}(\mathbf{x}, \hat{\mathbf{x}}) = \|\mathbf{x} - \hat{\mathbf{x}}\|^2$

### Industrial Applications:
- Denoising noisy images / audio
- Unsupervised anomaly and fraud detection
- Non-linear dimensionality reduction and latent clustering

```python
import torch
import torch.nn as nn

class DeepAutoencoder(nn.Module):
    def __init__(self, input_dim=784, latent_dim=32):
        super(DeepAutoencoder, self).__init__()
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.ReLU(),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Linear(128, latent_dim)
        )
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 256),
            nn.ReLU(),
            nn.Linear(256, input_dim),
            nn.Sigmoid()
        )

    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

ae = DeepAutoencoder()
x_sample = torch.rand(4, 784)
reconstructed = ae(x_sample)
print("Input shape:         ", x_sample.shape)
print("Latent Bottle shape: ", ae.encoder(x_sample).shape)
print("Reconstructed shape: ", reconstructed.shape)
```
