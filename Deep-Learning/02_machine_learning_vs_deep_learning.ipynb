# 02. Machine Learning vs. Deep Learning

Understanding the fundamental distinction between classical Machine Learning and modern Deep Learning is essential for choosing the right architecture for production systems.

| Attribute | Classical Machine Learning | Deep Learning |
| :--- | :--- | :--- |
| **Feature Engineering** | Manual feature extraction (domain experts) | Learned automatically from raw data |
| **Data Requirements** | Performs well on small/medium tabular data ($10^2 - 10^5$) | Requires large data ($10^5 - 10^9+$ samples) |
| **Hardware** | Standard Multi-core CPU | GPUs / TPUs (Massive parallel tensor ops) |
| **Execution Time** | Minutes to hours to train | Hours to weeks of cluster training |
| **Interpretability** | High (Coefficients, Tree split paths) | Black-box / Post-hoc attribution (Grad-CAM) |
| **Optimal Domain** | Structured tabular databases | Unstructured data (Vision, Audio, NLP, Graphs) |

### Learning Objectives:
- Compare end-to-end learning against pipeline-based feature extraction
- Understand when to prefer XGBoost/Random Forest over Deep Neural Networks

```python
import pandas as pd

comparison_df = pd.DataFrame({
    "Criterion": ["Data Volume", "Feature Engineering", "Interpretability", "Hardware", "Dominant Modality"],
    "Classical ML": ["Small to Medium (<100K)", "Human Domain Expertise Required", "High (Tree branches, Linear weights)", "Standard CPU", "Structured Tabular Databases"],
    "Deep Learning": ["Massive (>100K to Billions)", "End-to-End Automatic Representation", "Low (Black-Box Distributed Latents)", "NVIDIA CUDA GPUs / Google TPUs", "Images, Video, Speech, Raw Text"]
})
print(comparison_df.to_string(index=False))
```
