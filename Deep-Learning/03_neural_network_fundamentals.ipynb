# 03. Neural Network Fundamentals & Architecture

An Artificial Neural Network (ANN) consists of organized layers of interconnected processing nodes (neurons). Each connection transmits a signal, modulated by a learnable **weight** ($W$) and an additive **bias** ($b$).

### Network Topology:
- **Input Layer**: Transmits feature vector $\mathbf{x} \in \mathbb{R}^d$ without computation.
- **Hidden Layer(s)**: Computes affine transformation followed by non-linear activation: $\mathbf{h} = \sigma(\mathbf{W} \mathbf{x} + \mathbf{b})$.
- **Output Layer**: Produces target predictions $\hat{\mathbf{y}}$ (logits, probabilities, or continuous scalars).

### Mathematical Formulation for a 2-Layer Feedforward Network:
$$\mathbf{z}^{[1]} = \mathbf{W}^{[1]} \mathbf{x} + \mathbf{b}^{[1]}$$
$$\mathbf{a}^{[1]} = \text{ReLU}(\mathbf{z}^{[1]})$$
$$\mathbf{z}^{[2]} = \mathbf{W}^{[2]} \mathbf{a}^{[1]} + \mathbf{b}^{[2]}$$
$$\hat{\mathbf{y}} = \sigma(\mathbf{z}^{[2]})$$

```python
import numpy as np

# Define network dimensional parameters
n_features = 4  # Input dimensionality
n_hidden = 5    # Hidden layer nodes
n_outputs = 1   # Single binary classification output

np.random.seed(42)
# Xavier/Glorot weight initialization
W1 = np.random.randn(n_hidden, n_features) * np.sqrt(2 / (n_features + n_hidden))
b1 = np.zeros((n_hidden, 1))
W2 = np.random.randn(n_outputs, n_hidden) * np.sqrt(2 / (n_hidden + n_outputs))
b2 = np.zeros((n_outputs, 1))

print("Weight Matrix W1 shape:", W1.shape)
print("Bias Vector b1 shape:   ", b1.shape)
print("Weight Matrix W2 shape:", W2.shape)
print("Total Learnable Parameters:", W1.size + b1.size + W2.size + b2.size)
```

## 🎯 Exercises & Practice
1. Calculate the total number of trainable parameters in a network with an input dimension of 784 (MNIST 28x28), a hidden layer of 128 units, a second hidden layer of 64 units, and an output layer of 10 classes.

```python
# Exercise Solution
layer1_params = 784 * 128 + 128
layer2_params = 128 * 64 + 64
output_params = 64 * 10 + 10
total_params = layer1_params + layer2_params + output_params
print(f"Layer 1: {layer1_params:,}")
print(f"Layer 2: {layer2_params:,}")
print(f"Output:  {output_params:,}")
print(f"Total Parameters: {total_params:,}")
```
