# 05. The Rosenblatt Perceptron

Invented by Frank Rosenblatt in 1957, the **Perceptron** is the earliest supervised binary classification model. It computes a linear boundary: $\hat{y} = 1$ if $\mathbf{w}^T \mathbf{x} + b \ge 0$, else $0$.

### Perceptron Learning Rule:
For every misclassified instance $(\mathbf{x}, y)$ with learning rate $\eta$:
$$\mathbf{w} \leftarrow \mathbf{w} + \eta (y - \hat{y}) \mathbf{x}$$
$$b \leftarrow b + \eta (y - \hat{y})$$

### The XOR Problem:
In 1969, Minsky & Papert proved that a single-layer perceptron can only learn **linearly separable** logic functions (AND, OR), but fundamentally cannot solve non-linear problems like **XOR** without a hidden layer!

```python
import numpy as np

class Perceptron:
    def __init__(self, lr=0.1, n_epochs=10):
        self.lr = lr
        self.n_epochs = n_epochs
        self.weights = None
        self.bias = 0.0

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0.0
        
        for _ in range(self.n_epochs):
            for xi, target in zip(X, y):
                linear_output = np.dot(xi, self.weights) + self.bias
                y_pred = 1 if linear_output >= 0 else 0
                update = self.lr * (target - y_pred)
                self.weights += update * xi
                self.bias += update
        return self

    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        return np.where(linear_output >= 0, 1, 0)

# Train Perceptron on logical AND gate
X_and = np.array([[0,0], [0,1], [1,0], [1,1]])
y_and = np.array([0, 0, 0, 1])

p = Perceptron(lr=0.1, n_epochs=10).fit(X_and, y_and)
print("AND Gate Predictions:", p.predict(X_and))
print("Trained Weights:", p.weights, "Bias:", p.bias)
```
