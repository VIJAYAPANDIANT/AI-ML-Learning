# 04. Python Operators

Comprehensive guide to Arithmetic, Comparison, Logical, Bitwise, Assignment, Identity, and Membership operators in Python.


## 1. Arithmetic Operators
`+`, `-`, `*`, `/` (float division), `//` (floor division), `%` (modulus), `**` (exponentiation)


```python
a, b = 15, 4
print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}")
print(f"{a} // {b} = {a // b}")
print(f"{a} % {b} = {a % b}")
print(f"{a} ** {b} = {a ** b}")
```

## 2. Comparison & Logical Operators
Comparison: `==`, `!=`, `>`, `<`, `>=`, `<=`
Logical: `and`, `or`, `not`


```python
x, y = 10, 20
print("x == y:", x == y)
print("x < y:", x < y)
print("(x < 15) and (y > 15):", (x < 15) and (y > 15))
print("not (x == y):", not (x == y))
```

## 3. Membership & Identity Operators
- **Membership**: `in`, `not in`
- **Identity**: `is`, `is not` (checks object identity in memory)


```python
fruits = ["apple", "banana", "cherry"]
print("banana in fruits:", "banana" in fruits)
print("grape not in fruits:", "grape" not in fruits)

list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

print("list1 == list2:", list1 == list2)  # Same content
print("list1 is list2:", list1 is list2)   # Different objects in memory
print("list1 is list3:", list1 is list3)   # Same object
```

## 4. Bitwise Operators
`&` (AND), `|` (OR), `^` (XOR), `~` (NOT), `<<` (Left shift), `>>` (Right shift)


```python
p, q = 6, 3  # 6 = 0b0110, 3 = 0b0011
print("p & q:", p & q)   # 0b0010 = 2
print("p | q:", p | q)   # 0b0111 = 7
print("p ^ q:", p ^ q)   # 0b0101 = 5
print("p << 1:", p << 1) # 12
```
