# 03. Input and Output in Python

Master reading user input using `input()` and formatting console outputs using modern Python f-strings.


## 1. Taking User Input
`input()` always returns input as a string.


```python
# Simulating user input
user_name = "Alex"
age_str = "24"

# Convert to required types
age = int(age_str)
print(f"Hello, {user_name}! Next year you will be {age + 1} years old.")
```

## 2. String Formatting Methods
Modern Python supports multiple ways to format strings. **f-strings** (formatted string literals) are fast and preferred.


```python
item = "GPU"
price = 699.99
quantity = 2

# 1. f-strings (Recommended)
print(f"Total for {quantity} x {item}: ${price * quantity:.2f}")

# 2. str.format()
print("Total for {} x {}: ${:.2f}".format(quantity, item, price * quantity))

# 3. % operator (Legacy)
print("Total for %d x %s: $%.2f" % (quantity, item, price * quantity))
```

## 3. Advanced print() Arguments
- `sep`: Separator between items
- `end`: Character printed at the end (default is `\n`)


```python
print("Python", "AI", "ML", sep=" | ")
print("Loading", end="...")
print(" Done!")
```
