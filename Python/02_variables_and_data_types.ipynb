# 02. Variables and Built-in Data Types

Learn about variables, dynamic typing, type casting, immutability, and built-in Python data types.


## 1. Variables in Python
Variables store data in memory. In Python, variables are dynamically typed (no explicit type declaration required).


```python
age = 25          # Integer
pi = 3.14159      # Float
name = "Alice"    # String
is_active = True  # Boolean
nothing = None    # NoneType

print(age, type(age))
print(pi, type(pi))
print(name, type(name))
print(is_active, type(is_active))
print(nothing, type(nothing))
```

## 2. Naming Conventions
- Must start with a letter or underscore `_`
- Cannot start with a number
- Case-sensitive (`total` vs `Total` are different)
- Recommended: `snake_case` for variables and functions.


```python
student_count = 100
_private_var = "hidden"
print(student_count, _private_var)
```

## 3. Type Conversion (Type Casting)
Explicitly converting one data type to another using `int()`, `float()`, `str()`, `bool()`.


```python
num_str = "123"
num_int = int(num_str)
num_float = float(num_str)

print(num_int, type(num_int))
print(num_float, type(num_float))

# Converting float to int truncates decimal
print(int(9.99))  # Output: 9
```

## 4. Memory Address & Immutability
Use `id()` to inspect memory address of objects.


```python
a = 5
print(f"Initial ID of a: {id(a)}")
a = a + 1
print(f"New ID of a after modification: {id(a)}")
```
