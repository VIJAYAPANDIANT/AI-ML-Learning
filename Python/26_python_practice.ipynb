## Problem 1: Word Frequency Counter
Write a function that accepts a string paragraph and returns the top 3 most frequent words.


```python
from collections import Counter
import re

def top_words(text, n=3):
    words = re.findall(r'\b\w+\b', text.lower())
    return Counter(words).most_common(n)

sample_text = "Python is great for AI. Python is easy to learn and Python is powerful."
print("Top 3 words:", top_words(sample_text))
```

## Problem 2: Matrix Transpose using List Comprehension


```python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]
transposed = [[row[i] for row in matrix] for i in range(len(matrix[0]))]
print("Original:", matrix)
print("Transposed:", transposed)
```

## Problem 3: Custom Exception and Validation
Validate password strength (must be at least 8 chars, contain uppercase, lowercase, and digit).


```python
def validate_password(pwd):
    if len(pwd) < 8:
        return False, "Length must be >= 8"
    if not any(c.isupper() for c in pwd):
        return False, "Must contain uppercase letter"
    if not any(c.islower() for c in pwd):
        return False, "Must contain lowercase letter"
    if not any(c.isdigit() for c in pwd):
        return False, "Must contain digit"
    return True, "Password valid!"

print(validate_password("Pass1234"))
print(validate_password("weak"))
```
