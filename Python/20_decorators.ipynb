# 20. Decorators & Higher-Order Functions

Master Python decorators: function wrappers, `@functools.wraps`, decorators with parameters, and real-world logging/timing decorators.


## 1. Higher-Order Functions & Closures


```python
def make_multiplier(factor):
    def multiply(number):
        return number * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)
print("Double 5:", double(5))
print("Triple 5:", triple(5))
```

## 2. Building a Custom Decorator


```python
import time
from functools import wraps

def timer_decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        print(f"[TIMER] '{func.__name__}' executed in {(end_time - start_time)*1000:.4f} ms")
        return result
    return wrapper

@timer_decorator
def compute_heavy_sum(n):
    return sum(i * i for i in range(n))

res = compute_heavy_sum(500000)
print("Result sum:", res)
```
