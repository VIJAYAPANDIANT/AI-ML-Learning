# 24. Introduction to NumPy

Introduction to high-performance array computing with NumPy (`ndarray`, operations, indexing, and linear algebra basics).


## 1. Creating NumPy Arrays


```python
import numpy as np

arr1d = np.array([1, 2, 3, 4, 5])
arr2d = np.array([[1, 2, 3], [4, 5, 6]])

print("1D Array:", arr1d)
print("2D Array:\n", arr2d)
print("Shape:", arr2d.shape)
print("Data Type:", arr2d.dtype)
```

## 2. Array Generation Functions


```python
zeros = np.zeros((2, 3))
ones = np.ones((3, 2))
arange = np.arange(0, 10, 2)
linspace = np.linspace(0, 1, 5)

print("Zeros:\n", zeros)
print("Arange:", arange)
print("Linspace:", linspace)
```

## 3. Vectorized Operations & Matrix Multiplication


```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("a + b:", a + b)
print("a * b (element-wise):", a * b)
print("Dot Product:", np.dot(a, b))
```
