# 15. Modules & Packages

Learn how to import standard modules, create custom modules, inspect `sys.path`, and structure Python packages.


## 1. Importing Built-in Modules


```python
import math
import random
from datetime import datetime

print("Square root of 64:", math.sqrt(64))
print("Pi:", math.pi)
print("Random choice:", random.choice(["A", "B", "C"]))
print("Current timestamp:", datetime.now())
```

## 2. Exploring Standard Library Highlights
- `sys`: Interpreter variables and interaction
- `os`: Operating system interface
- `pathlib`: Modern OOP file paths


```python
import sys
import os
from pathlib import Path

print("Python Version:", sys.version.split()[0])
print("Current Working Dir:", os.getcwd())
print("Home Dir:", Path.home())
```

## 3. Custom Modules & `__name__ == '__main__'`
When a python file is run directly, `__name__` is set to `'__main__'`. If imported, `__name__` is the module's file name.


```python
if __name__ == '__main__':
    print("Executed as main script!")
```
