# 11. Dictionaries & Key-Value Mapping

Master Python dictionaries, key-value lookups, methods, merging, `defaultdict`, and `Counter`.


## 1. Creating & Accessing Dictionaries


```python
student = {
    "id": 101,
    "name": "David",
    "courses": ["ML", "NLP"],
    "gpa": 3.8
}

print("Name:", student["name"])
print("GPA (using get):", student.get("gpa", 0.0))
print("Age (missing key safe get):", student.get("age", "Not Provided"))
```

## 2. Dictionary Methods & Iteration


```python
for key in student.keys():
    print("Key:", key)

for value in student.values():
    print("Value:", value)

for key, value in student.items():
    print(f"{key} -> {value}")
```

## 3. Merging Dictionaries & Collections Helpers


```python
d1 = {"a": 1, "b": 2}
d2 = {"b": 99, "c": 3}

# Merge using | operator (Python 3.9+)
merged = d1 | d2
print("Merged dict:", merged)

from collections import Counter, defaultdict

# Counter
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
word_counts = Counter(words)
print("Word Counts:", word_counts)

# defaultdict
groups = defaultdict(list)
groups["even"].extend([2, 4, 6])
print("defaultdict:", dict(groups))
```

2. Invert a dictionary (swap keys and values).


```python
text = "abracadabra"
freq = {}
for char in text:
    freq[char] = freq.get(char, 0) + 1
print("Char frequencies:", freq)
```
