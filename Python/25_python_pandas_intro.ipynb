Learn foundational tabular data processing using Pandas Series & DataFrames.


## 1. Creating Series and DataFrames


```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana"],
    "Age": [25, 30, 35, 28],
    "Score": [88.5, 92.0, 79.5, 95.0],
    "Passed": [True, True, True, True]
}

df = pd.DataFrame(data)
print(df)
```

## 2. DataFrame Inspection & Selection


```python
print("--- Info ---")
print(df.info())
print("\n--- Descriptive Stats ---")
print(df.describe())

# Selecting columns and rows
print("\nNames:", df["Name"].values)
print("\nHigh Scorers (Score > 85):")
print(df[df["Score"] > 85])
```
