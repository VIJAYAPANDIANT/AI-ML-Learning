# 08. Lists & Dynamic Sequences

Master Python lists: creation, indexing, slicing, modification, searching, sorting, and copy semantics.


## 1. Creating & Accessing Lists
Lists are ordered, mutable, and allow duplicate elements.


```python
data = [10, 20.5, "Python", True, [1, 2]]
print("List length:", len(data))
print("Element at index 2:", data[2])
print("Nested list element:", data[4][1])
```

## 2. Modifying Lists (Adding & Removing)


```python
items = ["cat", "dog"]
items.append("elephant")        # Add to end
items.insert(1, "rabbit")      # Insert at index
items.extend(["fox", "wolf"])    # Add multiple items
print("After additions:", items)

popped = items.pop(2)           # Remove by index
print("Popped item:", popped)
items.remove("fox")             # Remove by value
print("After removals:", items)
```

## 3. Sorting & Searching


```python
nums = [42, 12, 89, 5, 23]
print("Original:", nums)

# Temporary sorted copy
print("Sorted copy:", sorted(nums))

# In-place sort
nums.sort(reverse=True)
print("In-place sorted (descending):", nums)
print("Index of 42:", nums.index(42))
```

## 4. Shallow vs Deep Copying


```python
import copy

original = [1, [2, 3]]
shallow = original.copy()
deep = copy.deepcopy(original)

original[1][0] = 99
print("Original:", original)
print("Shallow copy (affected):", shallow)
print("Deep copy (unaffected):", deep)
```

2. Remove duplicates from a list while maintaining order.


```python
numbers = [15, 3, 42, 8, 23, 42, 15]

# Remove duplicates keeping order
unique_nums = []
for n in numbers:
    if n not in unique_nums:
        unique_nums.append(n)
print("Unique numbers:", unique_nums)
```
