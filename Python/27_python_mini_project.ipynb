## Project Overview: Student Grade & Performance Analytics System

This mini project demonstrates an end-to-end Python application that manages student record performance, calculates statistics, and saves data in JSON format.


```python
import json
from datetime import datetime

class Student:
    def __init__(self, student_id: int, name: str):
        self.student_id = student_id
        self.name = name
        self.scores = {}
        
    def add_score(self, subject: str, score: float):
        if 0 <= score <= 100:
            self.scores[subject] = score
        else:
            raise ValueError("Score must be between 0 and 100")
            
    def calculate_average(self) -> float:
        if not self.scores:
            return 0.0
        return sum(self.scores.values()) / len(self.scores)
        
    def to_dict(self):
        return {
            "student_id": self.student_id,
            "name": self.name,
            "scores": self.scores,
            "average": round(self.calculate_average(), 2)
        }

class GradeBook:
    def __init__(self):
        self.students = {}
        
    def add_student(self, student: Student):
        self.students[student.student_id] = student
        
    def get_class_average(self) -> float:
        if not self.students:
            return 0.0
        averages = [s.calculate_average() for s in self.students.values()]
        return sum(averages) / len(averages)
        
    def export_report(self, filepath: str):
        report = {
            "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "class_average": round(self.get_class_average(), 2),
            "students": [s.to_dict() for s in self.students.values()]
        }
        with open(filepath, "w") as f:
            json.dump(report, f, indent=4)
        print(f"Report exported to {filepath}")

# Demo run
gb = GradeBook()

s1 = Student(1, "Alice")
s1.add_score("Python", 95)
s1.add_score("Math", 90)
gb.add_student(s1)

s2 = Student(2, "Bob")
s2.add_score("Python", 82)
s2.add_score("Math", 78)
gb.add_student(s2)

print(f"Class Average: {gb.get_class_average():.2f}")
gb.export_report("gradebook_report.json")
```
