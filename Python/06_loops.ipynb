# 06. Loops & Iteration in Python

Learn `while` and `for` loops, loop control (`break`, `continue`, `pass`), `range()`, `enumerate()`, `zip()`, and `else` with loops.


## 1. for Loops and range()


```python
# Iterating over range
print("Range 1 to 5:")
for i in range(1, 6):
    print(i, end=" ")
print()

# Step argument
print("Even numbers 2 to 10:")
for i in range(2, 11, 2):
    print(i, end=" ")
print()
```

## 2. enumerate() and zip() helpers
- `enumerate()` provides both index and item.
- `zip()` combines multiple iterables in parallel.


```python
models = ["Linear Regression", "Decision Tree", "Random Forest"]
accuracies = [0.82, 0.89, 0.94]

# Enumerate
for idx, model in enumerate(models, 1):
    print(f"{idx}. {model}")

print("-----")
# Zip
for model, acc in zip(models, accuracies):
    print(f"{model}: {acc*100:.1f}%")
```

## 3. while Loop & Loop Control Statements
`break` terminates loop, `continue` skips iteration, `pass` is placeholder.


```python
count = 0
while count < 10:
    count += 1
    if count == 3:
        continue  # Skip 3
    if count == 7:
        break     # Stop loop at 7
    print(count, end=" ")
print()
```

## 4. for...else & while...else
The `else` block executes when the loop finishes normally (without encountering `break`).


```python
numbers = [2, 4, 6, 8, 9, 10]
for n in numbers:
    if n % 2 != 0:
        print(f"Found odd number: {n}")
        break
else:
    print("All numbers were even.")
```

## 🎯 Exercises
1. Write a loop to calculate the factorial of a number $N$.
2. Print the Fibonacci sequence up to 10 terms.


```python
# Factorial
N = 6
fact = 1
for i in range(1, N + 1):
    fact *= i
print(f"Factorial of {N} = {fact}")

# Fibonacci
fib = [0, 1]
for _ in range(8):
    fib.append(fib[-1] + fib[-2])
print("Fibonacci sequence:", fib)
```
