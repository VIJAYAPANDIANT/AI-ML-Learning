# 05. Conditional Statements

Master decision-making using `if`, `elif`, `else`, ternary expressions, and structural pattern matching (`match-case`).


## 1. basic if-elif-else Structure


```python
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'

print(f"Score: {score} -> Grade: {grade}")
```

## 2. Ternary Operator (Conditional Expression)
Syntax: `value_if_true if condition else value_if_false`


```python
age = 20
status = "Adult" if age >= 18 else "Minor"
print(f"Age {age}: {status}")
```

## 3. Structural Pattern Matching (`match-case` in Python 3.10+)


```python
command = "start"

match command:
    case "start":
        print("System starting...")
    case "stop":
        print("System stopping...")
    case "pause":
        print("System paused.")
    case _:
        print("Unknown command.")
```

## 🎯 Exercises
1. Write a program to check if a year is a leap year.
2. Write a BMI calculator and categorize into Underweight, Normal, Overweight, Obese.


```python
year = 2024
is_leap = (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)
print(f"Is {year} a leap year? {is_leap}")

weight, height = 70, 1.75  # kg, meters
bmi = weight / (height ** 2)
if bmi < 18.5:
    cat = "Underweight"
elif bmi < 25:
    cat = "Normal weight"
elif bmi < 30:
    cat = "Overweight"
else:
    cat = "Obese"
print(f"BMI: {bmi:.1f} ({cat})")
```
