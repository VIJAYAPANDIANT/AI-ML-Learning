# 09. Tuples & Immutable Sequences

Learn tuple packing, unpacking, immutability benefits, methods, and `namedtuple`.


## 1. Creating & Unpacking Tuples
Tuples are ordered and **immutable**.


```python
point = (10, 20, 30)
print("Tuple point:", point)

# Unpacking
x, y, z = point
print(f"x={x}, y={y}, z={z}")

# Extended Unpacking (*rest)
first, *middle, last = (1, 2, 3, 4, 5)
print(f"First: {first}, Middle: {middle}, Last: {last}")
```

## 2. Single-element Tuples & Immutability


```python
single_tuple = (42,)  # Comma is mandatory
not_a_tuple = (42)    # Just integer in parens

print(type(single_tuple), type(not_a_tuple))

# Immutability
tup = (10, [20, 30])
# tup[0] = 100 # Raises TypeError
tup[1][0] = 200 # Mutable element inside tuple CAN change
print("Modified tuple:", tup)
```

## 3. Namedtuples from collections


```python
from collections import namedtuple

Student = namedtuple("Student", ["id", "name", "gpa"])
s1 = Student(id=101, name="Sarah", gpa=3.9)
print(f"Student: {s1.name}, GPA: {s1.gpa}")
```
