## 1. Function Basics & Default Parameters


```python
def greet(name: str, greeting: str = "Hello") -> str:
    """Generates a custom greeting message."""
    return f"{greeting}, {name}!"

print(greet("Alice"))
print(greet("Bob", greeting="Welcome"))
```

## 2. Variable-length Arguments (*args and **kwargs)


```python
def compute_stats(*args, **kwargs):
    print("Positional args (*args):", args)
    print("Keyword args (**kwargs):", kwargs)
    total = sum(args)
    return total

result = compute_stats(10, 20, 30, operation="sum", author="AI Agent")
print("Result:", result)
```

## 3. Scope & LEGB Rule
LEGB = Local -> Enclosing -> Global -> Built-in


```python
global_var = 100

def outer_function():
    enclosing_var = 50
    
    def inner_function():
        nonlocal enclosing_var
        global global_var
        enclosing_var += 10
        global_var += 5
        print(f"Inner: enclosing={enclosing_var}, global={global_var}")
        
    inner_function()

outer_function()
```

## 🎯 Exercises
1. Write a function `is_prime(n)` returning True if n is prime.
2. Write a function that accepts any number of positional numerical arguments and returns their average.


```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print("Is 17 prime?", is_prime(17))
print("Is 20 prime?", is_prime(20))
```
