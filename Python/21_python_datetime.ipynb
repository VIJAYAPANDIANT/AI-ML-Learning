# 21. Python Datetime Module

Learn working with dates, times, time intervals (`timedelta`), strftime formatting, and strptime parsing.


## 1. Datetime Objects


```python
from datetime import datetime, date, time, timedelta

now = datetime.now()
print("Current datetime:", now)
print("Date:", now.date())
print("Time:", now.time())
```

## 2. Formatting & Parsing (`strftime` & `strptime`)


```python
# Formatting: datetime -> string
formatted_str = now.strftime("%Y-%m-%d %H:%M:%S")
print("Formatted:", formatted_str)

# Parsing: string -> datetime
date_str = "2026-10-03 14:30:00"
parsed_dt = datetime.strptime(date_str, "%Y-%m-%d %H:%M:%S")
print("Parsed datetime:", parsed_dt)
```

## 3. Timedelta Arithmetic


```python
future_date = now + timedelta(days=30, hours=5)
print("30 days from now:", future_date)

diff = future_date - now
print("Difference in days:", diff.days)
```
