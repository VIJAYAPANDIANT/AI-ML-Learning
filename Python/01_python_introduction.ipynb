## 1. What is Python?
Python is a high-level, interpreted, dynamically typed, and object-oriented programming language created by **Guido van Rossum** in 1991.

### Key Features:
- **Simple & Readable Syntax**: Looks almost like pseudo-code.
- **Extensive Ecosystem**: Powerful libraries like NumPy, Pandas, PyTorch, TensorFlow, Scikit-learn.
- **Cross-Platform**: Runs seamlessly on Windows, macOS, Linux.
- **Interpreted**: Code is executed line by line, making debugging easier.


```python
# First Python Program
print("Hello, World!")
print("Welcome to AI & Machine Learning with Python!")
```

## 2. Comments in Python
Comments are ignored by the Python interpreter and help document code.


```python
# Single line comment

"""
This is a multi-line comment / docstring.
It can span across multiple lines.
"""

x = 10  # Comment at the end of a statement
print(f"Value of x: {x}")
```

## 3. Python Indentation
Unlike C/C++ or Java that use curly braces `{}` to define code blocks, Python uses **indentation** (standard is 4 spaces).


```python
if True:
    print("Indented line 1")
    print("Indented line 2")

print("Outside if block")
```
