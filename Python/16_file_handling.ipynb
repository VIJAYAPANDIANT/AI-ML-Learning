# 16. File Handling & I/O

Learn reading and writing text files safely using Python's context manager (`with open(...)`), line-by-line processing, and `pathlib`.


## 1. Writing to a File


```python
filename = "sample_data.txt"

with open(filename, "w", encoding="utf-8") as f:
    f.write("Line 1: Python File Handling\n")
    f.write("Line 2: Machine Learning Data Log\n")
    f.writelines([
        "Line 3: Epoch 1 Loss 0.45\n",
        "Line 4: Epoch 2 Loss 0.28\n"
    ])

print(f"File '{filename}' written successfully.")
```

## 2. Reading from a File


```python
with open(filename, "r", encoding="utf-8") as f:
    content = f.read()
    print("--- Entire Content ---")
    print(content)

with open(filename, "r", encoding="utf-8") as f:
    lines = f.readlines()
    print("--- Readlines ---")
    for idx, line in enumerate(lines, 1):
        print(f"{idx}: {line.strip()}")
```

## 3. Cleaning Up Test File


```python
import os
if os.path.exists(filename):
    os.remove(filename)
    print(f"Removed temporary file: {filename}")
```
