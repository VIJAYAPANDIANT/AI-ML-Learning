# 13. Anonymous Lambda Functions

Learn how to create anonymous single-line lambda functions and use them with `map()`, `filter()`, `reduce()`, and sorting.


## 1. Lambda Function Syntax
Syntax: `lambda arguments: expression`


```python
square = lambda x: x ** 2
add = lambda a, b: a + b

print("Square of 5:", square(5))
print("Sum of 3 & 7:", add(3, 7))
```

## 2. Lambda with map(), filter(), reduce()


```python
from functools import reduce

nums = [1, 2, 3, 4, 5, 6]

# map(): Apply function to all items
squared = list(map(lambda x: x**2, nums))
print("Mapped (squared):", squared)

# filter(): Keep items matching condition
evens = list(filter(lambda x: x % 2 == 0, nums))
print("Filtered (evens):", evens)

# reduce(): Aggregate list into single value
product = reduce(lambda x, y: x * y, nums)
print("Reduced (product):", product)
```

## 3. Custom Sorting with Lambda


```python
students = [("Alice", 88), ("Bob", 95), ("Charlie", 78)]
# Sort by score ascending
students_sorted = sorted(students, key=lambda s: s[1], reverse=True)
print("Sorted by score:", students_sorted)
```

```python
words = ["AI", "Machine", "Deep", "Learning", "Data", "Science"]
long_words = list(filter(lambda w: len(w) > 4, words))
print("Words > 4 chars:", long_words)
```
