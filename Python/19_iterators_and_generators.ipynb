# 19. Iterators and Generators

Understand the Iterator Protocol (`__iter__`, `__next__`), `yield` keyword, Generator Expressions, and memory efficiency in processing large datasets.


## 1. Iterator Protocol


```python
nums = [10, 20, 30]
it = iter(nums)
print(next(it))
print(next(it))
print(next(it))
# next(it) # StopIteration exception
```

## 2. Generator Functions (`yield`)


```python
def fibonacci_generator(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

print("Fibonacci (first 7):", list(fibonacci_generator(7)))
```

## 3. Generator Expressions vs List Comprehension
Generator expressions use parentheses `()` instead of brackets `[]` and yield values lazily on demand.


```python
import sys

# Memory comparison
list_comp = [x for x in range(100000)]
gen_exp = (x for x in range(100000))

print(f"List Comp Size: {sys.getsizeof(list_comp)} bytes")
print(f"Gen Exp Size:   {sys.getsizeof(gen_exp)} bytes")
```
