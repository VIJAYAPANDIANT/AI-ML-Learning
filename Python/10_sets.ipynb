# 10. Sets & Set Operations

Understand unordered sets, uniqueness property, set operations (union, intersection, difference), and `frozenset`.


## 1. Creating Sets & Uniqueness


```python
nums_set = {1, 2, 2, 3, 4, 4, 4, 5}
print("Unique elements set:", nums_set)

# Empty set must be created with set(), not {}
empty_set = set()
print(type(empty_set))
```

## 2. Set Mathematical Operations


```python
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print("Union (A | B):", set_a | set_b)
print("Intersection (A & B):", set_a & set_b)
print("Difference (A - B):", set_a - set_b)
print("Symmetric Diff (A ^ B):", set_a ^ set_b)
```

## 3. Set Modification & Frozenset


```python
s = {10, 20}
s.add(30)
s.update([40, 50])
s.discard(20)  # Safe removal (doesn't throw error if missing)
print("Set after modifications:", s)

# Immutable set: frozenset
f_set = frozenset([1, 2, 3])
print("Frozenset:", f_set)
```
