Explore Python string indexing, slicing, immutability, built-in string methods, and escape sequences.


## 1. String Indexing & Slicing
Syntax: `string[start:stop:step]`


```python
text = "Artificial Intelligence"
print("First char:", text[0])
print("Last char:", text[-1])
print("First word:", text[:10])
print("Second word:", text[11:])
print("Reversed string:", text[::-1])
```

## 2. Essential String Methods


```python
msg = "  hello world of machine learning!  "
clean_msg = msg.strip()

print("Stripped:", f"'{clean_msg}'")
print("Title Case:", clean_msg.title())
print("Upper Case:", clean_msg.upper())
print("Replace:", clean_msg.replace("machine learning", "deep learning"))
print("Starts with 'hello':", clean_msg.startswith("hello"))

words = clean_msg.split()
print("Split into words:", words)
joined_str = "-".join(words)
print("Joined with hyphen:", joined_str)
```

## 3. String Immutability & Raw Strings


```python
# Strings cannot be modified in place
s = "Python"
# s[0] = 'J' # Would raise TypeError!
s = "J" + s[1:]  # Creates new string
print(s)

# Raw string (useful for regex & windows file paths)
path = r"C:\Users\Data\new_folder"
print(path)
```

## 🎯 Exercises
1. Check if a given string is a palindrome (e.g. "racecar").
2. Count occurrences of vowels in a given sentence.


```python
word = "racecar"
is_palindrome = word.lower() == word.lower()[::-1]
print(f"Is '{word}' palindrome? {is_palindrome}")

sentence = "Deep Learning and Neural Networks"
vowels = "aeiouAEIOU"
v_count = sum(1 for char in sentence if char in vowels)
print(f"Vowel count in sentence: {v_count}")
```
