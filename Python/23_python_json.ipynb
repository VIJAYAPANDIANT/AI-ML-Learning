# 23. Python JSON Processing

Learn encoding Python objects to JSON string/file (`dumps`/`dump`) and parsing JSON to Python dictionaries (`loads`/`load`).


## 1. Serialization & Deserialization


```python
import json

data = {
    "model_name": "Transformer",
    "hyperparameters": {"batch_size": 32, "lr": 0.001},
    "layers": ["embedding", "attention", "dense"],
    "trained": True
}

# Python dict -> JSON string
json_str = json.dumps(data, indent=2)
print("JSON String:")
print(json_str)

# JSON string -> Python dict
parsed_data = json.loads(json_str)
print("Parsed Model Name:", parsed_data["model_name"])
```

## 2. Reading & Writing JSON Files


```python
filename = "model_config.json"

# Write JSON file
with open(filename, "w") as f:
    json.dump(data, f, indent=4)

# Read JSON file
with open(filename, "r") as f:
    loaded_data = json.load(f)
    print("Loaded from file:", loaded_data["hyperparameters"])

# Cleanup
import os
os.remove(filename)
```
