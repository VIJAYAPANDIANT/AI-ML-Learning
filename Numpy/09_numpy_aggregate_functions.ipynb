# 09 Numpy Aggregate Functions

Summary statistics, aggregation functions, and working across axes.


## Aggregate Statistical Functions
Compute global or axis-wise statistics such as sum, mean, median, standard deviation, min, max, argmin, and argmax.

```python
import numpy as np

data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Total Sum:", np.sum(data))
print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))
print("Min:", np.min(data), "Max:", np.max(data))
print("Argmax (index of max element):", np.argmax(data))
```

### Aggregations Along Axes (`axis=0` columns, `axis=1` rows)

```python
print("Column-wise sum (axis=0):", np.sum(data, axis=0))
print("Row-wise sum (axis=1):", np.sum(data, axis=1))
print("Column-wise mean (axis=0):", np.mean(data, axis=0))
```
