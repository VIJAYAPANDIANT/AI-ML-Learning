# 13 Numpy Array Attributes

Inspecting ndarray properties: shape, ndim, size, dtype, itemsize, nbytes.


## NumPy Array Attributes
Key attributes that define the structure and memory characteristics of `ndarray` objects.

```python
import numpy as np

matrix = np.array([
    [1.5, 2.5, 3.5, 4.5],
    [5.5, 6.5, 7.5, 8.5]
], dtype='float64')

print(f"Dimensions (.ndim): {matrix.ndim}")
print(f"Shape (.shape): {matrix.shape}")
print(f"Total Elements (.size): {matrix.size}")
print(f"Data Type (.dtype): {matrix.dtype}")
print(f"Bytes per Element (.itemsize): {matrix.itemsize} bytes")
print(f"Total Memory (.nbytes): {matrix.nbytes} bytes")
```
