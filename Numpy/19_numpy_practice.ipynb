### Exercise 1: Standardize a 2D Dataset
Given a 2D matrix of shape `(5, 3)`, calculate column means and standard deviations, then z-score normalize the matrix ($Z = \frac{X - \mu}{\sigma}$).

```python
import numpy as np

np.random.seed(42)
X = np.random.randint(10, 100, size=(5, 3))
print("Original X:\n", X)

mu = np.mean(X, axis=0)
sigma = np.std(X, axis=0)

Z = (X - mu) / sigma
print("Z-score Normalized X:\n", Z)
```

### Exercise 2: Find Chessboard Pattern
Create an $8 \times 8$ 2D array with a checkerboard pattern of 0s and 1s.

```python
chessboard = np.zeros((8, 8), dtype=int)
chessboard[1::2, ::2] = 1
chessboard[::2, 1::2] = 1
print("Chessboard Pattern:\n", chessboard)
```
