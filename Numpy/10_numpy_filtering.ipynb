# 10 Numpy Filtering

Boolean indexing, masks, and conditional assignment with np.where.


## Filtering & Boolean Indexing
Filtering allows you to extract elements based on explicit logical criteria.

```python
import numpy as np

arr = np.array([12, 45, 7, 89, 23, 56, 3, 90])

mask = arr > 30
print("Boolean Mask:", mask)

filtered = arr[mask]
print("Filtered Elements (> 30):", filtered)
```

### Combining Conditions & `np.where()`

```python
combined = arr[(arr > 10) & (arr < 60)]
print("10 < x < 60:", combined)

replaced = np.where(arr > 30, 999, arr)
print("Replaced >30 with 999:\n", replaced)
```
