# 16 Numpy Sorting Searching

Sorting arrays, argsort, searchsorted, non-zero elements, and searching.


## Sorting and Searching
Efficient array sorting, index tracking (`argsort`), binary search (`searchsorted`), and finding indices (`where`).

```python
import numpy as np

arr = np.array([40, 10, 50, 20, 30])

sorted_arr = np.sort(arr)
sort_indices = np.argsort(arr)

print("Original:", arr)
print("Sorted:", sorted_arr)
print("Sort Indices:", sort_indices)
```

### Searching with `np.searchsorted()` & `np.where()`

```python
sorted_seq = np.array([10, 20, 30, 40, 50])
idx = np.searchsorted(sorted_seq, 25)
print("Insertion index for 25:", idx)

vals = np.array([0, 5, 0, 12, 0])
print("Indices of non-zero elements:", np.nonzero(vals)[0])
```
