# 18 Numpy Linear Algebra

Matrix operations: dot product, transpose, inverse, determinant, eigenvalues.


## Linear Algebra (`np.linalg`)
Linear algebra operations fundamental for Machine Learning & Neural Networks.

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

dot_product = np.dot(A, B)
print("Matrix Product A @ B:\n", dot_product)

print("Transpose of A:\n", A.T)
```

### Determinant, Inverse, and Eigenvalues

```python
det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)
eigvals, eigvecs = np.linalg.eig(A)

print(f"Determinant of A: {det_A:.2f}")
print("Inverse of A:\n", inv_A)
print("Eigenvalues:", eigvals)
```
