# 04 Numpy Reshape

Reshaping, flattening, and expanding array dimensions.


## Array Reshaping & Flattening
Reshaping means changing the shape of an array without changing its data.

```python
import numpy as np

arr = np.arange(12)
print("Original 1D:", arr)

# Reshape to 3x4 matrix
matrix_3x4 = arr.reshape(3, 4)
print("Reshaped to 3x4:\n", matrix_3x4)

# Reshape using -1 for dynamic dimension calculation
matrix_2x6 = arr.reshape(2, -1)
print("Reshaped with -1 (2x6):\n", matrix_2x6)
```

### Flattening Arrays: `.flatten()` vs `.ravel()`

```python
flat = matrix_3x4.flatten()
rav = matrix_3x4.ravel()

print("Flattened:", flat)
print("Ravel:", rav)
```
