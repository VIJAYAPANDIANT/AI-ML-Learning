# 17 Numpy Copy Vs View

Understanding differences between copies and views, memory pointers, and .base.


## Copy vs View in NumPy
- **View**: A new array object referring to the same underlying buffer data.
- **Copy**: An independent clone of the data in memory.

```python
import numpy as np

original = np.array([1, 2, 3, 4, 5])

view_arr = original[1:4]
copy_arr = original[1:4].copy()

view_arr[0] = 999
print("Original after view mutation:", original)

copy_arr[1] = 888
print("Original after copy mutation:", original)
```

### Checking Array Ownership with `.base`

```python
print("view_arr base is original:", view_arr.base is original)
print("copy_arr base is None:", copy_arr.base is None)
```
