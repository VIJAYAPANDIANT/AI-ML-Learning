# 03 Numpy Multi Dimensional Arrays

Working with 1D, 2D, 3D, and N-dimensional arrays.


## Multi-Dimensional Arrays in NumPy
Arrays can have 0 dimensions (scalars), 1 dimension (vectors), 2 dimensions (matrices), or $N$ dimensions (tensors).

```python
import numpy as np

# 1D Vector
a_1d = np.array([1, 2, 3, 4])
# 2D Matrix
a_2d = np.array([[1, 2, 3], [4, 5, 6]])
# 3D Tensor
a_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print("1D Dimensions:", a_1d.ndim, "Shape:", a_1d.shape)
print("2D Dimensions:", a_2d.ndim, "Shape:", a_2d.shape)
print("3D Dimensions:", a_3d.ndim, "Shape:", a_3d.shape)
```

### Indexing Multi-Dimensional Arrays

```python
# Accessing element in 2D array: arr[row, column]
print("Element at row 1, col 2 of 2D array:", a_2d[1, 2])

# Accessing element in 3D array: arr[depth, row, column]
print("Element at 3D [1, 0, 1]:", a_3d[1, 0, 1])
```
