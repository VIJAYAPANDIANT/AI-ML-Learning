# 11 Numpy Random Numbers

Random sampling, probability distributions, shuffling, and setting seeds.


## Random Number Generation
NumPy includes `np.random` for generating random arrays from various statistical distributions.

```python
import numpy as np

np.random.seed(42)

rand_uniform = np.random.rand(3, 3)
rand_normal = np.random.randn(3, 3)
rand_ints = np.random.randint(10, 50, size=(2, 4))

print("Uniform Random:\n", rand_uniform)
print("Standard Normal:\n", rand_normal)
print("Random Integers:\n", rand_ints)
```

### Random Choice & Shuffling

```python
items = np.array(['Red', 'Green', 'Blue', 'Yellow'])
chosen = np.random.choice(items, size=5, replace=True)
print("Random Choice:", chosen)

arr_to_shuffle = np.array([1, 2, 3, 4, 5])
np.random.shuffle(arr_to_shuffle)
print("Shuffled Array:", arr_to_shuffle)
```
