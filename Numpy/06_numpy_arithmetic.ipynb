# 06 Numpy Arithmetic

Vectorized element-wise operations and mathematical functions.


## Element-Wise Arithmetic Operations
NumPy provides vectorization, allowing operations directly across arrays without standard loops.

```python
import numpy as np

a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("Addition (a + b):", a + b)
print("Subtraction (a - b):", a - b)
print("Multiplication (a * b):", a * b)
print("Division (a / b):", a / b)
print("Exponentiation (a ** 2):", a ** 2)
```

### Universal Mathematical Functions (ufuncs)

```python
x = np.array([0, np.pi/2, np.pi])
print("Sine values:", np.sin(x))
print("Square root:", np.sqrt(np.array([1, 4, 9, 16])))
print("Exponential:", np.exp(np.array([1, 2])))
```
