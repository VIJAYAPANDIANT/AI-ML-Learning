# 02 Numpy Data Types

Understanding NumPy dtypes, type conversion, and memory layout.


## NumPy Data Types (`dtype`)
Unlike Python lists which can store mixed types, NumPy arrays are homogeneous. Common NumPy types include `int32`, `int64`, `float32`, `float64`, `bool_`, `complex128`, and `string_`.

```python
import numpy as np

# Default data types
int_arr = np.array([1, 2, 3])
float_arr = np.array([1.0, 2.5, 3.8])
str_arr = np.array(['apple', 'banana', 'cherry'])

print("int_arr dtype:", int_arr.dtype)
print("float_arr dtype:", float_arr.dtype)
print("str_arr dtype:", str_arr.dtype)
```

### Explicitly Setting Data Types & Converting with `.astype()`

```python
# Explicit dtype
arr_f32 = np.array([1, 2, 3], dtype='float32')
print("Explicit float32:", arr_f32, arr_f32.dtype)

# Converting integer array to float using astype
converted = int_arr.astype(np.float64)
print("Converted array:", converted, converted.dtype)
```
