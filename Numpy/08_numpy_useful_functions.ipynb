# 08 Numpy Useful Functions

Essential creation and utility functions like zeros, ones, arange, and linspace.


## Array Creation Utilities
NumPy provides convenient functions to quickly initialize arrays filled with specific values or sequences.

```python
import numpy as np

print("Zeros (2x3):\n", np.zeros((2, 3)))
print("Ones (3x2):\n", np.ones((3, 2)))
print("Full (2x2 filled with 7):\n", np.full((2, 2), 7))
print("Identity Matrix (3x3):\n", np.eye(3))
```

### Sequence Generators: `np.arange()` vs `np.linspace()`

```python
seq_arange = np.arange(0, 10, 2)
seq_linspace = np.linspace(0, 1, 5)

print("arange(0, 10, 2):", seq_arange)
print("linspace(0, 1, 5):", seq_linspace)
```
