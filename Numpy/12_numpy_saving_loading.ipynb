# 12 Numpy Saving Loading

Saving arrays to disk (.npy, .npz, .csv) and loading them back.


## Saving & Loading NumPy Arrays
NumPy provides binary format (`.npy`, `.npz`) and text file utilities (`savetxt`, `loadtxt`).

```python
import numpy as np
import os

data = np.array([[10, 20, 30], [40, 50, 60]])

np.save('sample_array.npy', data)
loaded_data = np.load('sample_array.npy')
print("Loaded Binary Array:\n", loaded_data)
```

### Saving & Loading Compressed `.npz` and CSV Files

```python
a = np.arange(5)
b = np.arange(10, 15)
np.savez('multiple_arrays.npz', arr_a=a, arr_b=b)

loaded_npz = np.load('multiple_arrays.npz')
print("Array A from npz:", loaded_npz['arr_a'])

if os.path.exists('sample_array.npy'): os.remove('sample_array.npy')
if os.path.exists('multiple_arrays.npz'): os.remove('multiple_arrays.npz')
```
