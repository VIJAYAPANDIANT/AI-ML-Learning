# 15 Numpy Splitting Arrays

Dividing arrays into multiple sub-arrays with split, array_split, vsplit, hsplit.


## Splitting Arrays
Splitting is the reverse of joining, breaking a single array into multiple smaller arrays.

```python
import numpy as np

arr = np.arange(1, 10)
splits = np.split(arr, 3)
print("Splitting 1D array into 3 parts:")
for idx, sub in enumerate(splits):
    print(f" Part {idx+1}:", sub)
```

### 2D Splitting: `vsplit` & `hsplit`

```python
matrix = np.arange(16).reshape(4, 4)

upper, lower = np.vsplit(matrix, 2)
print("Upper 2 rows:\n", upper)
print("Lower 2 rows:\n", lower)
```
