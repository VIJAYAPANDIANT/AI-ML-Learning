# 02. Figure and Axes Deep-Dive

Mastering the **Figure** and **Axes** relationship is the key to creating customized, multi-panel publications and advanced dashboards in Matplotlib.

### Learning Objectives:
- Understand Figure canvas dimensions, DPI (Dots Per Inch), and background colors
- Create multiple axes with `add_subplot()` and `add_axes()`
- Implement **inset plots** (zoomed-in picture-in-picture)
- Control plot margins and spines

## 1. Controlling Figure Size, DPI & Backgrounds
- `figsize=(width, height)`: Defined in inches.
- `dpi=...`: Dots per inch; higher DPI produces sharper graphics for publications (standard screen: 100, publication: 300).
- `facecolor`: Figure canvas background color.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 5, 100)
y = np.exp(x)

# High-resolution figure with custom background
fig = plt.figure(figsize=(8, 4), dpi=120, facecolor='#f4f5f7')
ax = fig.add_subplot(1, 1, 1)

ax.plot(x, y, color='#8e44ad', linewidth=2.5, label=r'$y = e^x$')
ax.set_facecolor('#ffffff') # Axes plotting area background
ax.set_title("Exponential Growth with Custom Figure Styling", fontsize=12, fontweight='bold')
ax.set_xlabel("Time (t)")
ax.set_ylabel("Growth Metric")
ax.legend()
ax.grid(True, linestyle=':', alpha=0.6)

plt.show()
```

## 2. Inset Plots using `fig.add_axes()`
`fig.add_axes([left, bottom, width, height])` allows placing an Axes anywhere on the canvas using normalized coordinates ($0.0$ to $1.0$). This is ideal for **inset magnification** of specific anomalies or peaks.

```python
fig = plt.figure(figsize=(9, 5), dpi=100)

# Master Axes spanning 80% of canvas
ax_main = fig.add_axes([0.1, 0.1, 0.8, 0.8])

x_dense = np.linspace(0, 10, 500)
y_signal = np.sin(x_dense) * np.exp(-0.2 * x_dense) + 0.05 * np.random.normal(size=500)

ax_main.plot(x_dense, y_signal, color='#2980b9', lw=1.5, label='Damped Oscillatory Signal')
ax_main.set_title("Main Signal with Zoomed Inset Area", fontsize=13, fontweight='bold')
ax_main.set_xlabel("Timestamp (ms)")
ax_main.set_ylabel("Amplitude (mV)")
ax_main.grid(True, alpha=0.4)
ax_main.legend(loc='upper right')

# Inset Axes: Zoomed view of peak at x in [1.0, 2.0]
ax_inset = fig.add_axes([0.52, 0.45, 0.35, 0.38])
ax_inset.plot(x_dense, y_signal, color='#e74c3c', lw=2)
ax_inset.set_xlim(1.0, 2.0)
ax_inset.set_ylim(0.4, 1.1)
ax_inset.set_title("Zoomed Peak [1.0s - 2.0s]", fontsize=10, fontweight='bold')
ax_inset.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

## 🎯 Exercises & Practice
1. Create a figure with a resolution of `dpi=110` and dimensions `figsize=(8, 4.5)`.
2. Plot the cubic polynomial $f(x) = x^3 - 3x^2 + 2x$ over $x \in [-1, 3]$.
3. Add an inset axes focusing on the local minimum region $x \in [1.2, 2.0]$ with an orange highlight curve.

```python
# Exercise Solution
x_poly = np.linspace(-1, 3, 200)
y_poly = x_poly**3 - 3*x_poly**2 + 2*x_poly

fig = plt.figure(figsize=(8, 4.5), dpi=110)
ax1 = fig.add_axes([0.12, 0.12, 0.78, 0.78])
ax1.plot(x_poly, y_poly, color='#2c3e50', lw=2.2, label=r'$f(x) = x^3 - 3x^2 + 2x$')
ax1.axhline(0, color='gray', linestyle=':', lw=1)
ax1.set_title("Cubic Function with Inset Local Minimum", fontsize=12, fontweight='bold')
ax1.set_xlabel("x")
ax1.set_ylabel("f(x)")
ax1.legend(loc='upper left')
ax1.grid(True, alpha=0.4)

# Inset
ax2 = fig.add_axes([0.50, 0.22, 0.36, 0.36])
ax2.plot(x_poly, y_poly, color='#e67e22', lw=2.5)
ax2.set_xlim(1.2, 2.0)
ax2.set_ylim(-1.5, 0.5)
ax2.set_title("Local Minimum Region", fontsize=9, fontweight='bold')
ax2.grid(True, linestyle='--', alpha=0.5)

plt.show()
```
