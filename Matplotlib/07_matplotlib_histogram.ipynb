# 07. Histograms & Distribution Analysis

Histograms display the frequency distribution of continuous datasets, allowing analysts to assess skewness, bimodality, spread, and outlier presence.

### Learning Objectives:
- Construct histograms with `ax.hist()` controlling bin counts and bin edges
- Overlay theoretical Probability Density Functions (PDF) with `density=True`
- Compare overlapping distributions using transparency (`alpha`)
- Add vertical statistical markers (Mean, Median, Standard Deviation)

## 1. Histogram with Density Normalization & Gaussian Fit
Setting `density=True` scales bar heights so that total bar area equals 1.0, enabling direct comparison against theoretical probability density functions.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
mu, sigma = 100, 15
data = np.random.normal(mu, sigma, 1000)

fig, ax = plt.subplots(figsize=(8.5, 4.5), dpi=100)

# Plot histogram with normalized density
count, bins, ignored = ax.hist(data, bins=30, density=True, color='#3498db', edgecolor='white', alpha=0.7, label='Sample Distribution')

# Overlay theoretical Normal Bell Curve
x_pdf = np.linspace(50, 150, 200)
pdf = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(- (x_pdf - mu)**2 / (2 * sigma**2))
ax.plot(x_pdf, pdf, color='#e74c3c', lw=2.5, label='Theoretical PDF: $\\mathcal{N}(100, 15^2)$')

# Mean and Median indicators
ax.axvline(np.mean(data), color='black', lw=1.5, ls='--', label=f'Mean = {np.mean(data):.1f}')
ax.axvline(np.median(data), color='orange', lw=1.5, ls=':', label=f'Median = {np.median(data):.1f}')

ax.set_title("Customer IQ Distribution with Normal Curve Overlay", fontsize=13, fontweight='bold')
ax.set_xlabel("Value")
ax.set_ylabel("Probability Density")
ax.legend(loc='upper right')
ax.grid(True, alpha=0.3)

plt.show()
```

## 2. Comparing Two Overlapping Distributions
Overlapping histograms allow evaluating differences in distribution between two cohorts (e.g. Treatment vs. Control group in A/B testing).

```python
group_a = np.random.normal(loc=70, scale=10, size=500)
group_b = np.random.normal(loc=82, scale=12, size=500)

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)

ax.hist(group_a, bins=25, alpha=0.55, color='#e67e22', label='Group A (Baseline)', edgecolor='black', lw=0.5)
ax.hist(group_b, bins=25, alpha=0.55, color='#27ae60', label='Group B (New Treatment)', edgecolor='black', lw=0.5)

ax.set_title("A/B Testing: Metric Score Distribution Comparison", fontsize=12, fontweight='bold')
ax.set_xlabel("Performance Score")
ax.set_ylabel("Frequency Count")
ax.legend()
ax.grid(True, linestyle='--', alpha=0.4)

plt.show()
```

## 🎯 Exercises & Practice
1. Simulate a right-skewed dataset of 1,000 household incomes using an exponential distribution (`np.random.exponential(scale=50000, size=1000)`).
2. Plot a histogram with 40 bins.
3. Plot the **Mean** and **Median** with vertical dashed lines to demonstrate the effect of skewness.

```python
# Exercise Solution
np.random.seed(42)
incomes = np.random.exponential(scale=50000, size=1000) + 15000

mean_inc = np.mean(incomes)
med_inc = np.median(incomes)

fig, ax = plt.subplots(figsize=(8.5, 4.5), dpi=100)
ax.hist(incomes / 1000, bins=40, color='#8e44ad', edgecolor='white', alpha=0.8)

ax.axvline(mean_inc / 1000, color='#e74c3c', lw=2, ls='--', label=f'Mean: ${mean_inc/1000:.1f}k')
ax.axvline(med_inc / 1000, color='#f39c12', lw=2, ls='-', label=f'Median: ${med_inc/1000:.1f}k')

ax.set_title("Right-Skewed Household Income Distribution", fontsize=12, fontweight='bold')
ax.set_xlabel("Annual Income ($ Thousands)")
ax.set_ylabel("Number of Households")
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
```
