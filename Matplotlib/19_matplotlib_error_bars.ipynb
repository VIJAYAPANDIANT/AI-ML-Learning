# 19. Error Bars & Experimental Uncertainty

In empirical science and Machine Learning evaluations, presenting a single mean metric without error bounds is misleading. Error bars communicate variance, measurement noise, and statistical confidence intervals.

### Learning Objectives:
- Construct error bar line plots using `ax.errorbar()`
- Configure symmetric vs asymmetric error bounds (`yerr`, `xerr`)
- Customize error caps: `capsize`, `capthick`, and `ecolor`
- Add error bars to vertical bar charts

## 1. Error Bars in Line Series (Symmetric vs Asymmetric)
`yerr` accepts a scalar, a 1D array (symmetric errors), or a $2 \times N$ array for asymmetric errors ($[\text{lower\_errors}, \text{upper\_errors}]$).

```python
import matplotlib.pyplot as plt
import numpy as np

dosages = np.array([10, 25, 50, 100, 200]) # mg
response_mean = np.array([12.5, 24.8, 55.2, 78.6, 92.1])
response_std = np.array([2.1, 3.4, 6.2, 5.1, 3.8]) # Standard deviation

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)

ax.errorbar(
    dosages, response_mean, 
    yerr=response_std, 
    fmt='-o', 
    color='#2c3e50', 
    ecolor='#e74c3c', 
    elinewidth=1.8, 
    capsize=5, 
    capthick=1.5, 
    markersize=6, 
    label='Mean Efficacy ± 1 SD'
)

ax.set_title("Dose-Response Curve with Measurement Uncertainty", fontsize=12, fontweight='bold')
ax.set_xlabel("Dose (mg)")
ax.set_ylabel("Patient Response Rate (%)")
ax.set_ylim(0, 110)
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
```

## 2. Bar Charts with Statistical Error Bars
Frequently used in academic papers to report algorithm benchmark mean and standard deviation.

```python
models = ['Naive Bayes', 'Decision Tree', 'Random Forest', 'Gradient Boosting']
means = [74.2, 81.5, 89.4, 91.8]
stds = [3.8, 4.2, 1.8, 1.5]

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
bars = ax.bar(
    models, means, 
    yerr=stds, 
    capsize=6, 
    color='#3498db', 
    edgecolor='black', 
    lw=0.8, 
    width=0.55
)

ax.set_title("Model Benchmark Accuracy Across 10-Fold CV (Mean ± SD)", fontsize=11, fontweight='bold')
ax.set_ylabel("Accuracy Score (%)")
ax.set_ylim(60, 100)
ax.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```

## 🎯 Exercises & Practice
1. Given 4 test temperatures: `[100, 150, 200, 250]` with yields `[42, 68, 85, 72]` and asymmetric error bars: lower errors `[3, 5, 4, 8]`, upper errors `[5, 4, 6, 3]`.
2. Plot the series using `ax.errorbar` with asymmetric errors.

```python
# Exercise Solution
temps = [100, 150, 200, 250]
yields = [42, 68, 85, 72]
lower_err = [3, 5, 4, 8]
upper_err = [5, 4, 6, 3]
asym_err = [lower_err, upper_err]

fig, ax = plt.subplots(figsize=(7.5, 4), dpi=100)
ax.errorbar(temps, yields, yerr=asym_err, fmt='-s', color='#16a085', ecolor='#c0392b', capsize=5, lw=2)
ax.set_title("Chemical Yield under Temperature Variations (Asymmetric Errors)", fontsize=11, fontweight='bold')
ax.set_xlabel("Temperature (°C)")
ax.set_ylabel("Reaction Yield (%)")
ax.grid(True, alpha=0.3)
plt.show()
```
