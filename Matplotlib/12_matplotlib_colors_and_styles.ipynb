# 12. Color Palettes, Colormaps & Built-In Styles

Color choices dictate readability and emotional impact. Matplotlib provides pre-built style sheets and an extensive suite of sequential, diverging, and qualitative colormaps.

### Learning Objectives:
- Inspect and activate pre-packaged styles with `plt.style.use()`
- Use style contexts safely with `with plt.style.context():`
- Master Sequential (`viridis`), Diverging (`coolwarm`), and Qualitative colormaps
- Create correlation matrix heatmaps with `ax.imshow()` and centered zero normalization

## 1. Exploring Built-In Styles
Matplotlib includes dozens of themes: `'seaborn-v0_8'`, `'ggplot'`, `'dark_background'`, `'fivethirtyeight'`.

```python
import matplotlib.pyplot as plt
import numpy as np

# Listing a sample of available style themes
print("Available styles (sample):", plt.style.available[:8])

x = np.linspace(0, 10, 50)

# Using dark_background within a temporary context manager
with plt.style.context('dark_background'):
    fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
    ax.plot(x, np.sin(x), color='#00ffcc', lw=2.5, label='Sine Wave')
    ax.plot(x, np.cos(x), color='#ff007f', lw=2.5, ls='--', label='Cosine Wave')
    ax.set_title("Dark Theme Context", fontsize=12, fontweight='bold')
    ax.legend()
    ax.grid(True, alpha=0.2)
    plt.show()
```

## 2. Feature Correlation Heatmap with Diverging Colormap
A 2D matrix heatmap using a diverging colormap (`'coolwarm'` or `'RdBu'`) centered at zero.

```python
from matplotlib.colors import TwoSlopeNorm

# Synthetic 5x5 Correlation Matrix
features = ['Age', 'Income', 'Education', 'Spending', 'Credit_Score']
np.random.seed(42)
corr = np.array([
    [ 1.00,  0.42,  0.15, -0.32,  0.28],
    [ 0.42,  1.00,  0.65,  0.58,  0.49],
    [ 0.15,  0.65,  1.00,  0.35,  0.12],
    [-0.32,  0.58,  0.35,  1.00, -0.18],
    [ 0.28,  0.49,  0.12, -0.18,  1.00]
])

fig, ax = plt.subplots(figsize=(7, 6), dpi=100)

# Center diverging colormap at correlation = 0.0
norm = TwoSlopeNorm(vmin=-1, vcenter=0, vmax=1)
cax = ax.imshow(corr, cmap='coolwarm', norm=norm)

# Add numerical labels inside each heatmap cell
for i in range(len(features)):
    for j in range(len(features)):
        val = corr[i, j]
        text_col = 'white' if abs(val) > 0.5 else 'black'
        ax.text(j, i, f"{val:.2f}", ha='center', va='center', color=text_col, fontweight='bold')

ax.set_xticks(range(len(features)))
ax.set_yticks(range(len(features)))
ax.set_xticklabels(features, rotation=45, ha='right')
ax.set_yticklabels(features)
ax.set_title("Feature Correlation Matrix", fontsize=12, fontweight='bold')

fig.colorbar(cax, ax=ax, fraction=0.046, pad=0.04, label='Pearson Correlation')
plt.show()
```

## 🎯 Exercises & Practice
1. Use `plt.style.context('ggplot')` to plot a multi-line comparison of 3 normal distributions with different standard deviations ($\\sigma = 1, 2, 3$).

```python
# Exercise Solution
x_eval = np.linspace(-6, 6, 200)

with plt.style.context('ggplot'):
    fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
    for s in [1, 2, 3]:
        pdf = (1 / (s * np.sqrt(2 * np.pi))) * np.exp(-x_eval**2 / (2 * s**2))
        ax.plot(x_eval, pdf, lw=2.2, label=f'σ = {s}')
    
    ax.set_title("Gaussian PDFs under ggplot Theme", fontsize=12, fontweight='bold')
    ax.set_xlabel("x")
    ax.set_ylabel("Probability Density")
    ax.legend()
    plt.show()
```
