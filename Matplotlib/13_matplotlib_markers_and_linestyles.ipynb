# 13. Markers, Linestyles & Line Aesthetics

Distinguishing multiple data series visually without relying purely on color is critical for accessibility, black-and-white printing, and publication clarity.

### Learning Objectives:
- Master marker symbols (`'o'`, `'s'`, `'^'`, `'D'`, `'*'`, `'x'`)
- Configure marker face colors (`mfc`), edge colors (`mec`), and edge widths (`mew`)
- Customize linestyles (solid, dashed, dash-dot, dotted) and custom dash patterns
- Implement stepped lines with `drawstyle='steps-post'`

## 1. Marker Catalog & Detailed Aesthetics
Combining distinct markers with clear edge boundaries ensures accessibility.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(1, 10, 10)

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)

ax.plot(x, x, marker='o', markersize=9, mfc='#ffffff', mec='#1f77b4', mew=2, lw=2, label='Circle (hollow)')
ax.plot(x, x * 1.5, marker='s', markersize=8, mfc='#2ca02c', mec='black', mew=1, ls='--', lw=2, label='Square')
ax.plot(x, x * 2.0, marker='^', markersize=9, mfc='#d62728', mec='white', mew=1, ls='-.', lw=2, label='Triangle')
ax.plot(x, x * 2.5, marker='D', markersize=7, mfc='#9467bd', mec='black', mew=1, ls=':', lw=2, label='Diamond')

ax.set_title("Marker Styling & Linestyle Variety", fontsize=12, fontweight='bold')
ax.set_xlabel("Sample Index")
ax.set_ylabel("Metric Value")
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
```

## 2. Stepped Lines (`drawstyle='steps-post'`)
Used to plot discrete state changes, digital signal triggers, or Kaplan-Meier survival curves.

```python
time_steps = np.arange(0, 12)
state_levels = [1, 1, 2, 2, 4, 3, 3, 5, 2, 2, 1, 1]

fig, ax = plt.subplots(figsize=(8.5, 4), dpi=100)

ax.plot(time_steps, state_levels, drawstyle='steps-post', color='#2c3e50', lw=2.5, label='Digital State Transition')
ax.scatter(time_steps, state_levels, color='#e74c3c', s=50, zorder=3, label='Clock Triggers')

ax.set_title("Discrete System State Changes (Stepped Plot)", fontsize=12, fontweight='bold')
ax.set_xlabel("Time (Clock Ticks)")
ax.set_ylabel("Operational State")
ax.set_yticks([1, 2, 3, 4, 5])
ax.legend()
ax.grid(True, linestyle=':', alpha=0.5)

plt.show()
```

## 🎯 Exercises & Practice
1. Plot two time series: Model A and Model B accuracy across 8 checkpoint epochs.
2. Model A should use dashed blue lines with white hollow circle markers.
3. Model B should use dotted red lines with solid black square markers.

```python
# Exercise Solution
epochs = np.arange(1, 9)
acc_a = [0.65, 0.72, 0.78, 0.81, 0.84, 0.86, 0.87, 0.88]
acc_b = [0.60, 0.70, 0.79, 0.85, 0.89, 0.91, 0.92, 0.93]

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
ax.plot(epochs, acc_a, ls='--', lw=2, color='#1f77b4', marker='o', mfc='white', mec='#1f77b4', mew=2, label='Model A')
ax.plot(epochs, acc_b, ls=':', lw=2, color='#d62728', marker='s', mfc='black', mec='black', label='Model B')

ax.set_title("Checkpoint Validation Accuracy Comparison", fontsize=11, fontweight='bold')
ax.set_xlabel("Epoch")
ax.set_ylabel("Accuracy")
ax.legend()
ax.grid(True, alpha=0.4)
plt.show()
```
