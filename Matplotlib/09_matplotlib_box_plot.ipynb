# 09. Box Plots & Statistical Five-Number Summaries

Box plots (Box-and-Whisker plots) provide a compact visualization of the five-number summary: **Minimum, First Quartile (Q1), Median, Third Quartile (Q3), and Maximum**, alongside algorithmic detection of outliers based on the Interquartile Range ($IQR = Q3 - Q1$).

### Learning Objectives:
- Construct box plots using `ax.boxplot()`
- Understand notch confidence intervals and `patch_artist=True`
- Compare performance distributions across multiple Machine Learning models
- Style box elements: medians, whiskers, caps, and flier outlier markers

## 1. Five-Number Summary & Multi-Group Box Plot
By setting `patch_artist=True`, Matplotlib generates filled polygon boxes whose colors can be customized.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
# Simulating 5-fold cross-validation scores across 4 ML models
rf_scores = np.random.normal(loc=0.88, scale=0.03, size=30)
xgb_scores = np.random.normal(loc=0.91, scale=0.02, size=30)
svm_scores = np.random.normal(loc=0.84, scale=0.05, size=30)
lr_scores = np.random.normal(loc=0.79, scale=0.04, size=30)

scores_data = [rf_scores, xgb_scores, svm_scores, lr_scores]
model_labels = ['Random Forest', 'XGBoost', 'Support Vector', 'Logistic Reg']

fig, ax = plt.subplots(figsize=(8.5, 5), dpi=100)

bplot = ax.boxplot(
    scores_data, 
    labels=model_labels, 
    patch_artist=True, 
    notch=False, 
    vert=True,
    flierprops=dict(marker='o', markerfacecolor='#e74c3c', markersize=6, linestyle='none'),
    medianprops=dict(color='black', linewidth=2)
)

# Customizing fill colors for each model box
palette = ['#3498db', '#2ecc71', '#9b59b6', '#f39c12']
for patch, color in zip(bplot['boxes'], palette):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)

ax.set_title("Machine Learning Cross-Validation F1-Scores", fontsize=13, fontweight='bold')
ax.set_ylabel("F1 Score", fontsize=11)
ax.set_ylim(0.70, 0.98)
ax.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

## 2. Horizontal Box Plots for Feature Distributions
Horizontal orientation (`vert=False`) is useful when evaluating skewed continuous features across categories.

```python
dept_eng = np.random.normal(120, 15, 60)
dept_sales = np.random.normal(95, 25, 60)
dept_hr = np.random.normal(80, 10, 60)

salaries = [dept_hr, dept_sales, dept_eng]
depts = ['Human Resources', 'Sales', 'Engineering']

fig, ax = plt.subplots(figsize=(9, 4), dpi=100)
bplot_h = ax.boxplot(salaries, labels=depts, vert=False, patch_artist=True, notch=True)

for patch in bplot_h['boxes']:
    patch.set_facecolor('#16a085')
    patch.set_alpha(0.6)

ax.set_title("Departmental Salary Distributions (Notched at Median 95% CI)", fontsize=11, fontweight='bold')
ax.set_xlabel("Annual Salary ($ Thousands)")
ax.grid(axis='x', linestyle=':', alpha=0.5)

plt.show()
```

## 🎯 Exercises & Practice
1. Generate three synthetic datasets with varying variance: Dataset A ($\sigma=5$), Dataset B ($\sigma=15$), Dataset C ($\sigma=25$).
2. Plot a boxplot comparing the three, highlighting outlier fliers in red stars (`marker='*'`).

```python
# Exercise Solution
np.random.seed(123)
data_a = np.random.normal(100, 5, 100)
data_b = np.random.normal(100, 15, 100)
data_c = np.random.normal(100, 25, 100)

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
bp = ax.boxplot(
    [data_a, data_b, data_c], 
    labels=['Low Variance (σ=5)', 'Mid Variance (σ=15)', 'High Variance (σ=25)'],
    patch_artist=True,
    flierprops=dict(marker='*', markerfacecolor='red', markersize=8, markeredgecolor='black')
)

colors = ['#a8dadc', '#457b9d', '#1d3557']
for patch, col in zip(bp['boxes'], colors):
    patch.set_facecolor(col)
    patch.set_alpha(0.8)

ax.set_title("Impact of Variance on Whisker Span and Outlier Detection", fontsize=11, fontweight='bold')
ax.set_ylabel("Value Scale")
ax.grid(axis='y', linestyle='--', alpha=0.4)
plt.show()
```
