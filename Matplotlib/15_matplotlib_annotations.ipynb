# 15. Annotations, Arrows & Reference Highlights

Guiding the viewer's eye to critical data points, outliers, local optima, and significant regime shifts transforms an ordinary chart into an insightful analytical story.

### Learning Objectives:
- Place descriptive text boxes with `ax.text()`
- Draw precision callout arrows with `ax.annotate()`
- Highlight threshold boundaries with horizontal (`ax.axhline`) and vertical lines (`ax.axvline`)
- Emphasize event spans using `ax.axvspan()` and `ax.axhspan()`

## 1. Precision Callout Arrows with `ax.annotate()`
`ax.annotate()` connects an arbitrary text position (`xytext`) to an exact target data coordinate (`xy`) with an arrow.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 200)
y = np.sin(x) * np.exp(-0.15 * x)

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)
ax.plot(x, y, color='#2c3e50', lw=2.2, label='Sensor Signal')

# Find the primary peak point
peak_idx = np.argmax(y)
x_peak, y_peak = x[peak_idx], y[peak_idx]

# Add Annotation Arrow to peak
ax.annotate(
    f'Maximum Peak\n({x_peak:.2f}s, {y_peak:.2f}V)', 
    xy=(x_peak, y_peak), 
    xytext=(x_peak + 1.5, y_peak + 0.25),
    arrowprops=dict(arrowstyle='->', connectionstyle='arc3,rad=-0.2', color='#c0392b', lw=2),
    bbox=dict(boxstyle='round,pad=0.5', facecolor='#fadbd8', edgecolor='#c0392b', lw=1.2),
    fontsize=10, fontweight='bold'
)

ax.set_title("Damped Signal with Annotated Peak Optimum", fontsize=12, fontweight='bold')
ax.set_xlabel("Time (s)")
ax.set_ylabel("Signal (V)")
ax.set_ylim(-0.8, 1.2)
ax.grid(True, alpha=0.3)
ax.legend()

plt.show()
```

## 2. Event Span Highlighting with `ax.axvspan()`
Highlighting time periods (e.g. Market Recessions, System Outages, Maintenance Windows).

```python
days = np.arange(1, 61)
latency = 45 + np.random.normal(0, 5, size=60)
# Simulate an incident spike between days 25 and 32
latency[24:32] += np.random.uniform(40, 70, size=8)

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)
ax.plot(days, latency, color='#2980b9', lw=1.8, label='API Response Time')

# Threshold reference line
ax.axhline(60, color='#e67e22', ls='--', lw=1.5, label='SLA Target (60ms)')

# Highlight incident period with a colored vertical span
ax.axvspan(24, 32, color='#e74c3c', alpha=0.2, label='Service Degradation Incident')

ax.text(28, 110, "Network Switch Outage\n(P1 Incident)", ha='center', fontsize=9, fontweight='bold', color='#922b21')

ax.set_title("API Service Latency Monitoring & SLA Tracking", fontsize=12, fontweight='bold')
ax.set_xlabel("Day of Month")
ax.set_ylabel("Latency (ms)")
ax.legend(loc='upper right')
ax.grid(True, alpha=0.3)

plt.show()
```

## 🎯 Exercises & Practice
1. Plot a continuous stock price series over 50 days.
2. Annotate the single lowest price point with a green callout arrow saying `'Optimal Buy Opportunity'`.
3. Add a horizontal dashed line representing the stop-loss limit.

```python
# Exercise Solution
np.random.seed(99)
t = np.arange(1, 51)
price = 150 + np.cumsum(np.random.normal(0.5, 2.0, 50))

min_idx = np.argmin(price)

fig, ax = plt.subplots(figsize=(8.5, 4.2), dpi=100)
ax.plot(t, price, color='#2c3e50', lw=2)
ax.axhline(140, color='#e74c3c', ls='--', label='Stop-Loss Limit ($140)')

ax.annotate(
    'Optimal Buy Opportunity',
    xy=(t[min_idx], price[min_idx]),
    xytext=(t[min_idx] + 5, price[min_idx] - 6),
    arrowprops=dict(facecolor='#27ae60', edgecolor='#27ae60', shrink=0.05, width=1.5, headwidth=8),
    bbox=dict(boxstyle='square', fc='#d4efdf', ec='#27ae60')
)

ax.set_title("Stock Price Volatility with Buy Signal", fontsize=11, fontweight='bold')
ax.set_xlabel("Trading Day")
ax.set_ylabel("Price ($)")
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```
