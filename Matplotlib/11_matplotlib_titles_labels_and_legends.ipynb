## 1. LaTeX Math Formatting & Typography
Matplotlib has a built-in TeX expression parser. Wrapping text in `$...` renders mathematical equations.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-4, 4, 200)
# Standard normal PDF
pdf = (1 / np.sqrt(2 * np.pi)) * np.exp(-x**2 / 2)
# CDF
cdf = 0.5 * (1 + np.vectorize(np.math.erf)(x / np.sqrt(2)))

fig, ax = plt.subplots(figsize=(8.5, 4.5), dpi=100)

ax.plot(x, pdf, color='#1f77b4', lw=2.5, label=r'PDF: $f(x) = \frac{1}{\sqrt{2\pi}} e^{-\frac{x^2}{2}}$')
ax.plot(x, cdf, color='#ff7f0e', lw=2.5, ls='--', label=r'CDF: $\Phi(x) = \frac{1}{2} \left[1 + \text{erf}\left(\frac{x}{\sqrt{2}}\right)\right]$')

ax.set_title(r"Gaussian Normal Distribution: $mu = 0, sigma^2 = 1$", fontsize=13, fontweight='bold', pad=12)
ax.set_xlabel(r"Standard Deviations ($z$-score)", fontsize=11)
ax.set_ylabel(r"Probability / Cumulative Density", fontsize=11)
ax.grid(True, linestyle=':', alpha=0.6)

# Advanced Legend Customization
ax.legend(
    loc='upper left', 
    fontsize=10, 
    frameon=True, 
    facecolor='#f8f9fa', 
    edgecolor='#cccccc', 
    framealpha=0.95,
    title='Theoretical Curves', 
    title_fontsize=11
)

plt.show()
```

## 2. External Legends using `bbox_to_anchor`
When charts have dense data, placing the legend outside the plotting bounds prevents visual occlusion.

```python
x = np.linspace(0, 10, 50)

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
for alpha in [0.5, 1.0, 1.5, 2.0]:
    ax.plot(x, np.sin(x) * np.exp(-0.2 * x * alpha), lw=2, label=f'Damping factor $\\alpha = {alpha}$')

ax.set_title("Harmonic Oscillators with External Legend", fontsize=12, fontweight='bold')
ax.set_xlabel("Time (s)")
ax.set_ylabel("Amplitude")
ax.grid(True, alpha=0.3)

# Move legend outside to the right
ax.legend(bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0, frameon=True)

fig.tight_layout()
plt.show()
```

## 🎯 Exercises & Practice
1. Plot the sigmoid activation function $S(z) = \frac{1}{1 + e^{-z}}$ over $z \in [-6, 6]$.
2. Add a LaTeX title, an axis label for $z$, and a legend located at `loc='lower right'` with a drop shadow (`shadow=True`).

```python
# Exercise Solution
z = np.linspace(-6, 6, 200)
sigmoid = 1 / (1 + np.exp(-z))

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
ax.plot(z, sigmoid, color='#8e44ad', lw=2.5, label=r'$sigma(z) = rac{1}{1 + e^{-z}}$')
ax.axhline(0.5, color='gray', linestyle='--', lw=1, alpha=0.7)
ax.axvline(0, color='gray', linestyle='--', lw=1, alpha=0.7)

ax.set_title(r"Neural Network Sigmoid Activation Function", fontsize=12, fontweight='bold')
ax.set_xlabel(r"Logit Input ($z$)", fontsize=11)
ax.set_ylabel(r"Activated Output ($sigma(z)$)", fontsize=11)
ax.legend(loc='lower right', shadow=True, fontsize=11)
ax.grid(True, alpha=0.3)

plt.show()
```
