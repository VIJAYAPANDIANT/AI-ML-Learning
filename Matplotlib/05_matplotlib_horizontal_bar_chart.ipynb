# 05. Horizontal Bar Charts & Feature Importance

Horizontal bar charts (`ax.barh()`) are ideal when categorical labels are long, for leaderboard rankings, and for machine learning feature importance visualizers.

### Learning Objectives:
- Construct horizontal bar charts using `ax.barh()`
- Sort categories descending and invert axes with `ax.invert_yaxis()`
- Build a production-grade **Machine Learning Feature Importance Plot**
- Add direct value labels at the end of horizontal bars

## 1. Constructing Horizontal Bars with Ranked Sorting
When plotting rankings, sorting from lowest to highest in data and calling `ax.invert_yaxis()` ensures the #1 item sits at the top.

```python
import matplotlib.pyplot as plt
import numpy as np

countries = ['United States', 'China', 'Japan', 'Germany', 'India', 'United Kingdom', 'France']
gdp_trillions = [28.78, 18.53, 4.11, 4.59, 3.94, 3.50, 3.13]

# Sort descending
sorted_indices = np.argsort(gdp_trillions)
countries_sorted = [countries[i] for i in sorted_indices]
gdp_sorted = [gdp_trillions[i] for i in sorted_indices]

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)
bars = ax.barh(countries_sorted, gdp_sorted, color='#2980b9', edgecolor='black', height=0.6)

ax.bar_label(bars, fmt='$%.2f T', padding=5, fontsize=9, fontweight='bold')
ax.set_title("Top Global Economies by Nominal GDP (Trillions USD)", fontsize=13, fontweight='bold')
ax.set_xlabel("GDP ($ Trillions)")
ax.set_xlim(0, 33)
ax.grid(axis='x', linestyle='--', alpha=0.5)

plt.show()
```

## 2. Machine Learning: Random Forest Feature Importance
A standard technique in ML model explainability is visualizing Gini or Permutation feature importances.

```python
features = ['Credit_Score', 'Annual_Income', 'Debt_to_Income', 'Loan_Amount', 'Employment_Years', 'Home_Ownership', 'Open_Accounts']
importances = np.array([0.28, 0.22, 0.18, 0.14, 0.09, 0.05, 0.04])

# Sort ascending so the most important feature appears at the top
sort_idx = np.argsort(importances)
features_sorted = [features[i] for i in sort_idx]
importances_sorted = importances[sort_idx]

# Highlight top 2 features with a distinct color
colors = ['#95a5a6' if imp < 0.20 else '#e74c3c' for imp in importances_sorted]

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)
bars = ax.barh(features_sorted, importances_sorted, color=colors, height=0.65, edgecolor='#2c3e50', lw=0.8)

ax.bar_label(bars, fmt='%.1f%%', labels=[f"{v*100:.1f}%" for v in importances_sorted], padding=5, fontsize=9)
ax.set_title("Random Forest: Feature Importance Ranking (Credit Risk Model)", fontsize=12, fontweight='bold')
ax.set_xlabel("Relative Importance Weight", fontsize=10)
ax.set_xlim(0, 0.35)
ax.grid(axis='x', linestyle=':', alpha=0.5)

plt.show()
```

## 🎯 Exercises & Practice
1. Given 6 deep learning models: `['GPT-4o', 'Claude 3.5 Sonnet', 'Gemini 1.5 Pro', 'Llama 3 70B', 'Mistral Large', 'Gemma 2']` and their benchmark MMLU scores: `[88.7, 88.3, 85.9, 82.0, 81.2, 75.7]`.
2. Plot a horizontal bar chart sorted from highest to lowest with `ax.invert_yaxis()`.
3. Add value labels and color the top 2 models in royal blue and the rest in slate gray.

```python
# Exercise Solution
models = ['GPT-4o', 'Claude 3.5 Sonnet', 'Gemini 1.5 Pro', 'Llama 3 70B', 'Mistral Large', 'Gemma 2']
scores = [88.7, 88.3, 85.9, 82.0, 81.2, 75.7]

colors = ['#1a73e8' if i < 2 else '#70757a' for i in range(len(models))]

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)
bars = ax.barh(models, scores, color=colors, height=0.6, edgecolor='black', lw=0.6)
ax.invert_yaxis() # Top score at top

ax.bar_label(bars, fmt='%.1f', padding=4, fontsize=9, fontweight='bold')
ax.set_title("LLM Benchmark Comparison: Massive Multitask Language Understanding (MMLU)", fontsize=11, fontweight='bold')
ax.set_xlabel("Score (%)")
ax.set_xlim(60, 95)
ax.grid(axis='x', linestyle='--', alpha=0.4)

plt.show()
```
