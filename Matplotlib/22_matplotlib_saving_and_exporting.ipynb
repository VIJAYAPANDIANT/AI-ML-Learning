# 22. Production Figure Exporting & Publication Quality

Exporting graphics at high resolution without clipped labels is necessary for academic papers, enterprise executive decks, and web deployment.

### Learning Objectives:
- Master the `fig.savefig()` parameters
- Choose between Raster formats (PNG, JPG) and Vector formats (PDF, SVG, EPS)
- Prevent clipped labels with `bbox_inches='tight'`
- Configure print resolution with `dpi=300` and transparent backgrounds

## 1. High-Resolution Figure Export Pipeline
Key parameters for publication-grade exports:
- `dpi=300`: Standard for academic journals and print graphics
- `bbox_inches='tight'`: Automatically trims excess whitespace borders
- `transparent=True`: Renders transparent backgrounds for web presentation decks

```python
import matplotlib.pyplot as plt
import numpy as np
import os

x = np.linspace(0, 10, 100)
y = np.sin(x) * np.exp(-0.2 * x)

fig, ax = plt.subplots(figsize=(7, 3.8), dpi=100)
ax.plot(x, y, color='#2c3e50', lw=2.2, label='Model Response')
ax.set_title("Production Figure for Publication", fontsize=11, fontweight='bold')
ax.set_xlabel("Input Feature (x)")
ax.set_ylabel("Response (y)")
ax.legend()
ax.grid(True, alpha=0.3)

# Export to high-res PNG
output_png = 'sample_export_300dpi.png'
fig.savefig(output_png, dpi=300, bbox_inches='tight')
print(f"Exported: {output_png} (Exists: {os.path.exists(output_png)})")

plt.show()
```

## 2. Vector Graphics Export (SVG / PDF)
Vector formats store shapes mathematically rather than pixel bitmaps, allowing infinite zoom without pixelation.

```python
# Export to Scalable Vector Graphics (SVG)
output_svg = 'sample_vector.svg'
fig.savefig(output_svg, format='svg', bbox_inches='tight')
print(f"Exported Vector Graphic: {output_svg} (Exists: {os.path.exists(output_svg)})")

# Clean up demo files
if os.path.exists(output_png): os.remove(output_png)
if os.path.exists(output_svg): os.remove(output_svg)
print("Cleaned up temporary demonstration artifacts.")
```
