## 2. Pyplot Interface vs. Object-Oriented (OO) Interface
Matplotlib provides two distinct coding paradigms:
1. **Pyplot State-Machine (`plt.plot`)**: Implicitly tracks the "current" figure and axes. Convenient for rapid exploratory sketches.
2. **Object-Oriented API (`fig, ax = plt.subplots()`)**: Explicitly instantiates Figure and Axes objects. Strongly recommended for production pipelines, complex dashboards, and clean software architecture.

```python
# Approach 1: Pyplot (Procedural / State-Machine)
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, color='#1f77b4', linewidth=2, label='Sine Wave')
plt.title("Pyplot Procedural Interface Demo")
plt.xlabel("Input X")
plt.ylabel("sin(X)")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# Approach 2: Object-Oriented Interface (Recommended Standard)
fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(x, y, color='#2ca02c', linewidth=2.5, linestyle='-', label=r'$y = sin(x)$')
ax.set_title("Object-Oriented Interface (fig, ax)", fontsize=13, fontweight='bold', pad=10)
ax.set_xlabel("Time (seconds)", fontsize=11)
ax.set_ylabel("Amplitude", fontsize=11)
ax.legend(loc='upper right', frameon=True)
ax.grid(True, linestyle=':', alpha=0.7)

plt.show()
```

## 3. Anatomy of a Matplotlib Figure
A Matplotlib visualization consists of a hierarchy:
- **Figure**: The top-level bounding canvas holding everything (axes, titles, legends, colorbars).
- **Axes**: The actual plotting coordinate system (one figure can contain multiple axes).
- **Axis**: The numerical scales (X-axis, Y-axis) controlling limits, ticks, and tick labels.
- **Artist**: Everything visible on the canvas (lines, text, patches, markers).

```python
# Exploring Figure and Axes properties
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot([1, 2, 3, 4], [10, 20, 25, 30], marker='o', color='#d62728', label='Series A')

print("Figure Object:", type(fig))
print("Axes Object:", type(ax))
print("X-Axis Object:", type(ax.xaxis))
print("Lines on Axes:", ax.lines)

ax.set_title("Inspecting Matplotlib Objects")
ax.legend()
plt.show()
```

## 🎯 Exercises & Practice
1. Generate an array of 200 points from $-2\pi$ to $2\pi$.
2. Compute both $\sin(x)$ and $\cos(x)$.
3. Using the **Object-Oriented interface**, plot both curves on the same axes with custom colors, dashed grid, and a descriptive legend.

```python
# Exercise Solution
x_vals = np.linspace(-2 * np.pi, 2 * np.pi, 200)
sin_vals = np.sin(x_vals)
cos_vals = np.cos(x_vals)

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)
ax.plot(x_vals, sin_vals, color='#007acc', linewidth=2, label=r'$sin(x)$')
ax.plot(x_vals, cos_vals, color='#ff7f0e', linewidth=2, linestyle='--', label=r'$cos(x)$')

ax.set_title(r'Harmonic Functions: $sin(x)$ vs $cos(x)$', fontsize=13, fontweight='bold')
ax.set_xlabel('Angle (radians)', fontsize=11)
ax.set_ylabel('Function Value', fontsize=11)
ax.set_xlim(-2 * np.pi, 2 * np.pi)
ax.axhline(0, color='gray', linestyle='-', linewidth=0.8, alpha=0.7)
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(loc='lower left', framealpha=0.9)

plt.show()
```
