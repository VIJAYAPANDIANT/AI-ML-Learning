# 08. Pie Charts & Donut Visualizations

Pie and Donut charts visualize part-to-whole proportions and categorical composition.

### Learning Objectives:
- Construct pie charts with `ax.pie()`
- Format slice percentages using `autopct`
- Explode key segments with `explode` to draw visual attention
- Create modern **Donut Charts** by layering a center circle patch
- Learn best practices for when to use pie charts vs bar charts

## 1. Standard Exploded Pie Chart
`explode` accepts a tuple of float offsets pulling selected wedges away from the center.

```python
import matplotlib.pyplot as plt

segments = ['Cloud Infrastructure', 'Digital Advertising', 'Subscription Services', 'Hardware Devices', 'Other']
revenue_share = [42.5, 28.0, 16.5, 9.0, 4.0]
colors = ['#2980b9', '#27ae60', '#f39c12', '#e74c3c', '#95a5a6']

# Explode the 1st segment ('Cloud Infrastructure')
explode = (0.08, 0, 0, 0, 0)

fig, ax = plt.subplots(figsize=(7, 6), dpi=100)
wedges, texts, autotexts = ax.pie(
    revenue_share, 
    explode=explode, 
    labels=segments, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=colors,
    shadow=True,
    textprops=dict(fontsize=10)
)

# Make percentage numbers bold and white
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontweight('bold')

ax.set_title("Enterprise Revenue Composition", fontsize=13, fontweight='bold')
plt.show()
```

## 2. Modern Donut Chart with Center KPI
A Donut Chart is created by adding a white `plt.Circle((0,0), radius=0.7, fc='white')` inside the axes.

```python
categories = ['Search Ads', 'Video & Media', 'Cloud Apps', 'Hardware']
shares = [55, 20, 15, 10]
palette = ['#4285f4', '#ea4335', '#fbbc05', '#34a853']

fig, ax = plt.subplots(figsize=(6.5, 6.5), dpi=100)

wedges, texts, autotexts = ax.pie(
    shares, 
    labels=categories, 
    autopct='%1.0f%%', 
    pctdistance=0.82, 
    startangle=90, 
    colors=palette,
    wedgeprops=dict(width=0.35, edgecolor='white', lw=2)
)

for at in autotexts:
    at.set_fontweight('bold')
    at.set_fontsize(10)

# Center summary text
ax.text(0, 0.05, '$320B', ha='center', va='center', fontsize=18, fontweight='bold', color='#2c3e50')
ax.text(0, -0.12, 'Total Revenue', ha='center', va='center', fontsize=10, color='#7f8c8d')

ax.set_title("Global Revenue Breakdown by Vertical", fontsize=12, fontweight='bold')
plt.show()
```

## 🎯 Exercises & Practice
1. Create a Donut chart showing Global Mobile OS Market Share: `Android (71.2%)`, `iOS (28.1%)`, `Others (0.7%)`.
2. Add a center annotation displaying the total active device count: `'3.8 Billion Active Devices'`.

```python
# Exercise Solution
os_names = ['Android', 'iOS', 'Other']
os_shares = [71.2, 28.1, 0.7]
os_colors = ['#3ddc84', '#555555', '#d1d5db']

fig, ax = plt.subplots(figsize=(6, 6), dpi=100)
wedges, texts, autotexts = ax.pie(
    os_shares, 
    labels=os_names, 
    autopct='%1.1f%%', 
    pctdistance=0.8, 
    startangle=45, 
    colors=os_colors,
    wedgeprops=dict(width=0.36, edgecolor='white', lw=2)
)

for at in autotexts:
    at.set_fontweight('bold')

ax.text(0, 0.05, '3.8B', ha='center', va='center', fontsize=20, fontweight='bold', color='#1f2937')
ax.text(0, -0.10, 'Active Devices', ha='center', va='center', fontsize=10, color='#6b7280')

ax.set_title("Global Mobile OS Market Share", fontsize=12, fontweight='bold')
plt.show()
```
