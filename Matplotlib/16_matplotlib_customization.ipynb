# 16. Advanced Customization & Publication-Quality Themes

Clean, minimalist graphics communicate insights faster. In this notebook, we apply **Edward Tufte's data-ink ratio principles**, modify figure spines, and configure global `rcParams`.

### Learning Objectives:
- Remove unnecessary visual clutter by hiding top and right spines
- Modify spine thickness, offset, and color
- Configure global chart defaults using `plt.rcParams`
- Build a custom publication-grade minimalist theme

## 1. Spine Manipulation: Eliminating Data-Ink Waste
Hiding the top and right enclosing borders creates an open, modern design.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 50)
y = 2.5 * x + np.random.normal(0, 2, 50)

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
ax.scatter(x, y, color='#2b5c8f', alpha=0.8, s=40, edgecolors='white', label='Observations')
ax.plot(x, 2.5 * x, color='#e056fd', lw=2, label='Model Fit')

# Hide top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Customize left and bottom spines
ax.spines['left'].set_color('#444444')
ax.spines['bottom'].set_color('#444444')
ax.spines['left'].set_linewidth(1.2)
ax.spines['bottom'].set_linewidth(1.2)

ax.set_title("Minimalist Scatter: Clean Spines", fontsize=13, fontweight='bold', loc='left', pad=15)
ax.set_xlabel("Predictor Feature (X)", fontsize=10)
ax.set_ylabel("Target Variable (Y)", fontsize=10)
ax.legend(frameon=False, loc='upper left')
ax.grid(axis='y', linestyle=':', alpha=0.5)

plt.show()
```

## 2. Global Styling with `plt.rcParams`
Modifying `rcParams` applies styling choices consistently across all subsequent figures in your session.

```python
# Setting global styling preferences
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.edgecolor'] = '#555555'
plt.rcParams['axes.linewidth'] = 0.8
plt.rcParams['axes.titlesize'] = 12
plt.rcParams['xtick.direction'] = 'out'
plt.rcParams['ytick.direction'] = 'out'

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
ax.bar(['Category 1', 'Category 2', 'Category 3', 'Category 4'], [45, 62, 85, 30], color='#00b894', width=0.5)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_title("Standardized rcParams Appearance", fontweight='bold', loc='left')
ax.grid(axis='y', alpha=0.3)

plt.show()

# Reset rcParams back to default when done
import matplotlib as mpl
mpl.rcdefaults()
```

## 🎯 Exercises & Practice
1. Create a bar chart comparing 5 programming languages: `['Python', 'JavaScript', 'C++', 'Java', 'Go']`.
2. Apply the minimalist spine pattern (hide top and right spines, remove vertical gridlines, keep subtle horizontal gridlines).
3. Title the chart left-aligned with `loc='left'`.

```python
# Exercise Solution
languages = ['Python', 'JavaScript', 'C++', 'Java', 'Go']
popularity = [92, 84, 68, 62, 55]

fig, ax = plt.subplots(figsize=(8, 4.2), dpi=100)
bars = ax.bar(languages, popularity, color='#0984e3', width=0.55)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.bar_label(bars, fmt='%d%%', padding=3, fontweight='bold')

ax.set_title("Developer Language Adoption Rate 2026", fontsize=12, fontweight='bold', loc='left', pad=12)
ax.set_ylabel("Adoption Index (%)")
ax.set_ylim(0, 105)
ax.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```
