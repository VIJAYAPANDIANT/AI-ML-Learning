# 10. Multi-Panel Subplots & Complex Grid Layouts

Real-world analytical dashboards and scientific publications require presenting multiple interrelated charts within a single figure canvas.

### Learning Objectives:
- Create regular grid subplots using `plt.subplots(nrows, ncols)`
- Index 1D arrays vs 2D axes matrices and iterate using `axes.flat`
- Share axes with `sharex=True` and `sharey=True`
- Construct asymmetric multi-panel dashboards using `matplotlib.gridspec.GridSpec`
- Adjust spacing with `fig.tight_layout()` and `fig.subplots_adjust()`

## 1. A 2x2 Subplot Grid Iteration with `axes.flat`
When `nrows > 1` and `ncols > 1`, `axes` is a 2D NumPy array. `axes.flat` allows clean 1D iteration.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 100)

fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 7), dpi=100)

funcs = [
    (np.sin(x), r'Sine: $sin(x)$', '#3498db'),
    (np.cos(x), r'Cosine: $cos(x)$', '#e74c3c'),
    (np.sin(2 * x), r'Higher Frequency: $sin(2x)$', '#2ecc71'),
    (np.sin(x)**2, r'Squared: $sin^2(x)$', '#9b59b6')
]

for ax, (y, title, color) in zip(axes.flat, funcs):
    ax.plot(x, y, color=color, lw=2)
    ax.set_title(title, fontsize=11, fontweight='bold')
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.set_xlim(0, 2 * np.pi)

fig.suptitle("Trigonometric Harmonic Overview (2x2 Panel)", fontsize=14, fontweight='bold')
fig.tight_layout()
plt.show()
```

## 2. Asymmetric Dashboard Layouts with `GridSpec`
`GridSpec` allows an individual subplot to span multiple rows or columns (e.g. 1 full-width top chart and 2 smaller bottom charts).

```python
import matplotlib.gridspec as gridspec

fig = plt.figure(figsize=(10, 6), dpi=100)
gs = gridspec.GridSpec(nrows=2, ncols=2, height_ratios=[1.2, 1])

# Top plot spans both columns (gs[0, :])
ax_top = fig.add_subplot(gs[0, :])
t = np.arange(1, 100)
price = 100 + np.cumsum(np.random.normal(0.2, 1.5, size=99))
ax_top.plot(t, price, color='#2c3e50', lw=2, label='Asset Price Index')
ax_top.set_title("Executive Market Overview", fontsize=12, fontweight='bold')
ax_top.set_ylabel("Price ($)")
ax_top.legend(loc='upper left')
ax_top.grid(True, alpha=0.3)

# Bottom Left: Daily returns histogram
ax_bl = fig.add_subplot(gs[1, 0])
returns = np.diff(price)
ax_bl.hist(returns, bins=20, color='#3498db', edgecolor='black', alpha=0.7)
ax_bl.set_title("Return Volatility Distribution", fontsize=10, fontweight='bold')
ax_bl.set_xlabel("Daily Return ($)")
ax_bl.grid(True, alpha=0.3)

# Bottom Right: Moving average comparison
ax_br = fig.add_subplot(gs[1, 1])
ma_10 = np.convolve(price, np.ones(10)/10, mode='valid')
ax_br.plot(price[9:], ma_10, color='#e67e22', lw=2, label='10-Day MA')
ax_br.set_title("Moving Average Trend", fontsize=10, fontweight='bold')
ax_br.legend()
ax_br.grid(True, alpha=0.3)

fig.tight_layout()
plt.show()
```

## 🎯 Exercises & Practice
1. Create a 1x3 horizontal panel using `plt.subplots(1, 3, figsize=(12, 4))`.
2. Plot a Line plot, a Bar chart, and a Scatter plot on axes 0, 1, and 2 respectively.

```python
# Exercise Solution
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(12, 3.8), dpi=100)

# Panel 1: Line
ax1.plot([1, 2, 3, 4], [10, 24, 36, 40], 'o-', color='#16a085', lw=2)
ax1.set_title("Trend Series")
ax1.grid(True, alpha=0.4)

# Panel 2: Bar
ax2.bar(['A', 'B', 'C'], [45, 78, 52], color='#2980b9', width=0.5)
ax2.set_title("Categorical Totals")
ax2.grid(axis='y', alpha=0.4)

# Panel 3: Scatter
ax3.scatter(np.random.rand(30), np.random.rand(30), color='#e74c3c', alpha=0.7)
ax3.set_title("Feature Correlations")
ax3.grid(True, alpha=0.4)

fig.tight_layout()
plt.show()
```
