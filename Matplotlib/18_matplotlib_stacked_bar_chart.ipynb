# 18. Stacked & 100% Normalized Bar Charts

Stacked bar charts show both aggregate categorical volume and the constituent decomposition within each segment.

### Learning Objectives:
- Construct vertical stacked bars using the `bottom` parameter
- Construct horizontal stacked bars using the `left` parameter
- Build **100% Normalized Stacked Bar Charts** to show percentage composition
- Display segment labels and clean multi-category legends

## 1. Vertical Stacked Bar Chart with Totals
Each segment is stacked on top of previous layers by passing the cumulative heights to `bottom`.```python

````python
import matplotlib.pyplot as plt
import numpy as np

quarters = ['Q1', 'Q2', 'Q3', 'Q4']
software = np.array([45, 52, 60, 72])
services = np.array([30, 35, 42, 50])
hardware = np.array([25, 22, 18, 20])

fig, ax = plt.subplots(figsize=(8, 5), dpi=100)

# Stack bars
p1 = ax.bar(quarters, software, width=0.5, label='Software', color='#2980b9')
p2 = ax.bar(quarters, services, width=0.5, bottom=software, label='Services', color='#27ae60')
p3 = ax.bar(quarters, hardware, width=0.5, bottom=software + services, label='Hardware', color='#f39c12')

# Total values on top of each stack
totals = software + services + hardware
for i, total in enumerate(totals):
    ax.text(i, total + 3, f"${total}M", ha='center', fontweight='bold', fontsize=10)

ax.set_title("Quarterly Enterprise Revenue Decomposition", fontsize=13, fontweight='bold')
ax.set_ylabel("Revenue ($ Millions)")
ax.set_ylim(0, 165)
ax.legend(loc='upper left')
ax.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
````

## 2. 100% Normalized Stacked Bar Chart (Survey Sentiment)
Normalizing each row to 100% allows comparing proportion compositions across groups regardless of total sample size.

````python
departments = ['Engineering', 'Marketing', 'Sales', 'Customer Support']
promoters = np.array([65, 48, 55, 40])
passives = np.array([25, 32, 28, 35])
detractors = np.array([10, 20, 17, 25])

totals = promoters + passives + detractors
pct_prom = promoters / totals * 100
pct_pass = passives / totals * 100
pct_detr = detractors / totals * 100

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)

ax.barh(departments, pct_prom, color='#2ecc71', height=0.55, label='Promoters')
ax.barh(departments, pct_pass, left=pct_prom, color='#f1c40f', height=0.55, label='Passives')
ax.barh(departments, pct_detr, left=pct_prom + pct_pass, color='#e74c3c', height=0.55, label='Detractors')

ax.set_title("Employee Net Promoter Score (eNPS) Composition", fontsize=12, fontweight='bold')
ax.set_xlabel("Percentage Share (%)")
ax.set_xlim(0, 100)
ax.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
ax.grid(axis='x', linestyle=':', alpha=0.5)

plt.show()
````

## 🎯 Exercises & Practice
1. Plot a 100% stacked vertical bar chart comparing Desktop vs Mobile web traffic across 4 websites: `['Site A', 'Site B', 'Site C', 'Site D']`.
2. Desktop: `[60, 40, 30, 75]`, Mobile: `[40, 60, 70, 25]`.

````python
# Exercise Solution
sites = ['Site A', 'Site B', 'Site C', 'Site D']
desktop = np.array([60, 40, 30, 75])
mobile = np.array([40, 60, 70, 25])

fig, ax = plt.subplots(figsize=(7, 4.5), dpi=100)
ax.bar(sites, desktop, label='Desktop', color='#34495e', width=0.5)
ax.bar(sites, mobile, bottom=desktop, label='Mobile', color='#3498db', width=0.5)

ax.set_title("Web Traffic Source Composition", fontsize=12, fontweight='bold')
ax.set_ylabel("Traffic Share (%)")
ax.set_ylim(0, 100)
ax.legend()
ax.grid(axis='y', alpha=0.3)
plt.show()
````
