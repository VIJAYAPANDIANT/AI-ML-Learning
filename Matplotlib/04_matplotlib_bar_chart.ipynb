# 04. Vertical Bar Charts & Grouped Comparisons

Vertical bar charts are essential for visualizing discrete categorical data, comparing performance metrics, and presenting aggregated group sums or averages.

### Learning Objectives:
- Create single vertical bar charts with custom widths, edgecolors, and color palettes
- Automatically display value labels on bars using `ax.bar_label()`
- Implement **grouped / clustered bar charts** comparing multiple series
- Add pattern hatching (`hatch`) and error bar indications

## 1. Single Bar Chart with Automatic Value Labels
`ax.bar_label()` formats and places text annotations directly above or inside bars.

```python
import matplotlib.pyplot as plt
import numpy as np

frameworks = ['TensorFlow', 'PyTorch', 'Scikit-Learn', 'Keras', 'JAX']
stars_k = [185, 88, 59, 61, 31]
colors = ['#ff6f00', '#ee4c2c', '#f7931e', '#d00000', '#4285f4']

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
bars = ax.bar(frameworks, stars_k, color=colors, edgecolor='#333333', linewidth=1, width=0.55)

# Adding automated value labels on top of each bar
ax.bar_label(bars, fmt='%dK', padding=4, fontsize=10, fontweight='bold')

ax.set_title("AI/ML Framework GitHub Stars (Thousands)", fontsize=13, fontweight='bold', pad=12)
ax.set_ylabel("Stars in Thousands", fontsize=11)
ax.set_ylim(0, 215)
ax.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()
```

## 2. Grouped (Clustered) Bar Charts
To display multiple metrics per category side-by-side, offset the x-positions by the bar width.

```python
categories = ['NLP', 'Computer Vision', 'Tabular ML', 'Reinforcement Learning']
accuracy_v1 = [84.2, 79.5, 91.0, 72.3]
accuracy_v2 = [89.7, 86.1, 93.4, 81.0]

x = np.arange(len(categories))
width = 0.35

fig, ax = plt.subplots(figsize=(9, 5), dpi=100)

rects1 = ax.bar(x - width/2, accuracy_v1, width, label='Model v1.0', color='#3498db', edgecolor='black', lw=0.8)
rects2 = ax.bar(x + width/2, accuracy_v2, width, label='Model v2.0 (Optimized)', color='#2ecc71', edgecolor='black', lw=0.8)

ax.bar_label(rects1, fmt='%.1f%%', padding=3, fontsize=9)
ax.bar_label(rects2, fmt='%.1f%%', padding=3, fontsize=9, fontweight='bold')

ax.set_title("Model Accuracy Improvement Across Problem Domains", fontsize=13, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(categories, fontsize=10)
ax.set_ylabel("Top-1 Accuracy (%)", fontsize=11)
ax.set_ylim(60, 100)
ax.legend(loc='lower right')
ax.grid(axis='y', linestyle=':', alpha=0.5)

plt.show()
```

## 🎯 Exercises & Practice
1. Create a grouped bar chart comparing Quarterly Revenue ($ Millions) across 3 regions ('North America', 'Europe', 'Asia-Pacific') for Q1, Q2, and Q3.
2. Apply `ax.bar_label()` to format the values with `$` prefixes and two decimal places.

```python
# Exercise Solution
regions = ['North America', 'Europe', 'Asia-Pacific']
q1 = [12.5, 9.2, 14.8]
q2 = [14.1, 10.5, 17.2]
q3 = [16.8, 12.0, 19.5]

x = np.arange(len(regions))
w = 0.25

fig, ax = plt.subplots(figsize=(9, 5), dpi=100)
b1 = ax.bar(x - w, q1, w, label='Q1', color='#34495e')
b2 = ax.bar(x, q2, w, label='Q2', color='#16a085')
b3 = ax.bar(x + w, q3, w, label='Q3', color='#f39c12')

ax.bar_label(b1, fmt='$%.1fM', padding=3, fontsize=8)
ax.bar_label(b2, fmt='$%.1fM', padding=3, fontsize=8)
ax.bar_label(b3, fmt='$%.1fM', padding=3, fontsize=8, fontweight='bold')

ax.set_title("Quarterly Regional Revenue Growth", fontsize=12, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(regions, fontsize=10)
ax.set_ylabel("Revenue ($ Millions)", fontsize=11)
ax.set_ylim(0, 24)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```
