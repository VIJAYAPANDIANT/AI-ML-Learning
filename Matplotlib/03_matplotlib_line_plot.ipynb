# 03. Line Plots & Multi-Series Trends

Line plots are the primary visualization tool for continuous sequential trends, time series data, and machine learning training dynamics (loss and metric progression across epochs).

### Learning Objectives:
- Create single and multi-line series with custom linestyles, colors, and markers
- Visualize Machine Learning model training loss vs. validation loss
- Use logarithmic scaling on axes with `ax.set_yscale('log')`
- Set custom limits and axis boundaries

## 1. Multiple Lines & Aesthetic Parameters
Parameters available in `ax.plot()`: `color`, `linewidth` (`lw`), `linestyle` (`ls`), `marker`, `markersize`, `alpha`, `label`.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 20)
y1 = x
y2 = x ** 1.5
y3 = x ** 2

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)

ax.plot(x, y1, color='#1abc9c', lw=2, ls='-', marker='o', label='Linear: O(n)')
ax.plot(x, y2, color='#3498db', lw=2, ls='--', marker='s', label=r'Superlinear: $O(n^{1.5})$')
ax.plot(x, y3, color='#e74c3c', lw=2, ls='-.', marker='^', label=r'Quadratic: $O(n^2)$')

ax.set_title("Algorithm Growth Comparison", fontsize=13, fontweight='bold')
ax.set_xlabel("Input Size (n)", fontsize=11)
ax.set_ylabel("Operations Count", fontsize=11)
ax.legend(loc='upper left', frameon=True)
ax.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

## 2. Machine Learning Training Dynamics (Loss Curves)
Tracking training loss and validation loss across epochs is essential to diagnose underfitting, convergence, and overfitting.

```python
np.random.seed(42)
epochs = np.arange(1, 51)

# Simulated training loss decaying smoothly
train_loss = 1.8 * np.exp(-0.08 * epochs) + 0.15 + np.random.normal(0, 0.015, size=50)
# Simulated validation loss bottoming out and showing slight overfitting
val_loss = 1.9 * np.exp(-0.07 * epochs) + 0.22 + 0.003 * np.maximum(0, epochs - 30) + np.random.normal(0, 0.02, size=50)

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)
ax.plot(epochs, train_loss, color='#2980b9', lw=2.5, label='Training Loss')
ax.plot(epochs, val_loss, color='#e67e22', lw=2.5, ls='--', label='Validation Loss')

# Mark best epoch checkpoint
best_epoch = np.argmin(val_loss) + 1
ax.axvline(best_epoch, color='#27ae60', ls=':', lw=2, label=f'Best Checkpoint (Epoch {best_epoch})')

ax.set_title("Deep Learning Model Convergence Curve", fontsize=13, fontweight='bold')
ax.set_xlabel("Epoch", fontsize=11)
ax.set_ylabel("Cross-Entropy Loss", fontsize=11)
ax.set_ylim(0, 2.2)
ax.grid(True, alpha=0.4)
ax.legend(loc='upper right')

plt.show()
```

## 🎯 Exercises & Practice
1. Plot the exponential decay function $N(t) = N_0 e^{-\lambda t}$ with $N_0 = 1000$ and decay rates $\lambda = [0.1, 0.2, 0.5]$ over time $t \in [0, 25]$.
2. Set the Y-axis to logarithmic scale (`ax.set_yscale('log')`) to demonstrate how exponential curves become linear under log scaling.

```python
# Exercise Solution
t = np.linspace(0, 25, 100)
N0 = 1000
lambdas = [0.1, 0.2, 0.5]
colors = ['#3498db', '#9b59b6', '#e74c3c']

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5), dpi=100)

for lam, col in zip(lambdas, colors):
    N = N0 * np.exp(-lam * t)
    ax1.plot(t, N, lw=2.2, color=col, label=f'λ = {lam}')
    ax2.plot(t, N, lw=2.2, color=col, label=f'λ = {lam}')

ax1.set_title("Linear Scale Decay", fontsize=11, fontweight='bold')
ax1.set_xlabel("Time (t)")
ax1.set_ylabel("Quantity N(t)")
ax1.legend()
ax1.grid(True, alpha=0.5)

ax2.set_yscale('log')
ax2.set_title("Logarithmic Scale Decay", fontsize=11, fontweight='bold')
ax2.set_xlabel("Time (t)")
ax2.set_ylabel("Quantity N(t) [Log]")
ax2.legend()
ax2.grid(True, alpha=0.5, which='both')

fig.tight_layout()
plt.show()
```
