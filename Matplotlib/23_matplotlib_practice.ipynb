## Problem 1: Dual Y-Axis Machine Learning Training Dynamics
Plot Loss (0 to 2.5) on the left Y-axis and Validation Accuracy (0 to 100%) on the right twin axis across 30 epochs.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
epochs = np.arange(1, 31)
loss = 2.0 * np.exp(-0.12 * epochs) + 0.15 + np.random.normal(0, 0.02, 30)
accuracy = (1 - np.exp(-0.10 * epochs)) * 88 + 10 + np.random.normal(0, 0.8, 30)

fig, ax_loss = plt.subplots(figsize=(8.5, 4.5), dpi=100)

color_loss = '#c0392b'
ax_loss.plot(epochs, loss, color=color_loss, lw=2.2, label='Cross-Entropy Loss')
ax_loss.set_xlabel("Epoch", fontweight='bold')
ax_loss.set_ylabel("Loss", color=color_loss, fontweight='bold')
ax_loss.tick_params(axis='y', labelcolor=color_loss)
ax_loss.set_ylim(0, 2.5)
ax_loss.grid(True, alpha=0.3)

ax_acc = ax_loss.twinx()
color_acc = '#27ae60'
ax_acc.plot(epochs, accuracy, color=color_acc, lw=2.2, ls='--', marker='o', markersize=4, label='Accuracy (%)')
ax_acc.set_ylabel("Accuracy (%)", color=color_acc, fontweight='bold')
ax_acc.tick_params(axis='y', labelcolor=color_acc)
ax_acc.set_ylim(50, 100)

plt.title("Problem 1: Dual-Scale Model Training Performance", fontsize=12, fontweight='bold', pad=12)
plt.show()
```

## Problem 2: Algorithmic Big-O Runtime Complexity Comparison

```python
n = np.linspace(1, 20, 100)

fig, ax = plt.subplots(figsize=(8.5, 4.5), dpi=100)
ax.plot(n, np.ones_like(n), label=r'$O(1)$ Constant', lw=2)
ax.plot(n, np.log2(n), label=r'$O(log n)$ Logarithmic', lw=2)
ax.plot(n, n, label=r'$O(n)$ Linear', lw=2)
ax.plot(n, n * np.log2(n), label=r'$O(n log n)$ Linearithmic', lw=2)
ax.plot(n, n**2, label=r'$O(n^2)$ Quadratic', lw=2)

ax.set_ylim(0, 50)
ax.set_title("Problem 2: Big-O Computational Growth Rates", fontsize=12, fontweight='bold')
ax.set_xlabel("Input Size (n)")
ax.set_ylabel("Operations Count")
ax.legend(loc='upper left')
ax.grid(True, alpha=0.3)

plt.show()
```

## Problem 3: Ranked Feature Importance with Color Gradient

```python
features = ['Age', 'Credit_Score', 'DTI_Ratio', 'Income', 'Loan_Tenure', 'Dependents']
weights = [0.12, 0.28, 0.22, 0.25, 0.08, 0.05]

sort_idx = np.argsort(weights)
f_sorted = [features[i] for i in sort_idx]
w_sorted = [weights[i] for i in sort_idx]

# Colormap mapping
cmap = plt.cm.Blues
colors = [cmap(0.3 + 0.7 * (w / max(w_sorted))) for w in w_sorted]

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
bars = ax.barh(f_sorted, w_sorted, color=colors, edgecolor='black', lw=0.6)
ax.bar_label(bars, fmt='%.2f', padding=4, fontweight='bold')
ax.set_xlim(0, 0.35)
ax.set_title("Problem 3: Machine Learning Feature Weights", fontsize=11, fontweight='bold')
ax.grid(axis='x', alpha=0.3)
plt.show()
```

## Problem 4: Customer RFM Segmentation Bubble Chart

```python
np.random.seed(42)
n_cust = 80
recency = np.random.uniform(5, 365, n_cust) # Days
monetary = np.random.uniform(100, 5000, n_cust) # Spend $
frequency = np.random.randint(1, 25, n_cust) # Orders count
clv_score = (monetary / 5000) * 50 + (frequency / 25) * 50

fig, ax = plt.subplots(figsize=(8.5, 5), dpi=100)
scatter = ax.scatter(recency, monetary, s=frequency * 12, c=clv_score, cmap='plasma', alpha=0.75, edgecolors='black', lw=0.5)

cbar = fig.colorbar(scatter, ax=ax)
cbar.set_label("Customer Health Index (CLV)", fontsize=10)

ax.set_title("Problem 4: Customer RFM Multivariate Scatter", fontsize=12, fontweight='bold')
ax.set_xlabel("Recency (Days Since Last Order)")
ax.set_ylabel("Monetary Spend ($)")
ax.grid(True, alpha=0.3)
plt.show()
```

## Problem 5: Multi-Class Probability Density Histograms

```python
np.random.seed(11)
c0 = np.random.normal(30, 8, 400)
c1 = np.random.normal(55, 10, 400)

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
ax.hist(c0, bins=25, density=True, alpha=0.6, color='#2980b9', label='Class 0 (Benign)', edgecolor='white')
ax.hist(c1, bins=25, density=True, alpha=0.6, color='#e74c3c', label='Class 1 (Malignant)', edgecolor='white')

ax.set_title("Problem 5: Diagnostic Risk Score Probability Density", fontsize=11, fontweight='bold')
ax.set_xlabel("Predicted Risk Score")
ax.set_ylabel("Density")
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

## Problem 6: Stock Price & Volume Dual-Panel Time-Series Dashboard

```python
import matplotlib.gridspec as gridspec

t = np.arange(1, 41)
price = 120 + np.cumsum(np.random.normal(0.4, 1.5, 40))
volume = np.random.randint(15, 60, 40)

fig = plt.figure(figsize=(9, 5), dpi=100)
gs = gridspec.GridSpec(2, 1, height_ratios=[2, 1])

ax0 = fig.add_subplot(gs[0])
ax0.plot(t, price, color='#2c3e50', lw=2)
ax0.set_title("Problem 6: Equity Price & Trading Volume", fontsize=11, fontweight='bold')
ax0.set_ylabel("Price ($)")
ax0.grid(True, alpha=0.3)

ax1 = fig.add_subplot(gs[1], sharex=ax0)
ax1.bar(t, volume, color='#3498db', width=0.6)
ax1.set_xlabel("Trading Day")
ax1.set_ylabel("Volume (K)")
ax1.grid(True, alpha=0.3)

fig.tight_layout()
plt.show()
```

## Problem 7: Global Market Share Donut Chart

```python
brands = ['Brand Alpha', 'Brand Beta', 'Brand Gamma', 'Others']
shares = [44, 28, 18, 10]

fig, ax = plt.subplots(figsize=(6, 6), dpi=100)
ax.pie(shares, labels=brands, autopct='%1.0f%%', startangle=60, colors=['#3498db', '#2ecc71', '#e67e22', '#95a5a6'], wedgeprops=dict(width=0.38, edgecolor='white', lw=2))
ax.text(0, 0, '100%\nMarket', ha='center', va='center', fontweight='bold', fontsize=12)
ax.set_title("Problem 7: Industry Market Share", fontsize=11, fontweight='bold')
plt.show()
```

## Problem 8: Departmental Salary Distribution Boxplots

```python
d1 = np.random.normal(70, 8, 40)
d2 = np.random.normal(85, 12, 40)
d3 = np.random.normal(110, 18, 40)

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
bp = ax.boxplot([d1, d2, d3], labels=['Support', 'Operations', 'Engineering'], patch_artist=True)
for box, col in zip(bp['boxes'], ['#1abc9c', '#f39c12', '#9b59b6']):
    box.set_facecolor(col)
    box.set_alpha(0.7)

ax.set_title("Problem 8: Salary Benchmark Distribution ($K)", fontsize=11, fontweight='bold')
ax.set_ylabel("Annual Compensation ($K)")
ax.grid(axis='y', alpha=0.3)
plt.show()
```

## Problem 9: 2x2 Publication Dashboard with Shared Styling

```python
fig, axes = plt.subplots(2, 2, figsize=(9, 6), dpi=100)

axes[0, 0].plot([1, 2, 3], [10, 20, 15], 'o-', color='#e74c3c')
axes[0, 0].set_title("Metric Trend")
axes[0, 1].bar(['A', 'B'], [50, 70], color='#2980b9')
axes[0, 1].set_title("Cohort Volume")
axes[1, 0].hist(np.random.randn(100), color='#27ae60', edgecolor='white')
axes[1, 0].set_title("Residuals")
axes[1, 1].scatter(np.random.rand(30), np.random.rand(30), color='#8e44ad')
axes[1, 1].set_title("Feature Spread")

for ax in axes.flat:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.grid(True, alpha=0.3)

fig.suptitle("Problem 9: Multi-Panel Analytics Dashboard", fontsize=13, fontweight='bold')
fig.tight_layout()
plt.show()
```

## Problem 10: 3D Optimization Surface & Gradient Descent Trajectory

```python
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-3, 3, 50)
y = np.linspace(-3, 3, 50)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2 # Convex Bowl

# Simulated gradient descent steps
traj_x = np.array([2.5, 1.8, 1.2, 0.7, 0.3, 0.05])
traj_y = np.array([2.5, 1.8, 1.1, 0.6, 0.2, 0.02])
traj_z = traj_x**2 + traj_y**2

fig = plt.figure(figsize=(8, 6), dpi=100)
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='coolwarm', alpha=0.5, edgecolor='none')
ax.plot(traj_x, traj_y, traj_z, color='red', marker='o', lw=2.5, label='Gradient Descent Path')

ax.set_title("Problem 10: 3D Gradient Descent Convergence", fontsize=11, fontweight='bold')
ax.legend()
plt.show()
```
