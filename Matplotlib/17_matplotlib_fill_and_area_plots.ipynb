# 17. Area Plots & Confidence Interval Shading

Area plots and region fills communicate cumulative shares over time and visualize statistical uncertainty bands (confidence intervals) around predictive curves.

### Learning Objectives:
- Shade regions between lines with `ax.fill_between()`
- Visualize $\pm 1.96 \sigma$ (95%) confidence intervals around regression forecasts
- Conditionally shade regions with `where=...` (e.g. positive vs negative profit)
- Construct cumulative multi-series area plots with `ax.stackplot()`

## 1. Regression Forecast with 95% Confidence Interval
Shading the upper and lower bounds communicates model uncertainty.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(1, 10, 50)
y_true = 2.0 * x + 5.0
# Forecast with increasing uncertainty over time
uncertainty = 0.8 + 0.35 * x

fig, ax = plt.subplots(figsize=(8.5, 4.5), dpi=100)

ax.plot(x, y_true, color='#2980b9', lw=2.5, label='Predicted Mean ($\\hat{y}$)')
ax.fill_between(
    x, 
    y_true - 1.96 * uncertainty, 
    y_true + 1.96 * uncertainty, 
    color='#3498db', 
    alpha=0.25, 
    label='95% Confidence Interval'
)

ax.set_title("Machine Learning Forecast with Expanding Prediction Horizon", fontsize=12, fontweight='bold')
ax.set_xlabel("Forecast Horizon (Months)")
ax.set_ylabel("Revenue ($ Millions)")
ax.legend(loc='upper left')
ax.grid(True, alpha=0.3)

plt.show()
```

## 2. Conditional Shading (Profit vs Loss)
Using the `where` parameter in `fill_between` allows separate color shading for positive gains and negative deficits.

```python
t = np.linspace(0, 12, 100)
net_income = 15 * np.sin(t) - 2

fig, ax = plt.subplots(figsize=(9, 4), dpi=100)
ax.plot(t, net_income, color='#2c3e50', lw=2)
ax.axhline(0, color='black', lw=1, ls='--')

# Shade profit in green
ax.fill_between(t, net_income, 0, where=(net_income >= 0), color='#2ecc71', alpha=0.4, label='Net Surplus')
# Shade deficit in red
ax.fill_between(t, net_income, 0, where=(net_income < 0), color='#e74c3c', alpha=0.4, label='Net Deficit')

ax.set_title("Monthly Net Income Surplus vs Deficit", fontsize=12, fontweight='bold')
ax.set_xlabel("Month")
ax.set_ylabel("Operating Margin ($K)")
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
```

## 🎯 Exercises & Practice
1. Generate an array $x \in [0, 10]$ and plot a sinusoidal wave.
2. Shade the region between the curve and the baseline $y = 0$ in sky blue with 30% opacity.

```python
# Exercise Solution
x_eval = np.linspace(0, 10, 100)
y_eval = np.sin(x_eval)

fig, ax = plt.subplots(figsize=(8, 3.8), dpi=100)
ax.plot(x_eval, y_eval, color='#0984e3', lw=2)
ax.fill_between(x_eval, y_eval, 0, color='#74b9ff', alpha=0.35)
ax.axhline(0, color='gray', lw=0.8)
ax.set_title("Sinusoidal Wave Area Fill", fontsize=11, fontweight='bold')
ax.grid(True, alpha=0.3)
plt.show()
```
