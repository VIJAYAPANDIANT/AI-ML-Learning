# 14. Grids, Ticks & Dual-Axis (Twinx) Charts

Controlling tick marks, tick frequency, custom formatters, and secondary axes is essential for presenting financial, engineering, and multi-scale data.

### Learning Objectives:
- Configure major and minor grids with custom intervals and linestyles
- Format tick values with `PercentFormatter` and `FuncFormatter`
- Rotate categorical tick labels to avoid collisions
- Construct **Dual-Axis charts** with `ax.twinx()` (e.g. Revenue Bar vs Margin % Line)

## 1. Major and Minor Grid Configuration & Formatters
Matplotlib's ticker module gives granular control over tick intervals.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import MultipleLocator, PercentFormatter

x = np.linspace(0, 100, 50)
y = 0.4 + 0.5 * (1 - np.exp(-0.05 * x))

fig, ax = plt.subplots(figsize=(8.5, 4.5), dpi=100)
ax.plot(x, y, color='#2980b9', lw=2.5, label='Efficiency Yield')

# Configure X-axis tick locators
ax.xaxis.set_major_locator(MultipleLocator(20))
ax.xaxis.set_minor_locator(MultipleLocator(5))

# Configure Y-axis format as percentages
ax.yaxis.set_major_formatter(PercentFormatter(xmax=1.0))

# Major & Minor Grids
ax.grid(which='major', color='#bdc3c7', linestyle='-', linewidth=0.8)
ax.grid(which='minor', color='#ecf0f1', linestyle=':', linewidth=0.5)

ax.set_title("Conversion Efficiency with Major/Minor Grids", fontsize=12, fontweight='bold')
ax.set_xlabel("Compute Resources Allocated (Units)")
ax.set_ylabel("Yield Percentage")
ax.legend()

plt.show()
```

## 2. Dual-Axis Chart with `ax.twinx()`
A twin axis shares the X-axis while providing an independent Y-axis on the right side.

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug']
revenue = [120, 135, 148, 162, 175, 190, 205, 220] # in $K
margin_pct = [18.2, 19.1, 18.8, 21.0, 22.4, 23.1, 22.8, 24.5] # in %

x = np.arange(len(months))

fig, ax1 = plt.subplots(figsize=(9, 4.5), dpi=100)

# Primary Axis: Revenue Bar Chart
color1 = '#3498db'
bars = ax1.bar(x, revenue, width=0.45, color=color1, alpha=0.8, label='Revenue ($K)')
ax1.set_xlabel("Fiscal Month", fontweight='bold')
ax1.set_ylabel("Revenue ($ Thousands)", color=color1, fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(months)
ax1.tick_params(axis='y', labelcolor=color1)
ax1.set_ylim(0, 260)

# Secondary Axis: Margin % Line Chart
ax2 = ax1.twinx()
color2 = '#e67e22'
line = ax2.plot(x, margin_pct, color=color2, lw=2.5, marker='o', label='Operating Margin (%)')
ax2.set_ylabel("Margin Percentage (%)", color=color2, fontweight='bold')
ax2.tick_params(axis='y', labelcolor=color2)
ax2.set_ylim(10, 30)

plt.title("Financial Growth: Revenue vs. Operating Margin", fontsize=13, fontweight='bold', pad=12)
plt.show()
```

## 🎯 Exercises & Practice
1. Plot temperature ($^{\circ}\text{C}$) on the left Y-axis and Humidity (\%) on a twin right Y-axis for 7 consecutive days.
2. Use distinct colors for lines and axis tick labels.

```python
# Exercise Solution
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temp = [22, 24, 21, 26, 28, 25, 23]
humidity = [65, 58, 72, 50, 45, 60, 68]

fig, ax_temp = plt.subplots(figsize=(8, 4), dpi=100)
ax_temp.plot(days, temp, color='#e74c3c', marker='o', lw=2, label='Temp (°C)')
ax_temp.set_ylabel("Temperature (°C)", color='#e74c3c', fontweight='bold')
ax_temp.tick_params(axis='y', labelcolor='#e74c3c')
ax_temp.set_ylim(15, 35)

ax_hum = ax_temp.twinx()
ax_hum.plot(days, humidity, color='#2980b9', marker='s', ls='--', lw=2, label='Humidity (%)')
ax_hum.set_ylabel("Humidity (%)", color='#2980b9', fontweight='bold')
ax_hum.tick_params(axis='y', labelcolor='#2980b9')
ax_hum.set_ylim(30, 85)

plt.title("Weekly Weather Metrics (Dual Axis)", fontsize=11, fontweight='bold')
plt.show()
```
