# 06. Scatter Plots & Multivariate Encodings

Scatter plots display pairwise relationships between continuous variables, detect clusters, and encode multi-dimensional data via position ($X, Y$), size ($S$), and color ($C$).

### Learning Objectives:
- Construct scatter plots using `ax.scatter()`
- Map a 3rd continuous variable to marker size and a 4th to color with a `colorbar`
- Overlay regression trendlines on scatter distributions
- Visualize Machine Learning clustering partitions (e.g., K-Means)

## 1. Multi-Dimensional Bubble Scatter
In a single plot, we can encode 4 dimensions:
1. **X-axis**: House Size (SqFt)
2. **Y-axis**: House Price ($)
3. **Bubble Size (`s`)**: Number of Bedrooms
4. **Color (`c` & `cmap`)**: Property Age (Years)

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
n_houses = 120

sqft = np.random.uniform(900, 3800, n_houses)
price = sqft * 220 + np.random.normal(0, 45000, n_houses) + 50000
bedrooms = np.random.choice([2, 3, 4, 5], size=n_houses, p=[0.2, 0.4, 0.3, 0.1])
age_years = np.random.uniform(1, 40, n_houses)

fig, ax = plt.subplots(figsize=(9, 5), dpi=100)

scatter = ax.scatter(
    sqft, price / 1000, 
    s=bedrooms * 30, # Marker size mapped to bedrooms
    c=age_years,     # Color mapped to house age
    cmap='viridis', 
    alpha=0.75, 
    edgecolors='black', 
    linewidth=0.5
)

# Add continuous colorbar
cbar = fig.colorbar(scatter, ax=ax)
cbar.set_label("Property Age (Years)", fontsize=10)

# Overlay linear trendline
m, b = np.polyfit(sqft, price / 1000, 1)
ax.plot(sqft, m * sqft + b, color='#e74c3c', lw=2, linestyle='--', label=f'Trendline: ${m:.2f}k/sqft')

ax.set_title("Real Estate Valuation: Price vs Size (Multivariate)", fontsize=13, fontweight='bold')
ax.set_xlabel("Living Area (Sq Ft)", fontsize=11)
ax.set_ylabel("Price ($ Thousands)", fontsize=11)
ax.legend(loc='upper left')
ax.grid(True, alpha=0.3)

plt.show()
```

## 2. Visualizing Machine Learning Clusters
Scatter plots are universally used to visualize 2D PCA or t-SNE projections of high-dimensional ML clusters.

```python
# Generating 3 synthetic clusters
cluster_1 = np.random.normal(loc=[2.0, 3.0], scale=0.6, size=(60, 2))
cluster_2 = np.random.normal(loc=[6.0, 7.0], scale=0.7, size=(60, 2))
cluster_3 = np.random.normal(loc=[7.5, 2.5], scale=0.5, size=(60, 2))

centroids = np.array([[2.0, 3.0], [6.0, 7.0], [7.5, 2.5]])

fig, ax = plt.subplots(figsize=(8, 5), dpi=100)

ax.scatter(cluster_1[:, 0], cluster_1[:, 1], color='#3498db', label='Cluster 0 (Low Spend / Low Freq)', alpha=0.7)
ax.scatter(cluster_2[:, 0], cluster_2[:, 1], color='#2ecc71', label='Cluster 1 (High Spend / High Freq)', alpha=0.7)
ax.scatter(cluster_3[:, 0], cluster_3[:, 1], color='#9b59b6', label='Cluster 2 (High Spend / Low Freq)', alpha=0.7)

# Mark cluster centroids
ax.scatter(centroids[:, 0], centroids[:, 1], s=180, c='red', marker='X', edgecolors='black', label='Centroids')

ax.set_title("Customer Segmentation via K-Means Clustering", fontsize=12, fontweight='bold')
ax.set_xlabel("Normalized Annual Spend", fontsize=10)
ax.set_ylabel("Normalized Purchase Frequency", fontsize=10)
ax.legend(loc='upper left', framealpha=0.9)
ax.grid(True, linestyle=':', alpha=0.5)

plt.show()
```

## 🎯 Exercises & Practice
1. Simulate 150 test samples with advertising spend ($x$) and revenue ($y$) having a non-linear relationship: $y = 50 + 10\sqrt{x} + \epsilon$.
2. Create a scatter plot with alpha transparency and plot a quadratic polynomial fit line.

```python
# Exercise Solution
np.random.seed(101)
ad_spend = np.random.uniform(10, 500, 150)
revenue = 50 + 10 * np.sqrt(ad_spend) + np.random.normal(0, 15, 150)

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
ax.scatter(ad_spend, revenue, color='#16a085', alpha=0.6, edgecolors='none', label='Observations')

# Fit 2nd order polynomial
p = np.poly1d(np.polyfit(ad_spend, revenue, 2))
x_line = np.linspace(10, 500, 100)
ax.plot(x_line, p(x_line), color='#c0392b', lw=2.5, label='Fitted Curve (Diminishing Returns)')

ax.set_title("Advertising Spend vs Revenue Response", fontsize=12, fontweight='bold')
ax.set_xlabel("Ad Spend ($ Thousands)")
ax.set_ylabel("Generated Revenue ($ Thousands)")
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
```
