# 21. 3D Plotting & Spatial Visualizations

Matplotlib's `mplot3d` toolkit enables rendering 3D scatter plots, line trajectories, wireframes, and loss function optimization surfaces.

### Learning Objectives:
- Initialize 3D axes using `fig.add_subplot(projection='3d')`
- Construct 3D scatter plots with color mapping
- Build 3D continuous surface plots with `ax.plot_surface()` and `np.meshgrid()`
- Adjust elevation and azimuthal camera viewing angles using `ax.view_init()`

## 1. Machine Learning Loss Function 3D Surface
Visualizing a 2D optimization loss surface: $z = \sin\left(\sqrt{x^2 + y^2}\right)$.

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

x = np.linspace(-4, 4, 80)
y = np.linspace(-4, 4, 80)
X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

fig = plt.figure(figsize=(9, 6), dpi=100)
ax = fig.add_subplot(111, projection='3d')

# Surface plot with viridis colormap
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', alpha=0.9)

# Adjust viewing perspective: elevation=35, azimuth=45
ax.view_init(elev=35, azim=45)

ax.set_title("3D Loss Surface: Non-Convex Landscape", fontsize=13, fontweight='bold', pad=15)
ax.set_xlabel("Parameter θ₁", labelpad=8)
ax.set_ylabel("Parameter θ₂", labelpad=8)
ax.set_zlabel("Loss Value J(θ)", labelpad=8)

fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10, label='Loss Magnitude')
plt.show()
```

## 2. 3D Scatter Plot with Cluster Visualizations
Plotting 3 distinct clusters in three-dimensional feature space.

```python
np.random.seed(42)
n = 80
c1 = np.random.normal(loc=[2, 2, 2], scale=0.6, size=(n, 3))
c2 = np.random.normal(loc=[6, 6, 6], scale=0.7, size=(n, 3))
c3 = np.random.normal(loc=[2, 7, 3], scale=0.5, size=(n, 3))

fig = plt.figure(figsize=(8, 6), dpi=100)
ax = fig.add_subplot(111, projection='3d')

ax.scatter(c1[:, 0], c1[:, 1], c1[:, 2], color='#3498db', label='Cluster 0', s=35, alpha=0.8)
ax.scatter(c2[:, 0], c2[:, 1], c2[:, 2], color='#e74c3c', label='Cluster 1', s=35, alpha=0.8)
ax.scatter(c3[:, 0], c3[:, 1], c3[:, 2], color='#2ecc71', label='Cluster 2', s=35, alpha=0.8)

ax.set_title("3D Feature Space Clustering Projection", fontsize=12, fontweight='bold')
ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
ax.set_zlabel("Feature 3")
ax.legend()

plt.show()
```

## 🎯 Exercises & Practice
1. Create a 3D helix trajectory: $x = \cos(t)$, $y = \sin(t)$, $z = t$ for $t \in [0, 6\pi]$.
2. Plot using `ax.plot3D()`.

```python
# Exercise Solution
t_vals = np.linspace(0, 6 * np.pi, 300)
x_h = np.cos(t_vals)
y_h = np.sin(t_vals)
z_h = t_vals

fig = plt.figure(figsize=(7, 5), dpi=100)
ax = fig.add_subplot(111, projection='3d')
ax.plot(x_h, y_h, z_h, color='#e67e22', lw=2.5)
ax.set_title("3D Parametric Spiral Helix", fontsize=11, fontweight='bold')
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z (Time)")
plt.show()
```
