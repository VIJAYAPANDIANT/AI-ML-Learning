# 20. Time Series & Date/Time Plotting

Handling temporal axes requires specialized date formatters, automatic tick rotation, and moving average overlays.

### Learning Objectives:
- Plot temporal data using Pandas timestamps and `matplotlib.dates`
- Format date axes using `DateFormatter`, `MonthLocator`, and `DayLocator`
- Use `fig.autofmt_xdate()` for clean angle rotations
- Plot stock prices with 20-day and 50-day moving averages (SMA)

## 1. Stock Price Series with Dual Moving Averages
Combining daily prices with smoothed moving average trendlines.

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import numpy as np
import pandas as pd

np.random.seed(42)
dates = pd.date_range(start='2026-01-01', periods=120, freq='B') # Business days
price = 100 + np.cumsum(np.random.normal(0.3, 1.8, len(dates)))
df = pd.DataFrame({'Price': price}, index=dates)
df['SMA_15'] = df['Price'].rolling(window=15).mean()
df['SMA_30'] = df['Price'].rolling(window=30).mean()

fig, ax = plt.subplots(figsize=(10, 4.5), dpi=100)

ax.plot(df.index, df['Price'], color='#bdc3c7', lw=1.2, label='Daily Close')
ax.plot(df.index, df['SMA_15'], color='#2980b9', lw=2.0, label='15-Day SMA')
ax.plot(df.index, df['SMA_30'], color='#e67e22', lw=2.0, label='30-Day SMA')

# Configure date locators and formatters
ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
ax.xaxis.set_minor_locator(mdates.DayLocator(interval=7))

# Automatically rotate and align date labels
fig.autofmt_xdate()

ax.set_title("Equities Trend Analysis: Price & Moving Average Crossover", fontsize=12, fontweight='bold')
ax.set_ylabel("Price ($)")
ax.legend(loc='upper left')
ax.grid(True, alpha=0.3)

plt.show()
```

## 🎯 Exercises & Practice
1. Generate 60 days of daily website visitors starting from `'2026-06-01'`.
2. Plot the daily traffic and apply `mdates.DateFormatter('%d %b')` with `fig.autofmt_xdate()`.

```python
# Exercise Solution
dates_ex = pd.date_range('2026-06-01', periods=60)
visitors = 5000 + np.cumsum(np.random.normal(50, 300, 60))

fig, ax = plt.subplots(figsize=(9, 4), dpi=100)
ax.plot(dates_ex, visitors, color='#8e44ad', lw=2)
ax.xaxis.set_major_formatter(mdates.DateFormatter('%d %b'))
fig.autofmt_xdate()

ax.set_title("Daily Active Users Traffic Trend", fontsize=11, fontweight='bold')
ax.set_ylabel("Active Users")
ax.grid(True, alpha=0.3)
plt.show()
```
