# 12. Multiple Linear Regression & Regularization (Ridge & Lasso)

Multiple linear regression predicts target $y$ from $p$ features: $y = \beta_0 + \sum_{j=1}^p \beta_j x_j$. When features are collinear, standard OLS overfits, requiring $L_2$ (Ridge) or $L_1$ (Lasso) penalties.

### Regularization Types:
- **Ridge Regression ($L_2$)**: Adds $\alpha \sum \beta_j^2$ penalty. Shrinks weights toward zero; handles multicollinearity.
- **Lasso Regression ($L_1$)**: Adds $\alpha \sum |\beta_j|$ penalty. Drives uninformative weights exactly to zero, performing automatic feature selection.

```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score

housing = fetch_california_housing(as_frame=True)
X, y = housing.data, housing.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
scaler = StandardScaler()
X_tr_scaled = scaler.fit_transform(X_train)
X_te_scaled = scaler.transform(X_test)

# Train OLS, Ridge, and Lasso
ols = LinearRegression().fit(X_tr_scaled, y_train)
ridge = Ridge(alpha=10.0).fit(X_tr_scaled, y_train)
lasso = Lasso(alpha=0.05).fit(X_tr_scaled, y_train)

print(f"OLS Test R²:   {r2_score(y_test, ols.predict(X_te_scaled)):.4f}")
print(f"Ridge Test R²: {r2_score(y_test, ridge.predict(X_te_scaled)):.4f}")
print(f"Lasso Test R²: {r2_score(y_test, lasso.predict(X_te_scaled)):.4f}")
print(f"Lasso Non-Zero Features: {np.sum(lasso.coef_ != 0)} of {X.shape[1]}")
```
