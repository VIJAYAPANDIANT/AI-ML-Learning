# 29. K-Fold Cross-Validation

A single train/test split can produce variable results depending on which samples randomly land in the test fold. **$K$-Fold Cross-Validation** partitions data into $K$ equal subsets, training on $K-1$ and validating on the remaining 1, rotating through all folds.

### Cross-Validation Strategies:
- **KFold**: Standard continuous regression partitioning
- **StratifiedKFold**: Ensures equal class representation across all folds (classification standard)
- `cross_val_score()`: Evaluates a single performance metric across folds
- `cross_validate()`: Evaluates multiple metrics and returns fit times

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier
import numpy as np

iris = load_iris()
X, y = iris.data, iris.target

rf = RandomForestClassifier(n_estimators=50, random_state=42)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = cross_val_score(rf, X, y, cv=cv, scoring='accuracy')

print(f"5-Fold Accuracy Scores: {scores.round(4)}")
print(f"Mean CV Accuracy:      {scores.mean() * 100:.2f}% (± {scores.std() * 100:.2f}%)")
```

```python
# Exercise Solution
from sklearn.model_selection import cross_validate
multi_scores = cross_validate(rf, X, y, cv=cv, scoring=['accuracy', 'f1_macro'])
print("Test Accuracy Mean:", multi_scores['test_accuracy'].mean().round(4))
print("Test F1-Macro Mean:", multi_scores['test_f1_macro'].mean().round(4))
```
