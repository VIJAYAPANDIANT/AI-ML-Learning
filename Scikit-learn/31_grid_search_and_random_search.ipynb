# 31. Automated Tuning: `GridSearchCV` vs `RandomizedSearchCV`

Automating the search across hyperparameter combinations is crucial for maximizing predictive power.

### Comparison:
- **GridSearchCV**: Exhaustively evaluates every combination on the Cartesian product grid. Guarantees finding the best configuration within the grid, but computationally expensive.
- **RandomizedSearchCV**: Samples fixed $N$ iterations from specified parameter distributions. Much faster and finds near-optimal parameters with fraction of compute time.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
from scipy.stats import randint

cancer = load_breast_cancer()
X_tr, X_te, y_tr, y_te = train_test_split(cancer.data, cancer.target, test_size=0.25, random_state=42)

# 1. RandomizedSearchCV across distributions
param_dist = {
    'n_estimators': randint(50, 200),
    'max_depth': randint(2, 12),
    'min_samples_split': randint(2, 10)
}

rs = RandomizedSearchCV(RandomForestClassifier(random_state=42), param_dist, n_iter=15, cv=3, random_state=42, n_jobs=-1)
rs.fit(X_tr, y_tr)

print(f"Best Parameters from Randomized Search: {rs.best_params_}")
print(f"Best CV Score: {rs.best_score_ * 100:.2f}%")
print(f"Holdout Test Accuracy: {rs.score(X_te, y_te) * 100:.2f}%")
```

## 🎯 Exercises & Practice
1. Narrow down the search around `rs.best_params_['n_estimators']` using a focused `GridSearchCV`.

```python
# Exercise Solution
best_n = rs.best_params_['n_estimators']
grid_params = {'n_estimators': [best_n - 10, best_n, best_n + 10]}
gs = GridSearchCV(RandomForestClassifier(random_state=42), grid_params, cv=3)
gs.fit(X_tr, y_tr)
print(f"Refined Best N: {gs.best_params_['n_estimators']} (CV Score: {gs.best_score_*100:.2f}%)")
```
