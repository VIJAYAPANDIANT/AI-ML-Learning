# 03. Loading & Generating Datasets in Scikit-Learn

Scikit-learn provides built-in toy datasets, real-world data loaders, and synthetic dataset generators under `sklearn.datasets`.

### Learning Objectives:
- Load benchmark datasets (`load_iris`, `load_wine`, `load_diabetes`, `load_breast_cancer`)
- Return datasets directly as Pandas DataFrames using `as_frame=True`
- Generate synthetic classification and regression datasets with `make_classification` and `make_regression`
- Generate clustering blobs with `make_blobs`

```python
from sklearn.datasets import load_breast_cancer, make_classification, make_regression
import pandas as pd

# Load Breast Cancer benchmark as a clean Pandas DataFrame
cancer_data = load_breast_cancer(as_frame=True)
df_cancer = cancer_data.frame
print("Cancer Dataset Shape:", df_cancer.shape)
print("Target classes:", cancer_data.target_names)
df_cancer.head(3)
```

## 1. Generating Controlled Synthetic Datasets
Synthetic data generators allow testing algorithms under controlled levels of noise, feature collinearity, and class separability.

```python
# Generate synthetic binary classification dataset
X_clf, y_clf = make_classification(
    n_samples=500,
    n_features=6,
    n_informative=4,
    n_redundant=1,
    n_classes=2,
    random_state=42
)

print(f"Synthetic Classification: X={X_clf.shape}, y={y_clf.shape}")

# Generate synthetic continuous regression dataset
X_reg, y_reg = make_regression(
    n_samples=300,
    n_features=5,
    noise=10.0,
    random_state=42
)
print(f"Synthetic Regression: X={X_reg.shape}, y={y_reg.shape}")
```
