# 33. Model Persistence: Serializing Models with Joblib

After training an optimal pipeline, it must be persisted to disk so production APIs and microservices can load and execute predictions without retraining.

### Learning Objectives:
- Save trained pipelines and estimators using `joblib.dump()`
- Load saved models into memory with `joblib.load()`
- Verify identical predictions before and after serialization

```python
import joblib
import os
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression

iris = load_iris()
clf = LogisticRegression(max_iter=500, random_state=42).fit(iris.data, iris.target)

# Save model to disk
model_path = 'iris_model.joblib'
joblib.dump(clf, model_path)
print(f"Model saved to: {model_path} (File size: {os.path.getsize(model_path)} bytes)")

# Reload model from disk
loaded_clf = joblib.load(model_path)

# Verify predictions match exactly
original_preds = clf.predict(iris.data[:5])
loaded_preds = loaded_clf.predict(iris.data[:5])
print(f"Original predictions: {original_preds}")
print(f"Loaded predictions:   {loaded_preds}")
print(f"Exact match verified: {np.array_equal(original_preds, loaded_preds)}")

# Cleanup temporary file
if os.path.exists(model_path): os.remove(model_path)
print("Cleaned up temporary demonstration file.")
```
