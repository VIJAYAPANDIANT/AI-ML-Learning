# 30. Hyperparameter Tuning & Validation Curves

Hyperparameters (e.g. tree `max_depth`, SVM `C`, KNN `n_neighbors`) are set prior to training and dictate model learning capacity. `validation_curve` visualizes how performance changes across hyperparameter values.

### Learning Objectives:
- Differentiate internal model parameters (weights learned during `.fit()`) from external hyperparameters
- Plot a **Validation Curve** to identify the sweet spot balancing Bias and Variance
- Plot a **Learning Curve** to assess if more training data will improve generalization

```python
from sklearn.datasets import load_digits
from sklearn.svm import SVC
from sklearn.model_selection import validation_curve
import numpy as np
import matplotlib.pyplot as plt

digits = load_digits()
X, y = digits.data[:500], digits.target[:500] # Subsample for speed

param_range = np.logspace(-5, -1, 5)
train_scores, test_scores = validation_curve(
    SVC(kernel='rbf', random_state=42), 
    X, y, 
    param_name='gamma', 
    param_range=param_range, 
    cv=3, 
    scoring='accuracy'
)

train_mean = np.mean(train_scores, axis=1)
test_mean = np.mean(test_scores, axis=1)

plt.figure(figsize=(8, 4.5), dpi=100)
plt.semilogx(param_range, train_mean, label='Training Score', color='#2980b9', lw=2)
plt.semilogx(param_range, test_mean, label='Cross-Validation Score', color='#e74c3c', lw=2)
plt.title("Validation Curve: SVM RBF Gamma Parameter", fontweight='bold')
plt.xlabel("Gamma Parameter (log scale)")
plt.ylabel("Accuracy Score")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
