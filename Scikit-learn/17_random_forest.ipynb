# 17. Random Forest: Bagging & Ensemble Learning

Random Forest is an ensemble of decision trees trained on bootstrap samples of the data with random feature subsets at each split (Bagging + Feature Subspacing). This substantially reduces variance without increasing bias.

### Learning Objectives:
- Train `RandomForestClassifier` with `n_estimators`
- Understand Out-of-Bag error evaluation with `oob_score=True`
- Compute and visualize Gini feature importances across the ensemble

```python
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import numpy as np

cancer = load_breast_cancer()
X_tr, X_te, y_tr, y_te = train_test_split(cancer.data, cancer.target, test_size=0.25, random_state=42)

# Train Random Forest with Out-Of-Bag evaluation
rf = RandomForestClassifier(n_estimators=100, max_features='sqrt', oob_score=True, random_state=42)
rf.fit(X_tr, y_tr)

print(f"OOB Evaluation Score: {rf.oob_score_ * 100:.2f}%")
print(f"Holdout Test Accuracy: {rf.score(X_te, y_te) * 100:.2f}%")

# Top 8 Feature Importances
importances = rf.feature_importances_
indices = np.argsort(importances)[-8:]

plt.figure(figsize=(8, 4), dpi=100)
plt.barh(range(8), importances[indices], color='#27ae60', edgecolor='black')
plt.yticks(range(8), [cancer.feature_names[i] for i in indices])
plt.title("Random Forest: Top 8 Feature Importances", fontweight='bold')
plt.xlabel("Gini Importance Weight")
plt.grid(axis='x', alpha=0.3)
plt.show()
```
