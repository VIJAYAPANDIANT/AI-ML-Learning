# 13. Polynomial Regression

When data exhibits non-linear curvature, `PolynomialFeatures` constructs interaction terms and polynomial powers ($x, x^2, x^3, \dots$), allowing a linear model to capture non-linear phenomena.

### Learning Objectives:
- Generate polynomial feature powers using `PolynomialFeatures(degree=...)`
- Fit a linear model on expanded polynomial features
- Visualize the **Bias-Variance tradeoff** (Underfitting with degree 1, optimal with degree 3, overfitting with degree 15)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

np.random.seed(42)
X = np.sort(np.random.uniform(0, 5, 30))[:, np.newaxis]
y = np.sin(X).ravel() + np.random.normal(0, 0.2, 30)
X_plot = np.linspace(0, 5, 100)[:, np.newaxis]

plt.figure(figsize=(9, 4.5), dpi=100)
plt.scatter(X, y, color='black', label='Observations')

for deg, col in zip([1, 3, 10], ['#e74c3c', '#2ecc71', '#3498db']):
    model = make_pipeline(PolynomialFeatures(degree=deg), LinearRegression())
    model.fit(X, y)
    plt.plot(X_plot, model.predict(X_plot), color=col, lw=2, label=f'Degree {deg}')

plt.title("Polynomial Regression: Underfitting (Deg 1) vs Overfitting (Deg 10)", fontweight='bold')
plt.ylim(-2, 2)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 🎯 Exercises & Practice
1. Calculate the training $R^2$ score for degree 1 vs degree 3.

```python
# Exercise Solution
m1 = make_pipeline(PolynomialFeatures(1), LinearRegression()).fit(X, y)
m3 = make_pipeline(PolynomialFeatures(3), LinearRegression()).fit(X, y)
print(f"Degree 1 Train R²: {m1.score(X, y):.4f}")
print(f"Degree 3 Train R²: {m3.score(X, y):.4f}")
```
