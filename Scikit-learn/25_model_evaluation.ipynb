# 25. Model Evaluation & Diagnostics

Model evaluation assesses generalization performance on unseen data, diagnosing whether a model suffers from High Bias (Underfitting) or High Variance (Overfitting).

### Core Diagnostic Concepts:
- **High Bias (Underfitting)**: Both training and test errors are high. The model is too simple.
- **High Variance (Overfitting)**: Training error is very low, but holdout test error is substantially higher. The model memorized noise.

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

digits = load_digits()
X_tr, X_te, y_tr, y_te = train_test_split(digits.data, digits.target, test_size=0.3, random_state=42)

# Unconstrained tree (Overfitting example)
dt_overfit = DecisionTreeClassifier(max_depth=None, random_state=42)
dt_overfit.fit(X_tr, y_tr)

print(f"Unconstrained Tree Train Accuracy: {dt_overfit.score(X_tr, y_tr) * 100:.2f}% (Memorized)")
print(f"Unconstrained Tree Test Accuracy:  {dt_overfit.score(X_te, y_te) * 100:.2f}% (Generalization Gap)")

# Pruned tree
dt_pruned = DecisionTreeClassifier(max_depth=6, random_state=42)
dt_pruned.fit(X_tr, y_tr)
print(f"
Pruned Tree Train Accuracy: {dt_pruned.score(X_tr, y_tr) * 100:.2f}%")
print(f"Pruned Tree Test Accuracy:  {dt_pruned.score(X_te, y_te) * 100:.2f}%")
```
