# 22. Hierarchical Agglomerative Clustering

Hierarchical clustering builds a bottom-up tree (Dendrogram) where every data point begins in its own cluster and successive pairs merge based on proximity.

### Linkage Criteria:
- **Ward**: Minimizes within-cluster variance (default in scikit-learn)
- **Complete**: Minimizes maximum distance between points in two clusters
- **Average**: Minimizes average pairwise distances

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt

X, _ = make_blobs(n_samples=50, centers=3, cluster_std=0.8, random_state=42)

# Compute linkage matrix using SciPy for dendrogram plotting
Z = linkage(X, method='ward')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5), dpi=100)

# Dendrogram
dendrogram(Z, ax=ax1, truncate_mode='lastp', p=12)
ax1.set_title("Hierarchical Clustering Dendrogram (Ward Linkage)", fontweight='bold')
ax1.set_ylabel("Euclidean Distance Threshold")

# Agglomerative Clustering in Scikit-Learn
agg = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
labels = agg.fit_predict(X)

ax2.scatter(X[:, 0], X[:, 1], c=labels, cmap='tab10', s=45)
ax2.set_title("Agglomerative Clusters (K=3)", fontweight='bold')

fig.tight_layout()
plt.show()
```
