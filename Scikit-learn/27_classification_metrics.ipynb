# 27. Classification Metrics: Precision, Recall, F1 & ROC-AUC

When class distributions are imbalanced (e.g. Fraud Detection 99.5% legitimate vs 0.5% fraud), raw Accuracy is misleading.

### Key Metrics:
- **Precision**: $\frac{TP}{TP + FP}$ — Of all predicted positives, how many are truly positive? (Minimizes False Alarms)
- **Recall (Sensitivity)**: $\frac{TP}{TP + FN}$ — Of all actual positives, how many did we catch? (Minimizes Misses)
- **F1-Score**: Harmonic mean of Precision and Recall ($2 \times \frac{P \times R}{P + R}$)
- **ROC-AUC**: Area Under the Receiver Operating Characteristic Curve across all decision thresholds

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, RocCurveDisplay, roc_auc_score
import matplotlib.pyplot as plt

cancer = load_breast_cancer()
X_tr, X_te, y_tr, y_te = train_test_split(cancer.data, cancer.target, test_size=0.25, random_state=42)

scaler = StandardScaler()
X_tr_sc = scaler.fit_transform(X_tr)
X_te_sc = scaler.transform(X_te)

clf = LogisticRegression(max_iter=1000, random_state=42).fit(X_tr_sc, y_tr)
y_pred = clf.predict(X_te_sc)
y_prob = clf.predict_proba(X_te_sc)[:, 1]

print("=== Comprehensive Classification Report ===")
print(classification_report(y_te, y_pred, target_names=cancer.target_names))
print(f"ROC-AUC Score: {roc_auc_score(y_te, y_prob):.4f}")

# Plot ROC Curve
fig, ax = plt.subplots(figsize=(6, 4.5), dpi=100)
RocCurveDisplay.from_estimator(clf, X_te_sc, y_te, ax=ax, color='#2980b9', lw=2)
ax.plot([0, 1], [0, 1], 'k--', lw=1, label='Random Chance (AUC = 0.50)')
ax.set_title("Receiver Operating Characteristic (ROC) Curve", fontweight='bold')
ax.legend()
plt.show()
```
