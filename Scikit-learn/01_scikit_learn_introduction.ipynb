# 01. Introduction to Scikit-Learn

Welcome to the **Scikit-Learn for AI, Data Science & Machine Learning** masterclass! Scikit-Learn is Python's industry-standard library for classical machine learning, providing efficient tools for predictive modeling, regression, classification, clustering, dimensionality reduction, and preprocessing.

### The Core Scikit-Learn Philosophy:
1. **Consistent Estimator Interface**: Every model shares identical method verbs:
   - `estimator.fit(X, y)`: Fits model parameters from training data
   - `estimator.predict(X)`: Predicts target values for new observations
   - `transformer.fit_transform(X)`: Computes parameters and transforms features
2. **Stateless vs Stateful Separation**: Objects are parameterized upon initialization, learn parameters during `.fit()`, and apply learned attributes (`coef_`, `intercept_`, `mean_`) during inference.
3. **Tight Integration**: Seamless interoperability with NumPy ndarrays, Pandas DataFrames, and SciPy sparse matrices.

```python
import sklearn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print(f"Scikit-Learn Version: {sklearn.__version__}")
print(f"NumPy Version: {np.__version__}")
print(f"Pandas Version: {pd.__version__}")
```

## 1. The Universal Estimator API in Action
Let's train a simple Linear Regression model demonstrating the fundamental `fit` and `predict` lifecycle.

```python
from sklearn.linear_model import LinearRegression

# 1. Prepare synthetic features (2D array X) and target (1D vector y)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2.2, 3.9, 6.1, 7.8, 10.2])

# 2. Instantiate the model with hyperparameters
model = LinearRegression()

# 3. Fit the model (learn parameters)
model.fit(X, y)

# 4. Inspect learned internal parameters (ending with a trailing underscore)
print(f"Learned Slope (Weight / Coefficient): {model.coef_[0]:.4f}")
print(f"Learned Intercept (Bias): {model.intercept_:.4f}")

# 5. Make predictions on unseen observations
X_new = np.array([[6], [7]])
y_pred = model.predict(X_new)
print(f"Predictions for X=[6, 7]: {y_pred.round(2)}")
```

## 🎯 Exercises & Practice
1. Instantiate a `LinearRegression` model.
2. Fit it on $X = [[10], [20], [30], [40]]$ and $y = [15, 25, 35, 45]$.
3. Predict the target value for $X = [[50], [60]]$.

```python
# Exercise Solution
X_ex = np.array([[10], [20], [30], [40]])
y_ex = np.array([15, 25, 35, 45])
reg = LinearRegression().fit(X_ex, y_ex)
preds = reg.predict([[50], [60]])
print(f"Predicted outputs: {preds}")
print(f"Slope: {reg.coef_[0]}, Intercept: {reg.intercept_}")
```
