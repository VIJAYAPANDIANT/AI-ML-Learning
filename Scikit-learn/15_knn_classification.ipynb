# 15. K-Nearest Neighbors (KNN) Classification

KNN is a non-parametric, instance-based (lazy learning) algorithm. An unseen sample is classified by taking the majority vote of its $k$ nearest neighbors in Euclidean feature space.

### Learning Objectives:
- Understand how distance metrics (Euclidean, Manhattan) influence classification
- Train `KNeighborsClassifier` with varying $k$
- Evaluate the critical importance of feature scaling for distance-based algorithms

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

iris = load_iris()
X_tr, X_te, y_tr, y_te = train_test_split(iris.data, iris.target, test_size=0.3, random_state=42)

k_values = range(1, 25)
accuracies = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_tr, y_tr)
    accuracies.append(accuracy_score(y_te, knn.predict(X_te)))

plt.figure(figsize=(8, 4), dpi=100)
plt.plot(k_values, accuracies, marker='o', color='#2c3e50', lw=2)
plt.title("KNN Classification Accuracy vs Number of Neighbors (K)", fontweight='bold')
plt.xlabel("K Neighbors")
plt.ylabel("Holdout Accuracy")
plt.grid(True, alpha=0.3)
plt.show()
```
