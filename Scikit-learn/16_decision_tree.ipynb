# 16. Decision Tree Classification & Regression

Decision Trees recursively partition feature space into orthogonal rectangular regions by maximizing Information Gain (minimizing Gini Impurity or Entropy).

### Learning Objectives:
- Train `DecisionTreeClassifier` using Gini and Entropy criteria
- Prune trees to prevent overfitting using `max_depth`, `min_samples_split`, and `min_samples_leaf`
- Visualize tree branching decisions with `sklearn.tree.plot_tree`
- Extract feature importances with `tree.feature_importances_`

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

iris = load_iris()
X, y = iris.data, iris.target

# Train a depth-constrained decision tree to prevent overfitting
dt = DecisionTreeClassifier(max_depth=3, criterion='gini', random_state=42)
dt.fit(X, y)

fig = plt.figure(figsize=(12, 7), dpi=100)
plot_tree(
    dt, 
    feature_names=iris.feature_names, 
    class_names=iris.target_names, 
    filled=True, 
    rounded=True, 
    fontsize=10
)
plt.title("Decision Tree Flowchart (Max Depth = 3)", fontweight='bold')
plt.show()
```

## 🎯 Exercises & Practice
1. Inspect feature importances from the decision tree and print the most dominant feature.

```python
# Exercise Solution
for name, imp in zip(iris.feature_names, dt.feature_importances_):
    print(f"{name:20s}: {imp:.4f}")
```
