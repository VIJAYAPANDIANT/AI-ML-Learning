# 19. Naive Bayes Classification

Naive Bayes classifiers are probabilistic models based on Bayes' Theorem: $P(y|X) = \frac{P(X|y) P(y)}{P(X)}$, under the "naive" assumption that all features are conditionally independent given the class label.

### Model Variants:
- **GaussianNB**: For continuous features following a normal distribution.
- **MultinomialNB**: For discrete text word count frequencies (NLP/Spam Filtering).
- **BernoulliNB**: For binary boolean features.

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

wine = load_wine()
X_tr, X_te, y_tr, y_te = train_test_split(wine.data, wine.target, test_size=0.25, random_state=42)

gnb = GaussianNB()
gnb.fit(X_tr, y_tr)
y_pred = gnb.predict(X_te)

print(f"Wine Classification Accuracy: {accuracy_score(y_te, y_pred) * 100:.2f}%")
print(f"Learned Class Priors: {gnb.class_prior_.round(3)}")
```
