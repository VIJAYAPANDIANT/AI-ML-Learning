# 11. Simple Linear Regression

Simple Linear Regression models the relationship between a single independent continuous predictor $x$ and a continuous dependent target $y$ via Ordinary Least Squares (OLS): $y = \beta_0 + \beta_1 x + \epsilon$.

### Learning Objectives:
- Fit `LinearRegression` using Ordinary Least Squares
- Extract the slope coefficient (`coef_`) and $y$-intercept (`intercept_`)
- Compute the coefficient of determination ($R^2$ score)
- Visualize fitted regression lines using Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# Generate synthetic data with noise
np.random.seed(42)
X = np.random.uniform(10, 80, 60).reshape(-1, 1) # Advertising spend ($k)
y = 2.5 * X.squeeze() + 30 + np.random.normal(0, 15, 60) # Sales ($k)

# Fit model
reg = LinearRegression()
reg.fit(X, y)
y_pred = reg.predict(X)

print(f"Regression Equation: Sales = {reg.intercept_:.2f} + {reg.coef_[0]:.2f} * AdSpend")
print(f"R² Score: {r2_score(y, y_pred):.4f}")
print(f"Root Mean Squared Error (RMSE): {np.sqrt(mean_squared_error(y, y_pred)):.2f}")

# Visualize
plt.figure(figsize=(7.5, 4), dpi=100)
plt.scatter(X, y, color='#2980b9', alpha=0.7, label='Actual Observations')
plt.plot(X, y_pred, color='#e74c3c', lw=2.5, label='OLS Fit Line')
plt.title("Simple Linear Regression: Ad Spend vs Sales", fontweight='bold')
plt.xlabel("Ad Spend ($ Thousands)")
plt.ylabel("Sales ($ Thousands)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
