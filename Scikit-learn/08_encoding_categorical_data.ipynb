# 08. Encoding Categorical Variables

Machine learning algorithms require numerical input. Categorical variables must be converted to numbers without introducing artificial ordinal rankings.

### Encoding Techniques:
- **OneHotEncoder**: Converts nominal categories into sparse binary indicator columns (0/1). Use `handle_unknown='ignore'` to prevent inference crashes on unseen test categories.
- **OrdinalEncoder**: Encodes ordered categorical variables (e.g. Low=0, Medium=1, High=2).
- **LabelEncoder**: Exclusively designed for encoding 1D target labels ($y$).

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
import pandas as pd

df = pd.DataFrame({
    'Device': ['Mobile', 'Desktop', 'Tablet', 'Mobile'],
    'Satisfaction': ['Low', 'High', 'Medium', 'High']
})

# 1. OneHotEncoder for Nominal 'Device'
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
device_encoded = ohe.fit_transform(df[['Device']])
df_ohe = pd.DataFrame(device_encoded, columns=ohe.get_feature_names_out(['Device']))

# 2. OrdinalEncoder for Ordered 'Satisfaction'
ord_encoder = OrdinalEncoder(categories=[['Low', 'Medium', 'High']])
sat_encoded = ord_encoder.fit_transform(df[['Satisfaction']])

df_encoded = pd.concat([df_ohe, pd.DataFrame(sat_encoded, columns=['Satisfaction_Rank'])], axis=1)
print("Encoded Feature Matrix:")
df_encoded
```
