# 23. DBSCAN: Density-Based Spatial Clustering

DBSCAN (Density-Based Spatial Clustering of Applications with Noise) groups points that are closely packed together, identifying non-linear shapes and marking isolated outliers as **noise** (`label = -1`).

### Core Parameters:
- `eps`: Maximum radius neighborhood distance
- `min_samples`: Minimum number of neighbor points required to form a dense core

```python
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN, KMeans
import matplotlib.pyplot as plt

X_moons, _ = make_moons(n_samples=300, noise=0.08, random_state=42)

# Compare KMeans (fails on non-convex shapes) vs DBSCAN (succeeds)
km = KMeans(n_clusters=2, random_state=42).fit(X_moons)
db = DBSCAN(eps=0.22, min_samples=5).fit(X_moons)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4.5), dpi=100)
ax1.scatter(X_moons[:, 0], X_moons[:, 1], c=km.labels_, cmap='Set1', s=35)
ax1.set_title("K-Means (Fails on Non-Spherical Moons)", fontweight='bold')

ax2.scatter(X_moons[:, 0], X_moons[:, 1], c=db.labels_, cmap='Set1', s=35)
ax2.set_title("DBSCAN (Correctly Identifies Arbitrary Density Moons)", fontweight='bold')

fig.tight_layout()
plt.show()
```
