# 02. The End-to-End Machine Learning Workflow

A production machine learning pipeline follows an ordered lifecycle from data ingestion to model deployment.

### The 7-Step Machine Learning Pipeline:
1. **Problem Formulation**: Framing business objective (Regression vs. Classification vs. Clustering)
2. **Data Ingestion & Inspection**: Loading tabular records and verifying shapes, distributions, and types
3. **Feature Preprocessing**: Imputing missing values, encoding categorical strings, scaling numerical scales
4. **Train / Test Partitioning**: Isolating validation holds to prevent data leakage
5. **Model Training (Estimation)**: Fitting candidate algorithms on training partitions
6. **Model Evaluation**: Assessing performance on holdout data using statistical error and accuracy metrics
7. **Hyperparameter Tuning & Deployment**: Optimizing model parameters and persisting pipelines to disk

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

# Step 1 & 2: Ingest Data
iris = load_iris(as_frame=True)
X, y = iris.data, iris.target
print(f"Loaded Dataset: {X.shape[0]} samples, {X.shape[1]} features.")

# Step 3 & 4: Train/Test Split (Preventing Data Leakage)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

# Step 5: Preprocessing (Standard Scaling)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test) # Transform only using train parameters!

# Step 6: Model Training
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)

# Step 7: Inference & Evaluation
y_pred = knn.predict(X_test_scaled)
acc = accuracy_score(y_test, y_pred)
print(f"Model Holdout Accuracy: {acc * 100:.2f}%")
```
