# 14. Logistic Regression

Despite its name, Logistic Regression is a foundational linear algorithm for **binary and multi-class classification**. It passes a linear combination of inputs through the sigmoid function: $P(y=1|X) = \sigma(w^T X + b) = \frac{1}{1 + e^{-(w^T X + b)}}$.

### Learning Objectives:
- Train a `LogisticRegression` classifier
- Extract predicted classes with `.predict()` and probabilities with `.predict_proba()`
- Adjust decision classification thresholds
- Handle multi-class classification using One-vs-Rest (OvR) and Multinomial cross-entropy

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score

cancer = load_breast_cancer()
X_tr, X_te, y_tr, y_te = train_test_split(cancer.data, cancer.target, test_size=0.25, random_state=42, stratify=cancer.target)

scaler = StandardScaler()
X_tr_sc = scaler.fit_transform(X_tr)
X_te_sc = scaler.transform(X_te)

clf = LogisticRegression(C=1.0, max_iter=1000, random_state=42)
clf.fit(X_tr_sc, y_tr)

y_pred = clf.predict(X_te_sc)
y_prob = clf.predict_proba(X_te_sc)[:, 1]

print(f"Test Accuracy: {accuracy_score(y_te, y_pred) * 100:.2f}%")
print(f"Test ROC-AUC:  {roc_auc_score(y_te, y_prob):.4f}")
```
