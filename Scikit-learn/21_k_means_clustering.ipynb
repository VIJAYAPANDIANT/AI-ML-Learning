# 21. K-Means Clustering

K-Means is an unsupervised clustering algorithm that partitions $n$ observations into $k$ clusters by minimizing the within-cluster sum of squares (WCSS / Inertia): $\sum_{i=1}^k \sum_{x \in S_i} ||x - \mu_i||^2$.

### Learning Objectives:
- Fit `KMeans` with `n_clusters` and `init='k-means++'`
- Determine optimal $k$ using the **Elbow Method** (Inertia curve)
- Evaluate cluster compactness using **Silhouette Score** with `silhouette_score`
- Extract cluster centroids via `kmeans.cluster_centers_`

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Generate synthetic clusters
X, _ = make_blobs(n_samples=400, centers=4, cluster_std=0.7, random_state=42)

inertias = []
silhouettes = []
k_range = range(2, 9)

for k in k_range:
    km = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=42)
    labels = km.fit_predict(X)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X, labels))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4), dpi=100)
ax1.plot(k_range, inertias, marker='o', color='#2980b9', lw=2)
ax1.set_title("Elbow Method (Inertia vs K)", fontweight='bold')
ax1.set_xlabel("Number of Clusters (K)")
ax1.set_ylabel("Inertia (WCSS)")
ax1.grid(True, alpha=0.3)

ax2.plot(k_range, silhouettes, marker='s', color='#27ae60', lw=2)
ax2.set_title("Silhouette Score vs K", fontweight='bold')
ax2.set_xlabel("Number of Clusters (K)")
ax2.set_ylabel("Silhouette Score")
ax2.grid(True, alpha=0.3)

fig.tight_layout()
plt.show()
```

## 🎯 Exercises & Practice
1. Fit KMeans with $K=4$ and plot the clustered points along with the 4 centroid markers ('X').

```python
# Exercise Solution
km_optimal = KMeans(n_clusters=4, random_state=42).fit(X)
plt.figure(figsize=(7, 4), dpi=100)
plt.scatter(X[:, 0], X[:, 1], c=km_optimal.labels_, cmap='viridis', alpha=0.6, s=30)
plt.scatter(km_optimal.cluster_centers_[:, 0], km_optimal.cluster_centers_[:, 1], s=180, c='red', marker='X', edgecolors='black', label='Centroids')
plt.title("K-Means (K=4) Cluster Partitions", fontweight='bold')
plt.legend()
plt.show()
```
