# 06. Data Preprocessing Foundations

Raw real-world datasets are messy, incomplete, and heterogeneous. Preprocessing transforms raw tables into standardized numerical matrices required by mathematical estimators.

### The Core Preprocessing Pillars:
1. **Imputation**: Filling missing values (`SimpleImputer`, `KNNImputer`)
2. **Encoding**: Converting categorical text labels into numbers (`OneHotEncoder`, `OrdinalEncoder`)
3. **Scaling**: Normalizing numerical features onto comparable ranges (`StandardScaler`, `MinMaxScaler`)
4. **Selection**: Isolating the most predictive features (`SelectKBest`, `RFE`)

```python
import pandas as pd
import numpy as np

# Sample un-preprocessed customer table
raw_data = pd.DataFrame({
    'Age': [25, np.nan, 47, 52, 36],
    'Salary': [50000, 64000, np.nan, 95000, 72000],
    'Country': ['France', 'Spain', 'Germany', 'Spain', 'France'],
    'Purchased': ['No', 'Yes', 'No', 'Yes', 'No']
})
print("Raw Heterogeneous Table:")
raw_data
```
