# 10. Feature Selection: `SelectKBest` & `RFE`

Removing noisy, redundant, or irrelevant features prevents overfitting, reduces computational latency, and improves model interpretability.

### Feature Selection Methods:
1. **Filter Methods**: Univariate statistical tests (`SelectKBest`, `f_classif`, `mutual_info_classif`)
2. **Wrapper Methods**: Recursive Feature Elimination (`RFE`)
3. **Embedded Methods**: Tree importance (`SelectFromModel` with Random Forest or Lasso $L_1$ penalty)

```python
from sklearn.datasets import load_breast_cancer
from sklearn.feature_selection import SelectKBest, f_classif, RFE
from sklearn.ensemble import RandomForestClassifier

data = load_breast_cancer()
X, y = data.data, data.target
print(f"Original Feature Count: {X.shape[1]}")

# 1. Filter: Select Top 5 Features using ANOVA F-value
selector = SelectKBest(score_func=f_classif, k=5)
X_kbest = selector.fit_transform(X, y)
selected_features = data.feature_names[selector.get_support()]
print(f"Top 5 Features by ANOVA F-test: {list(selected_features)}")

# 2. Wrapper: Recursive Feature Elimination with Random Forest
rfe = RFE(estimator=RandomForestClassifier(n_estimators=50, random_state=42), n_features_to_select=5)
X_rfe = rfe.fit_transform(X, y)
print(f"Top 5 Features by RFE: {list(data.feature_names[rfe.support_])}")
```

## 🎯 Exercises & Practice
1. Use `SelectFromModel` with `RandomForestClassifier` to automatically select features exceeding the mean importance threshold.

```python
# Exercise Solution
from sklearn.feature_selection import SelectFromModel
rf = RandomForestClassifier(n_estimators=50, random_state=42).fit(X, y)
sfm = SelectFromModel(rf, threshold='mean', prefit=True)
print(f"Features selected by SelectFromModel: {sfm.get_support().sum()} out of {X.shape[1]}")
```
