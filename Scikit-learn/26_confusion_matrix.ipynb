# 26. Confusion Matrix & Error Analysis

A Confusion Matrix cross-tabulates model predictions against true ground truth labels, breaking down errors into False Positives (Type I) and False Negatives (Type II).

### Learning Objectives:
- Generate confusion matrices using `confusion_matrix`
- Render visual matrices with `ConfusionMatrixDisplay`
- Calculate True Positive Rate (Sensitivity) and True Negative Rate (Specificity)

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cancer = load_breast_cancer()
X_tr, X_te, y_tr, y_te = train_test_split(cancer.data, cancer.target, test_size=0.25, random_state=42, stratify=cancer.target)

clf = RandomForestClassifier(n_estimators=50, random_state=42).fit(X_tr, y_tr)
y_pred = clf.predict(X_te)

cm = confusion_matrix(y_te, y_pred)
print("Raw Confusion Matrix:\n", cm)

# Visual Display
fig, ax = plt.subplots(figsize=(6, 5), dpi=100)
ConfusionMatrixDisplay.from_predictions(
    y_te, y_pred, 
    display_labels=cancer.target_names, 
    cmap='Blues', 
    colorbar=False, 
    ax=ax
)
plt.title("Cancer Diagnosis Confusion Matrix", fontweight='bold')
plt.show()
```
