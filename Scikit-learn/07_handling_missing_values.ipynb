# 07. Handling Missing Values: `SimpleImputer` & `KNNImputer`

Scikit-Learn estimators cannot compute gradients or matrix multiplications with `NaN` (Not a Number) values. Imputation replaces missing entries with statistically inferred values.

### Learning Objectives:
- Use `SimpleImputer` with strategies: `'mean'`, `'median'`, `'most_frequent'`, `'constant'`
- Apply multi-variate imputation with `KNNImputer`
- Maintain strict separation: `fit_transform` on train, `transform` on test

```python
from sklearn.impute import SimpleImputer, KNNImputer
import numpy as np
import pandas as pd

X_raw = np.array([
    [25, 50000],
    [np.nan, 64000],
    [47, np.nan],
    [52, 95000],
    [36, 72000]
])

# 1. SimpleImputer with median strategy
imputer_median = SimpleImputer(strategy='median')
X_imputed = imputer_median.fit_transform(X_raw)

print("Imputed with Column Medians:")
print(pd.DataFrame(X_imputed, columns=['Age', 'Salary']))

# 2. KNNImputer (uses nearest neighbor Euclidean distances)
knn_imp = KNNImputer(n_neighbors=2)
X_knn_imputed = knn_imp.fit_transform(X_raw)
print("
Imputed with KNN Nearest Neighbors:")
print(pd.DataFrame(X_knn_imputed, columns=['Age', 'Salary']))
```

1. Impute missing values in categorical column `['Cat', 'Dog', np.nan, 'Cat', 'Bird']` using `SimpleImputer(strategy='most_frequent')`.

```python
# Exercise Solution
cats = np.array(['Cat', 'Dog', np.nan, 'Cat', 'Bird'], dtype=object).reshape(-1, 1)
imp_freq = SimpleImputer(strategy='most_frequent')
print("Most Frequent Imputation:\n", imp_freq.fit_transform(cats))
```
