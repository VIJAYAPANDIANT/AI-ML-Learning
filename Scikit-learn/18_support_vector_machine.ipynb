# 18. Support Vector Machines (SVM)

Support Vector Machines search for the optimal hyperplane that maximizes the geometric margin separating distinct classes.

### Core Concepts:
- **Support Vectors**: The critical training instances lying directly on the margin boundaries.
- **Regularization ($C$)**: Controls the trade-off between maximizing margin width and penalizing margin violations.
- **Kernel Trick**: Maps non-linear input features into higher-dimensional space where linear separation is possible (`'linear'`, `'poly'`, `'rbf'` with parameter $\gamma$).

```python
from sklearn.datasets import make_moons
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import numpy as np

# Generate non-linearly separable moon dataset
X, y = make_moons(n_samples=300, noise=0.25, random_state=42)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=42)

# Fit Radial Basis Function (RBF) Kernel SVM
svm_rbf = SVC(kernel='rbf', C=1.0, gamma='scale', probability=True, random_state=42)
svm_rbf.fit(X_tr, y_tr)

print(f"RBF SVM Holdout Accuracy: {svm_rbf.score(X_te, y_te) * 100:.2f}%")
print(f"Total Support Vectors Identified: {len(svm_rbf.support_vectors_)}")
```
