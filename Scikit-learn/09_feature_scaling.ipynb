# 09. Feature Scaling: `StandardScaler`, `MinMaxScaler` & `RobustScaler`

Distance-based algorithms (KNN, SVM, K-Means) and gradient descent solvers (Logistic Regression, Neural Networks) are biased if features have drastically different magnitudes (e.g., Age 20–70 vs Salary $20,000–$200,000).

### Scaler Comparison:
- **StandardScaler**: Standardizes features to mean $\mu=0$ and unit variance $\sigma=1$ ($z = \frac{x - \mu}{\sigma}$)
- **MinMaxScaler**: Scales features into $[0, 1]$ range ($x_{norm} = \frac{x - x_{min}}{x_{max} - x_{min}}$)
- **RobustScaler**: Uses median and IQR, resistant to extreme outliers

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
import numpy as np
import pandas as pd

X_raw = np.array([
    [25, 45000],
    [32, 62000],
    [45, 88000],
    [58, 125000],
    [29, 500000] # Extreme outlier in salary
])

s_scaler = StandardScaler()
m_scaler = MinMaxScaler()
r_scaler = RobustScaler()

print("=== StandardScaler (z-score) ===")
print(pd.DataFrame(s_scaler.fit_transform(X_raw), columns=['Age', 'Salary']).round(2))

print("
=== RobustScaler (outlier resistant) ===")
print(pd.DataFrame(r_scaler.fit_transform(X_raw), columns=['Age', 'Salary']).round(2))
```
