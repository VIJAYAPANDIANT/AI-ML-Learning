# 28. Regression Metrics: MAE, MSE, RMSE & $R^2$

Evaluating continuous predictions requires measuring the magnitude and distribution of residuals ($e_i = y_i - \hat{y}_i$).

### Metric Formulations:
- **Mean Absolute Error (MAE)**: $\frac{1}{n} \sum |y_i - \hat{y}_i|$ — Robust to outliers; linear penalty
- **Mean Squared Error (MSE)**: $\frac{1}{n} \sum (y_i - \hat{y}_i)^2$ — Heavily penalizes large errors
- **Root Mean Squared Error (RMSE)**: $\sqrt{\text{MSE}}$ — Expressed in identical units as target variable $y$
- **Coefficient of Determination ($R^2$)**: $1 - \frac{SS_{res}}{SS_{tot}}$ — Proportion of variance explained by model

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_true = np.array([100, 150, 200, 250, 300])
y_pred = np.array([105, 142, 210, 240, 290])

mae = mean_absolute_error(y_true, y_pred)
mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_true, y_pred)

print(f"Mean Absolute Error (MAE):     {mae:.2f}")
print(f"Mean Squared Error (MSE):      {mse:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R² Score (Variance Explained): {r2:.4f}")
```
