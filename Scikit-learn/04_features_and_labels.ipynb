# 04. Feature Matrix ($X$) & Target Vector ($y$)

In supervised machine learning, models map an input feature matrix $X$ to an output target vector $y$: $y = f(X) + \epsilon$.

### Architectural Rules in Scikit-Learn:
1. **Feature Matrix ($X$)**: Must always be **2-Dimensional** with shape `(n_samples, n_features)`.
2. **Target Vector ($y$)**: Must typically be **1-Dimensional** with shape `(n_samples,)`.
3. Passing a 1D vector to $X$ will raise a `ValueError`. You must reshape using `.reshape(-1, 1)`.

```python
import numpy as np
import pandas as pd

# Example tabular DataFrame
df = pd.DataFrame({
    'Square_Feet': [1200, 1850, 2400, 3100, 1500],
    'Bedrooms': [2, 3, 4, 4, 3],
    'Bathrooms': [1, 2, 2.5, 3, 2],
    'Price': [250000, 380000, 490000, 620000, 310000]
})

# Correct Extraction: X is 2D, y is 1D
X = df[['Square_Feet', 'Bedrooms', 'Bathrooms']] # 2D DataFrame / Matrix
y = df['Price']                                # 1D Series / Vector

print(f"X (Feature Matrix) Shape: {X.shape} (Dimensions: {X.ndim}D)")
print(f"y (Target Vector) Shape: {y.shape} (Dimensions: {y.ndim}D)")
```
