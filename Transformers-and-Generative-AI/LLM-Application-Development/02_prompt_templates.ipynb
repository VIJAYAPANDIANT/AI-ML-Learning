# 02. Prompt Templates & Dynamic Variable Injection

Hardcoded string prompts fail in production. **Prompt Templates** encapsulate parameterized formatting, few-shot examples, and system instructions into reusable software components.

```python
class PromptTemplate:
    def __init__(self, template_str, input_variables):
        self.template_str = template_str
        self.input_variables = input_variables

    def format(self, **kwargs):
        for var in self.input_variables:
            if var not in kwargs:
                raise ValueError(f"Missing required variable: {var}")
        return self.template_str.format(**kwargs)

support_template = PromptTemplate(
    template_str="You are a customer support agent for {company}. Tone: {tone}.\nCustomer Issue: {issue}\nProvide a helpful 2-sentence resolution:",
    input_variables=["company", "tone", "issue"]
)

rendered = support_template.format(
    company="CloudScale AI",
    tone="empathetic and professional",
    issue="Our database cluster encountered a failover timeout during peak traffic."
)
print(rendered)
```
