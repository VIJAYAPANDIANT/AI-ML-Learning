# 06. Retrieval-Augmented Generation (RAG) Fundamentals

**Retrieval-Augmented Generation (RAG)** (Lewis et al., 2020) grounds LLM responses in authoritative external knowledge bases, mitigating hallucinations, eliminating knowledge cutoffs, and keeping proprietary enterprise data secure.

### The 3 Core Stages of RAG:
1. **Ingestion & Indexing**: Chunk documents (e.g. 500 tokens with 50-token overlap), generate dense embeddings, and store in a vector database.
2. **Retrieval**: Embed user query, perform ANN similarity search, and retrieve top-$k$ most relevant context chunks.
3. **Synthesis & Generation**: Augment LLM prompt with retrieved context and generate a grounded, cited response.

```python
def naive_rag_prompt(user_query, retrieved_chunks):
    context_block = "\n\n".join([f"[Source {i+1}]: {chunk}" for i, chunk in enumerate(retrieved_chunks)])
    prompt = f"""You are a precise corporate intelligence assistant.
Answer the user's question using ONLY the provided sources below.
If the answer cannot be found in the sources, reply: 'I do not have sufficient information.'

--- CONTEXT ---
{context_block}

--- QUESTION ---
{user_query}

--- ANSWER ---"""
    return prompt

chunks = [
    "Project Titan was approved with a budget of $4.2M for Q3 FY2026.",
    "The primary objective of Project Titan is deploying autonomous edge inference nodes."
]
print(naive_rag_prompt("What is the Q3 budget for Project Titan?", chunks))
```
