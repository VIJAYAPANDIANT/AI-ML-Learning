# 04. Text Embeddings & Vector Similarity Search

Embedding models convert natural language text chunks into high-dimensional semantic vectors $\vec{v} \in \mathbb{R}^d$ ($d \in [384, 1536, 3072]$).

### Vector Distance Metrics:
- **Cosine Similarity**: $\cos(\theta) = \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{u}\| \|\mathbf{v}\|} \in [-1, 1]$. Invariant to vector magnitude.
- **Dot Product**: $\mathbf{u} \cdot \mathbf{v}$ (equivalent to cosine similarity if vectors are $L_2$-normalized).
- **Euclidean (L2) Distance**: $\|\mathbf{u} - \mathbf{v}\|_2 = \sqrt{\sum (u_i - v_i)^2}$.

```python
import numpy as np

def cosine_sim(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# Simulated 4-dimensional embeddings
documents = {
    "Doc A (AI Infrastructure)": np.array([0.9, 0.8, 0.1, 0.2]),
    "Doc B (Deep Learning Ops)":  np.array([0.85, 0.75, 0.15, 0.25]),
    "Doc C (Baking Italian Pizza)": np.array([0.1, 0.05, 0.95, 0.88])
}

query = np.array([0.92, 0.81, 0.05, 0.18])  # 'GPU cluster setup for AI'

print("--- Vector Similarity Ranking ---")
for name, vec in documents.items():
    score = cosine_sim(query, vec)
    print(f"{name:30s} -> Cosine Score: {score:.4f}")
```
