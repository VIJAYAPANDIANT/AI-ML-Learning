# 10. Autonomous AI Agents: Tools, Planning & Memory

An **AI Agent** combines an LLM core with planning loops, external tool calling (calculator, SQL, web search), and persistent memory to autonomously achieve multi-step goals.

### Core Components of an AI Agent:
1. **Planning**: Decomposes high-level objectives into sub-goals and self-reflects on errors.
2. **Memory**: Short-term conversational context and long-term vector database recall.
3. **Tools / Function Calling**: Executing external APIs (REST, Python sandbox, database queries).
4. **ReAct Paradigm**: Interleaves Reasoning ("Thought") and Action ("Act") loops.

```python
class MockAgent:
    def __init__(self, tools):
        self.tools = tools

    def execute(self, task):
        print(f"Agent Goal: {task}")
        # Simulated ReAct Loop
        print("Thought: I need to check the current stock price of NVDA.")
        print("Action: calling tool 'fetch_stock_price' with {'symbol': 'NVDA'}")
        observation = self.tools['fetch_stock_price']('NVDA')
        print(f"Observation: {observation}")
        print("Thought: I now have the data to formulate the final answer.")
        return f"The current trading price of NVDA is {observation['price']} USD."

tools = {
    'fetch_stock_price': lambda sym: {'symbol': sym, 'price': 135.50, 'currency': 'USD'}
}

agent = MockAgent(tools)
response = agent.execute("What is the current stock price of NVDA?")
print("Final Agent Response:", response)
```
