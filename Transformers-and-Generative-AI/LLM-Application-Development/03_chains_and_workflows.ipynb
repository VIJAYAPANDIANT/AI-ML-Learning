# 03. Chains, Pipelines & Declarative Workflows

Complex tasks require composing multiple LLM calls where the output of step $N$ serves as the input to step $N+1$ (e.g., Translate $\to$ Summarize $\to$ Extract Action Items).

```python
class SimpleChain:
    def __init__(self, steps):
        self.steps = steps

    def run(self, initial_input):
        data = initial_input
        for name, step_fn in self.steps:
            data = step_fn(data)
            print(f"Completed Step [{name}] -> Intermediate length: {len(str(data))}")
        return data

# Pipeline: Raw Text -> Extraction -> Markdown Table
step_clean = lambda text: text.strip().lower()
step_extract = lambda text: [w for w in text.split() if len(w) > 5]
step_format = lambda words: "\n".join([f"- **{w}**" for w in words[:4]])

pipeline = SimpleChain([
    ("Cleaning", step_clean),
    ("Keyword Extraction", step_extract),
    ("Markdown Formatting", step_format)
])

result = pipeline.run("Transformers and Generative Artificial Intelligence revolutionize engineering workflows.")
print("\nFinal Pipeline Output:\n" + result)
```
