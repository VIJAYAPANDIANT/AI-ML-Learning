# 11. LLM Evaluation & Safety Guardrails

Deploying LLMs in production requires rigorous evaluation metrics and safety guardrails to prevent hallucinations, prompt injection, and PII leakage.

### The RAGAS Evaluation Framework:
1. **Faithfulness**: Are claims in the response strictly supported by retrieved context? (Prevents hallucination)
2. **Answer Relevancy**: Does the response directly address the user's prompt?
3. **Context Precision**: Are relevant chunks ranked higher than irrelevant ones?

### Guardrail Frameworks (NeMo Guardrails, Llama-Guard):
Intercepts inputs and outputs with moderation classifiers to block jailbreaks and topical deviations.

```python
def evaluate_faithfulness(claim, context_chunks):
    # Simulated exact-grounding verification
    grounded = any(word in " ".join(context_chunks).lower() for word in claim.lower().split())
    return 1.0 if grounded else 0.0

context = ["The capital of France is Paris. Population is approximately 2.1 million."]
test_claim_1 = "Paris is the capital of France."
test_claim_2 = "The capital of France is Berlin."

print("Faithfulness Score (Valid Claim):", evaluate_faithfulness(test_claim_1, context))
print("Faithfulness Score (Hallucinated):", evaluate_faithfulness(test_claim_2, context))
```
