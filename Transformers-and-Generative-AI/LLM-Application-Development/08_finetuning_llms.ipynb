# 08. Fine-Tuning LLMs: Full Fine-Tuning vs PEFT

When in-context prompting is insufficient for specialized domains (legal jargon, custom code syntaxes, medical diagnostics), models require weight updates via fine-tuning.

### The Memory Cost of Full Fine-Tuning:
For a 70B parameter model in FP16:
- Model Weights: 140 GB
- Gradients: 140 GB
- Adam Optimizer States (Momentum + Variance): 560 GB
- Activations & Cache: ~100 GB+
**Total VRAM Required: > 940 GB (Requires multi-node 8x H100 GPU cluster!)**

```python
def calculate_finetuning_memory_gb(params_billions):
    weights_gb = params_billions * 2
    grads_gb = params_billions * 2
    adam_states_gb = params_billions * 8
    total_gb = weights_gb + grads_gb + adam_states_gb
    return total_gb

for p in [3, 7, 13, 70]:
    mem = calculate_finetuning_memory_gb(p)
    print(f"Model Size: {p:2d}B Params -> Optimizer + Weight Memory: {mem:4d} GB")
```
