# 05. Vector Databases & Approximate Nearest Neighbors (ANN)

Brute-force exact k-NN search requires comparing query vector $\mathbf{q}$ against all $N$ database vectors ($O(N \cdot d)$). For millions of vectors, **Approximate Nearest Neighbor (ANN)** indexing achieves sub-millisecond retrieval.

### Primary ANN Indexing Algorithms:
1. **HNSW (Hierarchical Navigable Small World)**: Multi-layer proximity graph with logarithmic search complexity ($O(\log N)$). Industry standard for accuracy and speed.
2. **IVF-PQ (Inverted File with Product Quantization)**: Clusters vectors into Voronoi cells and quantizes vectors into compact byte codes, saving massive RAM.
3. **Production Vector DBs**: ChromaDB, Pinecone, Milvus, Weaviate, Qdrant, pgvector.

```python
import numpy as np

# Exact k-NN Search Benchmark
N = 50000
d = 128
np.random.seed(42)
database = np.random.randn(N, d).astype(np.float32)
database /= np.linalg.norm(database, axis=1, keepdims=True)  # L2 normalize

query = np.random.randn(1, d).astype(np.float32)
query /= np.linalg.norm(query)

# Compute dot product across all 50,000 vectors
similarities = np.dot(database, query.T).squeeze()
top_5_idx = np.argsort(similarities)[-5:][::-1]

print("Top 5 Nearest Neighbor Indices:", top_5_idx)
print("Top 5 Similarity Scores:       ", similarities[top_5_idx].round(4))
```
