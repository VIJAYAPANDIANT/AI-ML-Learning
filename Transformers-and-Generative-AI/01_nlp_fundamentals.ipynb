# 01. Natural Language Processing (NLP) Fundamentals

Natural Language Processing (NLP) bridges human linguistic communication and computational machine learning algorithms.

### The Hierarchy of Linguistic Structure:
1. **Phonology & Morphology**: Subword roots, prefixes, suffixes, and inflections.
2. **Syntax**: Grammatical sentence structure, dependency trees, and parts of speech (POS).
3. **Semantics**: Contextual meaning of words, phrases, and sentences.
4. **Pragmatics & Discourse**: World knowledge, conversational intent, and sentiment.

### Learning Objectives:
- Understand the evolution from rule-based grammars to modern statistical representations
- Build a frequency distribution analyzer for raw text in pure Python

```python
from collections import Counter
import re

corpus = """
Artificial intelligence and natural language processing are transforming modern technology.
Transformers and generative AI enable computers to understand, reason, and generate human language.
"""

# Tokenize and compute word frequency
tokens = re.findall(r'\b[a-zA-Z]+\b', corpus.lower())
vocab = Counter(tokens)

print("Total Word Count: ", len(tokens))
print("Unique Vocabulary:", len(vocab))
print("Top 5 Most Frequent Words:", vocab.most_common(5))
```

## 🎯 Exercises & Practice
1. Calculate the Lexical Diversity (Type-Token Ratio: Unique Vocabulary / Total Words) of the corpus.

```python
# Exercise Solution
ttr = len(vocab) / len(tokens)
print(f"Type-Token Ratio (Lexical Diversity): {ttr:.4f}")
```
