# 06. Scaled Dot-Product Self-Attention & Multi-Head Attention

Unlike Seq2Seq attention between two separate sequences, **Self-Attention** relates different positions of a single sequence to compute a contextualized representation of that same sequence.

### Mathematical Formulation:
$$\text{Attention}(\mathbf{Q}, \mathbf{K}, \mathbf{V}) = \text{softmax}\left( \frac{\mathbf{Q}\mathbf{K}^T}{\sqrt{d_k}} \right) \mathbf{V}$$

### Multi-Head Attention:
Rather than performing a single attention function, Multi-Head Attention linearly projects $\mathbf{Q}, \mathbf{K}, \mathbf{V}$ into $h$ distinct subspace heads, computes attention in parallel, and concatenates the results:
$$\text{MultiHead}(\mathbf{Q}, \mathbf{K}, \mathbf{V}) = \text{Concat}(\text{head}_1, \dots, \text{head}_h) \mathbf{W}^O$$

```python
import torch
import torch.nn as nn

# MultiheadAttention primitive in PyTorch
mha = nn.MultiheadAttention(embed_dim=64, num_heads=4, batch_first=True)

# Sequence of 5 tokens, embedding dimension 64
x_tokens = torch.randn(2, 5, 64)
out, attn_map = mha(x_tokens, x_tokens, x_tokens)

print("Multi-Head Output Shape:   ", out.shape)
print("Attention Map Shape (B, L, L):", attn_map.shape)
```
