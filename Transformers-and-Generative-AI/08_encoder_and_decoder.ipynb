# 08. Encoder-Only vs. Decoder-Only vs. Encoder-Decoder

Depending on the task, Transformer architectures branch into three distinct families:

| Family | Archetype | Attention Type | Primary Use Cases |
| :--- | :--- | :--- | :--- |
| **Encoder-Only** | BERT, RoBERTa | Bidirectional (all tokens attend to all) | Classification, Named Entity Recognition, Extractive QA |
| **Decoder-Only** | GPT-2, GPT-4, LLaMA | Causal Masked (tokens attend only to left) | Text Generation, Code Synthesis, Conversational Chat |
| **Encoder-Decoder** | T5, BART | Bidirectional Encoder + Cross-Attention Decoder | Machine Translation, Text Summarization |

```python
import pandas as pd

df_arch = pd.DataFrame({
    "Family": ["Encoder-Only", "Decoder-Only", "Encoder-Decoder"],
    "Exemplar Models": ["BERT, RoBERTa, DeBERTa", "GPT-3/4, LLaMA, Mistral, Claude", "T5, BART, mT5"],
    "Attention Mask": ["Full Bidirectional (no mask)", "Causal Lower-Triangular Mask", "Bidirectional Encoder + Causal Cross-Attention"],
    "Core Objective": ["Masked LM (MLM)", "Causal Next-Token Prediction", "Sequence-to-Sequence Transformation"]
})
print(df_arch.to_string(index=False))
```
