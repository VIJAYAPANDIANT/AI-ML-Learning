# 01. Introduction to Generative AI: Discriminative vs. Generative Modeling

**Generative AI** encompasses algorithms capable of generating novel, high-fidelity synthetic content (text, imagery, audio, 3D structures, code) that reflects the underlying statistical distribution of training data.

### The Mathematical Divide:
- **Discriminative Models**: Learn conditional probability $P(y | \mathbf{x})$: Given features $\mathbf{x}$, predict class label or target value $y$. Focuses on decision boundaries.
- **Generative Models**: Learn joint probability $P(\mathbf{x}, y)$ or marginal data distribution $P(\mathbf{x})$: Models how data was generated, enabling unconditional or conditional sampling of synthetic instances $\mathbf{x} \sim P(\mathbf{x})$.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Discriminative vs Generative intuition
np.random.seed(42)
class_0 = np.random.normal([2, 2], 0.6, (100, 2))
class_1 = np.random.normal([5, 5], 0.6, (100, 2))

# Generative Sampling from learned distribution P(X|class_0)
synthetic_samples = np.random.normal([2, 2], 0.6, (30, 2))

plt.figure(figsize=(7, 4), dpi=100)
plt.scatter(class_0[:, 0], class_0[:, 1], color='#3498db', alpha=0.5, label='Actual Class 0 Data')
plt.scatter(class_1[:, 0], class_1[:, 1], color='#e74c3c', alpha=0.5, label='Actual Class 1 Data')
plt.scatter(synthetic_samples[:, 0], synthetic_samples[:, 1], color='#2ecc71', marker='*', s=120, label='Generated Synthetic Points')
plt.title("Generative Modeling: Sampling from Learned Density P(X)", fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
