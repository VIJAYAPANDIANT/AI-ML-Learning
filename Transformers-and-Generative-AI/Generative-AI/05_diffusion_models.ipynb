# 05. Diffusion Models: Forward Noising & Reverse Denoising

**Denoising Diffusion Probabilistic Models (DDPM)** (Sohl-Dickstein 2015, Ho et al. 2020) model generation as reversing a gradual thermodynamic diffusion process.

### The Two Processes:
1. **Forward Process ($q$)**: Gradually adds Gaussian noise $\epsilon \sim \mathcal{N}(0, \mathbf{I})$ across $T$ steps according to variance schedule $\beta_t$, turning any image into pure white noise:
   $$\mathbf{x}_t = \sqrt{\bar{\alpha}_t} \mathbf{x}_0 + \sqrt{1 - \bar{\alpha}_t} \epsilon$$
2. **Reverse Process ($p_\theta$)**: A neural network (typically a U-Net with cross-attention) learns to predict and subtract the added noise: $\epsilon_\theta(\mathbf{x}_t, t)$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulate forward diffusion on a synthetic 1D signal
T = 50
beta = np.linspace(0.0001, 0.02, T)
alpha = 1.0 - beta
alpha_bar = np.cumprod(alpha)

x_0 = np.array([2.5])  # Clean signal

noisy_trajectory = []
for t in range(T):
    noise = np.random.normal(0, 1)
    x_t = np.sqrt(alpha_bar[t]) * x_0 + np.sqrt(1 - alpha_bar[t]) * noise
    noisy_trajectory.append(float(x_t[0]))

plt.figure(figsize=(8, 3.5), dpi=100)
plt.plot(noisy_trajectory, color='#e67e22', lw=2)
plt.axhline(0, color='black', ls='--', lw=0.8)
plt.title("Forward Diffusion Process: Signal Decay into Pure Gaussian Noise", fontweight='bold')
plt.xlabel("Diffusion Time Step (t)")
plt.ylabel("Signal Value x_t")
plt.grid(True, alpha=0.3)
plt.show()
```
