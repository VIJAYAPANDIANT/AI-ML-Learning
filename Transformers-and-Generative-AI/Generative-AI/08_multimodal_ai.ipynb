# 08. Multimodal AI: Vision-Language Models & Contrastive Learning

**Multimodal AI** processes and aligns information across disparate sensory modalities (text, vision, audio).

### Contrastive Language-Image Pre-Training (CLIP) (Radford et al., OpenAI 2021):
CLIP trains an Image Encoder (Vision Transformer) and a Text Encoder (Transformer) jointly using symmetric cross-entropy loss over an $N \times N$ batch cosine similarity matrix, aligning visual concepts directly with natural language supervision.

### Vision-Language LLMs (VLMs: LLaVA, GPT-4o, Gemini):
Project visual patch tokens from a vision encoder into the text embedding space of a decoder-only LLM via a linear projection or cross-attention resampler.

```python
import torch
import torch.nn.functional as F

# Simulated CLIP Contrastive Similarity Matrix
N = 4 # Batch size of 4 paired image-caption examples
img_embeds = F.normalize(torch.randn(N, 64), dim=-1)
txt_embeds = F.normalize(torch.randn(N, 64), dim=-1)

# Compute pairwise cosine similarity matrix
temperature = 0.07
logits = torch.matmul(img_embeds, txt_embeds.t()) / temperature

labels = torch.arange(N) # Diagonal elements are matching positive pairs
loss_img = F.cross_entropy(logits, labels)
loss_txt = F.cross_entropy(logits.t(), labels)
clip_loss = (loss_img + loss_txt) / 2.0

print("CLIP Pairwise Similarity Logits Matrix (4x4):\n", logits.detach().numpy().round(2))
print("Symmetric Contrastive Loss:", round(clip_loss.item(), 4))
```
