# 03. Text Generation Dynamics & Beam Search

Autoregressive text generation constructs sequences token-by-token: $\hat{x}_{t} \sim P(\cdot \mid x_1, \dots, x_{t-1})$.

### Decoding Strategies:
1. **Greedy Search**: Selects $\arg\max P(w_t)$. Fast, but easily gets trapped in repetitive loops.
2. **Beam Search**: Maintains the top $B$ most probable candidate sequences across steps, pruning lower-probability paths.
3. **Stochastic Sampling**: Temperature, Top-K, Top-P for creative conversational fluidity.

```python
import numpy as np

# Beam Search Demonstration in Pure Python
def beam_search_decoder(predictions, beam_width=2):
    sequences = [[list(), 0.0]]
    for row in predictions:
        all_candidates = list()
        for seq, score in sequences:
            for j in range(len(row)):
                candidate = [seq + [j], score - np.log(row[j] + 1e-12)]
                all_candidates.append(candidate)
        # Order by score (lowest negative log likelihood)
        ordered = sorted(all_candidates, key=lambda tup: tup[1])
        sequences = ordered[:beam_width]
    return sequences

# Simulated token probabilities over 3 timesteps across 4 vocabulary tokens
mock_preds = np.array([
    [0.1, 0.7, 0.1, 0.1],
    [0.6, 0.2, 0.1, 0.1],
    [0.1, 0.1, 0.8, 0.0]
])

best_paths = beam_search_decoder(mock_preds, beam_width=2)
for idx, (seq, cost) in enumerate(best_paths):
    print(f"Beam {idx+1}: Token Sequence {seq} | Cumulative NLL Cost: {cost:.4f}")
```
