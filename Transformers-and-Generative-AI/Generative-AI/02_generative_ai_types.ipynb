# 02. Taxonomy of Generative AI Architectures

Generative AI encompasses several distinct mathematical paradigms:

| Family | Core Formulation | Generation Speed | Fidelity | Dominant Domain |
| :--- | :--- | :--- | :--- | :--- |
| **Autoregressive (LLMs)** | $P(x) = \prod_{t=1}^T P(x_t \mid x_{<t})$ | Sequential ($O(T)$ steps) | State-of-the-art | Text, Code, Music |
| **Diffusion Models** | Iterative reverse score denoising | Moderate ($20 - 50$ steps) | Photorealistic | Image, Video, 3D, Audio |
| **GANs** | Minimax Game: $\min_G \max_D V(D, G)$ | Fast ($1$ forward pass) | High / Prone to mode collapse | Real-time image generation |
| **VAEs** | Maximize Evidence Lower Bound (ELBO) | Fast ($1$ forward pass) | Smooth / Blurry | Latent compression, molecule design |
| **Flow-Based** | Invertible Jacobians | Fast / Invertible | Exact log-likelihood | Audio synthesis |

```python
import pandas as pd

model_types = pd.DataFrame({
    "Family": ["Autoregressive", "Diffusion", "GAN", "VAE"],
    "Sampling Steps": ["T tokens (Sequential)", "20-50 Denoising Steps", "Single Step (1 Forward Pass)", "Single Step (1 Forward Pass)"],
    "Training Stability": ["Extremely Stable (Cross-Entropy)", "Very Stable (L2 Denoising)", "Difficult (Non-convex Minimax)", "Stable (ELBO Reconstruction + KL)"],
    "Flagship Models": ["GPT-4, LLaMA-3, Claude", "Stable Diffusion, Midjourney, Flux", "StyleGAN-3, CycleGAN", "Nouveau VAE, Stable Diffusion VAE"]
})
print(model_types.to_string(index=False))
```
