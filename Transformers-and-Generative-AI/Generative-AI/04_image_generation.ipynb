# 04. Neural Image Generation Paradigms

Modern neural image synthesis models operate on **latent spaces** rather than raw pixels to drastically reduce compute requirements.

### Pixel-Space vs Latent-Space Generation:
- **Pixel-Space (e.g. PixelCNN, DALL-E 1)**: Directly predicts $512 \times 512 \times 3 = 786,432$ RGB values. Incredibly slow and memory-intensive.
- **Latent Diffusion (e.g. Stable Diffusion)**: Uses a pre-trained VAE encoder to compress the image $8\times$ spatially ($512 \times 512 \to 64 \times 64 \times 4$), performing the expensive generative diffusion process in compressed latent space!

```python
import numpy as np

# Calculate spatial compression ratio of Latent Diffusion
pixel_shape = (3, 512, 512)
latent_shape = (4, 64, 64)

pixel_elements = np.prod(pixel_shape)
latent_elements = np.prod(latent_shape)

print(f"Raw Pixel Space Dimensions:     {pixel_elements:,} floats")
print(f"Compressed Latent Dimensions:   {latent_elements:,} floats")
print(f"Compute Memory Reduction:       {pixel_elements / latent_elements:.1f}x smaller!")
```
