# 04. Distributed Word Embeddings: Word2Vec, GloVe & FastText

Word embeddings map discrete words into continuous dense vector spaces ($\mathbb{R}^d$, $d \approx 100 - 768$) where semantic relationships correspond to geometric proximity: $\vec{v}(\text{King}) - \vec{v}(\text{Man}) + \vec{v}(\text{Woman}) \approx \vec{v}(\text{Queen})$.

### Foundational Models:
- **Word2Vec (Mikolov et al., 2013)**: Continuous Bag of Words (CBOW: predict word from context) and Skip-Gram (predict context from target word).
- **GloVe (Pennington et al., Stanford 2014)**: Factorizes global word-word co-occurrence matrix logarithms.
- **FastText (Bojanowski et al., Meta 2016)**: Enriches words with character $n$-gram representations for morphological generalization.

```python
import numpy as np

def cosine_similarity(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

# Simulated semantic embedding space
embeddings = {
    "king":   np.array([0.9, 0.2, 0.8]),
    "queen":  np.array([0.88, 0.22, 0.75]),
    "man":    np.array([0.7, 0.1, 0.2]),
    "woman":  np.array([0.68, 0.12, 0.18]),
    "apple":  np.array([0.1, 0.9, 0.1])
}

print("Similarity(king, queen):", round(cosine_similarity(embeddings['king'], embeddings['queen']), 4))
print("Similarity(king, apple):", round(cosine_similarity(embeddings['king'], embeddings['apple']), 4))

# Analogy: King - Man + Woman
analogy = embeddings['king'] - embeddings['man'] + embeddings['woman']
print("Similarity(Analogy, queen):", round(cosine_similarity(analogy, embeddings['queen']), 4))
```
