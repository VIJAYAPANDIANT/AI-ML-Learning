# 03. The Lifecycle: Pre-Training vs. Fine-Tuning vs. Alignment

Understanding the exact transition from a base language model to an instruction-following conversational assistant.

### The 3 Stages:
1. **Base Pre-Training**: Learns internet text statistics by predicting $P(w_t | w_{<t})$.
2. **Supervised Fine-Tuning (SFT)**: Trains on paired `{"instruction": "...", "response": "..."}` datasets using cross-entropy loss.
3. **Preference Alignment**: Direct Preference Optimization (DPO) optimizes policy $\pi_\theta$ directly on paired preferences $(y_w \succ y_l)$ without a separate reward model!

```python
# SFT Data Format Example
sft_dataset = [
    {
        "messages": [
            {"role": "system", "content": "You are a helpful Python coding assistant."},
            {"role": "user", "content": "How do I reverse a list in Python in-place?"},
            {"role": "assistant", "content": "Use `my_list.reverse()`. It reverses the list in-place with O(N) time and O(1) auxiliary space."}
        ]
    }
]
print("Structured Chat Template Example:")
for msg in sft_dataset[0]['messages']:
    print(f"[{msg['role'].upper()}]: {msg['content']}")
```
