# 02. Modern LLM Architectures & Innovations

Modern open-weight LLMs (LLaMA-3, Mistral, Gemma) incorporate key architectural enhancements over the original 2017 Transformer:

### Modern Architectural Innovations:
1. **RMSNorm (Root Mean Square Normalization)**: Replaces LayerNorm by removing mean-centering: $\bar{a}_i = \frac{a_i}{\text{RMS}(\mathbf{a})} g_i$, reducing latency by ~15%.
2. **SwiGLU Activation Function**: Gated linear unit: $\text{SwiGLU}(x) = (x W \cdot \sigma(x W)) \cdot (x V)$, superior to ReLU and standard GELU.
3. **Rotary Position Embedding (RoPE)**: Relative distance encoding applied to $Q$ and $K$ via rotation matrices.
4. **Grouped-Query Attention (GQA)**: Shares key and value heads across query heads, dramatically slashing KV cache memory.

```python
import torch
import torch.nn as nn

class RMSNorm(nn.Module):
    def __init__(self, dim, eps=1e-6):
        super().__init__()
        self.eps = eps
        self.weight = nn.Parameter(torch.ones(dim))

    def forward(self, x):
        # RMS calculation
        rms = torch.sqrt(torch.mean(x ** 2, dim=-1, keepdim=True) + self.eps)
        return (x / rms) * self.weight

norm = RMSNorm(dim=64)
x = torch.randn(2, 4, 64)
print("RMSNorm Output Shape:", norm(x).shape)
```
