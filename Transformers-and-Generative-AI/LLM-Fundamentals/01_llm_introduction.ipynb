# 01. Introduction to Large Language Models (LLMs)

**Large Language Models (LLMs)** are foundation models trained on massive multimodal and textual corpora (trillions of tokens) exhibiting emergent capabilities such as in-context reasoning, code generation, and multi-step problem solving.

### The Three Stages of Modern LLM Development:
1. **Pre-training**: Next-token prediction on internet-scale raw text (unsupervised, costs millions of dollars in compute).
2. **Supervised Fine-Tuning (SFT)**: High-quality instruction-response demonstrations transforming a raw next-word predictor into an assistant.
3. **Preference Alignment (RLHF / DPO)**: Reinforcement Learning from Human Feedback (RLHF) or Direct Preference Optimization (DPO) to maximize helpfulness and eliminate harm.
