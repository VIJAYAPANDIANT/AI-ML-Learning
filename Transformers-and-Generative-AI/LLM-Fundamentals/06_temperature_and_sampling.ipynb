# 06. Generation Decoding: Temperature, Top-K, Top-P (Nucleus)

Autoregressive models output raw unnormalized logits $\mathbf{z}$ across vocabulary $V$. Decoding strategies sample next tokens from logits.

### Decoding Parameters:
- **Temperature ($T$)**: Rescales logits $z_i / T$. Low $T \to 0$ makes distribution greedy (argmax); high $T > 1$ flattens distribution (creative/diverse).
- **Top-K**: Truncates candidate pool to top $K$ highest-probability tokens.
- **Top-P (Nucleus Sampling)**: Selects smallest dynamic subset of tokens whose cumulative probability $\sum p_i \ge P$ (e.g. $P = 0.9$).

```python
import numpy as np

def sample_with_temperature(logits, temperature=1.0):
    if temperature <= 1e-4:
        return np.argmax(logits)  # Greedy
    scaled_logits = logits / temperature
    exp_logits = np.exp(scaled_logits - np.max(scaled_logits))
    probs = exp_logits / np.sum(exp_logits)
    return np.random.choice(len(probs), p=probs), probs

logits = np.array([2.0, 1.5, 0.5, 0.1])
for temp in [0.2, 0.7, 1.5]:
    _, probs = sample_with_temperature(logits, temperature=temp)
    print(f"Temperature {temp:3.1f} Probabilities:", probs.round(3))
```
