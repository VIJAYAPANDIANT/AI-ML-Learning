# 04. Advanced Prompt Engineering & In-Context Learning

Prompt engineering guides LLM reasoning without updating model weights through structured formatting and in-context conditioning.

### Key Techniques:
1. **Zero-Shot Prompting**: Direct task query without examples.
2. **Few-Shot Prompting**: Providing 2–5 input-output exemplars establishing format and style.
3. **Chain-of-Thought (CoT) (Wei et al., 2022)**: "Think step-by-step" prompts that elicit intermediate reasoning tokens.
4. **ReAct (Reason + Act) (Yao et al., 2022)**: Interleaving Thought, Action, and Observation tokens for tool-use agents.

```python
def format_cot_prompt(question, examples=None):
    prompt = "Solve the following problem. Explain your reasoning step-by-step before stating the final answer.\n\n"
    if examples:
        for ex in examples:
            prompt += f"Q: {ex['q']}\nReasoning: {ex['thought']}\nFinal Answer: {ex['ans']}\n\n"
    prompt += f"Q: {question}\nReasoning: Let's think step-by-step:"
    return prompt

example = {
    'q': 'A store sells 3 apples for $6. How much do 8 apples cost?',
    'thought': '1. Cost per apple = $6 / 3 = $2. 2. For 8 apples: 8 * $2 = $16.',
    'ans': '$16'
}

prompt_text = format_cot_prompt("If a train travels 120 miles in 2 hours, how far does it travel in 5 hours?", [example])
print(prompt_text)
```
