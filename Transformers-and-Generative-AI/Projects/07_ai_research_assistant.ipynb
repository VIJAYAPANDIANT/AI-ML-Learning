# Project 07: Autonomous Multi-Tool AI Research Assistant

### Objective:
Build an autonomous research agent equipped with web search, Python code execution, and dynamic note-taking tools using the ReAct (Reason + Act) loop.

```python
class ResearchAgent:
    def __init__(self):
        self.notes = []

    def search_arxiv(self, query):
        return f"Found paper 'Scaling Laws for Neural Language Models' (Kaplan et al., 2020) for query '{query}'"

    def save_note(self, note):
        self.notes.append(note)
        return f"Note saved ({len(self.notes)} total notes)"

    def run_research(self, topic):
        print(f"Starting research on: {topic}")
        paper = self.search_arxiv(topic)
        self.save_note(paper)
        return f"Research Complete. Synthesized Findings:\n- {paper}"

agent = ResearchAgent()
summary = agent.run_research("LLM Scaling Laws")
print(summary)
```
