# Project 01: Multi-Turn Enterprise Conversational Chatbot with Sliding Memory

### Objective:
Build a stateful conversational chatbot incorporating a sliding conversation window buffer, intent classification, and system prompt personality conditioning.

```python
class ConversationalMemory:
    def __init__(self, max_turns=3):
        self.max_turns = max_turns
        self.history = []

    def add_turn(self, role, content):
        self.history.append({"role": role, "content": content})
        if len(self.history) > self.max_turns * 2:
            self.history = self.history[-self.max_turns * 2:]

    def get_context(self):
        return self.history

mem = ConversationalMemory(max_turns=2)
mem.add_turn("user", "Hello, my name is Alex.")
mem.add_turn("assistant", "Nice to meet you, Alex! How can I assist you today?")
mem.add_turn("user", "I need help with my cloud billing invoice.")
mem.add_turn("assistant", "I would be happy to review your billing invoice.")
mem.add_turn("user", "What was my name again?")

print("Current Active Memory Window (Sliding):")
for turn in mem.get_context():
    print(f"[{turn['role'].upper()}]: {turn['content']}")
```
