# Project 02: Multi-Document Question Answering with Source Citations

### Objective:
Build an automated Question Answering engine that queries heterogeneous documents, extracts exact answers, and attaches verifiable source citations and page references.

```python
def qa_with_citations(question, corpus_documents):
    matched_sources = []
    for doc in corpus_documents:
        if any(w.lower() in doc['text'].lower() for w in question.split()):
            matched_sources.append(doc)
    
    response = f"Answer synthesized for: '{question}'\n\n"
    for i, s in enumerate(matched_sources):
        response += f"[{i+1}] {s['title']} (Page {s['page']}): "{s['text']}"\n"
    return response

docs = [
    {"title": "Q3 Financial Report", "page": 12, "text": "Operating margin expanded to 28.4% driven by cloud growth."},
    {"title": "Annual Sustainability Review", "page": 5, "text": "Datacenter renewable energy utilization reached 94%."}
]
print(qa_with_citations("What was the operating margin growth?", docs))
```
