# Project 05: Long-Form Document Summarization (Map-Reduce & Refine)

### Objective:
Implement hierarchical summarization for documents that exceed the model's single context window using **Map-Reduce** and **Refine** chunking algorithms.

```python
def map_reduce_summarizer(chunks, summarize_fn):
    # 1. Map: Summarize each individual chunk
    intermediate_summaries = [summarize_fn(chunk) for chunk in chunks]
    print(f"Generated {len(intermediate_summaries)} chunk summaries.")
    
    # 2. Reduce: Combine and generate final executive synthesis
    combined = " ".join(intermediate_summaries)
    final_summary = f"Executive Summary: {combined[:120]}..."
    return final_summary

mock_chunks = ["Financial revenue grew 12% in Q1.", "Cloud operations expanded globally in Q2.", "AI adoption reached 40% in Q3."]
exec_summary = map_reduce_summarizer(mock_chunks, lambda c: c.replace(".", ";"))
print(exec_summary)
```
