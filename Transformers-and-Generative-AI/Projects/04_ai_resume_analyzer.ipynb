# Project 04: AI Resume Analyzer & Job Description Matcher

### Objective:
Extract candidate work experience, technical skills, and educational qualifications from unstructured resumes, computing a match score against job specifications using structured LLM schemas.

```python
candidate_skills = {"python", "pytorch", "transformers", "fastapi", "docker", "kubernetes"}
job_requirements = {"python", "pytorch", "transformers", "distributed_training", "aws", "docker"}

matching_skills = candidate_skills.intersection(job_requirements)
missing_skills = job_requirements - candidate_skills
match_score = len(matching_skills) / len(job_requirements) * 100

print(f"Candidate Skill Match Score: {match_score:.1f}%")
print("Matching Proficiencies:    ", sorted(list(matching_skills)))
print("Skill Gaps to Upskill:     ", sorted(list(missing_skills)))
```
