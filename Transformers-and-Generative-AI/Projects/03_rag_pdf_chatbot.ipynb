```python
import numpy as np

# Simulated PDF Chunking & Vector Search
pdf_chunks = [
    {"id": "chunk_01", "text": "Section 4.1: The encryption protocol utilizes AES-256-GCM for all at-rest storage."},
    {"id": "chunk_02", "text": "Section 4.2: Maximum session idle timeout is enforced at 15 minutes across all tenants."},
    {"id": "chunk_03", "text": "Section 5.0: Disaster recovery recovery point objective (RPO) is guaranteed at 5 minutes."}
]

# Vector similarity lookup simulation
def search_pdf(query, chunks):
    return [c for c in chunks if "encryption" in c['text'].lower()]

retrieved = search_pdf("What encryption is used for data at rest?", pdf_chunks)
print("Retrieved Document Evidence:")
for r in retrieved:
    print(f"- ({r['id']}): {r['text']}")
```
