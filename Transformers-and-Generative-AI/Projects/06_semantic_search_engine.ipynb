# Project 06: Production Semantic Search Engine with Bi-Encoder & Cross-Encoder

### Objective:
Construct a two-stage search pipeline:
1. Fast stage: Bi-Encoder vector retrieval retrieving top 20 candidates in $<10$ ms
2. Accurate stage: Cross-Encoder neural re-ranker ranking top 5 results with full cross-attention

```python
import numpy as np

# Stage 1: Fast Bi-Encoder Retrieval Simulation
candidates = [
    {"id": 1, "text": "Deep Learning Transformer architectures for natural language processing."},
    {"id": 2, "text": "Python recipes for baking sourdough bread."},
    {"id": 3, "text": "Attention mechanisms and self-attention in BERT and GPT models."}
]

# Stage 2: Cross-Encoder Precision Re-Ranking Scores (Simulated)
rerank_scores = [0.85, 0.02, 0.94]
for cand, score in zip(candidates, rerank_scores):
    cand['rerank_score'] = score

ranked_results = sorted(candidates, key=lambda c: c['rerank_score'], reverse=True)
print("Two-Stage Semantic Search Rankings:")
for rank, r in enumerate(ranked_results):
    print(f"Rank {rank+1} (Score {r['rerank_score']:.2f}): {r['text']}")
```
