# 10. Deep Dive: BERT vs. GPT Architecture

BERT (Devlin et al., 2018) and GPT (Radford et al., 2018) established the modern pre-train & fine-tune paradigm in NLP.

### Architectural Contrast:
- **BERT (Bidirectional Encoder Representations from Transformers)**:
  - Training Objective: Masked Language Modeling (MLM, masking 15% of tokens) + Next Sentence Prediction (NSP)
  - Context: Full bidirectional context visible simultaneously
- **GPT (Generative Pre-trained Transformer)**:
  - Training Objective: Causal Language Modeling (CLM: predict $x_t$ given $x_{<t}$)
  - Context: Left-to-right causal context only

```python
import torch
import torch.nn as nn

# Demonstrating causal masking for GPT vs unmasked BERT
seq_len = 4
bert_mask = torch.ones((seq_len, seq_len))
gpt_mask = torch.tril(torch.ones((seq_len, seq_len)))

print("BERT Attention Visibility (Full Bidirectional):\n", bert_mask.numpy().astype(int))
print("\nGPT Attention Visibility (Causal Lower-Triangular):\n", gpt_mask.numpy().astype(int))
```
