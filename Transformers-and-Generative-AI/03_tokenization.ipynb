# 03. Tokenization Algorithms: Subword, BPE & WordPiece

Tokenization partitions continuous text into discrete atomic indices. Modern LLMs use **Subword Tokenization** to eliminate Out-Of-Vocabulary (OOV) tokens while keeping vocabulary sizes compact ($32,000 - 128,000$).

### Primary Algorithms:
- **Byte-Pair Encoding (BPE)** (Used by GPT-2, GPT-4, LLaMA): Iteratively merges the most frequent pairs of bytes/characters.
- **WordPiece** (Used by BERT): Merges pairs that maximize likelihood under a language model.
- **SentencePiece**: Language-independent subword tokenizer treating whitespace as an explicit character (`_`).

```python
from collections import defaultdict

# Toy Byte-Pair Encoding (BPE) Demonstration
def get_stats(vocab):
    pairs = defaultdict(int)
    for word, freq in vocab.items():
        symbols = word.split()
        for i in range(len(symbols) - 1):
            pairs[symbols[i], symbols[i+1]] += freq
    return pairs

def merge_vocab(pair, v_in):
    v_out = {}
    bigram = ' '.join(pair)
    replacement = ''.join(pair)
    for word in v_in:
        w_out = word.replace(bigram, replacement)
        v_out[w_out] = v_in[word]
    return v_out

vocab = {'l o w </w>': 5, 'l o w e r </w>': 2, 'n e w e s t </w>': 6, 'w i d e s t </w>': 3}
pairs = get_stats(vocab)
best = max(pairs, key=pairs.get)
print(f"Top bigram pair to merge: {best} (Frequency: {pairs[best]})")
vocab = merge_vocab(best, vocab)
print("Vocabulary after merge:\n", vocab)
```
