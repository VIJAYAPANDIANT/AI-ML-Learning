# 02. Text Preprocessing & Cleaning Pipelines

Raw text is unstructured and noisy (HTML tags, punctuation, special symbols, casing inconsistencies). Preprocessing prepares text for numerical representation.

### Core Cleaning Steps:
- **Normalization**: Lowercasing, Unicode NFKD normalization, stripping accents
- **Noise Removal**: Regex removal of URLs, email addresses, HTML tags
- **Stopword Removal**: Eliminating high-frequency non-informative tokens (the, is, at)
- **Stemming vs Lemmatization**: Heuristic rule truncation (Porter) vs vocabulary morphological reduction (WordNet)

```python
import re

def clean_text(raw_text):
    # 1. Remove HTML tags
    text = re.sub(r'<.*?>', '', raw_text)
    # 2. Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)
    # 3. Lowercase and keep letters only
    text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
    # 4. Remove extra whitespaces
    text = ' '.join(text.split())
    return text

sample_noisy = "<div>Check out https://ai.google for <b>deep learning</b> updates! #AI2026</div>"
cleaned = clean_text(sample_noisy)
print("Raw Input: ", sample_noisy)
print("Cleaned:   ", cleaned)
```
