# 03. Pandas DataFrame

A **DataFrame** is a 2-dimensional tabular data structure with labeled axes (rows and columns). It is the central object in data manipulation and ML pipelines.

## 1. DataFrame Creation Methods
DataFrames can be constructed from a dictionary of lists, a list of dictionaries, or a 2D NumPy array.

```python
import pandas as pd
import numpy as np

# Method: List of Dictionaries (Row-oriented records)
employees = [
    {'EmpID': 'E01', 'Name': 'Sarah', 'Dept': 'Engineering', 'Salary': 115000},
    {'EmpID': 'E02', 'Name': 'David', 'Dept': 'Marketing', 'Salary': 85000},
    {'EmpID': 'E03', 'Name': 'Elena', 'Dept': 'Engineering', 'Salary': 120000},
    {'EmpID': 'E04', 'Name': 'Marcus', 'Dept': 'Product', 'Salary': 95000}
]
df_emp = pd.DataFrame(employees)
print("Employee DataFrame:")
df_emp
```

## 2. Custom Indices & Column Selection

```python
# Set EmpID as row index
df_indexed = df_emp.set_index('EmpID')
print("Indexed DataFrame:")
print(df_indexed)

# Selecting single column yields Series; double bracket yields DataFrame
print(f"\nSalary Series:\n{df_indexed['Salary']}")
print(f"\nSalary DataFrame:\n{df_indexed[['Salary']]}")
```

## 3. Structural Attributes

```python
print(f"Shape: {df_emp.shape}")
print(f"Columns: {list(df_emp.columns)}")
print(f"Total Cells: {df_emp.size}")
print(f"Data Types:\n{df_emp.dtypes}")
```

## 🎯 Exercises
1. Create a DataFrame representing 5 store inventory items (`ProductID`, `Name`, `Price`, `Stock`).
2. Set `ProductID` as index.
3. Add a computed column `Inventory_Value = Price * Stock`.

```python
# Solution
items = {
    'ProductID': ['P101', 'P102', 'P103', 'P104', 'P105'],
    'Name': ['Keyboard', 'Monitor', 'Mouse', 'USB Cable', 'Webcam'],
    'Price': [89.99, 249.50, 29.99, 12.50, 75.00],
    'Stock': [45, 18, 120, 250, 32]
}
df_store = pd.DataFrame(items).set_index('ProductID')
df_store['Inventory_Value'] = df_store['Price'] * df_store['Stock']
print(df_store)
```
