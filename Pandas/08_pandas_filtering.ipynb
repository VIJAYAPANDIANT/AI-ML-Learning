# 08. Filtering Data in Pandas

Boolean masking is the cornerstone of data filtering. Learn single conditions, multi-condition logic (`&`, `|`, `~`), `.isin()`, and `.query()`.

## 1. Boolean Masking & Multi-Condition Logic
Always wrap individual conditions in parentheses `(cond1) & (cond2)`.

```python
import pandas as pd

customers = pd.DataFrame({
    'Name': ['Arthur', 'Beatrix', 'Cassian', 'Daphne', 'Ezekiel', 'Freya'],
    'Country': ['UK', 'USA', 'Germany', 'USA', 'Germany', 'UK'],
    'Age': [28, 42, 35, 19, 58, 31],
    'Spend_USD': [1200, 3400, 850, 450, 7200, 2100]
})

# Filter: Country is USA OR Spend > 2000
filtered = customers[(customers['Country'] == 'USA') | (customers['Spend_USD'] > 2000)]
print("High spenders or USA:")
print(filtered)
```

## 2. Set Membership (`.isin()`) & Ranges (`.between()`)

```python
# Filter by Country in list
europe_customers = customers[customers['Country'].isin(['UK', 'Germany'])]
print("European Customers:")
print(europe_customers)

# Filter by Age between 30 and 45 inclusive
mid_age = customers[customers['Age'].between(30, 45)]
print("\nAge 30 to 45 Customers:")
print(mid_age)
```

## 3. SQL-like Filtering with `.query()`
Provides readable, string-based conditional filtering.

```python
result = customers.query("Spend_USD > 1000 and Age < 40")
print("Queried Result:")
print(result)
```

## 🎯 Exercises
1. Filter customers from the UK with `Spend_USD >= 1500`.
2. Use `.query()` to find all customers whose age is NOT between 25 and 40.

```python
# Solution
uk_spenders = customers[(customers['Country'] == 'UK') & (customers['Spend_USD'] >= 1500)]
print("UK High Spenders:")
print(uk_spenders)

not_mid_age = customers.query("Age < 25 or Age > 40")
print("\nAge < 25 or > 40:")
print(not_mid_age)
```
