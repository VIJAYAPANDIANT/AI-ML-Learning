# 21. Pivot Tables in Pandas

Reshaping tabular data into multidimensional summary tables with `pd.pivot_table()` and reverting with `pd.melt()`.

## 1. Creating a Pivot Table: `pd.pivot_table()`

```python
import pandas as pd

sales_data = pd.DataFrame({
    'Region': ['East', 'East', 'West', 'West', 'East', 'West', 'East', 'West'],
    'Quarter': ['Q1', 'Q2', 'Q1', 'Q2', 'Q1', 'Q2', 'Q2', 'Q1'],
    'Product': ['Tech', 'Tech', 'Tech', 'Tech', 'Office', 'Office', 'Office', 'Office'],
    'Revenue': [12000, 15000, 11000, 18000, 4000, 5000, 4500, 6000]
})

# Multi-dimensional pivot table: Revenue across Region and Product
pivot_rev = pd.pivot_table(
    sales_data,
    values='Revenue',
    index='Region',
    columns='Product',
    aggfunc='sum',
    margins=True,       # Adds row/column totals ('All')
    margins_name='Total'
)
print("Revenue Pivot Table:")
print(pivot_rev)
```

## 2. Multi-Metric Pivot Tables

```python
multi_pivot = pd.pivot_table(
    sales_data,
    values='Revenue',
    index='Region',
    columns='Quarter',
    aggfunc=['mean', 'sum']
)
print("Multi-Metric Pivot Table:")
print(multi_pivot)
```

## 3. Unpivoting with `pd.melt()`

```python
wide_df = pd.DataFrame({
    'Quarter': ['Q1', 'Q2', 'Q3'],
    'Product_A': [100, 150, 200],
    'Product_B': [90, 110, 130]
})

long_df = pd.melt(wide_df, id_vars=['Quarter'], var_name='Product', value_name='Sales')
print("Melted (Long Format) DataFrame:")
print(long_df)
```

## 🎯 Exercises
1. Pivot `sales_data` to show average revenue for each `Product` across each `Quarter`.
2. Fill any missing combinations with 0 using `fill_value=0`.

```python
# Solution
q_pivot = pd.pivot_table(
    sales_data,
    values='Revenue',
    index='Product',
    columns='Quarter',
    aggfunc='mean',
    fill_value=0
)
print(q_pivot)
```
