# 07. Indexing and Slicing in Pandas

Master `.loc` (label-based), `.iloc` (integer-position-based), and scalar lookups (`.at`, `.iat`).

## 1. Deep Dive: `.loc[]` vs `.iloc[]`
- `.loc[row_labels, col_labels]`: inclusive of stop bounds!
- `.iloc[row_indices, col_indices]`: exclusive of stop bounds (Python slice standard).

```python
import pandas as pd

data = {
    'Revenue': [120, 150, 170, 200, 240],
    'Costs': [80, 95, 110, 130, 150],
    'Units': [500, 620, 710, 850, 990]
}
df = pd.DataFrame(data, index=['Q1', 'Q2', 'Q3', 'Q4', 'Q5_Proj'])
print("Financials DataFrame:")
print(df)

# .loc slices by label (Q1 to Q3 inclusive!)
print("\n.loc['Q1':'Q3', ['Revenue', 'Costs']]:")
print(df.loc['Q1':'Q3', ['Revenue', 'Costs']])

# .iloc slices by index (0 to 3 exclusive, i.e., 0, 1, 2)
print("\n.iloc[0:3, 0:2]:")
print(df.iloc[0:3, 0:2])
```

## 2. Fast Scalar Access: `.at` & `.iat`
Optimized for high-throughput single value retrieval.

```python
print(f"Scalar via .at['Q4', 'Revenue']: ${df.at['Q4', 'Revenue']}k")
print(f"Scalar via .iat[3, 0]: ${df.iat[3, 0]}k")
```

## 🎯 Exercises

2. Using `.iloc`, retrieve the very last row and last column cell.

```python
# Solution
print("Q2 to Q4 Units:")
print(df.loc['Q2':'Q4', ['Units']])
print(f"\nBottom-right cell: {df.iat[-1, -1]}")
```
