# 04. Reading Data into Pandas

Loading datasets efficiently is the critical entry point of every data science workflow. Pandas supports CSV, Excel, JSON, Parquet, SQL, and HTML.

## 1. Reading CSV Files (`pd.read_csv`)
Using `io.StringIO` to simulate file reading with parameters: `sep`, `index_col`, `usecols`, and `parse_dates`.

```python
import pandas as pd
import io

csv_text = """order_id,customer,order_date,amount,status
1001,John Doe,2026-01-15,149.50,Completed
1002,Jane Smith,2026-01-16,89.00,Completed
1003,Bob Johnson,2026-01-17,210.75,Pending
1004,Alice Brown,2026-01-18,45.20,Cancelled
1005,Charlie Black,2026-01-19,320.00,Completed
"""

df_orders = pd.read_csv(io.StringIO(csv_text), parse_dates=['order_date'])
print("Loaded CSV DataFrame:")
print(df_orders)
print("\nParsed Dtypes:")
print(df_orders.dtypes)
```

## 2. Selective Column Loading (`usecols`) & Row Limits (`nrows`)
Save memory when loading massive multi-gigabyte datasets.

```python
# Load only order_id and amount, limited to 3 rows
df_small = pd.read_csv(io.StringIO(csv_text), usecols=['order_id', 'amount'], nrows=3)
print("Optimized Subset:")
print(df_small)
```

## 3. Reading JSON Data (`pd.read_json`)

```python
json_payload = """[
    {"sensor": "temp_01", "reading": 24.3, "active": true},
    {"sensor": "temp_02", "reading": 26.1, "active": false},
    {"sensor": "temp_03", "reading": 21.8, "active": true}
]"""

df_sensors = pd.read_json(io.StringIO(json_payload))
print("Loaded JSON:")
print(df_sensors)
```

## 🎯 Exercises
1. Simulate a CSV with 4 transactions: `tx_id`, `user`, `timestamp`, `usd_value`.
2. Read it with `tx_id` as the index and parse `timestamp` as datetime.

```python
# Solution
tx_csv = """tx_id,user,timestamp,usd_value
TX01,usr_alpha,2026-02-01 10:15:00,450.00
TX02,usr_beta,2026-02-01 11:20:30,1200.50
TX03,usr_gamma,2026-02-01 12:05:10,32.75
TX04,usr_delta,2026-02-01 13:45:00,890.00
"""
df_tx = pd.read_csv(io.StringIO(tx_csv), index_col='tx_id', parse_dates=['timestamp'])
print(df_tx)
print(f"\nIndex: {df_tx.index}")
print(f"Timestamp dtype: {df_tx['timestamp'].dtype}")
```
