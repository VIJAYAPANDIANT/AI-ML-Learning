# 17. Apply, Map, and Lambda Functions

Custom transformations across Series and DataFrames using `.apply()`, `.map()`, and lambda expressions.

## 1. Applying Functions to Series: `.apply()` & `lambda`

```python
import pandas as pd

df = pd.DataFrame({
    'Product': ['Standard Laptop', 'Pro Monitor', 'Wireless Mouse'],
    'Price_USD': [1200, 450, 35],
    'Tax_Rate': [0.08, 0.08, 0.05]
})

# Series apply with lambda
df['Price_EUR'] = df['Price_USD'].apply(lambda usd: round(usd * 0.92, 2))
print(df)
```

## 2. Row-Wise DataFrame Apply (`axis=1`)
Useful when computation spans multiple columns simultaneously.

```python
# Total price with tax
df['Total_Price'] = df.apply(lambda row: row['Price_USD'] * (1 + row['Tax_Rate']), axis=1)
print("Calculated with row-wise apply:")
print(df)
```

## 3. Dictionary Mapping with `.map()` on Series

```python
tier_map = {1200: 'Premium', 450: 'Mid-Tier', 35: 'Budget'}
df['Category_Tier'] = df['Price_USD'].map(tier_map)
print("Mapped Categories:")
print(df[['Product', 'Price_USD', 'Category_Tier']])
```

## 🎯 Exercises
1. Write a custom Python function `classify_price(p)` that returns 'High' if price > 500, else 'Standard'.
2. Apply it to `Price_USD`.

```python
# Solution
def classify_price(p):
    return 'High' if p > 500 else 'Standard'

df['Price_Classification'] = df['Price_USD'].apply(classify_price)
print(df[['Product', 'Price_USD', 'Price_Classification']])
```
