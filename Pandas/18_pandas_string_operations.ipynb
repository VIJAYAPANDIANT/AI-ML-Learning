# 18. Vectorized String Operations (`.str`)

Text cleaning and manipulation via Pandas' `.str` accessor.

## 1. Case Conversion & Whitespace Cleaning

```python
import pandas as pd

emails = pd.Series([
    '  Alice.Smith@Domain.COM ',
    'bob_builder@WORK.net',
    ' Charlie_Brown@domain.com  ',
    'diana.prince@AMAZON.ORG'
])

# Strip whitespace and convert to lowercase
clean_emails = emails.str.strip().str.lower()
print("Cleaned Emails:")
print(clean_emails)
```

## 2. Pattern Matching: `.str.contains()` & `.str.startswith()`

```python
# Check if email belongs to domain.com
is_domain_com = clean_emails.str.contains('domain\.com', regex=True)
print("Belongs to domain.com:")
print(is_domain_com)
```

## 3. String Splitting & Regex Extraction

```python
# Split username and domain into separate columns
split_cols = clean_emails.str.split('@', expand=True)
split_cols.columns = ['Username', 'Domain']
print("\nExtracted Columns:")
print(split_cols)
```

## 🎯 Exercises
1. Extract usernames and replace any '.' or '_' with a single space.
2. Convert names to Title Case.

```python
# Solution
clean_names = split_cols['Username'].str.replace(r'[._]', ' ', regex=True).str.title()
print("Formatted Names:")
print(clean_names)
```
