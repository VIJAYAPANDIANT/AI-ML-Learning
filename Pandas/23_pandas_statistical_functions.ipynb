# 23. Statistical Functions in Pandas

Descriptive metrics, quantiles, correlation matrices, covariance, and cumulative window functions.

## 1. Summary Statistics & Quantiles

```python
import pandas as pd
import numpy as np

np.random.seed(42)
data = pd.DataFrame({
    'Marketing_Spend': [1000, 1500, 2000, 2500, 3000, 3500, 4000],
    'Sales_Units': [50, 75, 90, 120, 145, 170, 210],
    'Ad_Clicks': [120, 180, 240, 290, 350, 410, 520]
})

print("Statistical Summary:")
print(data.describe())

print("\nSkewness of features:")
print(data.skew())
```

## 2. Correlation & Covariance Matrix
Critical for multicollinearity detection and feature selection in machine learning.

```python
# Pearson correlation matrix
corr_matrix = data.corr(method='pearson')
print("Correlation Matrix:")
print(corr_matrix)
```

## 3. Cumulative Statistics: `.cumsum()`, `.cummax()`

```python
data['Cumulative_Sales'] = data['Sales_Units'].cumsum()
data['Peak_Ad_Clicks'] = data['Ad_Clicks'].cummax()
print("Cumulative Features:")
print(data[['Sales_Units', 'Cumulative_Sales', 'Ad_Clicks', 'Peak_Ad_Clicks']])
```

## 🎯 Exercises
1. Calculate the covariance between `Marketing_Spend` and `Sales_Units`.
2. Check the 90th percentile of `Sales_Units`.

```python
# Solution
cov_val = data['Marketing_Spend'].cov(data['Sales_Units'])
print(f"Covariance: {cov_val:,.2f}")
print(f"90th Percentile Sales: {data['Sales_Units'].quantile(0.90)}")
```
