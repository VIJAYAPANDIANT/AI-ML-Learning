### Problem 1: Employee Retention & Tenure Analysis

```python
import pandas as pd
import numpy as np

emp = pd.DataFrame({
    'EmpID': [101, 102, 103, 104, 105, 106],
    'Dept': ['Eng', 'HR', 'Eng', 'Sales', 'Sales', 'Eng'],
    'HireDate': pd.to_datetime(['2020-03-01', '2018-06-15', '2022-01-10', '2019-11-20', '2023-05-01', '2017-08-01']),
    'Salary': [110000, 68000, 95000, 82000, 75000, 130000]
})

# Calculate tenure in years relative to 2026-01-01
ref_date = pd.Timestamp('2026-01-01')
emp['Tenure_Years'] = ((ref_date - emp['HireDate']).dt.days / 365.25).round(1)
print("Tenure Analysis:")
print(emp)
```

### Problem 2: Department-Level Salary Gap

```python
dept_gap = emp.groupby('Dept')['Salary'].agg(Max_Salary='max', Min_Salary='min')
dept_gap['Salary_Spread'] = dept_gap['Max_Salary'] - dept_gap['Min_Salary']
print("Department Salary Spread:")
print(dept_gap)
```

### Problem 3: Rolling 3-Day Sales Average

```python
daily_sales = pd.DataFrame({
    'Date': pd.date_range('2026-01-01', periods=7, freq='D'),
    'Sales': [100, 140, 120, 200, 180, 220, 250]
}).set_index('Date')

daily_sales['Rolling_3D_Avg'] = daily_sales['Sales'].rolling(window=3).mean().round(2)
print("Rolling 3-Day Moving Average:")
print(daily_sales)
```

### Problem 4: Detecting Outliers with IQR (Interquartile Range)

```python
def find_iqr_outliers(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return series[(series < lower) | (series > upper)]

test_vals = pd.Series([10, 12, 11, 14, 13, 12, 95, 11, -40, 15])
print("Detected Outliers via IQR:")
print(find_iqr_outliers(test_vals))
```

### Problem 5: Complex Multi-Condition Customer Segmentation

```python
customers = pd.DataFrame({
    'CustID': [1, 2, 3, 4, 5],
    'Orders': [15, 2, 8, 25, 1],
    'TotalSpend': [3500, 150, 1200, 8500, 45]
})

conditions = [
    (customers['Orders'] >= 10) & (customers['TotalSpend'] >= 3000),
    (customers['Orders'] >= 5) & (customers['TotalSpend'] >= 1000)
]
choices = ['VIP Elite', 'Loyal Regular']
customers['Segment'] = np.select(conditions, choices, default='Occasional')
print("Segmented Customers:")
print(customers)
```
