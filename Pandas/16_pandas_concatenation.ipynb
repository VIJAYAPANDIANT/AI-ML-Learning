# 16. Concatenation in Pandas

Stacking DataFrames vertically (along rows, `axis=0`) or horizontally (along columns, `axis=1`) with `pd.concat()`.

## 1. Vertical Stacking (Appending Rows)

```python
import pandas as pd

# Q1 and Q2 sales
q1_sales = pd.DataFrame({
    'Rep': ['Tom', 'Sue', 'Alice'],
    'Sales': [15000, 22000, 18000]
})

q2_sales = pd.DataFrame({
    'Rep': ['Bob', 'Diana', 'Tom'],
    'Sales': [17000, 25000, 16000]
})

# Concatenate vertically with reset index
combined_sales = pd.concat([q1_sales, q2_sales], ignore_index=True)
print("Combined Vertical Concatenation:")
print(combined_sales)
```

## 2. Hierarchical Keys in Concatenation

```python
multi_idx = pd.concat([q1_sales, q2_sales], keys=['Q1', 'Q2'])
print("Concatenation with MultiIndex Keys:")
print(multi_idx)
print("\nExtracting Q1 data:")
print(multi_idx.loc['Q1'])
```

## 3. Horizontal Concatenation (`axis=1`)

```python
metrics = pd.DataFrame({'Target': [20000, 20000, 20000]}, index=q1_sales.index)
horiz = pd.concat([q1_sales, metrics], axis=1)
print("Horizontal Concatenation:")
print(horiz)
```

## 🎯 Exercises
1. Concatenate three DataFrames representing January, February, and March data.
2. Add keys `['Jan', 'Feb', 'Mar']` to identify source months.

```python
# Solution
m1 = pd.DataFrame({'Sales': [100, 120]})
m2 = pd.DataFrame({'Sales': [130, 110]})
m3 = pd.DataFrame({'Sales': [150, 160]})
q1_all = pd.concat([m1, m2, m3], keys=['Jan', 'Feb', 'Mar'])
print(q1_all)
```
