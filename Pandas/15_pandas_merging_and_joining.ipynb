# 15. Merging and Joining DataFrames

Relational database operations using `pd.merge()` and `.join()`: Inner, Left, Right, and Outer joins.

## 1. Join Types: Inner vs Left vs Outer

```python
import pandas as pd

# Customer table
customers = pd.DataFrame({
    'cust_id': [1, 2, 3, 4],
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'city': ['NY', 'SF', 'LA', 'Austin']
})

# Orders table
orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105],
    'cust_id': [1, 2, 2, 5, 1],
    'amount': [250, 400, 150, 700, 320]
})

# Inner Join
inner_merge = pd.merge(customers, orders, on='cust_id', how='inner')
print("--- Inner Merge (matching cust_id only) ---")
print(inner_merge)

# Left Join (keeps all customers)
left_merge = pd.merge(customers, orders, on='cust_id', how='left')
print("\n--- Left Merge (all customers preserved) ---")
print(left_merge)
```

## 2. Merging on Different Column Names & Suffixes

```python
profiles = pd.DataFrame({'id': [1, 2, 3], 'status': ['Active', 'VIP', 'Regular']})
merged_diff = pd.merge(customers, profiles, left_on='cust_id', right_on='id')
print("Merged on different key names:")
print(merged_diff)
```

## 🎯 Exercises
1. Perform an Outer Join between `customers` and `orders` to find customers without orders and orders without customers.
2. Calculate total order amount per customer name.

```python
# Solution
outer_df = pd.merge(customers, orders, on='cust_id', how='outer')
print("Outer Join:")
print(outer_df)

spend_by_name = pd.merge(customers, orders, on='cust_id', how='inner').groupby('name')['amount'].sum()
print("\nTotal Spend per Customer:")
print(spend_by_name)
```
