# 11. Handling Duplicates in Pandas

Duplicate entries distort model weights and inflate statistical metrics. Learn how to locate, count, and remove duplicate rows.

## 1. Locating Duplicates: `.duplicated()`
`keep='first'` (default), `keep='last'`, or `keep=False` (flags all duplicates).

```python
import pandas as pd

logs = pd.DataFrame({
    'User_ID': [101, 102, 101, 103, 102, 104],
    'Event': ['Login', 'Click', 'Login', 'Logout', 'Click', 'Purchase'],
    'Device': ['Mobile', 'Desktop', 'Mobile', 'Mobile', 'Desktop', 'Desktop']
})
print("Original Log Entries:")
print(logs)

print("\nDuplicate flags (keep='first'):")
print(logs.duplicated())
print(f"Total Duplicate Rows: {logs.duplicated().sum()}")
```

## 2. Dropping Duplicates: `.drop_duplicates()`
Can target all columns or a specific subset of columns.

```python
# Drop complete duplicates
clean_logs = logs.drop_duplicates()
print("Unique Full Records:")
print(clean_logs)

# Keep only one record per User_ID (subset)
unique_users = logs.drop_duplicates(subset=['User_ID'], keep='first')
print("\nUnique Users:")
print(unique_users)
```

## 🎯 Exercises
1. Find all rows that are duplicates based on `User_ID` and `Device`.
2. Keep only the last occurrence of duplicate records.

```python
# Solution
dupes_user_device = logs[logs.duplicated(subset=['User_ID', 'Device'], keep=False)]
print("All Duplicate User-Device Pairs:")
print(dupes_user_device)

last_kept = logs.drop_duplicates(subset=['User_ID', 'Device'], keep='last')
print("\nKeeping Last Occurrence:")
print(last_kept)
```
