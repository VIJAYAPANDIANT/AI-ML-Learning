# 10. Handling Missing Values in Pandas

Real-world datasets are rarely complete. Missing data (`NaN` or `None`) must be identified, evaluated, and either dropped or imputed before modeling.

## 1. Detecting Missing Values: `.isna()`, `.isnull()`, `.notna()`

```python
import pandas as pd
import numpy as np

data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Math': [92, np.nan, 85, np.nan, 78],
    'Physics': [88, 76, np.nan, 90, 84],
    'Chemistry': [95, 80, 82, 88, np.nan]
}
df = pd.DataFrame(data)
print("Original DataFrame with Missing Data:")
print(df)
print("\nMissing values count per column:")
print(df.isna().sum())
```

## 2. Dropping Missing Data: `.dropna()`
Options: `axis=0` (rows), `axis=1` (cols), `how='any'` vs `how='all'`, and `subset=['col']`.

```python
# Drop rows where ANY subject is NaN
df_dropped = df.dropna()
print("Rows with zero NaNs:")
print(df_dropped)

# Drop only if Math is NaN
df_math_clean = df.dropna(subset=['Math'])
print("\nDropped where Math is NaN:")
print(df_math_clean)
```

## 3. Imputing Missing Values: `.fillna()` & Column-Specific Means

```python
# Impute missing physics with mean physics score
mean_physics = df['Physics'].mean()
df_imputed = df.copy()
df_imputed['Physics'] = df_imputed['Physics'].fillna(mean_physics)

# Fill remaining with constant or forward-fill
df_imputed = df_imputed.fillna({'Math': 0, 'Chemistry': df['Chemistry'].median()})
print("Imputed DataFrame:")
print(df_imputed)
```

## 🎯 Exercises
1. Calculate the percentage of missing values in each column.
2. Impute `Math` with the median score and `Physics` with forward-fill (`ffill`).

```python
# Solution
missing_pct = (df.isna().mean() * 100).round(2)
print("Missing Percentages:")
print(missing_pct)

df_filled = df.assign(
    Math=df['Math'].fillna(df['Math'].median()),
    Physics=df['Physics'].ffill()
)
print("\nFilled Result:")
print(df_filled)
```
