# 22. Cross-Tabulation in Pandas (`pd.crosstab`)

Compute frequency tables and contingency matrices to analyze the relationship between two or more categorical features.

## 1. Frequency Cross-Tabulation

```python
import pandas as pd

survey = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Male', 'Female', 'Male', 'Female'],
    'Satisfaction': ['High', 'High', 'Medium', 'Low', 'High', 'High', 'Medium', 'Medium'],
    'Department': ['IT', 'HR', 'IT', 'Marketing', 'IT', 'Marketing', 'HR', 'IT']
})

# Contingency table: Gender vs Satisfaction
ct_freq = pd.crosstab(survey['Gender'], survey['Satisfaction'], margins=True)
print("Frequency Contingency Table:")
print(ct_freq)
```

## 2. Normalized Probabilities / Percentages
Options: `normalize='index'` (row percentages), `normalize='columns'`, or `normalize='all'`.

```python
# Normalized row-wise (within each gender)
ct_norm = pd.crosstab(survey['Gender'], survey['Satisfaction'], normalize='index').round(3) * 100
print("Row-Normalized Percentage Matrix (%):")
print(ct_norm)
```

## 🎯 Exercises
1. Generate a crosstab showing Department vs Satisfaction.
2. Compute overall column-normalized percentages.

```python
# Solution
dept_ct = pd.crosstab(survey['Department'], survey['Satisfaction'], normalize='columns').round(3) * 100
print("Department vs Satisfaction (Column Normalized %):")
print(dept_ct)
```
