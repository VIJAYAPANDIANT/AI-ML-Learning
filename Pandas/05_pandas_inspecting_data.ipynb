# 05. Inspecting Data in Pandas

Prior to feature engineering or model training, inspecting structural integrity, null values, and summary statistics is essential.

## 1. Quick Glance: `.head()`, `.tail()`, `.sample()`

```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 60
df = pd.DataFrame({
    'Age': np.random.randint(21, 65, size=n),
    'Department': np.random.choice(['AI/ML', 'DevOps', 'Data', 'Security'], size=n),
    'Salary': np.random.randint(60000, 160000, size=n),
    'Remote': np.random.choice([True, False], size=n),
    'Rating': np.random.choice([np.nan, 3.5, 4.0, 4.5, 5.0], size=n)
})

print("First 3 rows:")
print(df.head(3))
print("\nRandom sample of 2 rows:")
print(df.sample(2))
```

## 2. Structural Profiling: `.info()` & `.isna().sum()`

```python
print("=== DataFrame Info ===")
df.info()
print("\nMissing values count per column:")
print(df.isna().sum())
```

## 3. Statistical Distribution: `.describe()` & `.value_counts()`

```python
print("Numerical Summary:")
print(df.describe())

print("\nDepartment Distribution (normalized):")
print(df['Department'].value_counts(normalize=True))
```

## 🎯 Exercises
1. Find the unique count of departments (`nunique`).
2. Compute the 25th, 50th (median), and 75th percentiles of Salary.

```python
# Solution
print(f"Unique Departments: {df['Department'].nunique()}")
print("Salary Percentiles (25%, 50%, 75%):")
print(df['Salary'].quantile([0.25, 0.50, 0.75]))
```
