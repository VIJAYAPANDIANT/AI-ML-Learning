# 20. Renaming and Replacing in Pandas

Standardizing column names, re-indexing, and replacing categorical/sentinel values.

## 1. Renaming Columns and Index Labels

```python
import pandas as pd

df = pd.DataFrame({
    'First Name': ['John', 'Jane', 'Alice'],
    'Annual Salary ($)': [95000, 110000, 88000],
    'Dept Name': ['Engineering', 'Marketing', 'HR']
})
print("Original Messy Headers:")
print(df.columns.tolist())

# Rename via dictionary
clean_df = df.rename(columns={
    'First Name': 'first_name',
    'Annual Salary ($)': 'salary',
    'Dept Name': 'department'
})
print("\nRenamed Clean Headers:")
print(clean_df)
```

## 2. Programmatic Column Standardization

```python
# Lowercase and replace spaces with underscores across all columns
df.columns = df.columns.str.lower().str.replace(r'[^a-zA-Z0-9]', '_', regex=True).str.strip('_')
print("Programmatically Standardized Columns:")
print(df.columns.tolist())
```

## 3. Replacing Values: `.replace()`

```python
status_df = pd.DataFrame({
    'order_id': [1, 2, 3, 4],
    'code': ['CMP', 'PND', 'ERR', 'CMP']
})

mapping = {'CMP': 'Completed', 'PND': 'Pending', 'ERR': 'Failed'}
status_df['status_full'] = status_df['code'].replace(mapping)
print("Replaced Codes:")
print(status_df)
```

## 🎯 Exercises
1. Replace sentinel values like `-999` and `'?'` with `pd.NA` or `np.nan`.
2. Clean the column names of a DataFrame by stripping whitespace and replacing spaces with underscores.

```python
# Solution
import numpy as np
messy = pd.DataFrame({'age': [25, -999, 32, -999], 'income': [50000, 75000, '?', 90000]})
cleaned_vals = messy.replace([-999, '?'], np.nan)
print("Cleaned Sentinels:")
print(cleaned_vals)
```
