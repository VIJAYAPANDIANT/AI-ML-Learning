# 25. Pandas Optimization and Performance

Scaling Pandas to large datasets: memory profiling, downcasting numeric types, category dtypes, and chunked processing.

## 1. Profiling Memory: `.memory_usage(deep=True)`

```python
import pandas as pd
import numpy as np

n = 100_000
df = pd.DataFrame({
    'id': np.arange(n, dtype=np.int64),
    'value': np.random.randn(n),
    'status': np.random.choice(['pending', 'completed', 'cancelled', 'refunded'], size=n)
})

initial_mem = df.memory_usage(deep=True).sum() / (1024 ** 2)
print(f"Initial Memory Usage: {initial_mem:.2f} MB")
```

## 2. Optimization: Category Dtype & Downcasting
Converting repetitive string columns to `category` cuts memory footprint by up to 80%!

```python
# Optimize numeric types and category
df['id'] = pd.to_numeric(df['id'], downcast='integer')
df['value'] = pd.to_numeric(df['value'], downcast='float')
df['status'] = df['status'].astype('category')

optimized_mem = df.memory_usage(deep=True).sum() / (1024 ** 2)
print(f"Optimized Memory Usage: {optimized_mem:.2f} MB")
print(f"Memory Reduction: {((initial_mem - optimized_mem) / initial_mem) * 100:.1f}%")
```

## 🎯 Exercises
1. Generate a DataFrame with 10,000 rows containing repeated strings in a `City` column.
2. Convert it to `category` dtype and compare memory usage before and after.

```python
# Solution
cities = pd.Series(np.random.choice(['New York', 'London', 'Tokyo', 'Paris'], size=10000))
mem_before = cities.memory_usage(deep=True)
cities_cat = cities.astype('category')
mem_after = cities_cat.memory_usage(deep=True)
print(f"Memory Before: {mem_before:,} bytes | After: {mem_after:,} bytes")
print(f"Saved: {((mem_before - mem_after) / mem_before) * 100:.1f}%")
```
