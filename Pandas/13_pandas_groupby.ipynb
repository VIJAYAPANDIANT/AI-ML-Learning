# 13. Pandas GroupBy: Split-Apply-Combine

The GroupBy operation is the bedrock of exploratory data analysis, following Hadley Wickham's **Split-Apply-Combine** principle.

## 1. Single Column Grouping

```python
import pandas as pd

sales = pd.DataFrame({
    'Region': ['East', 'West', 'East', 'North', 'West', 'East', 'North'],
    'Rep': ['Tom', 'Sue', 'Tom', 'Bob', 'Sue', 'Alice', 'Bob'],
    'Units': [12, 18, 15, 8, 22, 19, 14],
    'Revenue': [1200, 1800, 1500, 800, 2200, 1900, 1400]
})

# Mean revenue per region
regional_mean = sales.groupby('Region')['Revenue'].mean()
print("Average Revenue per Region:")
print(regional_mean)
```

## 2. Multi-Column Grouping

```python
# Group by Region and Sales Rep
rep_summary = sales.groupby(['Region', 'Rep'])['Revenue'].sum()
print("Revenue by Region & Rep:")
print(rep_summary)
```

## 3. Group Properties & Group Inspection

```python
grouped = sales.groupby('Region')
print("Group Sizes:")
print(grouped.size())

print("\nInspecting 'East' group data:")
print(grouped.get_group('East'))
```

## 🎯 Exercises
1. Calculate the total Units sold per Region.
2. Find the region with the highest total Revenue.

```python
# Solution
units_per_region = sales.groupby('Region')['Units'].sum()
print("Units per Region:")
print(units_per_region)

top_rev_region = sales.groupby('Region')['Revenue'].sum().idxmax()
top_rev_val = sales.groupby('Region')['Revenue'].sum().max()
print(f"\nTop Revenue Region: {top_rev_region} with ${top_rev_val:,.2f}")
```
