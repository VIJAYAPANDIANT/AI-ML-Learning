# 12. Sorting Data in Pandas

Sorting tables by single or multiple columns in ascending/descending order, sorting by indices, and extracting top-N values with `.nlargest()`.

## 1. Sorting by Values: `.sort_values()`

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan', 'Fiona'],
    'Department': ['IT', 'HR', 'IT', 'Marketing', 'IT', 'Marketing'],
    'Salary': [95000, 72000, 110000, 85000, 110000, 92000],
    'Performance': [4.5, 3.8, 4.9, 4.2, 4.0, 4.7]
})

# Sort by Salary descending
sorted_salary = df.sort_values(by='Salary', ascending=False)
print("Sorted by Salary (Desc):")
print(sorted_salary)
```

## 2. Multi-Column Sorting with Mixed Order

```python
# Sort by Department ASC, then Salary DESC
multi_sorted = df.sort_values(by=['Department', 'Salary'], ascending=[True, False])
print("Department (Asc) then Salary (Desc):")
print(multi_sorted)
```

## 3. Fast Extrema: `.nlargest()` & `.nsmallest()`

```python
print("Top 2 Highest Performers:")
print(df.nlargest(2, 'Performance'))

print("\n2 Lowest Salaries:")
print(df.nsmallest(2, 'Salary'))
```

## 🎯 Exercises
1. Sort the employees by `Performance` descending, and reset the index.
2. Extract the 3 highest salaries.

```python
# Solution
perf_sorted = df.sort_values(by='Performance', ascending=False).reset_index(drop=True)
print("Sorted by Performance (Reset Index):")
print(perf_sorted)

print("\nTop 3 Salaries:")
print(df.nlargest(3, 'Salary')[['Name', 'Salary']])
```
