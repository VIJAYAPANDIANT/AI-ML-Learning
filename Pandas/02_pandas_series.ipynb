# 02. Pandas Series

A **Series** is a 1-dimensional labeled array capable of holding any data type (`int`, `float`, `string`, `datetime`, etc.).

## 1. Creating a Series
A Series can be created from a Python list, NumPy array, dictionary, or scalar.

```python
import pandas as pd
import numpy as np

# 1. From list
scores = pd.Series([88, 92, 79, 95, 84], name="Exam_Scores")
print("--- Series from List ---")
print(scores)

# 2. From Dictionary (keys become index labels)
gpa_dict = {'Alice': 3.9, 'Bob': 3.4, 'Charlie': 3.8, 'David': 3.2}
gpa_series = pd.Series(gpa_dict, name="GPA")
print("\n--- Series from Dictionary ---")
print(gpa_series)
```

## 2. Series Attributes & Indexing
Explore `.values`, `.index`, `.dtype`, and access values via position (`.iloc`) or label (`.loc`).

```python
print(f"Underlying Values: {gpa_series.values}")
print(f"Index Labels: {gpa_series.index}")
print(f"Data Type: {gpa_series.dtype}")

# Label-based vs Positional Access
print(f"Alice's GPA (by label): {gpa_series['Alice']}")
print(f"First Student's GPA (by position): {gpa_series.iloc[0]}")
```

## 3. Vectorized Operations & Filtering
Operations on Series are vectorized and blazing fast.

```python
# Vectorized bonus calculation
adjusted_scores = scores + 5
print("Original:", scores.tolist())
print("Adjusted:", adjusted_scores.tolist())

# Boolean mask filter
high_achievers = gpa_series[gpa_series >= 3.5]
print("\nHigh Achievers (GPA >= 3.5):")
print(high_achievers)
```

## 🎯 Exercises
1. Create a Series with monthly sales revenue for 6 months ('Jan' to 'Jun').
2. Compute the mean, maximum, and minimum revenue.
3. Filter for months that exceeded the mean revenue.

```python
# Solution
revenue = pd.Series([12000, 15500, 11200, 18400, 19100, 22000], 
                    index=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'], 
                    name="Revenue_USD")

avg_rev = revenue.mean()
print(f"Mean Revenue: ${avg_rev:,.2f}")
print(f"Peak Month: {revenue.idxmax()} (${revenue.max():,.2f})")
print("\nAbove Average Months:")
print(revenue[revenue > avg_rev])
```
