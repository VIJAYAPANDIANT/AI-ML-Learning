# 01. Introduction to Pandas

Welcome to the **Pandas for AI & Machine Learning** masterclass! Pandas (Python Data Analysis Library) is the foundational tool for data preparation, cleaning, exploration, and feature engineering in Python.

## 1. What is Pandas & Why Do We Use It?
Pandas is an open-source library built directly on top of **NumPy**.
- While NumPy provides fast $N$-dimensional numerical arrays, Pandas introduces **Series** (1D) and **DataFrame** (2D) objects with labeled axes (rows and columns).
- It bridges raw unstructured data into clean structured tabular formats required by Machine Learning algorithms.

```python
import pandas as pd
import numpy as np

print(f"Pandas Version: {pd.__version__}")
print(f"NumPy Version: {np.__version__}")
```

## 2. Your First Pandas DataFrame
A DataFrame represents a 2D table with rows and named columns.

```python
frameworks_data = {
    'Framework': ['Pandas', 'NumPy', 'Scikit-Learn', 'PyTorch', 'TensorFlow'],
    'Domain': ['Data Wrangling', 'Array Math', 'Classical ML', 'Deep Learning', 'Deep Learning'],
    'GitHub_Stars_K': [43, 26, 58, 85, 185]
}

df = pd.DataFrame(frameworks_data)
print("=== AI Frameworks DataFrame ===")
df
```

## 3. Basic DataFrame Inspection
Inspect the dimensions, column labels, index, and data types of your dataset.

```python
print(f"Shape (rows, columns): {df.shape}")
print(f"Columns: {list(df.columns)}")
print(f"Row Index: {df.index}")
print("\nData Types:")
print(df.dtypes)
```
