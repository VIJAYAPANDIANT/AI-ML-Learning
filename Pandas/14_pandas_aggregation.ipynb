# 14. Data Aggregation in Pandas

Master `.agg()`, named aggregations, column-specific mappings, and `.transform()` vs `.filter()`.

## 1. Multi-Metric Aggregation with `.agg()`

```python
import pandas as pd

df = pd.DataFrame({
    'Category': ['Tech', 'Furniture', 'Tech', 'Office', 'Furniture', 'Tech'],
    'Sales': [300, 150, 450, 80, 220, 600],
    'Profit': [60, 25, 110, 12, 40, 180]
})

# Multiple functions on single column
print("Sales Summary Metrics:")
print(df.groupby('Category')['Sales'].agg(['count', 'mean', 'std', 'min', 'max']))
```

## 2. Column-Specific Aggregation Mapping & Named Aggregations

```python
# Named aggregation (clean modern Pandas syntax)
summary = df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Avg_Profit=('Profit', 'mean'),
    Transaction_Count=('Sales', 'count')
)
print("Named Aggregation Summary:")
print(summary)
```

## 3. Group-Level Broadcast: `.transform()`
Returns a Series with the same shape as the original DataFrame!

```python
# Compute category-mean sales and broadcast back
df['Category_Avg_Sales'] = df.groupby('Category')['Sales'].transform('mean')
df['Sales_vs_Category_Avg'] = df['Sales'] - df['Category_Avg_Sales']
print("Transformed DataFrame:")
print(df)
```

## 🎯 Exercises
1. Aggregate `Profit` to calculate both `median` and `sum` per `Category`.
2. Filter categories whose total Sales exceed 400 using `.filter()`.

```python
# Solution
profit_stats = df.groupby('Category')['Profit'].agg(['median', 'sum'])
print("Profit Stats:")
print(profit_stats)

high_vol_categories = df.groupby('Category').filter(lambda g: g['Sales'].sum() > 400)
print("\nRecords from Categories with Total Sales > 400:")
print(high_vol_categories)
```
