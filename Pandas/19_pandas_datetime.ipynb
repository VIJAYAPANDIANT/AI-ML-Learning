# 19. DateTime Operations in Pandas

Temporal feature engineering: parsing dates, extracting time components with the `.dt` accessor, calculating time deltas, and time series resampling.

## 1. Parsing DateTime Strings with `pd.to_datetime()`

```python
import pandas as pd

transactions = pd.DataFrame({
    'txn_id': [101, 102, 103, 104],
    'raw_date': ['2026-01-12 08:30:00', '2026-02-14 14:15:30', '2026-03-22 18:45:00', '2026-04-05 09:10:15'],
    'amount': [450, 120, 890, 310]
})

# Convert to datetime series
transactions['timestamp'] = pd.to_datetime(transactions['raw_date'])
print(transactions.dtypes)
```

## 2. Extracting Date Components via `.dt` Accessor

```python
transactions['year'] = transactions['timestamp'].dt.year
transactions['month'] = transactions['timestamp'].dt.month_name()
transactions['day_of_week'] = transactions['timestamp'].dt.day_name()
transactions['quarter'] = transactions['timestamp'].dt.quarter
transactions['hour'] = transactions['timestamp'].dt.hour
print("Extracted Temporal Features:")
print(transactions[['txn_id', 'timestamp', 'month', 'day_of_week', 'quarter']])
```

## 3. Time Deltas & Elapsed Durations

```python
delivery_df = pd.DataFrame({
    'order_time': pd.to_datetime(['2026-01-01 10:00', '2026-01-02 12:30']),
    'delivery_time': pd.to_datetime(['2026-01-03 16:45', '2026-01-05 09:15'])
})

delivery_df['duration'] = delivery_df['delivery_time'] - delivery_df['order_time']
delivery_df['duration_hours'] = delivery_df['duration'].dt.total_seconds() / 3600
print("Delivery Durations:")
print(delivery_df)
```

## 🎯 Exercises
1. Given a sequence of order dates, find which day of the week generates the highest total transaction amount.
2. Calculate the number of days between each transaction and today's date.

```python
# Solution
day_summary = transactions.groupby('day_of_week')['amount'].sum()
print("Amount per Day of Week:")
print(day_summary)

now = pd.Timestamp.now()
transactions['days_ago'] = (now - transactions['timestamp']).dt.days
print("\nDays Ago Calculation:")
print(transactions[['txn_id', 'timestamp', 'days_ago']])
```
