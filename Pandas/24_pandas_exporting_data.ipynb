# 24. Exporting Data in Pandas

Persisting clean processed DataFrames to disk in various formats: CSV, JSON, Excel, and high-performance Parquet.

## 1. Exporting to CSV and JSON

```python
import pandas as pd
import io

df = pd.DataFrame({
    'user_id': [1, 2, 3],
    'name': ['Alex', 'Brooke', 'Cameron'],
    'active': [True, False, True],
    'balance': [1450.50, 200.00, 3120.75]
})

# Exporting to CSV string
csv_buffer = io.StringIO()
df.to_csv(csv_buffer, index=False)
print("CSV Output Preview:")
print(csv_buffer.getvalue())

# Exporting to JSON records
json_output = df.to_json(orient='records', indent=2)
print("JSON Output Preview:")
print(json_output)
```
