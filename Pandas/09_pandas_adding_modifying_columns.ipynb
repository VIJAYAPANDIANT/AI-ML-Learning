# 09. Adding and Modifying Columns

Feature engineering involves creating derived metrics, updating column values, and organizing DataFrame topology.

## 1. Adding Derived Columns & `.assign()`

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Item': ['Laptop', 'Desktop', 'Tablet', 'Smartphone'],
    'Cost': [600, 500, 200, 350],
    'Retail': [999, 799, 349, 599],
    'Quantity': [15, 8, 25, 40]
})

# Direct column creation
df['Profit_Per_Unit'] = df['Retail'] - df['Cost']
df['Total_Profit'] = df['Profit_Per_Unit'] * df['Quantity']
print("Updated Inventory:")
print(df)
```

## 2. Conditional Column Assignment (`np.where`)

```python
# High margin flag: profit per unit > 250
df['Margin_Tier'] = np.where(df['Profit_Per_Unit'] > 250, 'High', 'Standard')
print("\nTier Assigned:")
print(df[['Item', 'Profit_Per_Unit', 'Margin_Tier']])
```

## 3. Inserting at Specific Positions & Dropping Columns

```python
# Insert SKU code at position 0
df.insert(0, 'SKU', ['SKU_01', 'SKU_02', 'SKU_03', 'SKU_04'])

# Drop Cost and Retail columns (non-destructive)
lean_df = df.drop(columns=['Cost', 'Retail'])
print("\nLean DataFrame (Dropped Cost/Retail):")
print(lean_df)
```

## 🎯 Exercises
1. Add a column `Margin_Percentage = (Profit_Per_Unit / Retail) * 100` rounded to 1 decimal place.
2. Create a column `Stock_Status`: 'Low' if Quantity < 10, else 'Adequate'.

```python
# Solution
df['Margin_Pct'] = ((df['Profit_Per_Unit'] / df['Retail']) * 100).round(1)
df['Stock_Status'] = np.where(df['Quantity'] < 10, 'Low', 'Adequate')
print(df[['Item', 'Quantity', 'Stock_Status', 'Margin_Pct']])
```
