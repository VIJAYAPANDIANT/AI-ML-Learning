# 06. Data Selection in Pandas

Selecting subsets of columns and rows is the most frequent data manipulation task. Learn single vs multi-column extraction and scalar retrieval.

## 1. Selecting Columns: Series vs DataFrame
- `df['col']`: Returns a 1D `pd.Series`.
- `df[['col1', 'col2']]`: Returns a 2D `pd.DataFrame`.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan'],
    'Role': ['Scientist', 'Engineer', 'Analyst', 'Scientist', 'Manager'],
    'City': ['Seattle', 'Austin', 'Boston', 'Seattle', 'Chicago'],
    'Experience': [5, 3, 2, 8, 11]
})

# Series selection
roles = df['Role']
print(f"Type of df['Role']: {type(roles)}")

# DataFrame selection
subset = df[['Name', 'Experience']]
print(f"\nType of df[['Name', 'Experience']]: {type(subset)}")
print(subset)
```

## 2. Positional vs Label-based Selection Preview
- `.iloc[row_pos, col_pos]`
- `.loc[row_label, col_label]`

```python
# Get 1st row, all columns
print("First row via iloc:")
print(df.iloc[0])

# Get specific cell: row 2 ('Charlie'), column 'Role'
print(f"\nCell at row 2, col 1: {df.iloc[2, 1]}")
```

## 🎯 Exercises
1. Select the `City` and `Role` columns for the first 3 employees.
2. Extract the `Experience` value of 'Diana'.

```python
# Solution
print("First 3 employees City and Role:")
print(df.loc[0:2, ['City', 'Role']])

diana_exp = df.loc[df['Name'] == 'Diana', 'Experience'].values[0]
print(f"\nDiana Experience: {diana_exp} years")
```
