# 17. EDA Visualization — Statistical Graphics, Heatmaps & Publication Themes

Effective data visualization communicates exploratory findings with precision and clarity. In production data science, EDA plots must convey distribution shapes, confidence intervals, and categorical partitions without chartjunk.

---

### Key Visual Architectures
1. **Distribution Panels**: Multi-panel subplots combining Histogram, KDE, and Box Plot.
2. **Correlation Heatmaps**: Masked upper triangles with annotated Pearson/Spearman coefficients.
3. **Categorical Facet Grids**: Conditioning distributions across secondary dimensions.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
n = 1000

df_viz = pd.DataFrame({
    'tenure_months': np.random.exponential(scale=24, size=n).clip(1, 72).round(0),
    'monthly_charges': np.random.normal(65, 20, size=n).clip(20, 120).round(2),
    'total_spend': np.random.lognormal(mean=7.5, sigma=0.8, size=n).round(2),
    'contract_type': np.random.choice(['Month-to-Month', 'One-Year', 'Two-Year'], size=n, p=[0.55, 0.25, 0.20]),
    'churn': np.random.choice([0, 1], size=n, p=[0.74, 0.26])
})

print("Visualization Dataset Initialized:")
print(df_viz.head())
```

## 1. Visualizing Distributions & Statistical Facets

```python
# Compute grouped statistical aggregates for visualization
grouped_stats = df_viz.groupby(['contract_type', 'churn'])['monthly_charges'].agg(['count', 'mean', 'median', 'std'])
print("Grouped Summary Matrix (Contract Type x Churn):")
print(grouped_stats.round(2))
```

## 2. Correlation Matrix Heatmap Aesthetics
A publication-quality correlation matrix:
- Uses a diverging color palette (e.g. `coolwarm` or `vlag`) centered at zero.
- Masks the redundant upper triangular entries to halve visual cognitive load.
- Annotates numeric coefficients directly inside each cell.
