# 01. Exploratory Data Analysis (EDA) — Foundations & Methodology

Exploratory Data Analysis (**EDA**) is the critical investigative discipline of exploring, visualizing, and summarizing datasets before applying machine learning models or statistical hypotheses. Pioneered by mathematician **John W. Tukey** in 1977, EDA shifts data science from rigid assumptions to discovery-driven understanding.

---

### 🎯 Key Objectives of EDA
1. **Understand Structure & Scale**: Uncover dimensionalities, column types, record counts, and domain constraints.
2. **Diagnose Data Quality**: Detect missing records, corrupt values, inconsistent entries, and schema anomalies.
3. **Uncover Patterns & Relationships**: Identify linear and non-linear correlations, target dependencies, and feature interactions.
4. **Detect Anomalies & Outliers**: Flag extreme sensor spikes, fraudulent transactions, or measurement errors.
5. **Validate Modeling Hypotheses**: Verify normality, homoscedasticity, and multicollinearity before model training.

````python
# Synthesize an illustrative enterprise customer dataset
rng = np.random.default_rng(seed=42)
n_samples = 1000

df_sample = pd.DataFrame({
    'customer_id': [f"CUST_{i:04d}" for i in range(1, n_samples + 1)],
    'age': rng.integers(18, 75, size=n_samples),
    'annual_income': rng.normal(65000, 18000, size=n_samples).round(2),
    'credit_score': rng.integers(350, 850, size=n_samples),
    'spending_score': rng.integers(1, 100, size=n_samples),
    'education': rng.choice(['High School', 'Bachelor', 'Master', 'Doctorate'], size=n_samples, p=[0.3, 0.45, 0.2, 0.05]),
    'churned': rng.choice([0, 1], size=n_samples, p=[0.78, 0.22])
})

print("Dataset Overview:")
print(f"Total Observations: {df_sample.shape[0]:,}")
print(f"Total Features:     {df_sample.shape[1]}")
display_cols = list(df_sample.columns)
print(f"Columns: {display_cols}")
print("\nFirst 5 Records:")
print(df_sample.head())
````

## 3. Exploratory vs. Confirmatory Data Analysis

| Dimension | Exploratory Data Analysis (EDA) | Confirmatory Data Analysis (CDA) |
| :--- | :--- | :--- |
| **Philosophy** | Inductive: "What do the data tell us?" | Deductive: "Do the data support this hypothesis?" |
| **Prerequisites** | Few or no prior assumptions | Formal null and alternative hypotheses |
| **Primary Tools** | Visual distributions, boxplots, heatmaps | p-values, t-tests, ANOVA, confidence intervals |
| **Risk** | False discoveries from overfitting visual trends | Overlooking unexpected anomalies and non-linear patterns |
| **Stage** | Preliminary phase of model building | Formal post-hypothesis validation phase |
