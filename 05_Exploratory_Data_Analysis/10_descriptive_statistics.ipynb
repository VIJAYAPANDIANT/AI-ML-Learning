# 10. Descriptive Statistics — Central Tendency, Dispersion & Robust Estimators

Descriptive statistics summarize empirical data quantitatively. Selecting the proper estimator depends fundamentally on data distribution and outlier resistance.

---

### Estimator Taxonomy
| Metric | Parametric (Normal Data) | Non-Parametric / Robust (Skewed/Outliers) |
| :--- | :--- | :--- |
| **Center** | Arithmetic Mean ($\bar{x}$) | Median, Trimmed Mean, Huber M-Estimator |
| **Spread** | Standard Deviation ($\sigma$), Variance ($\sigma^2$) | Interquartile Range (IQR), Median Absolute Deviation (MAD) |
| **Relative Scale** | Coefficient of Variation ($CV = \sigma / \mu$) | Quartile Coefficient of Dispersion |

```python
import numpy as np
import pandas as pd
import scipy.stats as stats

# Generate standard data contaminated with heavy positive outliers
np.random.seed(42)
clean_data = np.random.normal(loc=100, scale=15, size=980)
outliers = np.random.uniform(300, 800, size=20)  # 2% severe extreme outliers
sample = np.concatenate([clean_data, outliers])

print(f"Total Sample Size: {len(sample)}")
print(f"Known Clean Elements: 980 (~N(100, 15)), Outliers: 20 (> 300)")
```

## 1. Comparing Parametric vs. Robust Estimators

```python
mean_val = np.mean(sample)
median_val = np.median(sample)
trimmed_mean_10 = stats.trim_mean(sample, proportiontocut=0.05)  # 5% trimmed mean

std_val = np.std(sample, ddof=1)
iqr_val = stats.iqr(sample)
mad_val = stats.median_abs_deviation(sample)

print("=== CENTRAL TENDENCY COMPARISON ===")
print(f"  Arithmetic Mean:       {mean_val:,.2f}  (Distorted by outliers!)")
print(f"  5% Trimmed Mean:       {trimmed_mean_10:,.2f}  (Robust alternative)")
print(f"  Median:                {median_val:,.2f}  (Resistant to extreme values)\n")

print("=== DISPERSION COMPARISON ===")
print(f"  Standard Deviation:    {std_val:,.2f}  (Inflated 4x by 2% outliers)")
print(f"  Interquartile Range:   {iqr_val:,.2f}  (Q3 - Q1)")
print(f"  Median Abs Dev (MAD):  {mad_val:,.2f}  (Consistent estimator for normal scale: MAD * 1.4826 = {mad_val*1.4826:.2f})")
```

## 2. Key Takeaway
- If data is symmetric without outliers, use **Mean and Standard Deviation**.
- If data is skewed or contains outliers, always report **Median and IQR / MAD**.
