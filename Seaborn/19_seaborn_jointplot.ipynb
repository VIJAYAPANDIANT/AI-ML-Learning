# 19. Joint Plots: `sns.jointplot()`

`sns.jointplot()` plots a bivariate relationship between two variables alongside their individual marginal univariate distributions on the top and right margins.

### Learning Objectives:
- Construct joint plots with `sns.jointplot()`
- Toggle bivariate representations: `kind='scatter'`, `kind='reg'`, `kind='hex'`, `kind='kde'`
- Color distributions by class using `hue`

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins')

# Joint plot with linear regression and marginal histograms
g = sns.jointplot(
    data=penguins,
    x='bill_length_mm',
    y='bill_depth_mm',
    hue='species',
    kind='scatter',
    palette='Dark2',
    height=6,
    marginal_kws=dict(common_norm=False, fill=True)
)
g.fig.suptitle("Penguins Bill Length vs Depth with Marginal Distributions", fontweight='bold', y=1.02)
plt.show()
```

## 1. Hexbin Joint Plot (`kind='hex'`)
Hexagonal binning is ideal for large datasets with dense point overlap.

```python
diamonds = sns.load_dataset('diamonds').sample(2500, random_state=42)
g = sns.jointplot(data=diamonds, x='carat', y='price', kind='hex', color='#2980b9')
g.fig.suptitle("Hexbin Joint Distribution (Carat vs Price)", fontweight='bold', y=1.02)
plt.show()
```
