# 04. Relational Plots: `sns.relplot()`

Relational plots visualize bivariate and multivariate relationships between quantitative continuous variables. The figure-level function `sns.relplot()` allows smooth switching between scatter and line plots with automatic multi-panel faceting.

### Learning Objectives:
- Understand `sns.relplot()` and its parameters: `x`, `y`, `hue`, `size`, `style`
- Toggle between `kind='scatter'` and `kind='line'`
- Create faceted multi-panel grids across categories with `col=...` and `row=...`

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='ticks')
tips = sns.load_dataset('tips')

# Faceting scatter plot across smoker status and time of day
g = sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='day',
    col='time',
    row='smoker',
    kind='scatter',
    palette='tab10',
    height=3.2,
    aspect=1.2
)
g.fig.subplots_adjust(top=0.9)
g.fig.suptitle("Faceted Relational Scatter (2x2 Grid)", fontsize=13, fontweight='bold')
plt.show()
```

## 1. Relational Line Plot with Facets
Switching to `kind='line'` aggregates continuous series over time.

```python
fmri = sns.load_dataset('fmri')

g = sns.relplot(
    data=fmri,
    x='timepoint',
    y='signal',
    hue='event',
    style='event',
    col='region',
    kind='line',
    height=3.5,
    aspect=1.3,
    errorbar=('ci', 95)
)
g.fig.subplots_adjust(top=0.85)
g.fig.suptitle("FMRI Neural Signal Progression by Brain Region", fontsize=12, fontweight='bold')
plt.show()
```

## 🎯 Exercises & Practice
1. In the `tips` dataset, use `sns.relplot()` to plot `total_bill` vs `tip`.
2. Color by `sex` and facet columns by `day`.
3. Wrap the column facets to 2 columns using `col_wrap=2`.

```python
# Exercise Solution
g = sns.relplot(
    data=tips, x='total_bill', y='tip', hue='sex', col='day', col_wrap=2, height=3.0, aspect=1.2
)
g.fig.subplots_adjust(top=0.9)
g.fig.suptitle("Bill vs Tip Faceted Across Weekdays (Wrapped)", fontweight='bold')
plt.show()
```
