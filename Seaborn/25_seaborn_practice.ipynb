## Problem 1: A/B Testing Conversion Rates (Barplot with 95% CI)

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

np.random.seed(42)
n = 200
df_ab = pd.DataFrame({
    'Variant': np.random.choice(['Control (A)', 'Treatment (B)'], size=n),
    'Converted': np.random.choice([0, 1], size=n, p=[0.78, 0.22]),
    'Device': np.random.choice(['Mobile', 'Desktop'], size=n)
})

fig, ax = plt.subplots(figsize=(7, 4), dpi=100)
sns.barplot(data=df_ab, x='Variant', y='Converted', hue='Device', palette='Set2', errorbar=('ci', 95), ax=ax)
ax.set_title("Problem 1: A/B Conversion Rate by Device (Mean ± 95% CI)", fontweight='bold')
ax.set_ylabel("Conversion Rate")
plt.show()
```

## Problem 2: Multi-Model Evaluation Loss with Error Bands

```python
epochs = np.tile(np.arange(1, 21), 6)
model_type = np.repeat(['ResNet-50', 'Vision Transformer'], 60)
loss_vals = 1.5 * np.exp(-0.15 * epochs) + np.random.normal(0, 0.08, len(epochs))

df_models = pd.DataFrame({'Epoch': epochs, 'Model': model_type, 'Validation_Loss': loss_vals})

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
sns.lineplot(data=df_models, x='Epoch', y='Validation_Loss', hue='Model', style='Model', markers=True, ax=ax)
ax.set_title("Problem 2: Model Validation Loss Convergence Across Replicates", fontweight='bold')
plt.show()
```

## Problem 3: Multi-Sector Return Volatility KDE Density

```python
ret_tech = np.random.normal(0.08, 0.20, 300)
ret_energy = np.random.normal(0.04, 0.12, 300)
ret_health = np.random.normal(0.06, 0.15, 300)

df_returns = pd.DataFrame({
    'Returns': np.concatenate([ret_tech, ret_energy, ret_health]),
    'Sector': ['Technology']*300 + ['Energy']*300 + ['Healthcare']*300
})

fig, ax = plt.subplots(figsize=(8.5, 4), dpi=100)
sns.kdeplot(data=df_returns, x='Returns', hue='Sector', fill=True, common_norm=False, alpha=0.35, palette='tab10', ax=ax)
ax.set_title("Problem 3: Annual Return Volatility Density across Sectors", fontweight='bold')
plt.show()
```

## Problem 4: Feature Correlation Heatmap with Upper Triangle Mask

```python
iris = sns.load_dataset('iris')
corr = iris.drop(columns='species').corr()
mask = np.triu(np.ones_like(corr, dtype=bool))

fig, ax = plt.subplots(figsize=(6, 5), dpi=100)
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='vlag', center=0, square=True, linewidths=1, ax=ax)
ax.set_title("Problem 4: Feature Correlation (Lower Triangle)", fontweight='bold')
plt.show()
```

## Problem 5: Customer RFM Segmentation Multivariate Scatter

```python
np.random.seed(42)
df_rfm = pd.DataFrame({
    'Recency_Days': np.random.uniform(5, 360, 100),
    'Monetary_Spend': np.random.uniform(100, 4000, 100),
    'Segment': np.random.choice(['Champions', 'Loyal', 'At Risk'], 100),
    'Orders_Count': np.random.randint(1, 20, 100)
})

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
sns.scatterplot(data=df_rfm, x='Recency_Days', y='Monetary_Spend', hue='Segment', size='Orders_Count', sizes=(30, 200), palette='Set1', alpha=0.8, ax=ax)
ax.set_title("Problem 5: Customer RFM Segmentation Scatter", fontweight='bold')
plt.show()
```

## Problem 6: Split Violin Plot (Compensation by Department and Gender)

```python
df_sal = pd.DataFrame({
    'Department': np.random.choice(['Engineering', 'Product', 'Sales'], 200),
    'Gender': np.random.choice(['Male', 'Female'], 200),
    'Salary_K': np.random.normal(110, 25, 200)
})

fig, ax = plt.subplots(figsize=(8, 4.2), dpi=100)
sns.violinplot(data=df_sal, x='Department', y='Salary_K', hue='Gender', split=True, inner='quartile', palette='pastel', ax=ax)
ax.set_title("Problem 6: Departmental Salary Distribution (Split Violin)", fontweight='bold')
plt.show()
```

## Problem 7: ECDF SLA Response Times Benchmark

```python
lat_v1 = np.random.exponential(scale=45, size=250)
lat_v2 = np.random.exponential(scale=28, size=250)
df_lat = pd.DataFrame({'Latency_ms': np.concatenate([lat_v1, lat_v2]), 'Version': ['API v1']*250 + ['API v2 (Cached)']*250})

fig, ax = plt.subplots(figsize=(7.5, 4), dpi=100)
sns.ecdfplot(data=df_lat, x='Latency_ms', hue='Version', palette=['#e74c3c', '#2ecc71'], lw=2.2, ax=ax)
ax.axvline(100, color='gray', linestyle='--', label='100ms SLA Target')
ax.set_title("Problem 7: API Latency SLA Compliance ECDF", fontweight='bold')
ax.legend()
plt.show()
```

## Problem 8: Custom PairGrid with Mixed Plots

```python
tips = sns.load_dataset('tips')[['total_bill', 'tip', 'size']]
g = sns.PairGrid(tips, height=2.2)
g.map_diag(sns.histplot, kde=True, color='#2980b9')
g.map_offdiag(sns.scatterplot, color='#e67e22', alpha=0.6)
g.fig.suptitle("Problem 8: Custom PairGrid Architecture", fontweight='bold', y=1.03)
plt.show()
```

## Problem 9: Logistic Regression Churn Prediction Probability

```python
age = np.random.uniform(20, 70, 200)
prob = 1 / (1 + np.exp(-(0.05 * age - 2.5)))
churn = (np.random.rand(200) < prob).astype(int)

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
sns.regplot(x=age, y=churn, logistic=True, y_jitter=0.03, color='#c0392b', scatter_kws={'alpha': 0.4}, ax=ax)
ax.set_title("Problem 9: Logistic Regression Fit for Customer Churn vs Age", fontweight='bold')
ax.set_xlabel("Customer Age")
ax.set_ylabel("Churn Probability")
plt.show()
```

## Problem 10: FacetGrid Multi-Panel Categorical Distribution

```python
tips = sns.load_dataset('tips')
g = sns.FacetGrid(tips, col='day', hue='time', palette='Dark2', height=3, aspect=1)
g.map(sns.scatterplot, 'total_bill', 'tip')
g.add_legend()
g.fig.subplots_adjust(top=0.85)
g.fig.suptitle("Problem 10: Weekday Tipping Scatter Faceted Across Days", fontweight='bold')
plt.show()
```
