# 23. Color Palettes & Custom Palettes

Color choices communicate information semantics. Seaborn organizes palettes into Qualitative, Sequential, and Diverging systems.

### Three Palette Types:
1. **Qualitative**: For categorical nominal data without inherent order (`'deep'`, `'muted'`, `'tab10'`, `'Set2'`)
2. **Sequential**: For continuous data moving from low to high (`'Blues'`, `'viridis'`, `'rocket'`, `'mako'`)
3. **Diverging**: For data centered at a critical midpoint like zero or mean (`'vlag'`, `'coolwarm'`, `'icefire'`)```python

````python
import seaborn as sns
import matplotlib.pyplot as plt

fig, axes = plt.subplots(3, 1, figsize=(8, 4), dpi=100)

# 1. Qualitative
sns.palplot(sns.color_palette("deep", 8), ax=axes[0])
axes[0].set_title("Qualitative: 'deep'", fontsize=10, fontweight='bold')

# 2. Sequential
sns.palplot(sns.color_palette("rocket", 8), ax=axes[1])
axes[1].set_title("Sequential: 'rocket'", fontsize=10, fontweight='bold')

# 3. Diverging
sns.palplot(sns.color_palette("vlag", 8), ax=axes[2])
axes[2].set_title("Diverging: 'vlag'", fontsize=10, fontweight='bold')

fig.tight_layout()
plt.show()
````

## 🎯 Exercises & Practice
1. Create a custom diverging palette with `sns.diverging_palette(220, 20, as_cmap=True)` and use it on a synthetic correlation matrix.

````python
# Exercise Solution
import numpy as np
cmap = sns.diverging_palette(240, 10, as_cmap=True)
mat = np.random.uniform(-1, 1, (4, 4))
fig, ax = plt.subplots(figsize=(5, 4), dpi=100)
sns.heatmap(mat, cmap=cmap, center=0, annot=True, fmt='.2f', ax=ax)
ax.set_title("Custom Diverging Palette Heatmap", fontweight='bold')
plt.show()
````
