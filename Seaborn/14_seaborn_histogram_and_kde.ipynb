# 14. Histograms & Kernel Density Estimation (KDE)

`sns.histplot()` and `sns.kdeplot()` provide deep statistical profiling of data distributions.

### Learning Objectives:
- Construct histograms with `sns.histplot()`
- Adjust binning, normalization (`stat='density'`, `stat='probability'`)
- Manage multi-category stacks with `multiple='dodge'` and `multiple='stack'`
- Smooth KDE curves with bandwidth adjustments (`bw_adjust`)
- Construct 2D Bivariate KDE density contours

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4.5), dpi=100)

# Dodged histogram comparing total bills by meal time
sns.histplot(
    data=tips,
    x='total_bill',
    hue='time',
    multiple='dodge',
    bins=15,
    palette='Set2',
    ax=ax1
)
ax1.set_title("Dodged Histogram: Lunch vs Dinner")

# 2D Bivariate KDE Density Contour Plot
sns.kdeplot(
    data=tips,
    x='total_bill',
    y='tip',
    cmap='mako',
    fill=True,
    thresh=0.05,
    ax=ax2
)
ax2.set_title("2D Bivariate KDE Density: Bill vs Tip")

fig.tight_layout()
plt.show()
```

## 🎯 Exercises & Practice
1. Plot `sns.kdeplot()` for diamond prices, comparing `cut` categories with `fill=True` and `alpha=0.3`.

```python
# Exercise Solution
diamonds = sns.load_dataset('diamonds').sample(1000, random_state=42)
fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
sns.kdeplot(data=diamonds, x='price', hue='cut', fill=True, alpha=0.3, common_norm=False, palette='viridis', ax=ax)
ax.set_title("Diamond Price Density by Cut Quality", fontweight='bold')
plt.show()
```
