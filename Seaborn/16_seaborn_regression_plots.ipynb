# 16. Regression & Statistical Modeling Plots

Seaborn integrates statistical model fitting directly into visualization. `sns.regplot()` (axes-level) and `sns.lmplot()` (figure-level) fit linear, polynomial, and logistic regression lines with bootstrap confidence bands.

### Learning Objectives:
- Construct linear regression plots with `sns.regplot()` and `sns.lmplot()`
- Fit higher-order polynomial curves with `order=2`
- Model binary probabilities with logistic regression (`logistic=True`)
- Apply robust estimators (`robust=True`) to downweight severe outliers

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

# Figure-level linear regression faceted by meal time
g = sns.lmplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='smoker',
    col='time',
    palette='Set1',
    height=4,
    aspect=1.2,
    scatter_kws={'alpha': 0.6, 's': 40}
)
g.fig.subplots_adjust(top=0.85)
g.fig.suptitle("Linear Regression: Tip vs Total Bill by Smoker & Time", fontweight='bold')
plt.show()
```

## 1. Logistic Regression Modeling with `logistic=True`
Fitting a sigmoid curve to predict binary classification probabilities.

```python
titanic = sns.load_dataset('titanic')

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
sns.regplot(
    data=titanic,
    x='age',
    y='survived',
    logistic=True,
    y_jitter=0.03,
    color='#2980b9',
    scatter_kws={'alpha': 0.3, 's': 25},
    ax=ax
)

ax.set_title("Logistic Regression: Survival Probability as a Function of Age", fontsize=12, fontweight='bold')
ax.set_ylabel("Predicted Survival Probability")
ax.set_ylim(-0.05, 1.05)
plt.show()
```
