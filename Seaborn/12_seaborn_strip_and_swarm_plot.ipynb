# 12. Strip & Swarm Plots: Categorical Scatter

When sample sizes are small to medium, plotting individual data observations provides maximum transparency without statistical abstraction.

### Learning Objectives:
- Construct jittered scatter plots with `sns.stripplot()`
- Build non-overlapping point distributions with `sns.swarmplot()`
- Layer swarm points on top of box or violin plots

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4.5), dpi=100)

# Strip plot with jitter
sns.stripplot(data=tips, x='day', y='tip', jitter=0.25, size=5, alpha=0.7, palette='dark', ax=ax1)
ax1.set_title("Strip Plot (with Jitter)")

# Swarm plot (points adjusted algorithmically so they do not overlap)
sns.swarmplot(data=tips, x='day', y='tip', hue='sex', palette='Set2', size=5, ax=ax2)
ax2.set_title("Swarm Plot (Non-Overlapping Points)")

fig.tight_layout()
plt.show()
```

## 1. Layering Swarm Points over Boxplots
Combining summary statistics with exact raw data points.

```python
fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
sns.boxplot(data=tips, x='day', y='total_bill', color='#ecf0f1', width=0.5, fliersize=0, ax=ax)
sns.stripplot(data=tips, x='day', y='total_bill', color='#2c3e50', alpha=0.6, jitter=0.2, size=5, ax=ax)
ax.set_title("Layered Hybrid: Boxplot with Raw Data Overlay", fontweight='bold')
plt.show()
```
