# 06. Line Plots & Time Series: `sns.lineplot()`

Unlike standard line charts, Seaborn's `sns.lineplot()` automatically aggregates multiple observations per $x$-value and displays a shaded confidence band representing experimental uncertainty.

### Learning Objectives:
- Plot single and multi-series sequential trends
- Understand automatic aggregation and confidence interval calculation (`errorbar='ci'` or `errorbar='sd'`)
- Customize line markers and line dash patterns

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='darkgrid')
fmri = sns.load_dataset('fmri')

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)

# By default, computes mean signal and 95% bootstrap confidence band
sns.lineplot(
    data=fmri,
    x='timepoint',
    y='signal',
    hue='region',
    style='event',
    markers=True,
    dashes=True,
    errorbar=('ci', 95),
    ax=ax
)

ax.set_title("FMRI Hemodynamic Response (Mean with 95% Confidence Band)", fontsize=12, fontweight='bold')
ax.set_xlabel("Timepoint (Seconds)")
ax.set_ylabel("Normalized BOLD Signal")
plt.show()
```
