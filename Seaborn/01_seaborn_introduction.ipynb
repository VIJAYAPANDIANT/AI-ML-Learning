# 01. Introduction to Seaborn

Welcome to the **Seaborn for AI, Data Science & Machine Learning** masterclass! Seaborn is a high-level statistical data visualization library built directly on top of Matplotlib and integrated closely with Pandas DataFrames.

### Why Seaborn for AI/ML?
- **Statistical by Default**: Automatically computes and displays aggregations, regression fits, and confidence intervals (e.g. bootstrap $\pm 95\%$ CI).
- **DataFrame Centric**: Understands tabular data structures natively; passing column names as `x`, `y`, `hue`, `style`, and `size` eliminates tedious indexing.
- **Concise API**: Tasks requiring 15–20 lines in Matplotlib can be accomplished in 1–2 expressive lines in Seaborn.
- **Aesthetic Themes**: Provides built-in color palettes and publication-grade typographic styling.

## 1. Comparing Matplotlib vs. Seaborn
Observe how Seaborn eliminates manual groupby aggregations and simplifies multi-dimensional grouping.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

%matplotlib inline

print(f"Seaborn Version: {sns.__version__}")
print(f"Pandas Version: {pd.__version__}")
```

```python
# Load sample dataset: tips collected in a restaurant
tips = sns.load_dataset('tips')

# Quick inspection of DataFrame structure
print("Dataset Shape:", tips.shape)
tips.head(5)
```

## 2. Your First Statistical Seaborn Plot
Notice how adding `hue='sex'` and `style='time'` encodes multiple dimensions effortlessly.

```python
fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)

sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='time',
    style='smoker',
    size='size',
    palette='deep',
    alpha=0.85,
    ax=ax
)

ax.set_title("Multi-Dimensional Relationship: Bill vs Tip", fontsize=12, fontweight='bold', pad=10)
ax.set_xlabel("Total Bill ($)")
ax.set_ylabel("Tip Amount ($)")
plt.show()
```

## 🎯 Exercises & Practice
1. In the `tips` dataset, plot total bill vs tip using `sns.scatterplot`.
2. Encode customer sex using `hue` and day of the week using `style`.
3. Add a clear title and inspect whether dinner tips differ from lunch tips.

```python
# Exercise Solution
fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='day', s=60, alpha=0.8, ax=ax)
ax.set_title("Exercise: Bill vs Tip Segmented by Gender & Weekday", fontsize=11, fontweight='bold')
plt.show()
```
