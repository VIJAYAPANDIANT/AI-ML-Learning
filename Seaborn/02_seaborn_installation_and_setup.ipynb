# 02. Installation, Environment & Figure-Level vs Axes-Level Functions

To master Seaborn, you must understand its two fundamental architectures: **Axes-level functions** and **Figure-level functions**.

### Learning Objectives:
- Differentiate between **Axes-level** functions (`sns.scatterplot`, `sns.barplot`, `sns.histplot`) and **Figure-level** functions (`sns.relplot`, `sns.catplot`, `sns.displot`)
- Understand when to pass `ax=ax` for Matplotlib integration
- Configure global Seaborn themes with `sns.set_theme()`

## 1. Axes-Level vs. Figure-Level Overview

| Function Type | Examples | Returns | Primary Benefit |
| :--- | :--- | :--- | :--- |
| **Axes-Level** | `sns.scatterplot`, `sns.boxplot`, `sns.lineplot` | `matplotlib.axes.Axes` | Easily embeds into existing Matplotlib subplots (`ax=ax`) |
| **Figure-Level** | `sns.relplot`, `sns.catplot`, `sns.displot` | `FacetGrid` | Automatic multi-panel column/row faceting across categories |

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Configure Seaborn global default theme
sns.set_theme(style='whitegrid', palette='muted')

penguins = sns.load_dataset('penguins')
penguins.head(4)
```

```python
# 1. Axes-level example: embedded in Matplotlib subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4.5), dpi=100)

sns.scatterplot(data=penguins, x='bill_length_mm', y='bill_depth_mm', hue='species', ax=ax1)
ax1.set_title("Axes-Level: Bill Dimensions")

sns.boxplot(data=penguins, x='species', y='body_mass_g', ax=ax2)
ax2.set_title("Axes-Level: Body Mass Distribution")

fig.tight_layout()
plt.show()
```

```python
# 2. Figure-level example: automatic multi-panel faceting
g = sns.relplot(
    data=penguins, 
    x='flipper_length_mm', 
    y='body_mass_g', 
    hue='species', 
    col='sex', # Automatically creates subpanels for each sex
    height=3.8, 
    aspect=1.2
)
g.fig.subplots_adjust(top=0.85)
g.fig.suptitle("Figure-Level (relplot): Faceted by Sex", fontsize=12, fontweight='bold')
plt.show()
```

## 🎯 Exercises & Practice
1. Use `sns.set_theme(style='darkgrid', palette='pastel')` to customize appearance.
2. Plot a boxplot of penguin `flipper_length_mm` by `island` using an axes-level function on a custom figure.

```python
# Exercise Solution
with sns.axes_style('darkgrid'):
    fig, ax = plt.subplots(figsize=(7, 4), dpi=100)
    sns.boxplot(data=penguins, x='island', y='flipper_length_mm', palette='Set2', ax=ax)
    ax.set_title("Flipper Length across Antarctic Islands", fontweight='bold')
    plt.show()
```
