# 24. Interfacing Seaborn with Matplotlib & Figure Customization

Because Seaborn is built directly on Matplotlib, you can seamlessly modify any Seaborn plot using standard Matplotlib `Axes` and `Figure` methods.

### Learning Objectives:
- Customize axes limits, labels, and ticks on Seaborn plots
- Reposition and customize Seaborn legends
- Save high-resolution publication figures with `plt.savefig(dpi=300)`

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid')
tips = sns.load_dataset('tips')

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)

# Create plot via Seaborn
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='time',
    style='time',
    s=70,
    palette='tab10',
    ax=ax
)

# Fine-grained customization via Matplotlib API
ax.set_title("Restaurant Tipping Dynamics: Lunch vs Dinner", fontsize=13, fontweight='bold', pad=12)
ax.set_xlabel("Check Total ($ USD)", fontsize=11)
ax.set_ylabel("Tip Gratuity ($ USD)", fontsize=11)
ax.set_xlim(0, 55)
ax.set_ylim(0, 11)

# Benchmark reference line
ax.axhline(5.0, color='#e74c3c', linestyle='--', linewidth=1.5, label='Standard $5 Gratuity')
ax.legend(loc='upper left', frameon=True, framealpha=0.9)

sns.despine(ax=ax)
plt.show()
```
