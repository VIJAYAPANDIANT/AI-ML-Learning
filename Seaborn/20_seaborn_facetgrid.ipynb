# 20. Multi-Plot Faceting: `sns.FacetGrid`

`sns.FacetGrid` is the foundational class behind figure-level functions like `relplot`, `catplot`, and `displot`. It allows mapping any arbitrary plotting function across slices of a dataset.

### Learning Objectives:
- Initialize a `FacetGrid` with categorical row and column dimensions
- Map plotting functions across subplots using `.map()` and `.map_dataframe()`
- Add custom titles and axis limits with `.set_axis_labels()` and `.set_titles()`

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

# Initialize FacetGrid across day and smoker status
g = sns.FacetGrid(tips, col='day', row='smoker', margin_titles=True, height=2.6, aspect=1.2)

# Map scatterplot onto each facet
g.map(sns.scatterplot, 'total_bill', 'tip', color='#2c3e50', alpha=0.7)

g.set_axis_labels("Total Bill ($)", "Tip ($)")
g.set_titles(col_template="Day: {col_name}", row_template="Smoker: {row_name}")
g.fig.suptitle("Custom FacetGrid Mapping (Day vs Smoker)", fontweight='bold', y=1.03)
plt.show()
```

## 🎯 Exercises & Practice
1. Initialize a `FacetGrid` on the `titanic` dataset with columns partitioned by `class`.
2. Map a histogram of passenger `age` using `plt.hist` with 15 bins.

```python
# Exercise Solution
titanic = sns.load_dataset('titanic')
g = sns.FacetGrid(titanic, col='class', height=3.5, aspect=1.1)
g.map(plt.hist, 'age', bins=15, color='#3498db', edgecolor='white')
g.fig.suptitle("Age Distribution Faceted Across Travel Classes", fontweight='bold', y=1.05)
plt.show()
```
