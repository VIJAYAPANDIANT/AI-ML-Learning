# 21. Pair Grids: `sns.PairGrid`

While `sns.pairplot()` is automated, `sns.PairGrid` gives complete manual control over diagonal, upper-triangle, and lower-triangle visualization types.

### Learning Objectives:
- Initialize a `PairGrid` across numerical variables
- Map different visualizations to different regions: `.map_diag()`, `.map_upper()`, `.map_lower()`
- Combine KDE density contours, scatter plots, and histograms on a single canvas

```python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset('iris')

g = sns.PairGrid(iris, hue='species', palette='Set1', corner=False, height=2.2)

# Diagonal: KDE density curves
g.map_diag(sns.kdeplot, fill=True)

# Upper Triangle: Scatter plots
g.map_upper(sns.scatterplot, alpha=0.7, s=25)

# Lower Triangle: Bivariate density contours
g.map_lower(sns.kdeplot, alpha=0.6)

g.add_legend()
g.fig.subplots_adjust(top=0.92)
g.fig.suptitle("Custom PairGrid: Scatter (Upper) + KDE (Lower) + Density (Diag)", fontweight='bold')
plt.show()
```

## 🎯 Exercises & Practice
1. Create a `PairGrid` on `tips[['total_bill', 'tip', 'size']]`.
2. Map histograms to the diagonal and scatter plots to the off-diagonals with `.map_offdiag()`.

```python
# Exercise Solution
tips = sns.load_dataset('tips')
g = sns.PairGrid(tips[['total_bill', 'tip', 'size']])
g.map_diag(plt.hist, color='#2ecc71', edgecolor='black')
g.map_offdiag(sns.scatterplot, color='#34495e', alpha=0.6)
plt.show()
```
