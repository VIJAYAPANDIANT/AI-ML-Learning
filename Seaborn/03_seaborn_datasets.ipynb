```python
import seaborn as sns
import pandas as pd
import numpy as np

# Get total catalog of available datasets
available = sns.get_dataset_names()
print(f"Total Seaborn Datasets Available: {len(available)}")
print("Featured catalog:", available[:12])
```

## 1. Inspecting Key Datasets
Let's inspect the canonical `titanic` and `diamonds` datasets frequently utilized in ML classification and regression benchmarks.

```python
# Titanic: Binary Survival Classification
titanic = sns.load_dataset('titanic')
print("=== Titanic Dataset ===")
print("Shape:", titanic.shape)
print(titanic[['survived', 'pclass', 'sex', 'age', 'fare', 'class', 'embark_town']].head(3))
```

```python
# Flights: Time Series Matrix
flights = sns.load_dataset('flights')
print("
=== Flights Dataset ===")
print("Shape:", flights.shape)
print(flights.head(3))
```

## 2. Converting Wide-Form Data to Tidy Long-Form Data
Seaborn functions perform best with **tidy / long-form data** where each variable is a column and each observation is a row.

```python
# Create Wide-Form DataFrame
wide_df = pd.DataFrame({
    'Model': ['Model_A', 'Model_B', 'Model_C'],
    'Train_Acc': [0.95, 0.92, 0.88],
    'Val_Acc': [0.89, 0.91, 0.85],
    'Test_Acc': [0.87, 0.90, 0.84]
})

# Convert to Long-Form using pd.melt
long_df = pd.melt(wide_df, id_vars=['Model'], var_name='Metric', value_name='Score')
print("Long-Form Tidy Data for Seaborn:")
long_df.head(6)
```
