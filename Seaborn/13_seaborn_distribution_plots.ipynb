# 13. Distribution Plots Overview: `sns.displot()`

`sns.displot()` is the figure-level function for visualizing univariate and bivariate distributions of continuous data variables.

### Learning Objectives:
- Understand `sns.displot()` and its representations (`kind='hist'`, `kind='kde'`, `kind='ecdf'`)
- Facet distributions across columns with `col=...`
- Compare multi-cohort distributions using `hue` and `multiple='stack'` or `multiple='dodge'`

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins')

# Figure-level displot with KDE overlay and column faceting
g = sns.displot(
    data=penguins,
    x='flipper_length_mm',
    hue='species',
    col='sex',
    kde=True,
    height=3.6,
    aspect=1.2
)
g.fig.subplots_adjust(top=0.85)
g.fig.suptitle("Flipper Length Distribution Faceted by Sex", fontweight='bold')
plt.show()
```

## 🎯 Exercises & Practice
1. Use `sns.displot()` with `kind='kde'` to plot the distribution of `total_bill` in `tips`, colored by `time` and filled with `fill=True`.

```python
# Exercise Solution
tips = sns.load_dataset('tips')
g = sns.displot(data=tips, x='total_bill', hue='time', kind='kde', fill=True, height=3.5, aspect=1.4, palette='tab10')
g.fig.suptitle("Total Bill Continuous KDE Density", fontweight='bold', y=1.02)
plt.show()
```
