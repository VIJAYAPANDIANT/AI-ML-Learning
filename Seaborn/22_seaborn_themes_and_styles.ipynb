# 22. Themes, Styles & Despining

Seaborn provides 5 built-in theme presets, 4 context scalings for presentations/publications, and automatic axis despining.

### Learning Objectives:
- Apply the 5 core themes: `'darkgrid'`, `'whitegrid'`, `'dark'`, `'white'`, `'ticks'`
- Scale typography for presentations with `sns.set_context('notebook'/'talk'/'poster'/'paper')`
- Remove top and right spines with `sns.despine()`
- Apply temporary styles using `with sns.axes_style():`

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

# Apply clean white theme with ticks
sns.set_style('ticks')
sns.set_context('notebook', font_scale=1.1)

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
sns.boxplot(data=tips, x='day', y='total_bill', palette='pastel', ax=ax)

# Remove top and right borders (despine)
sns.despine(top=True, right=True, offset=5, trim=True)

ax.set_title("Publication-Grade Minimalist Theme (Despined)", fontsize=13, fontweight='bold', pad=12)
plt.show()
```

## 🎯 Exercises & Practice
1. Use `sns.set_context('poster')` inside a context manager to display a high-visibility bar plot suited for conference presentation slides.

```python
# Exercise Solution
with sns.plotting_context('talk'):
    fig, ax = plt.subplots(figsize=(7, 4), dpi=100)
    sns.barplot(data=tips, x='sex', y='tip', palette='Set2', ax=ax)
    sns.despine()
    ax.set_title("Talk Context: Bold Visual Hierarchy", fontweight='bold')
    plt.show()
```
