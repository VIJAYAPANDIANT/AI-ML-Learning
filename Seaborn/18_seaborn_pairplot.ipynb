# 18. Pair Plots: `sns.pairplot()`

`sns.pairplot()` generates a full grid of pairwise scatter plots across all numerical features in a dataset, with univariate distributions along the diagonal. It is a cornerstone tool for initial Exploratory Data Analysis (EDA).

### Learning Objectives:
- Construct complete pairwise feature grids with `sns.pairplot()`
- Segment observations across target classes with `hue='species'`
- Toggle diagonal plots with `diag_kind='kde'` vs `diag_kind='hist'`
- Remove redundant upper triangle with `corner=True`

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins').dropna()

# Corner pairplot colored by species
g = sns.pairplot(
    data=penguins,
    hue='species',
    diag_kind='kde',
    corner=True, # Display only lower triangle
    palette='Dark2',
    plot_kws={'alpha': 0.7, 's': 30}
)
g.fig.subplots_adjust(top=0.92)
g.fig.suptitle("Penguins Pairwise Feature Grid (Corner View)", fontsize=13, fontweight='bold')
plt.show()
```

## 🎯 Exercises & Practice
1. In `iris`, generate a pairplot colored by `species` with `diag_kind='hist'`.

```python
# Exercise Solution
iris = sns.load_dataset('iris')
g = sns.pairplot(iris, hue='species', diag_kind='hist', palette='tab10')
g.fig.suptitle("Iris Pairwise Scatter Matrix", fontweight='bold', y=1.02)
plt.show()
```
