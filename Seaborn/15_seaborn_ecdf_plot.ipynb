# 15. Empirical Cumulative Distribution Functions (ECDF)

An **ECDF** plots the proportion of data points less than or equal to each value, avoiding arbitrary binning bias found in histograms.

### Learning Objectives:
- Construct ECDFs with `sns.ecdfplot()`
- Evaluate percentiles and threshold coverage (e.g. "What % of transactions are < $20?")
- Compare cumulative distribution shapes across cohorts

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)

sns.ecdfplot(data=tips, x='total_bill', hue='time', lw=2.5, palette='tab10', ax=ax)

# Benchmark line at 80th percentile
ax.axhline(0.80, color='gray', linestyle='--', lw=1)
ax.text(5, 0.82, "80th Percentile Threshold", color='#555555', fontweight='bold', fontsize=9)

ax.set_title("Empirical Cumulative Distribution: Lunch vs Dinner Bills", fontsize=12, fontweight='bold')
ax.set_xlabel("Total Bill ($)")
ax.set_ylabel("Cumulative Probability (Proportion)")
plt.show()
```

## 🎯 Exercises & Practice
1. In the `titanic` dataset, plot an ECDF of `age` comparing survivors vs non-survivors.

```python
# Exercise Solution
titanic = sns.load_dataset('titanic')
fig, ax = plt.subplots(figsize=(7.5, 4), dpi=100)
sns.ecdfplot(data=titanic, x='age', hue='survived', palette=['#e74c3c', '#2ecc71'], lw=2.2, ax=ax)
ax.set_title("Age Cumulative Distribution: Non-Survivors (0) vs Survivors (1)", fontweight='bold')
plt.show()
```
