# 05. Scatter Plots: `sns.scatterplot()`

`sns.scatterplot()` is the axes-level workhorse for examining relationships, correlations, and cluster separations.

### Learning Objectives:
- Map continuous and categorical dimensions to `hue`, `size`, `sizes`, and `style`
- Adjust marker alpha transparency to handle dense overlapping points
- Embed scatter plots directly into Matplotlib axes hierarchies

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid')
diamonds = sns.load_dataset('diamonds').sample(n=1500, random_state=42)

fig, ax = plt.subplots(figsize=(9, 5), dpi=100)

sns.scatterplot(
    data=diamonds,
    x='carat',
    y='price',
    hue='clarity',
    size='depth',
    sizes=(20, 150),
    palette='viridis',
    alpha=0.7,
    ax=ax
)

ax.set_title("Diamond Valuation: Carat vs Price with Clarity & Depth Encodings", fontsize=12, fontweight='bold')
ax.set_xlabel("Carat Weight")
ax.set_ylabel("Price ($ USD)")
plt.show()
```

## 🎯 Exercises & Practice
1. Load `iris` dataset.
2. Create a scatter plot of `sepal_length` vs `petal_length`.
3. Encode `species` with `hue` and `petal_width` with `size`.

```python
# Exercise Solution
iris = sns.load_dataset('iris')
fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
sns.scatterplot(data=iris, x='sepal_length', y='petal_length', hue='species', size='petal_width', palette='Set1', ax=ax)
ax.set_title("Iris Floral Dimensions Multivariate Scatter", fontweight='bold')
plt.show()
```
