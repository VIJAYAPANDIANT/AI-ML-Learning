# 17. Heatmaps & Correlation Matrices: `sns.heatmap()`

Heatmaps visualize 2D tabular matrices through color encoding. In Machine Learning, they are universally used for feature correlation analysis, confusion matrices, and hierarchical clustering (`sns.clustermap`).

### Learning Objectives:
- Compute and visualize feature correlation matrices
- Annotate numerical values with `annot=True` and `fmt='%.2f'`
- Mask the redundant upper triangle using `np.triu()`
- Center diverging colormaps with `center=0`
- Perform hierarchical dendrogram clustering with `sns.clustermap()`

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Load iris dataset and compute correlation matrix of numerical features
iris = sns.load_dataset('iris')
corr = iris.drop(columns='species').corr()

# Mask upper triangle to avoid redundant duplicated information
mask = np.triu(np.ones_like(corr, dtype=bool))

fig, ax = plt.subplots(figsize=(7, 5.5), dpi=100)
sns.heatmap(
    corr, 
    mask=mask,
    annot=True, 
    fmt='.2f', 
    cmap='coolwarm', 
    vmin=-1, 
    vmax=1, 
    center=0,
    square=True, 
    linewidths=1.5,
    cbar_kws={'label': 'Pearson Correlation', 'shrink': 0.8},
    ax=ax
)

ax.set_title("Iris Floral Dimensions Correlation Matrix (Masked)", fontsize=12, fontweight='bold', pad=12)
plt.show()
```

## 1. Hierarchical Clustermap (`sns.clustermap`)
Groups similar features and observations together using hierarchical agglomerative clustering.

```python
flights = sns.load_dataset('flights')
flights_pivot = flights.pivot(index='month', columns='year', values='passengers')

# Clustermap with dendrogram tree
g = sns.clustermap(
    flights_pivot, 
    cmap='magma', 
    figsize=(8, 7), 
    standard_scale=1, 
    linewidths=0.5
)
g.fig.suptitle("Hierarchical Clustermap: Flight Passenger Seasonality", fontweight='bold', y=1.02)
plt.show()
```

1. In `tips`, compute the correlation matrix of numerical features and plot a heatmap with `annot=True` and colormap `'mako'`.

```python
# Exercise Solution
tips = sns.load_dataset('tips')
num_corr = tips[['total_bill', 'tip', 'size']].corr()
fig, ax = plt.subplots(figsize=(6, 4.5), dpi=100)
sns.heatmap(num_corr, annot=True, cmap='mako', vmin=0, vmax=1, ax=ax)
ax.set_title("Tips Dataset Correlation Heatmap", fontweight='bold')
plt.show()
```
