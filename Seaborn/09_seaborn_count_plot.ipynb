# 09. Count Plots: `sns.countplot()`

`sns.countplot()` displays frequency distributions (the number of observations) across categorical groups. It acts like a histogram for discrete categorical variables.

### Learning Objectives:
- Compute and visualize categorical frequency counts
- Order categories descending using `order=...`
- Compare subgroups with `hue`
- Implement horizontal count plots for long text labels

```python
import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset('titanic')

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)

# Sort categories by frequency
embark_order = titanic['embark_town'].value_counts().index

sns.countplot(
    data=titanic,
    x='embark_town',
    order=embark_order,
    hue='class',
    palette='tab10',
    ax=ax
)

ax.set_title("Passenger Embarkation Town Frequency by Ticket Class", fontsize=12, fontweight='bold')
ax.set_xlabel("Port of Embarkation")
ax.set_ylabel("Passenger Count")
plt.show()
```

## 🎯 Exercises & Practice
1. Plot a horizontal count plot (`y='species'`) of penguin species ordered by count.

```python
# Exercise Solution
penguins = sns.load_dataset('penguins')
fig, ax = plt.subplots(figsize=(7, 3.5), dpi=100)
sns.countplot(data=penguins, y='species', order=penguins['species'].value_counts().index, palette='Blues_r', ax=ax)
ax.set_title("Penguin Species Observation Counts", fontweight='bold')
plt.show()
```
