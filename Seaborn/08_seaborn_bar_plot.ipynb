# 08. Bar Plots & Point Estimates: `sns.barplot()`

Unlike Matplotlib's `plt.bar` which merely draws raw scalar values, Seaborn's `sns.barplot()` operates on full DataFrames, automatically calculating an aggregated point estimate (mean by default) and drawing an error bar representing confidence intervals.

### Learning Objectives:
- Construct bar plots comparing category means
- Switch aggregation estimators with `estimator=np.median`
- Implement horizontal bar charts by setting `orient='h'`
- Group bars across sub-categories with `hue`

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

titanic = sns.load_dataset('titanic')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4.5), dpi=100)

# Mean survival rate by class and gender
sns.barplot(data=titanic, x='class', y='survived', hue='sex', palette='muted', errorbar=('ci', 95), ax=ax1)
ax1.set_title("Survival Rate by Class & Sex (Mean ± 95% CI)", fontweight='bold')
ax1.set_ylabel("Survival Probability")

# Median fare paid by class (using custom estimator)
sns.barplot(data=titanic, x='class', y='fare', estimator=np.median, palette='viridis', errorbar=None, ax=ax2)
ax2.set_title("Median Ticket Fare by Class", fontweight='bold')
ax2.set_ylabel("Median Fare ($)")

fig.tight_layout()
plt.show()
```

## 🎯 Exercises & Practice
1. In the `tips` dataset, plot average `total_bill` across weekdays (`day`).
2. Group by customer gender (`sex`) with custom palette `'crest'`.

```python
# Exercise Solution
fig, ax = plt.subplots(figsize=(7, 4), dpi=100)
tips = sns.load_dataset('tips')
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', palette='crest', ax=ax)
ax.set_title("Average Total Bill by Weekday and Gender", fontweight='bold')
plt.show()
```
