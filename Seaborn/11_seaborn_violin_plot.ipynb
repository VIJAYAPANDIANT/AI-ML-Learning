# 11. Violin Plots & Kernel Density: `sns.violinplot()`

Violin plots combine a standard box plot with a **Kernel Density Estimate (KDE)** on each side, revealing multimodal distribution peaks that box plots conceal.

### Learning Objectives:
- Construct violin plots with `sns.violinplot()`
- Use `split=True` for compact side-by-side comparison of binary subgroups
- Display quartiles inside the violin body with `inner='quartile'`

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=100)

# Split violin plot comparing male vs female on left and right halves
sns.violinplot(
    data=tips,
    x='day',
    y='total_bill',
    hue='sex',
    split=True,
    inner='quartile',
    palette='muted',
    ax=ax
)

ax.set_title("Total Bill Distribution by Day: Male vs Female (Split Violin)", fontsize=12, fontweight='bold')
ax.set_ylabel("Total Bill ($)")
plt.show()
```

## 🎯 Exercises & Practice
1. In the `titanic` dataset, plot a violin plot of passenger `age` across `class`.
2. Split by survival status (`hue='survived', split=True`).

```python
# Exercise Solution
titanic = sns.load_dataset('titanic')
fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)
sns.violinplot(data=titanic, x='class', y='age', hue='survived', split=True, inner='quartile', palette='Set1', ax=ax)
ax.set_title("Titanic Age Distribution by Class and Survival", fontweight='bold')
plt.show()
```
