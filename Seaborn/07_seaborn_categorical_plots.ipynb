# 07. Categorical Plots Overview: `sns.catplot()`

Categorical plots analyze relationships involving at least one discrete grouping variable. `sns.catplot()` is the figure-level gateway to all categorical representations.

### Three Categorical Plot Families:
1. **Categorical Scatterplots**: `kind='strip'` (default), `kind='swarm'`
2. **Categorical Distribution Plots**: `kind='box'`, `kind='violin'`, `kind='boxen'`
3. **Categorical Estimate Plots**: `kind='bar'`, `kind='count'`, `kind='point'`

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

# Faceting box plots across days
g = sns.catplot(
    data=tips,
    x='time',
    y='total_bill',
    hue='sex',
    col='day',
    kind='box',
    col_wrap=2,
    height=3.2,
    aspect=1.2,
    palette='Set2'
)
g.fig.subplots_adjust(top=0.9)
g.fig.suptitle("Faceted Categorical Distribution (catplot)", fontweight='bold')
plt.show()
```
