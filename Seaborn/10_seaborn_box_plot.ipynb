# 10. Statistical Box Plots: `sns.boxplot()`

`sns.boxplot()` visualizes the distribution of quantitative data across categories through quartiles, identifying skewness and flier outliers.

### Learning Objectives:
- Construct statistical box plots comparing numerical distributions across categorical groups
- Add nested subgroups with `hue`
- Modify whisker bounds and outlier appearance

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

fig, ax = plt.subplots(figsize=(8.5, 4.5), dpi=100)
sns.boxplot(
    data=tips,
    x='day',
    y='total_bill',
    hue='smoker',
    palette='Set2',
    width=0.6,
    fliersize=4,
    ax=ax
)

ax.set_title("Customer Spending Distribution by Day & Smoker Status", fontsize=12, fontweight='bold')
ax.set_xlabel("Weekday")
ax.set_ylabel("Total Bill ($)")
plt.show()
```

## 🎯 Exercises & Practice
1. Plot a boxplot of penguin `body_mass_g` across `species`.
2. Further divide each species by `sex` using `hue` and custom palette `'pastel'`.

```python
# Exercise Solution
penguins = sns.load_dataset('penguins')
fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
sns.boxplot(data=penguins, x='species', y='body_mass_g', hue='sex', palette='pastel', ax=ax)
ax.set_title("Penguin Body Mass across Species and Gender", fontweight='bold')
plt.show()
```
