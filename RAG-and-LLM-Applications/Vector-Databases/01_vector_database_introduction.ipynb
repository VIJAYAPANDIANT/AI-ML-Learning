## Vector Databases: 01 - Introduction to Vector Databases & ANN

### Theoretical Overview
Traditional relational databases index scalar values (B-Trees for integers, strings). But searching high-dimensional vectors ($D \in [384, 1536]$) requires calculating distances against millions of entries.
- **Exact Search ($k$-NN)**: Linear scan with $O(N \cdot D)$ complexity. Unviable for millions of vectors.
- **Approximate Nearest Neighbors (ANN)**: Trades sub-1% recall accuracy for $100\times - 1,000\times$ latency speedups using specialized graph and tree data structures:
  1. **HNSW (Hierarchical Navigable Small World)**: Multi-layer proximity graph navigation.
  2. **IVF (Inverted File Index)**: Voronoi cell partitioning via $K$-Means centroids.
  3. **Product Quantization (PQ)**: Compressing vector chunks into compact byte codes.

```python
import numpy as np
import time

def exact_knn_scan(vectors: np.ndarray, query: np.ndarray, k: int = 5):
    """Performs exact linear scan brute-force search."""
    # Compute squared Euclidean distances
    dists = np.sum((vectors - query) ** 2, axis=1)
    top_indices = np.argsort(dists)[:k]
    return top_indices, dists[top_indices]

# Benchmark 50,000 vectors in 128 dimensions
np.random.seed(42)
corpus_vectors = np.random.randn(50000, 128).astype(np.float32)
query_vector = np.random.randn(128).astype(np.float32)

t0 = time.perf_counter()
top_idx, top_dist = exact_knn_scan(corpus_vectors, query_vector, k=5)
elapsed_ms = (time.perf_counter() - t0) * 1000

print(f"Exact k-NN search over 50,000 vectors: {elapsed_ms:.2f} ms")
print(f"Top-5 Nearest Indices: {top_idx}")
print(f"Top-5 Distances: {np.round(top_dist, 2)}")
```
