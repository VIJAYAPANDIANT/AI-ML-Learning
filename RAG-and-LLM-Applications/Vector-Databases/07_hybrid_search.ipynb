## Vector Databases: 07 - Hybrid Search & Reciprocal Rank Fusion (RRF)

### Theoretical Overview
Vector search struggles with exact keyword hits (e.g. part numbers `XYZ-9042`, acronyms). Sparse search (BM25) struggles with synonyms.
**Reciprocal Rank Fusion (RRF)** combines dense and sparse rankings without requiring score normalization:$$RRF(d) = \sum_{m \in M} \frac{1}{k + r_m(d)}$$where $k \approx 60$, and $r_m(d)$ is the 1-based rank of document $d$ in system $m$.

```python
def reciprocal_rank_fusion(dense_ranks: list, sparse_ranks: list, k: int = 60) -> list:
    """Fuses dense and sparse rankings using Reciprocal Rank Fusion."""
    scores = {}
    for rank, doc_id in enumerate(dense_ranks, start=1):
        scores[doc_id] = scores.get(doc_id, 0.0) + (1.0 / (k + rank))
    for rank, doc_id in enumerate(sparse_ranks, start=1):
        scores[doc_id] = scores.get(doc_id, 0.0) + (1.0 / (k + rank))
    
    ranked = sorted(scores.items(), key=lambda x: x[1], reverse=True)
    return ranked

dense_results = ["doc_A", "doc_B", "doc_C", "doc_D"]
sparse_results = ["doc_C", "doc_A", "doc_E", "doc_B"]

fused = reciprocal_rank_fusion(dense_results, sparse_results, k=60)
print("RRF Combined Rankings:")
for rank, (doc_id, score) in enumerate(fused, start=1):
    print(f"Rank {rank}: {doc_id} (RRF Score: {score:.5f})")
```
