## Vector Databases: 05 - Document Chunk Storage & CRUD Lifecycles

### Theoretical Overview
A production vector store must manage full CRUD operations:
- **Create / Upsert**: Batch insertions with deduplication hashes.
- **Read / Query**: K-NN proximity queries with score thresholds.
- **Update**: Versioned chunk overwrites when documents are re-indexed.
- **Delete**: Soft/hard deletes via source document IDs.

```python
import hashlib
import uuid

def generate_chunk_id(doc_name: str, chunk_index: int, text_content: str) -> str:
    """Generates a deterministic hash UUID for chunk deduplication."""
    hash_input = f"{doc_name}:{chunk_index}:{text_content}".encode("utf-8")
    return hashlib.sha256(hash_input).hexdigest()[:16]

chunk1_id = generate_chunk_id("q3_report.pdf", 0, "Revenue increased by 14% to $2.1B.")
chunk2_id = generate_chunk_id("q3_report.pdf", 1, "Operating margin improved by 220 bps.")
print(f"Deterministic Chunk IDs:\n- Chunk 0: {chunk1_id}\n- Chunk 1: {chunk2_id}")
```
