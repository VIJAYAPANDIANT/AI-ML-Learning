## Vector Databases: 06 - Advanced Metadata Filtering

### Theoretical Overview
Filtering approaches in vector search:
1. **Pre-Filtering**: Evaluates metadata filters first, then scans the filtered subset.
2. **Post-Filtering**: Retrieves Top-$K$ semantic matches first, then removes non-matching metadata (can result in fewer than $K$ results).
3. **Single-Stage Filtered ANN**: Modern vector DBs traverse the HNSW graph while skipping nodes that fail the predicate in real time.

```python
class MetadataFilterEngine:
    @staticmethod
    def matches_predicate(metadata: dict, filter_expr: dict) -> bool:
        for field, condition in filter_expr.items():
            val = metadata.get(field)
            if isinstance(condition, dict):
                if "$eq" in condition and val != condition["$eq"]: return False
                if "$in" in condition and val not in condition["$in"]: return False
                if "$gte" in condition and (val is None or val < condition["$gte"]): return False
            else:
                if val != condition: return False
        return True

doc_meta = {"dept": "Engineering", "security_clearance": 3, "tags": ["python", "rag"]}
query_filter = {"dept": {"$eq": "Engineering"}, "security_clearance": {"$gte": 2}}
print("Predicate matches:", MetadataFilterEngine.matches_predicate(doc_meta, query_filter))
```
