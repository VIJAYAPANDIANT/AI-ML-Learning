## Project 07: Cross-Document Multi-Corpus RAG Engine with Hierarchical Routing

### Project Objective
Implement query routing across disparate enterprise corpora (HR Policies, Financial Filings, Engineering Architecture).

```python
def route_query_to_corpus(query: str) -> str:
    q = query.lower()
    if any(k in q for k in ["salary", "vacation", "benefits", "leave"]):
        return "Corpus: HR_Policies_VectorStore"
    elif any(k in q for k in ["revenue", "ebitda", "margin", "q3"]):
        return "Corpus: Financial_Filings_VectorStore"
    return "Corpus: General_Engineering_VectorStore"

print(route_query_to_corpus("How many days of sick leave are allowed?"))
print(route_query_to_corpus("What was the operating profit margin?"))
```
