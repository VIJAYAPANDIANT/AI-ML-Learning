This production capstone integrates all core competencies from Module 9 into a unified, high-throughput Enterprise Knowledge Intelligence Platform:

```
+------------------------------------------------------------------------------------------------+

| 6. Observability: OpenTelemetry Distributed Spans, Latency Profiling & Token Billing           |
```

````python
import time
import json
import hashlib
from typing import List, Dict, Any

class EnterpriseKnowledgePlatform:
    """Production-grade Enterprise Knowledge Intelligence Platform."""
    def __init__(self):
        self.knowledge_base = [
            {"id": "DOC-001", "dept": "engineering", "text": "Kubernetes clusters auto-scale worker nodes using the Cluster Autoscaler based on pending pod metrics.", "source": "k8s_runbook.md"},
            {"id": "DOC-002", "dept": "finance", "text": "Corporate travel expenses exceeding $500 require VP-level sign-off prior to booking.", "source": "expense_policy.pdf"},
            {"id": "DOC-003", "dept": "hr", "text": "Annual open enrollment for health and dental insurance benefits occurs from October 15 through November 15.", "source": "benefits_guide.pdf"},
        ]
        self.execution_audit_log = []
        
    def sanitize_input(self, query: str) -> str:
        # Ingress security guardrail
        if "ignore previous" in query.lower() or "override" in query.lower():
            raise ValueError("SECURITY ALERT: Prompt injection attempt detected and blocked.")
        return query.strip()
        
    def hybrid_retrieve(self, query: str, top_k: int = 2) -> List[Dict[str, Any]]:
        q_terms = set(query.lower().split())
        scored = []
        for doc in self.knowledge_base:
            doc_terms = set(doc["text"].lower().split())
            sparse_score = len(q_terms & doc_terms) / (len(q_terms) + 1)
            scored.append((doc, sparse_score))
        scored.sort(key=lambda x: x[1], reverse=True)
        return [s[0] for s in scored[:top_k]]
        
    def generate_grounded_response(self, query: str, context_docs: List[Dict[str, Any]]) -> Dict[str, Any]:
        t0 = time.perf_counter()
        clean_q = self.sanitize_input(query)
        context_str = " ".join([d["text"] for d in context_docs])
        
        # Simulated generation grounded in context
        response_text = f"Based on enterprise policy: {context_docs[0]['text']}"
        citations = [{"source": d["source"], "doc_id": d["id"]} for d in context_docs]
        
        latency_ms = (time.perf_counter() - t0) * 1000
        audit_record = {
            "query": clean_q,
            "latency_ms": round(latency_ms, 2),
            "citations": citations,
            "faithfulness_score": 0.98,
            "status": "SUCCESS"
        }
        self.execution_audit_log.append(audit_record)
        return {
            "answer": response_text,
            "citations": citations,
            "faithfulness": 0.98,
            "latency_ms": audit_record["latency_ms"]
        }

platform = EnterpriseKnowledgePlatform()
query = "When is annual open enrollment for benefits?"
docs = platform.hybrid_retrieve(query)
result = platform.generate_grounded_response(query, docs)

print("Enterprise Knowledge Platform Response:")
print(json.dumps(result, indent=2))
print("\nExecution Audit Trail:")
print(json.dumps(platform.execution_audit_log, indent=2))
````
