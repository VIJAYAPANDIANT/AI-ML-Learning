## Project 01: Multi-Turn Conversational Assistant with Context Management

### Project Objective
Build an interactive conversational chatbot featuring rolling context buffers, system personality framing, and token usage tracking.

```python
class ProductionChatbot:
    def __init__(self, system_persona: str, max_history_turns: int = 4):
        self.system_persona = system_persona
        self.max_turns = max_history_turns
        self.history = []
        
    def chat(self, user_input: str) -> str:
        self.history.append({"role": "user", "content": user_input})
        # Enforce history limit
        if len(self.history) > self.max_turns * 2:
            self.history = self.history[-(self.max_turns * 2):]
        bot_reply = f"Bot ({self.system_persona}): Processed '{user_input}' with {len(self.history)} messages in context."
        self.history.append({"role": "assistant", "content": bot_reply})
        return bot_reply

bot = ProductionChatbot(system_persona="FinTech Support Specialist")
print(bot.chat("Hello! What are current prime loan rates?"))
print(bot.chat("How do I qualify for a preferred rate?"))
```
