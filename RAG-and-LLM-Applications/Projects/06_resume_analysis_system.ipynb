## Project 06: Automated Talent Intelligence & Resume Matching Pipeline

### Project Objective
Extract structured candidate profiles using Pydantic, embed resume skills, and compute semantic fit scores against target job descriptions.

```python
def compute_job_match(resume_skills: set, job_skills: set) -> float:
    intersection = resume_skills & job_skills
    return len(intersection) / len(job_skills)

cand_skills = {"Python", "PyTorch", "Docker", "RAG", "ChromaDB"}
target_reqs = {"Python", "RAG", "ChromaDB", "Kubernetes"}

score = compute_job_match(cand_skills, target_reqs)
print(f"Talent Match Score: {score:.1%} | Matched: {cand_skills & target_reqs}")
```
