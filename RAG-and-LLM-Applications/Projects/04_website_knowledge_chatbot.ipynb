## Project 04: Web Crawler & Sitemap Knowledge Ingestion RAG Assistant

### Project Objective
Simulate crawling a website sitemap, parsing HTML pages, stripping navigation noise, and answering live customer inquiries.

```python
def mock_crawl_sitemap(sitemap_urls: list):
    return [{"url": url, "content": f"Documentation text from {url}"} for url in sitemap_urls]

sitemap = ["https://docs.acme.ai/api", "https://docs.acme.ai/quickstart"]
print("Crawled Knowledge Pages:", mock_crawl_sitemap(sitemap))
```
