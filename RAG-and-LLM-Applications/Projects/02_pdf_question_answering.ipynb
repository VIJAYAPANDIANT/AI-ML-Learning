## Project 02: Multi-PDF Question Answering with Exact Page Citations

### Project Objective
Ingest multi-page corporate PDFs, extract text with page metadata, build an in-memory vector index, and generate answers with exact source page citations.

```python
def mock_pdf_qa(query: str, pdf_database: list):
    # Find best matching chunk
    match = max(pdf_database, key=lambda doc: len(set(query.lower().split()) & set(doc["text"].lower().split())))
    return {
        "answer": f"According to page {match['page']} of {match['file']}: {match['text']}",
        "citation": {"file": match["file"], "page": match["page"]}
    }

pdf_db = [
    {"file": "10K_Annual.pdf", "page": 42, "text": "Net income grew to $3.2B in 2023 driven by cloud revenue."},
    {"file": "10K_Annual.pdf", "page": 88, "text": "Capital expenditures reached $1.1B for datacenter expansion."}
]
res = mock_pdf_qa("What was the 2023 cloud net income?", pdf_db)
print("QA Output with Page Citation:")
print(f"Answer: {res['answer']}")
print(f"Citation: {res['citation']}")
```
