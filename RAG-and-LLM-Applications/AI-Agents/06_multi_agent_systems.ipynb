### Theoretical Overview
Complex problems exceed the capability of a single agent. **Multi-Agent Systems (MAS)** decompose tasks across specialized agents:
- **Supervisor / Router**: Distributes sub-tasks to worker agents.
- **Worker Agents**: Specialized in Research, Code Generation, Critical Review, or Formatting.
- **Consensus / Synthesis**: Merges worker deliverables into a final output.

```python
class MultiAgentOrchestrator:
    def __init__(self):
        self.researcher = lambda topic: f"Research notes on {topic}"
        self.coder = lambda notes: f"Python script implementing: {notes}"
        self.critic = lambda code: f"Code verified. 0 bugs found in {code[:30]}..."
        
    def run(self, task: str):
        print(f"[Orchestrator] Starting task: {task}")
        notes = self.researcher(task)
        print(f"[Researcher] Produced: {notes}")
        code = self.coder(notes)
        print(f"[Coder] Produced: {code}")
        review = self.critic(code)
        print(f"[Critic] Produced: {review}")
        return {"task": task, "status": "COMPLETED", "code": code}

mas = MultiAgentOrchestrator()
result = mas.run("Vector DB Search Engine")
```
