## AI Agents: 02 - Standardized Tool Schemas & JSON Specifications

### Theoretical Overview
To enable LLMs to invoke tools reliably, tools must be declared with strict JSON Schema definitions specifying parameter types, descriptions, and required fields.

```python
search_tool_schema = {
    "type": "function",
    "function": {
        "name": "search_enterprise_knowledge",
        "description": "Searches internal company knowledge base for technical documentation.",
        "parameters": {
            "type": "object",
            "properties": {
                "query": {"type": "string", "description": "The search query string"},
                "dept": {"type": "string", "enum": ["eng", "hr", "sales"], "description": "Target department"}
            },
            "required": ["query"]
        }
    }
}
print("Tool JSON Schema Definition:")
print(json.dumps(search_tool_schema, indent=2))
```
