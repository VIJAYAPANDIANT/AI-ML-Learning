### Theoretical Overview
An **AI Agent** is an autonomous system that uses an LLM as its central reasoning engine to perceive environments, formulate multi-step plans, execute external tools, and observe intermediate outcomes until achieving a goal.

```
+---------------------------------------------------------+
|                       ReAct Loop                        |
+---------------------------------------------------------+
| 1. Thought: Reason about the current state & goal       |
| 2. Action: Choose tool to call with specific parameters |
| 3. Observation: Ingest tool execution output            |
| 4. Repeat until Final Answer is determined             |
+---------------------------------------------------------+
```

````python
class ReActAgentSimulator:
    def __init__(self, tools: dict):
        self.tools = tools
        self.trajectory = []
        
    def step(self, thought: str, action: str, action_input: str):
        self.trajectory.append(f"Thought: {thought}")
        self.trajectory.append(f"Action: {action}({action_input})")
        # Execute tool
        tool_fn = self.tools.get(action)
        obs = tool_fn(action_input) if tool_fn else f"Error: Tool {action} not found."
        self.trajectory.append(f"Observation: {obs}")
        return obs

tools = {
    "lookup_stock_price": lambda symbol: f"{symbol}: $182.40 USD",
    "currency_converter": lambda val: f"{float(val.split()[0]) * 83.0:.2f} INR"
}

agent = ReActAgentSimulator(tools)
obs1 = agent.step("Need to find AAPL stock price", "lookup_stock_price", "AAPL")
obs2 = agent.step("Need to convert $182.40 USD to INR", "currency_converter", "182.40")

for line in agent.trajectory:
    print(line)
````
