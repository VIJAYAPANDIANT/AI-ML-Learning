## AI Agents: 03 - Native Function Calling & Execution Loops

### Theoretical Overview
Modern LLMs emit native `tool_calls` payloads instead of unstructured text. The host application catches the tool call, executes the Python function, and returns the result in a `tool` role message.

```python
# Simulating native tool calling execution
def execute_tool_call(call_payload: dict, tool_registry: dict):
    fn_name = call_payload["name"]
    fn_args = call_payload["arguments"]
    if fn_name in tool_registry:
        res = tool_registry[fn_name](**fn_args)
        return {"role": "tool", "name": fn_name, "content": json.dumps(res)}
    return {"role": "tool", "name": fn_name, "content": "Tool not found"}

registry = {"add_numbers": lambda a, b: {"sum": a + b}}
simulated_call = {"name": "add_numbers", "arguments": {"a": 42, "b": 58}}
print("Tool Execution Response:", execute_tool_call(simulated_call, registry))
```
