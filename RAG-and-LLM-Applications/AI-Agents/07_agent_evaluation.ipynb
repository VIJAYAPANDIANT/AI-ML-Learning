## AI Agents: 07 - Agent Evaluation & Trajectory Benchmarking

### Theoretical Overview
Evaluating agents requires metrics beyond simple token accuracy:
1. **Goal Completion Rate**: Did the agent solve the user problem?
2. **Trajectory Efficiency**: Number of tool execution steps taken vs optimal path.
3. **Tool Call Accuracy**: Did the agent pass valid JSON schemas and appropriate arguments?

```python
def evaluate_agent_trajectory(steps_taken: int, optimal_steps: int, goal_achieved: bool):
    efficiency = min(1.0, optimal_steps / max(1, steps_taken))
    return {
        "goal_achieved": goal_achieved,
        "steps_taken": steps_taken,
        "optimal_steps": optimal_steps,
        "trajectory_efficiency": round(efficiency, 2)
    }

print("Trajectory Eval 1 (Optimal):", evaluate_agent_trajectory(3, 3, True))
print("Trajectory Eval 2 (Inefficient):", evaluate_agent_trajectory(9, 3, True))
```
