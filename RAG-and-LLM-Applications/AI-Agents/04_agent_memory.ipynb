## AI Agents: 04 - Agent Memory & Scratchpads

### Theoretical Overview
Agents require multiple memory tiers:
- **Short-Term Scratchpad**: Holds current multi-step reasoning thoughts and intermediate observations.
- **Episodic Memory**: Remembers past task executions and outcomes to avoid repeating dead-end paths.
- **Semantic Long-Term Memory**: Vector database of persistent facts.

```python
class AgentScratchpad:
    def __init__(self):
        self.steps = []
    def add_thought(self, thought: str): self.steps.append(f"Thought: {thought}")
    def add_action(self, action: str): self.steps.append(f"Action: {action}")
    def add_observation(self, obs: str): self.steps.append(f"Observation: {obs}")
    def compile(self) -> str: return "\n".join(self.steps)

pad = AgentScratchpad()
pad.add_thought("Database connection timeout. Let me retry with alternate replica.")
pad.add_action("connect_replica(port=5433)")
pad.add_observation("Connected successfully to replica.")
print("Compiled Agent Scratchpad:\n" + pad.compile())
```
