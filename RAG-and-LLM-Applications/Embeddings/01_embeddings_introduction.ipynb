## Embeddings: 01 - Introduction to Vector Embeddings

### Theoretical Overview
Vector embeddings map unstructured discrete data (words, sentences, documents) into a continuous $D$-dimensional metric space:
$$\mathbf{x} \in \mathbb{R}^D$$
In this latent space, geometric proximity corresponds directly to semantic similarity. Key mathematical properties include:
- **Directional Semantics**: Vector offsets reflect conceptual relationships (e.g., $\vec{v}_{\text{king}} - \vec{v}_{\text{man}} + \vec{v}_{\text{woman}} \approx \vec{v}_{\text{queen}}$).
- **Dense vs Sparse**: Unlike sparse bag-of-words or one-hot vectors ($100,000+$ dims, mostly zeros), dense embeddings (typically $384, 768, 1536$ dims) capture latent continuous features.

```python
import numpy as np

# Geometric simulation of word embeddings in 4 dimensions
mock_vocab = {
    "king":   np.array([0.91, 0.12, 0.85, 0.05]),
    "man":    np.array([0.88, 0.10, 0.15, 0.02]),
    "woman":  np.array([0.12, 0.89, 0.18, 0.04]),
    "queen":  np.array([0.15, 0.92, 0.88, 0.07]),
    "apple":  np.array([0.05, 0.02, 0.01, 0.95]),
}

def cosine_similarity(u: np.ndarray, v: np.ndarray) -> float:
    return float(np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v)))

print("Semantic Vector Analogy Simulation:")
predicted_queen = mock_vocab["king"] - mock_vocab["man"] + mock_vocab["woman"]
sim_queen = cosine_similarity(predicted_queen, mock_vocab["queen"])
sim_apple = cosine_similarity(predicted_queen, mock_vocab["apple"])

print(f"Similarity (king - man + woman, queen): {sim_queen:.4f}")
print(f"Similarity (king - man + woman, apple): {sim_apple:.4f}")
```

### Dimensionality and Metric Properties
Embeddings map heterogeneous documents to uniform vectors, enabling instantaneous $O(D)$ pairwise distance checks.
