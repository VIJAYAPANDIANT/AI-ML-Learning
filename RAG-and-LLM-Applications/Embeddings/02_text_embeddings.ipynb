## Embeddings: 02 - Text Embeddings & Normalization

### Theoretical Overview
When generating embeddings from production APIs (e.g. `text-embedding-3-small`, `text-embedding-3-large`), $L_2$ normalization is standard practice:
$$\hat{\mathbf{v}} = \frac{\mathbf{v}}{\|\mathbf{v}\|_2} = \frac{\mathbf{v}}{\sqrt{\sum_{i=1}^D v_i^2}}$$
When vectors are normalized to unit length ($\|\hat{\mathbf{v}}\|_2 = 1$), the cosine similarity reduces to a fast dot product:
$$\cos(\theta) = \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{u}\|_2 \|\mathbf{v}\|_2} = \hat{\mathbf{u}} \cdot \hat{\mathbf{v}}$$

```python
import numpy as np

def l2_normalize(matrix: np.ndarray) -> np.ndarray:
    """Normalizes rows of a matrix to unit L2 norm."""
    norms = np.linalg.norm(matrix, axis=1, keepdims=True)
    return matrix / np.maximum(norms, 1e-12)

# Generate synthetic batch of 5 document embeddings (dim=8)
np.random.seed(42)
raw_embeddings = np.random.randn(5, 8)
norm_embeddings = l2_normalize(raw_embeddings)

print("Norm check for each row (should be exactly 1.0):")
print(np.linalg.norm(norm_embeddings, axis=1))

# Fast dot product matrix multiplication
similarity_matrix = np.dot(norm_embeddings, norm_embeddings.T)
print("\nPairwise Cosine Similarity Matrix (5x5):")
print(np.round(similarity_matrix, 3))
```
