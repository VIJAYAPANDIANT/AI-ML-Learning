## Embeddings: 04 - Embedding Model Benchmarks & Selection (MTEB)

### Theoretical Overview
Selecting an embedding model requires evaluating:
1. **MTEB Score (Massive Text Embedding Benchmark)**: Classification, Clustering, Retrieval, and Reranking metrics.
2. **Context Window**: 512 tokens (standard SBERT) vs 8,192 tokens (BGE-M3, OpenAI text-embedding-3).
3. **Execution Mode**: Local CPU/GPU inference (free, private, sub-millisecond) vs Hosted API (managed, zero infra).

| Model | Dimensions | Max Tokens | MTEB Retrieval | Best Suited For |
| :--- | :--- | :--- | :--- | :--- |
| `all-MiniLM-L6-v2` | 384 | 256 | 41.95 | Fast edge/CPU, real-time search |
| `BAAI/bge-small-en-v1.5` | 384 | 512 | 51.11 | High efficiency, top open-source |
| `BAAI/bge-m3` | 1024 | 8192 | 55.43 | Multilingual, multi-granularity |
| `text-embedding-3-small` | 1536 / 512 | 8192 | 55.10 | General enterprise SaaS apps |
| `text-embedding-3-large` | 3072 / 1024 | 8192 | 57.80 | Complex cross-domain reasoning |

```python
def recommend_embedding_model(tokens_per_doc: int, local_only: bool, ram_budget_gb: float) -> str:
    if local_only:
        if ram_budget_gb < 2.0 or tokens_per_doc <= 256:
            return "Recommended: all-MiniLM-L6-v2 (384-dim, fast CPU)"
        return "Recommended: BAAI/bge-small-en-v1.5 or bge-m3"
    else:
        if tokens_per_doc > 1000:
            return "Recommended: OpenAI text-embedding-3-small (8k context, Matryoshka 512)"
        return "Recommended: OpenAI text-embedding-3-small or Cohere Embed v3"

print(recommend_embedding_model(150, True, 1.0))
print(recommend_embedding_model(3500, False, 16.0))
```
