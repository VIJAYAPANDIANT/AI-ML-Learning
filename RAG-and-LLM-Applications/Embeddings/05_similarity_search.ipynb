## Embeddings: 05 - Similarity Metrics & Vector Distance Search

### Theoretical Overview
Vector search retrieves the top-$K$ nearest neighbors using distance metrics:
- **Cosine Distance**: $1 - \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{u}\|_2 \|\mathbf{v}\|_2}$ (scale-invariant).
- **Euclidean ($L_2$) Distance**: $\sqrt{\sum_{i=1}^D (u_i - v_i)^2}$.
- **Dot Product (Inner Product)**: $\mathbf{u} \cdot \mathbf{v}$ (optimal when vectors are normalized).

```python
import numpy as np

def compute_metrics(u: np.ndarray, v: np.ndarray):
    dot = np.dot(u, v)
    cos_sim = dot / (np.linalg.norm(u) * np.linalg.norm(v))
    l2_dist = np.linalg.norm(u - v)
    return {"dot_product": dot, "cosine_similarity": cos_sim, "l2_distance": l2_dist}

v1 = np.array([1.0, 2.0, 3.0])
v2 = np.array([2.0, 4.0, 6.0]) # Same direction, double magnitude
v3 = np.array([-1.0, -2.0, -3.0]) # Opposite direction

print("v1 vs v2 (collinear):", compute_metrics(v1, v2))
print("v1 vs v3 (opposite):", compute_metrics(v1, v3))
```
