## 05: Dynamic Prompt Templates & Composable Catalogs

### Theoretical Overview
In enterprise software engineering, prompts cannot be hard-coded strings. A robust prompt templating system guarantees:
- **Strict Variable Validation**: Catching missing parameters before network dispatch.
- **Modular Composition**: Reusing common preamble sections across specialized child prompts.
- **Serialization & Versioning**: Storing prompt templates in git or central prompt registries.

```python
import re
from typing import Set

class EnterprisePromptTemplate:
    def __init__(self, template: str):
        self.template = template
        self.input_variables: Set[str] = set(re.findall(r"{([a-zA-Z_][a-zA-Z0-9_]*)}", template))
        
    def format(self, **kwargs) -> str:
        provided_keys = set(kwargs.keys())
        missing_keys = self.input_variables - provided_keys
        if missing_keys:
            raise KeyError(f"Missing required prompt template variables: {missing_keys}")
        return self.template.format(**kwargs)

rag_template = EnterprisePromptTemplate(
    "You are an assistant answering questions based strictly on the provided context.
"
    "Context: {context}
"
    "User Query: {question}
"
    "Format answer as {format_style}."
)

print(f"Extracted variables: {rag_template.input_variables}")
formatted = rag_template.format(
    context="Quarterly earnings rose 14% to $2.4B.",
    question="What was the Q3 revenue growth?",
    format_style="concise bullet points"
)
print("
Formatted Prompt:
" + formatted)
```

### Composable Prompt Stacks

```python
class ComposablePrompt:
    def __init__(self, header: str, footer: str):
        self.header = header
        self.footer = footer
        
    def compose(self, body: str) -> str:
        return f"{self.header}

{body}

{self.footer}"

compliance_wrapper = ComposablePrompt(
    header="[SECURITY CLASSIFICATION: CONFIDENTIAL - HIPAA COMPLIANT]",
    footer="[DISCLAIMER: This output is AI-generated and requires clinician sign-off.]"
)

full_prompt = compliance_wrapper.compose("Patient summary: 45yo female with recurring migraine.")
print(full_prompt)
```
