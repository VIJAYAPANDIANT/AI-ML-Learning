## LangChain: 01 - Architecture & LangChain Expression Language (LCEL)

### Theoretical Overview
**LangChain** provides a unified composable interface for LLM applications. Modern LangChain is built entirely upon **LCEL (LangChain Expression Language)**:
- Pipe Operator (`|`): Chains runnables with seamless streaming, async support, and parallel execution.
- Standard Runnable Protocol: `invoke()`, `batch()`, `stream()`, `ainvoke()`, `abatch()`, `astream()`.

```python
# LCEL Mental Model Simulation
class MockRunnable:
    def __init__(self, func):
        self.func = func
    def __or__(self, other):
        return MockRunnable(lambda x: other.invoke(self.invoke(x)))
    def invoke(self, x):
        return self.func(x)

prompt_step = MockRunnable(lambda query: f"User asked: {query}")
llm_step = MockRunnable(lambda text: f"LLM_RESPONSE({text})")
parser_step = MockRunnable(lambda out: {"result": out.lower()})

chain = prompt_step | llm_step | parser_step
print("LCEL Chain Output:", chain.invoke("Explain LCEL"))
```
