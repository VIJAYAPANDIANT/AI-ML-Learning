## LangChain: 06 - Tool Calling & AgentExecutor

### Theoretical Overview
LangChain agents pair model reasoning with external tools using `create_tool_calling_agent`.
Tools are Python functions decorated with `@tool` carrying type hints and docstrings that serve as tool descriptions.

```python
def mock_tool(name: str, description: str):
    def decorator(func):
        func.name = name
        func.description = description
        return func
    return decorator

@mock_tool("calculate_tax", "Computes corporate tax on net profit.")
def calculate_tax(net_profit: float, rate: float = 0.21) -> float:
    return net_profit * rate

print(f"Tool Schema: Name={calculate_tax.name} | Doc={calculate_tax.description}")
print(f"Executed tool: Tax on $1,000,000 = ${calculate_tax(1000000):,.2f}")
```
