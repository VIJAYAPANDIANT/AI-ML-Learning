## LangChain: 03 - Advanced Runnables & Branching Chains

### Theoretical Overview
- `RunnableParallel`: Dispatches input concurrently to multiple independent branches.
- `RunnablePassthrough`: Forwards the original input dictionary unaltered.
- `.with_fallbacks()`: Automatically switches to backup models on rate limits or API outages.

```python
class MockRunnableParallel:
    def __init__(self, **branches):
        self.branches = branches
    def invoke(self, x):
        return {key: branch.invoke(x) for key, branch in self.branches.items()}

b1 = MockRunnable(lambda text: len(text.split()))
b2 = MockRunnable(lambda text: text.upper())
par = MockRunnableParallel(word_count=b1, uppercase=b2)
print("RunnableParallel Result:", par.invoke("LangChain simplifies complex chains"))
```
