## LangChain: 02 - Chat Models & ChatPromptTemplates

### Theoretical Overview
LangChain chat models operate on polymorphic message types:
- `SystemMessage`: Establishes system-level instructions and safety parameters.
- `HumanMessage`: Contains user inputs.
- `AIMessage`: Contains model responses, tool calls, and structured outputs.

```python
class MockChatPromptTemplate:
    def __init__(self, messages):
        self.messages = messages
    def format_messages(self, **kwargs):
        formatted = []
        for role, template in self.messages:
            formatted.append({"role": role, "content": template.format(**kwargs)})
        return formatted

prompt = MockChatPromptTemplate([
    ("system", "You are an expert tutor in {subject}."),
    ("human", "Explain the concept of {concept} simply.")
])
msgs = prompt.format_messages(subject="Computer Science", concept="Recursion")
for m in msgs:
    print(f"[{m['role'].upper()}]: {m['content']}")
```
