## LangChain: 05 - Modern LCEL RAG Pipelines

### Theoretical Overview
The canonical LCEL RAG idiom:
```python
rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
```

````python
def format_docs(docs):
    return "\n".join(docs)

class MockLCELRAG:
    def __init__(self, retriever, prompt_fn, llm_fn):
        self.retriever = retriever
        self.prompt_fn = prompt_fn
        self.llm_fn = llm_fn
    def invoke(self, question: str):
        ctx_docs = self.retriever(question)
        ctx = format_docs(ctx_docs)
        prompt = self.prompt_fn(context=ctx, question=question)
        return self.llm_fn(prompt)

mock_chain = MockLCELRAG(
    retriever=lambda q: ["[Doc 1]: Q3 net income $4.2B", "[Doc 2]: Gross margin 62%"],
    prompt_fn=lambda **kwargs: f"Context: {kwargs['context']}\nQ: {kwargs['question']}",
    llm_fn=lambda p: f"Answer: Based on context, Q3 net income was $4.2B with 62% gross margin."
)
print(mock_chain.invoke("What was the financial performance in Q3?"))
````
