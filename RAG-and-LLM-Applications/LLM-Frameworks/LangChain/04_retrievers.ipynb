## LangChain: 04 - Retrievers & Ensemble Systems

### Theoretical Overview
LangChain retrievers wrap heterogeneous search engines behind `get_relevant_documents(query)`.
- `EnsembleRetriever`: Combines dense semantic search and sparse BM25 search with weighted scoring.
- `ContextualCompressionRetriever`: Extracts only query-relevant sentences from retrieved documents.

```python
class MockEnsembleRetriever:
    def __init__(self, retrievers, weights=[0.5, 0.5]):
        self.retrievers = retrievers
        self.weights = weights
    def get_relevant_documents(self, query):
        docs = []
        for r in self.retrievers:
            docs.extend(r(query))
        return list(set(docs))

r_dense = lambda q: ["Doc A: Neural embeddings", "Doc B: Vector indexing"]
r_sparse = lambda q: ["Doc B: Vector indexing", "Doc C: Keyword BM25"]
ens = MockEnsembleRetriever([r_dense, r_sparse])
print("Ensemble Deduplicated Documents:", ens.get_relevant_documents("vector"))
```
