## LlamaIndex: 03 - VectorStoreIndex & SummaryIndex

### Theoretical Overview
- `VectorStoreIndex`: Embeds nodes for semantic retrieval.
- `SummaryIndex`: Stores nodes sequentially; reads all nodes for comprehensive summarization.

```python
class MockVectorStoreIndex:
    def __init__(self, nodes):
        self.nodes = nodes
    def as_retriever(self, similarity_top_k=2):
        return lambda query: self.nodes[:similarity_top_k]

index = MockVectorStoreIndex(["Node 1: AWS VPC architecture", "Node 2: Subnet routing tables"])
retriever = index.as_retriever(similarity_top_k=1)
print("Retrieved Node:", retriever("How do subnets route?"))
```
