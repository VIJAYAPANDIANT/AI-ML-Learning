### Theoretical Overview
Query engines transform retrieved nodes into final answers using synthesis modes:
- `refine`: Iteratively answers using the first node, then updates the answer using subsequent nodes.
- `compact`: Packs as many nodes as possible into each prompt to minimize LLM calls.
- `tree_summarize`: Recursively summarizes pairs of nodes in a hierarchical tree.

```python
def simulate_refine_synthesis(nodes: list, query: str) -> str:
    current_answer = f"Preliminary answer from {nodes[0]}"
    for n in nodes[1:]:
        current_answer += f" + refined with additional detail from {n}"
    return current_answer

print("Refined Synthesis Trace:")
print(simulate_refine_synthesis(["Section 1 (Core)", "Section 2 (Exceptions)"], "What is the policy?"))
```
