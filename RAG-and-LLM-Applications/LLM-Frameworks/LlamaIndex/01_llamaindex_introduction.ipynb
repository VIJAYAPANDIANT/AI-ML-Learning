## LlamaIndex: 01 - Architecture & Node Hierarchy

### Theoretical Overview
**LlamaIndex** is a data framework dedicated to connecting external enterprise data to LLMs.
- `Document`: Raw container for text, images, and metadata.
- `Node`: Atomic chunk of a Document with parent/child relationship pointers.
- `Index`: Data structures (Vector, Summary, Tree, Keyword) enabling optimized retrieval.

```python
from dataclasses import dataclass, field
from typing import List, Dict

@dataclass
class LlamaNode:
    id_: str
    text: str
    doc_id: str
    relationships: Dict[str, str] = field(default_factory=dict)

node1 = LlamaNode("node_01", "Text chunk 1", "doc_01", {"next": "node_02"})
node2 = LlamaNode("node_02", "Text chunk 2", "doc_01", {"prev": "node_01"})
print("Node Relationship Graph:", node1, node2)
```
