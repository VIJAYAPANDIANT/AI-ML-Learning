### Theoretical Overview
**LlamaHub** provides 300+ out-of-the-box data readers for SQL, Notion, Slack, Confluence, S3, and Google Drive, standardizing ingestion into `Document` objects.

```python
class MockSimpleDirectoryReader:
    def __init__(self, dir_path: str):
        self.dir_path = dir_path
    def load_data(self):
        # Emulates reading local directory
        return [
            {"doc_id": "file_1.txt", "text": "Company vacation policy guidelines.", "extra_info": {"path": self.dir_path}},
            {"doc_id": "file_2.txt", "text": "Health insurance benefits summary.", "extra_info": {"path": self.dir_path}}
        ]

reader = MockSimpleDirectoryReader("./knowledge_base")
print("Loaded Documents:", reader.load_data())
```
