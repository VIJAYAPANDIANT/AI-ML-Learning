## LlamaIndex: 05 - End-to-End LlamaIndex RAG Pipeline

### Theoretical Overview
Building a full RAG pipeline with NodeParsers, VectorStoreIndex, SimilarityPostprocessor, and source citation extraction.

```python
class LlamaIndexRAGDemo:
    def __init__(self, doc_text: str):
        self.doc_text = doc_text
    def query(self, q: str):
        return {
            "response": "The RAG pipeline successfully verified company guidelines.",
            "source_nodes": [{"node_id": "n01", "score": 0.89, "text": self.doc_text[:60]}]
        }

app = LlamaIndexRAGDemo("All employees must complete compliance training by Q4 annually.")
res = app.query("When is compliance training due?")
print("LlamaIndex Query Response:")
print(f"Answer: {res['response']}")
print(f"Source Node (Score {res['source_nodes'][0]['score']}): \"{res['source_nodes'][0]['text']}...\"")
```
