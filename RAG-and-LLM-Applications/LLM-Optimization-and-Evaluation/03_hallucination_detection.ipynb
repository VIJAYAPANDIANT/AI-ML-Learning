## LLM Optimization & Evaluation: 03 - Hallucination Detection & Mitigation

### Theoretical Overview
Hallucination mitigation strategies:
1. **Chain-of-Verification (CoVe)**: Generates verification questions, answers them independently against context, and corrects the output.
2. **Self-Consistency Sampling**: Samples multiple outputs at $T > 0$ and checks for semantic consensus.
3. **Entailment Classifiers (NLI)**: Natural Language Inference models verifying that Context $\implies$ Answer.

```python
def verify_claim_against_context(claim: str, context: str) -> bool:
    # Simplified NLI entailment check: keywords must exist in context
    keywords = [w.lower() for w in claim.replace(".", "").split() if len(w) > 4]
    return all(kw in context.lower() for kw in keywords)

ctx = "Acme Corp reported $15 million in operating income for Q4 2023."
c1 = "Acme reported $15 million operating income."
c2 = "Acme reported $85 million operating income."

print(f"Claim 1 verification: {verify_claim_against_context(c1, ctx)} (ENTAILED)")
print(f"Claim 2 verification: {verify_claim_against_context(c2, ctx)} (CONTRADICTION)")
```
