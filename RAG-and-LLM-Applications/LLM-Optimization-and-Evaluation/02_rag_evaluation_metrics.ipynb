## LLM Optimization & Evaluation: 02 - RAGAS Framework & Metrics

### Theoretical Overview
**RAGAS (Retrieval Augmented Generation Assessment)** defines mathematically rigorous metrics:
- **Faithfulness**: $\frac{|\text{Claims in Answer supported by Context}|}{|\text{Total Claims in Answer}|}$
- **Answer Relevance**: Semantic similarity between query and generated answer.
- **Context Precision**: Signal-to-noise ratio of relevant chunks in retrieved Top-$K$.
- **Context Recall**: Extent to which retrieved context covers all ground-truth facts.

```python
def compute_faithfulness_metric(claims_total: int, claims_supported: int) -> float:
    if claims_total == 0: return 1.0
    return round(claims_supported / claims_total, 4)

def compute_context_precision(ranked_relevance: list) -> float:
    precisions = []
    true_positives = 0
    for k, is_rel in enumerate(ranked_relevance, start=1):
        if is_rel:
            true_positives += 1
            precisions.append(true_positives / k)
    return round(sum(precisions) / max(1, len(precisions)), 4) if precisions else 0.0

print("Faithfulness Score (4/5 claims supported):", compute_faithfulness_metric(5, 4))
print("Context Precision (hits at ranks 1 & 3):", compute_context_precision([True, False, True, False]))
```
