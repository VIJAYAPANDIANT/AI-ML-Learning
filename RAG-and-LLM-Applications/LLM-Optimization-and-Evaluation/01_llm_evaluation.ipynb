## LLM Optimization & Evaluation: 01 - LLM Evaluation & LLM-as-a-Judge

### Theoretical Overview
Evaluating open-ended generative responses requires scalable automated judging:
- **LLM-as-a-Judge**: Using a high-capacity model (GPT-4o) with calibrated scoring rubrics to evaluate candidate outputs.
- **Pairwise Win-Rate**: Comparing two models head-to-head with position-bias swapping to eliminate ordering artifacts.
- **Standard Benchmarks**: MMLU (knowledge), GSM8K (multi-step math), HumanEval (code synthesis).

```python
import json

class LLMAsJudgeSimulator:
    @staticmethod
    def evaluate_response(query: str, ground_truth: str, candidate_answer: str) -> dict:
        # Calibrated rubric evaluation simulation
        has_key_facts = all(word in candidate_answer.lower() for word in ["rag", "retrieval"])
        score = 5 if has_key_facts else 3
        return {
            "score": score,
            "rubric_criterion": "Accuracy and factual consistency with reference",
            "explanation": "Candidate accurately cited core retrieval mechanisms without hallucination."
        }

q = "What does RAG stand for and do?"
ref = "RAG stands for Retrieval-Augmented Generation, combining search with generative models."
cand = "RAG is Retrieval-Augmented Generation. It enhances LLMs with external document retrieval."

result = LLMAsJudgeSimulator.evaluate_response(q, ref, cand)
print("Judge Evaluation Output:")
print(json.dumps(result, indent=2))
```
