## LLM Optimization & Evaluation: 07 - LLM Observability & OpenTelemetry Tracing

### Theoretical Overview
Production LLM tracing tracks execution spans across chains, retrievers, and tool invocations:
- Token consumption accounting (input vs output tokens, dollar cost).
- Per-span latency profiling (retrieval time vs model inference time).
- Complete execution audit logs.

```python
import time

class TraceSpan:
    def __init__(self, name: str):
        self.name = name
        self.start_time = None
        self.duration_ms = None
    def __enter__(self):
        self.start_time = time.perf_counter()
        return self
    def __exit__(self, exc_type, exc_val, exc_tb):
        self.duration_ms = (time.perf_counter() - self.start_time) * 1000

print("Simulating Execution Tracing:")
with TraceSpan("Vector Retrieval") as span_retrieval:
    time.sleep(0.04)
print(f"Span '{span_retrieval.name}' completed in {span_retrieval.duration_ms:.2f} ms")

with TraceSpan("LLM Generation") as span_gen:
    time.sleep(0.08)
print(f"Span '{span_gen.name}' completed in {span_gen.duration_ms:.2f} ms")
```
