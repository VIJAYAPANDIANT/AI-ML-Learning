## LLM Optimization & Evaluation: 04 - Latency, Throughput & Prompt Compression

### Theoretical Overview
Optimizing operational expenses and inference latency:
- **Prompt Compression**: Removing low-perplexity stopwords and syntactic noise (e.g. LLMLingua) to cut tokens by $30\% - 50\%$.
- **KV Caching**: Storing attention Key/Value tensors across multi-turn prompts.
- **Speculative Decoding**: Using a fast draft model (1B) verified by a large target model (70B).

```python
def simple_prompt_compressor(raw_prompt: str) -> str:
    # Removes redundant whitespace and boilerplate phrases
    boilerplate = ["please note that", "it is important to remember that", "as an ai model,"]
    compressed = raw_prompt
    for b in boilerplate:
        compressed = compressed.replace(b, "")
    return " ".join(compressed.split())

verbose_text = "Please note that it is important to remember that RAG minimizes model hallucinations."
compressed = simple_prompt_compressor(verbose_text)
print(f"Original: {len(verbose_text)} chars -> Compressed: {len(compressed)} chars")
print("Output:", compressed)
```
