## RAG: 04 - Document Parsing & Layout Analysis

### Theoretical Overview
Messy PDFs contain multi-column text, embedded tables, and headers/footers. Parsing techniques:
- **Layout-Aware Parsing**: Separates headers, paragraphs, and tables.
- **Table Linearization**: Converting tabular rows into markdown or HTML strings for LLM comprehension.

```python
def linearize_table(headers: list, rows: list) -> str:
    header_line = "| " + " | ".join(headers) + " |"
    sep_line = "| " + " | ".join(["---"] * len(headers)) + " |"
    row_lines = ["| " + " | ".join(str(cell) for cell in row) + " |" for row in rows]
    return "\n".join([header_line, sep_line] + row_lines)

table_md = linearize_table(["Metric", "Q2", "Q3"], [["Revenue ($M)", 140, 165], ["EBITDA ($M)", 32, 41]])
print("Linearized Markdown Table for LLM Context:")
print(table_md)
```
