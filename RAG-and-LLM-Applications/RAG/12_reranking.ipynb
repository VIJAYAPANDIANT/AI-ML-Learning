## RAG: 12 - Two-Stage Retrieval & Neural Re-Ranking

### Theoretical Overview
Bi-Encoders retrieve candidates fast, but suffer from lost semantic nuance. Cross-Encoders pass query and document jointly through full attention layers, providing precise relevance scoring.
- **Stage 1**: Broad retrieval of Top-50 candidates via vector search ($O(1)$ fast).
- **Stage 2**: Re-ranking Top-50 down to Top-5 via Cross-Encoder neural scoring.

```python
def mock_cross_encoder_rerank(query: str, candidates: list, top_k: int = 2):
    # Cross-encoders score (query, doc) pairs directly
    scored = []
    for c in candidates:
        score = sum(1 for word in query.lower().split() if word in c.lower()) / (len(query.split()) + 1)
        scored.append((c, score))
    scored.sort(key=lambda x: x[1], reverse=True)
    return scored[:top_k]

query = "vector database search"
docs = [
    "Relational SQL databases use B-Trees.",
    "Chroma is a vector database for semantic search.",
    "Deep learning requires matrix multiplications."
]
print("Reranked Candidates:", mock_cross_encoder_rerank(query, docs))
```
