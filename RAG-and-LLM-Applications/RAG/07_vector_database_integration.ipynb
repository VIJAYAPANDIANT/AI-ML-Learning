## RAG: 07 - Vector Database Ingestion & Metadata Tagging

### Theoretical Overview
Attaching comprehensive metadata (`file_name`, `page_number`, `created_at`, `section_heading`) enables precise pre-filtering during production retrieval.

```python
def create_payload(chunk_text: str, source_doc: str, page_num: int):
    return {
        "id": f"{source_doc}_p{page_num}_{hash(chunk_text) % 10000}",
        "text": chunk_text,
        "metadata": {"source": source_doc, "page": page_num}
    }

print("Sample Ingestion Payload:", create_payload("Net operating revenue increased 12%", "annual_report.pdf", 4))
```
