## RAG: 03 - Multi-Format Document Ingestion

### Theoretical Overview
Enterprise knowledge lives across PDFs, DOCX, Markdown, HTML, and JSON. Robust loaders extract clean text while preserving document structural metadata (source file, page numbers, author, creation date).

```python
from dataclasses import dataclass

@dataclass
class LoadedDocument:
    page_content: str
    metadata: dict

class SimpleMarkdownLoader:
    @staticmethod
    def load(file_path: str, raw_text: str):
        return LoadedDocument(
            page_content=raw_text.strip(),
            metadata={"source": file_path, "format": "markdown", "length": len(raw_text)}
        )

doc = SimpleMarkdownLoader.load("docs/arch.md", "# System Architecture\nThis system uses microservices.")
print("Loaded Document:", doc)
```
