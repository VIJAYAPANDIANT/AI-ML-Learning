## RAG: 14 - Conversational RAG with Chat History

### Theoretical Overview
In multi-turn chat, queries often contain pronouns: *"What was its revenue?"* (referring to Apple).
**Query Contextualization**: An LLM reformulates the user prompt and conversation history into a standalone retrieval query before database search.

```python
def contextualize_query(history: list, latest_query: str) -> str:
    # Simulates standalone query generation from dialogue
    if "its" in latest_query.lower() and history:
        return latest_query.replace("its", "Apple's") # Resolved standalone query
    return latest_query

chat_history = [("What company did Steve Jobs found?", "Steve Jobs founded Apple.")]
follow_up = "What was its 2023 revenue?"
standalone = contextualize_query(chat_history, follow_up)
print(f"Original follow-up: {follow_up}")
print(f"Reformulated query: {standalone}")
```
