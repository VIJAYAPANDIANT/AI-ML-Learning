## RAG: 06 - Batch Embedding Pipelines & Token Caching

### Theoretical Overview
Embedding millions of chunks requires batching to avoid API rate limits and token count pre-checks to avoid context truncation.

```python
def batch_chunks(chunks: list, batch_size: int = 3):
    for i in range(0, len(chunks), batch_size):
        yield chunks[i:i + batch_size]

mock_chunks = [f"Chunk content block #{i}" for i in range(8)]
for b_idx, batch in enumerate(batch_chunks(mock_chunks, batch_size=3)):
    print(f"Dispatched Batch {b_idx} with {len(batch)} items: {batch}")
```
