## RAG: 09 - Context Augmentation & Prompt Framing

### Theoretical Overview
Augmenting retrieved chunks into prompts must respect context limits and enforce strict anti-hallucination guardrails.

```python
def build_rag_prompt(query: str, retrieved_chunks: list) -> str:
    formatted_context = "\n\n".join([f"Source [{r['source']}]: {r['text']}" for r in retrieved_chunks])
    return (
        "You are an assistant answering questions based solely on the provided context.\n"
        "If the answer cannot be found in the context, respond: \"I do not have sufficient information to answer.\"\n\n"
        f"<context>\n{formatted_context}\n</context>\n\n"
        f"<question>\n{query}\n</question>\n\n"
        "Answer:"
    )

chunks = [{"source": "HR_Policy.pdf", "text": "Employees receive 20 days paid vacation annually."}]
print(build_rag_prompt("How many vacation days do employees get?", chunks))
```
