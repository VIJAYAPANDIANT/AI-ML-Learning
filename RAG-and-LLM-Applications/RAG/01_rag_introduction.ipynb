## RAG: 01 - Introduction to Retrieval-Augmented Generation

### Theoretical Overview
Retrieval-Augmented Generation (RAG) grounds language models in external knowledge.
Core benefits:
1. Eliminates hallucinations by anchoring responses in retrieved evidence.
2. Enables exact source citations and audit trails.
3. Updates knowledge dynamically without expensive model re-training.

```python
# Naive RAG Mental Model Demonstration
def simulate_naive_rag(query: str, mock_db: dict):
    # Step 1: Retrieval
    retrieved_chunk = mock_db.get(query.lower(), "No matching document found.")
    # Step 2: Augmentation
    prompt = f"Context: {retrieved_chunk}\nQuestion: {query}\nAnswer strictly based on context:"
    # Step 3: Generation (simulated)
    answer = f"According to context: {retrieved_chunk}"
    return {"query": query, "prompt": prompt, "answer": answer}

knowledge = {"what is rag?": "RAG is a technique that combines retrieval of documents with generative models."}
out = simulate_naive_rag("What is RAG?", knowledge)
print("RAG Execution Flow:")
print(out["prompt"])
print("\nFinal Answer:\n" + out["answer"])
```
