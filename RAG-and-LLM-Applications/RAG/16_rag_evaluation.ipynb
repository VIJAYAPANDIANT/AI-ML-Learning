## RAG: 16 - RAG Evaluation & The RAG Triad

### Theoretical Overview
Evaluating RAG systems requires the **RAG Triad**:
1. **Context Relevance**: Is the retrieved context relevant to the query?
2. **Groundedness / Faithfulness**: Is the generated answer strictly derived from the context (no hallucination)?
3. **Answer Relevance**: Does the generated answer address the user query?

```python
def compute_rag_triad_scores(context_rel: float, faithfulness: float, answer_rel: float):
    composite = (context_rel + faithfulness + answer_rel) / 3.0
    return {
        "Context Relevance": f"{context_rel:.2f}",
        "Faithfulness": f"{faithfulness:.2f}",
        "Answer Relevance": f"{answer_rel:.2f}",
        "Composite RAG Quality": f"{composite:.2f}"
    }

print("Sample RAGAS Metric Report:")
print(json.dumps(compute_rag_triad_scores(0.95, 0.98, 0.92), indent=2))
```
