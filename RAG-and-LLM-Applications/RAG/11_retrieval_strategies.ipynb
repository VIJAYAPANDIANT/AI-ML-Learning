## RAG: 11 - Advanced Retrieval Strategies

### Theoretical Overview
- **Multi-Query Retrieval**: Generates multiple paraphrased variants of the user query and retrieves the union of results.
- **Parent Document Retriever**: Embeds small chunks for fine-grained retrieval, but passes the parent document to the LLM for rich context.

```python
def generate_query_variations(query: str) -> list:
    # Simulates LLM generating multi-query variations
    return [
        query,
        f"What are the details regarding {query.lower()}?",
        f"Overview and explanation of {query.lower()}"
    ]

print("Query variations for 'Kubernetes pod lifecycle':")
for q in generate_query_variations("Kubernetes pod lifecycle"):
    print(f"- {q}")
```
