## RAG: 08 - Similarity Search & Score Thresholding

### Theoretical Overview
Blindly retrieving Top-$K$ items can inject irrelevant context if the query has no true matches. **Score thresholding** drops documents with cosine similarity below a cut-off (e.g. $< 0.70$).

```python
def filter_by_threshold(results: list, min_score: float = 0.75):
    return [r for r in results if r["score"] >= min_score]

raw_hits = [
    {"doc": "RAG architecture guide", "score": 0.89},
    {"doc": "Vector database benchmark", "score": 0.81},
    {"doc": "Pasta recipe for dinner", "score": 0.34}
]
print("Thresholded Hits (score >= 0.75):", filter_by_threshold(raw_hits, 0.75))
```
