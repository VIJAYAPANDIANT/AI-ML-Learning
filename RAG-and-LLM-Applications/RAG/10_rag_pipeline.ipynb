## RAG: 10 - End-to-End Modular RAG Pipeline Class

### Theoretical Overview
Encapsulating ingestion, retrieval, augmentation, and generation into a clean, testable object-oriented class.

```python
class ModularRAGPipeline:
    def __init__(self, retriever, llm_engine):
        self.retriever = retriever
        self.llm = llm_engine
        
    def answer(self, question: str, top_k: int = 3):
        docs = self.retriever(question, k=top_k)
        prompt = f"Context: {docs}\nQuestion: {question}"
        response = self.llm(prompt)
        return {"question": question, "answer": response, "sources": docs}

mock_retriever = lambda q, k: ["Context doc 1", "Context doc 2"]
mock_llm = lambda p: "Synthesized verified answer based on context."
pipeline = ModularRAGPipeline(mock_retriever, mock_llm)
print("Pipeline execution output:", pipeline.answer("What is the quarterly projection?"))
```
