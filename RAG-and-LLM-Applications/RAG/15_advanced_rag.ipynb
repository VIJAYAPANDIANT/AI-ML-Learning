## RAG: 15 - Advanced RAG Patterns (CRAG, Self-RAG, Adaptive RAG)

### Theoretical Overview
- **Corrective RAG (CRAG)**: Evaluates retrieved document relevance score. If confidence is low, falls back to web search.
- **Self-RAG**: The model outputs reflection tokens (`[Retrieve]`, `[IsRel]`, `[IsSup]`) to evaluate its own generations.

```python
def crag_routing(retrieval_confidence: float):
    if retrieval_confidence > 0.8:
        return "CORRECT: Proceed to generation with internal context."
    elif retrieval_confidence > 0.4:
        return "AMBIGUOUS: Combine internal context with web search query."
    else:
        return "INCORRECT: Discard context, fallback to web search engine."

print(crag_routing(0.92))
print(crag_routing(0.25))
```
